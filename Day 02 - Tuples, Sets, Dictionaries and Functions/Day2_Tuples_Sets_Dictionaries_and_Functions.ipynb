# 🎫 Chapter 1: Tuples — Event Pass

### 🎬 Story
Fest ka **Event Pass** print ho raha hai: *Hackathon • Block-A • 10:30*. Aarav ne ye data ek **list** mein rakha tha.
Lekin kisi ne galti se `venue` badal diya — aur 200 passes pe galat venue chhap gaya! 😱

Bhavna Didi: *"Jo data **kabhi nahi badalna chahiye**, usse list mein mat rakho. **Tuple** use kar — ye list ka **lock lagaya hua** (🔒) bhai hai!"*

````python
# ❌ List: koi bhi galti se badal sakta hai
pass_list = ["Hackathon", "Block-A", 10, 30]
pass_list[1] = "Block-Z"          # chup-chaap venue badal gaya!
print("List  :", pass_list)

# ✅ Tuple: badalne ki koshish karoge toh Python rok dega
event_pass = ("Hackathon", "Block-A", 10, 30)
print("Tuple :", event_pass)

try:
    event_pass[1] = "Block-Z"
except TypeError as e:
    print("🔒 TypeError:", e)
````

### 1.1 Tuple Kya Hota Hai?
**Tuple = list jaisa hi ordered collection, lekin ek baar ban gaya toh badal nahi sakta (immutable).**
Ye **round brackets `( )`** mein likha jaata hai.

*Real-life examples*: Date of birth `(15, 8, 2006)` 🎂, GPS location `(27.49, 77.67)` 📍, RGB colour `(255, 0, 0)` 🎨, Din ke naam 📅

| Property | Matlab |
|---|---|
| **Ordered** | Jis order mein daala, usi order mein rahega |
| **Indexed** | Har item ka index hota hai (0 se shuru) |
| **Immutable** 🔒 | Ek baar ban gaya toh item badal / add / delete **nahi** ho sakta |
| **Duplicates allowed** | Same value kai baar aa sakti hai |
| **Mixed types** | Alag-alag type ki values ek saath |

### 1.2 Tuple Banane ke Tarike
| Tarika | Example |
|---|---|
| Round brackets | `(10, 20, 30)` |
| Bina brackets ke (packing) | `10, 20, 30` |
| `tuple()` constructor | `tuple([1, 2, 3])`, `tuple("GLA")`, `tuple(range(5))` |
| Khaali tuple | `()` |
| ⚠️ **Ek item ki tuple** | `("GLA",)` ← **comma zaroori hai!** |

````python
t1 = (10, 20, 30)                       # 1. round brackets
t2 = 10, 20, 30                         # 2. bina brackets bhi tuple hi hai
t3 = tuple([1, 2, 3])                   # 3. list -> tuple
t4 = tuple("GLA")                       # 4. string -> tuple of characters
t5 = tuple(range(1, 6))                 # 5. range -> tuple
t6 = ()                                 # 6. khaali tuple

for t in (t1, t2, t3, t4, t5, t6):
    print(t, "->", type(t).__name__)

print("\n--- ⚠️ Single item trap ---")
not_tuple = ("GLA")          # ye sirf bracket mein string hai!
real_tuple = ("GLA",)        # comma hi tuple banata hai
print("('GLA')  ->", type(not_tuple).__name__)
print("('GLA',) ->", type(real_tuple).__name__)
print("5,       ->", type((5,)).__name__)
````

### 1.3 Indexing & Slicing — List Jaisa Hi! ✂️
Index `0` se shuru, `-1` last item. Slicing `tuple[start:stop:step]` — `stop` **excluded**. Slicing se **nayi tuple** milti hai.

| Item | Hackathon | Block-A | 10 | 30 |
|---|---|---|---|---|
| **Index** | 0 | 1 | 2 | 3 |
| **Negative** | -4 | -3 | -2 | -1 |

````python
event_pass = ("Hackathon", "Block-A", 10, 30)

print("event_pass[0]   ->", event_pass[0])
print("event_pass[-1]  ->", event_pass[-1])
print("event_pass[1:3] ->", event_pass[1:3])     # index 1,2
print("event_pass[:2]  ->", event_pass[:2])
print("event_pass[::-1]->", event_pass[::-1])    # ulti tuple

days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
print("\nWeekdays:", days[:5])
print("Weekend :", days[5:])
print("Har 2nd :", days[::2])
````

### 1.4 Immutability 🔒 — Badal Nahi Sakte, Par Naya Bana Sakte Ho
Tuple mein **item assign / append / remove** nahi ho sakta. Badalna ho toh: **tuple → list → badlo → wapas tuple.**

#### 🪤 Ek Chhota Trap
Tuple khud immutable hai, par agar uske **andar ek list** hai toh us list ke andar ka data badal sakta hai (list mutable hai!).

````python
event_pass = ("Hackathon", "Block-A", 10, 30)

# ❌ Seedha badalna
try:
    event_pass[1] = "Block-B"
except TypeError as e:
    print("⚠️ TypeError:", e)

# ✅ Tarika: tuple -> list -> badlo -> tuple
temp = list(event_pass)
temp[1] = "Block-B"
event_pass_new = tuple(temp)
print("\nPurana pass:", event_pass)
print("Naya pass  :", event_pass_new)

# 🪤 Tuple ke andar list
team = ("Aarav", ["Python", "Java"])
team[1].append("C++")            # tuple nahi badli, andar wali list badli
print("\nTuple ke andar list badal gayi:", team)
try:
    team[1] = ["Go"]             # tuple ka slot badalna manaa hai
except TypeError as e:
    print("⚠️ Slot nahi badal sakte:", e)
````

### 1.5 Tuple Operators & Loop: `+`, `*`, `in`, `for`
| Operator | Kaam | Example | Result |
|---|---|---|---|
| `+` | Tuples jodna | `(1,2) + (3,)` | `(1, 2, 3)` |
| `*` | Repeat | `(0,) * 3` | `(0, 0, 0)` |
| `in` | Item hai? | `2 in (1,2,3)` | `True` |
| `for` | Ek-ek item par chalna | `for x in t:` | — |

````python
a = ("Aarav", "Bhavna")
b = ("Chirag",)

print("a + b      :", a + b)
print("b * 3      :", b * 3)
print("'Aarav' in a ->", "Aarav" in a)
print("'Zoya' in a  ->", "Zoya" in a)
print("a == ('Aarav', 'Bhavna') ->", a == ("Aarav", "Bhavna"))

print("\nLoop se:")
for name in a + b:
    print(" ->", name)
````

### 1.6 Packing & Unpacking 🎁
- **Packing** = kai values ko ek tuple mein baandhna: `pass_info = "Quiz", "Hall-2", 12`
- **Unpacking** = tuple ko **alag-alag variables** mein khol dena: `event, venue, hour = pass_info`
- Variables ki ginti = items ki ginti (warna `ValueError`)
- `*` (star) ke saath **bacha hua hissa** ek list mein aa jaata hai.
- **Swap trick:** `a, b = b, a` ✨

````python
# Packing
pass_info = "Quiz", "Hall-2", 12
print("Packed  :", pass_info)

# Unpacking
event, venue, hour = pass_info
print("Event:", event, "| Venue:", venue, "| Hour:", hour)

# Swap — bina temp variable ke!
a, b = 10, 20
a, b = b, a
print("\nSwap ke baad: a =", a, ", b =", b)

# Star unpacking
first, *middle, last = (1, 2, 3, 4, 5)
print("\nfirst :", first)
print("middle:", middle)          # middle hamesha LIST hoti hai
print("last  :", last)

winner, *others = ("Aarav", "Bhavna", "Chirag")
print("winner:", winner, "| others:", others)

# Galat count -> ValueError
try:
    x, y = (1, 2, 3)
except ValueError as e:
    print("\n⚠️ ValueError:", e)
````

### 1.7 Nested Tuples — Tuple ke Andar Tuple 🧩
Fest ka **schedule**: har event ka `(naam, ghanta)` ek chhoti tuple, aur poora schedule ek badi tuple.

````python
schedule = (
    ("Hackathon", 10),
    ("Quiz", 12),
    ("Robo-Race", 15),
)

print("Pehla event          :", schedule[0])
print("Doosre event ka naam :", schedule[1][0])
print("Last event ka ghanta :", schedule[-1][1])

print("\nPoora schedule:")
for event_name, hour in schedule:          # loop mein hi unpacking!
    print(f"  {hour}:00  ->  {event_name}")
````

### 1.8 Tuple Kyun Use Karein? 🤔 (List vs Tuple)
| | **List** `[ ]` | **Tuple** `( )` |
|---|---|---|
| Mutable? | ✅ Haan | ❌ **Nahi** |
| Safety (data lock) | Kam | ✅ Zyada |
| Memory / speed | Thodi zyada | ✅ Thodi kam / fast |
| Dictionary **key** ban sakti hai? | ❌ Nahi | ✅ **Haan** |
| Set ke andar ja sakti hai? | ❌ Nahi | ✅ **Haan** |
| Kab use karein | Data badalta rahe (cart, attendance) | Data fixed ho (date, coordinates, config) |

````python
import sys

as_list = [1, 2, 3, 4, 5]
as_tuple = (1, 2, 3, 4, 5)
print("List  ka size (bytes):", sys.getsizeof(as_list))
print("Tuple ka size (bytes):", sys.getsizeof(as_tuple))

# Tuple ko dictionary ka KEY bana sakte hain (list ko nahi)
seat_map = {(1, 1): "Aarav", (1, 2): "Bhavna", (2, 1): "Chirag"}
print("\nRow 1, Seat 2 par kaun?", seat_map[(1, 2)])

try:
    bad = {[1, 1]: "Aarav"}
except TypeError as e:
    print("⚠️ List key nahi ban sakti:", e)
````

# 📊 Chapter 2: 12 Tuple Tools — Judges ka Score Sheet

### 🎬 Story
Hackathon khatam! 3 judges ne scores de diye. Score sheet ek **tuple** mein hai (kyunki judging ke baad score **badalna nahi chahiye**!).

Ab Aarav ko nikalna hai: *kisko kitne mile? Sabse zyada? Average? Kitne judges ne 9 diya? Sabne pass kiya ya nahi?*

> 🛠️ Tuple immutable hai, isliye iske paas **sirf 2 methods** hain (`count`, `index`). Baaki **10 tools Python ke built-in functions** hain jo tuple par bhi chalte hain.

### 🗂️ 12 Tools, 4 Groups
| Group | Tools |
|---|---|
| 🔍 Tuple ke apne methods | `count()`, `index()` |
| 🧮 Stats functions | `len()`, `min()`, `max()`, `sum()` |
| 🔀 Arrange & Convert | `sorted()`, `tuple()`, `reversed()` |
| ✅ Check & Number | `any()`, `all()`, `enumerate()` |

## 🔍 Group 1: Tuple ke Apne Methods
### 1️⃣ `count(x)` — Kitni baar aaya?
### 2️⃣ `index(x)` — Pehla index kya hai? (nahi mila toh `ValueError`)

````python
judge_scores = (8, 9, 7, 9, 10, 9)

print("9 kitni baar aaya?   ->", judge_scores.count(9))
print("10 kitni baar aaya?  ->", judge_scores.count(10))
print("5 kitni baar aaya?   ->", judge_scores.count(5))      # nahi hai toh 0

print("\n7 kis index par hai? ->", judge_scores.index(7))
print("9 pehli baar kahan?  ->", judge_scores.index(9))      # duplicates mein pehla hi milta hai

try:
    judge_scores.index(5)
except ValueError as e:
    print("⚠️ ValueError:", e)
````

## 🧮 Group 2: `len()`, `min()`, `max()`, `sum()`
| Function | Kaam |
|---|---|
| `len(t)` | Kitne items hain |
| `min(t)` | Sabse chhota |
| `max(t)` | Sabse bada |
| `sum(t)` | Saare numbers ka jod |

> `min()` / `max()` strings par bhi chalte hain (alphabetical order). `sum()` sirf numbers par.

````python
judge_scores = (8, 9, 7, 9, 10, 9)

print("3. len(t) :", len(judge_scores))
print("4. min(t) :", min(judge_scores))
print("5. max(t) :", max(judge_scores))
print("6. sum(t) :", sum(judge_scores))

average = sum(judge_scores) / len(judge_scores)
print("\nAverage score:", round(average, 2))

names = ("Deepika", "Aarav", "Chirag")
print("min(names):", min(names), "| max(names):", max(names))   # alphabetical
````

## 🔀 Group 3: `sorted()`, `tuple()`, `reversed()`
### 7️⃣ `sorted(t)` — Sorted **LIST** deta hai (tuple nahi!)
Tuple immutable hai, isliye `sort()` method nahi hai. `sorted()` **nayi list** return karta hai. `reverse=True` se descending.

### 8️⃣ `tuple(x)` — Kisi bhi iterable ko tuple bana do
List, string, range, set... sab ko tuple mein badalta hai. Sorted list ko wapas tuple banana ho toh bhi yahi.

### 9️⃣ `reversed(t)` — Ulta chalne wala **iterator**
Seedha print karoge toh `<reversed object>` dikhega. `tuple(reversed(t))` se dekho. *(Shortcut: `t[::-1]` bhi wahi kaam karta hai.)*

````python
judge_scores = (8, 9, 7, 9, 10, 9)

# 7. sorted() -> list
asc = sorted(judge_scores)
desc = sorted(judge_scores, reverse=True)
print("sorted()               ->", asc, "| type:", type(asc).__name__)
print("sorted(reverse=True)   ->", desc)
print("Original (safe)        ->", judge_scores)

# 8. tuple() -> conversion
sorted_tuple = tuple(sorted(judge_scores))
print("\ntuple(sorted(...))     ->", sorted_tuple)
print("tuple('GLA')           ->", tuple("GLA"))
print("tuple(range(3))        ->", tuple(range(3)))
print("tuple([1, 2])          ->", tuple([1, 2]))

# 9. reversed() -> iterator
r = reversed(judge_scores)
print("\nreversed() seedha      ->", type(r).__name__)
print("tuple(reversed(...))   ->", tuple(reversed(judge_scores)))
print("t[::-1] (shortcut)     ->", judge_scores[::-1])

# Top 3 scores
print("\nTop 3 scores:", sorted(judge_scores, reverse=True)[:3])
````

## ✅ Group 4: `any()`, `all()`, `enumerate()`
### 🔟 `any(t)` — **Koi ek bhi** `True` hai? ➜ `True`
### 1️⃣1️⃣ `all(t)` — **Saare** `True` hain? ➜ `True`

> Python mein `0`, `""`, `None`, `False`, khaali container = **falsy**. Baaki sab **truthy**.
> Condition check karni ho toh generator-style likho: `all(m >= 40 for m in marks)` — "kya **har** m, 40 ya zyada hai?"

### 1️⃣2️⃣ `enumerate(t, start=0)` — Index + Value saath mein

````python
# ---- any() / all() ----
attendance = (True, True, False, True)
print("any(attendance) ->", any(attendance), "   (koi toh aaya?)")
print("all(attendance) ->", all(attendance), "  (sab aaye?)")
print("any(())         ->", any(()), "   (khaali par)")
print("all(())         ->", all(()), "    (khaali par — mazedaar!)")

marks = (78, 92, 45, 88)
print("\nall(m >= 40 for m in marks) ->", all(m >= 40 for m in marks), " (sab pass?)")
print("any(m > 90 for m in marks)  ->", any(m > 90 for m in marks), " (koi 90+ laaya?)")
print("any(m < 40 for m in marks)  ->", any(m < 40 for m in marks), " (koi fail?)")

print("\nTruthy/Falsy: any((0, '', None)) ->", any((0, "", None)), "| any((0, 'a')) ->", any((0, "a")))

# ---- enumerate() ----
names = ("Aarav", "Bhavna", "Chirag")
print("\nenumerate default (0 se):")
for i, name in enumerate(names):
    print(" ", i, "->", name)

print("\nenumerate(start=1) -> Roll No.:")
for roll, name in enumerate(names, start=1):
    print(f"  Roll No. {roll}: {name}")

print("\ntuple(enumerate(...)) ->", tuple(enumerate(names, start=1)))
````

### 3.1 Set Kya Hota Hai?
**Set = unique items ka unordered collection.** Ye **curly brackets `{ }`** mein likha jaata hai.

*Real-life examples*: Ek class ke unique students 🎓, Kisi event mein aaye unique visitors 🎟️, Ek sentence ke unique words 📝

| Property | Matlab |
|---|---|
| **Unique** ✨ | Duplicate apne aap hat jaate hain |
| **Unordered** 🎲 | Koi fix order nahi — print ka order alag-alag ho sakta hai |
| **No indexing** 🚫 | `s[0]` nahi chalta (kyunki position hoti hi nahi) |
| **Mutable** ✏️ | Set mein items add / remove kar sakte ho |
| **Items immutable hone chahiye** 🔒 | Number, string, tuple ✅ • list, dict, set ❌ |

### 3.2 Set Banane ke Tarike
| Tarika | Example |
|---|---|
| Curly brackets | `{1, 2, 3}` |
| `set()` se (list / string / tuple / range) | `set([1, 2, 2, 3])`, `set("hello")` |
| ⚠️ **Khaali set** | `set()` ← `{}` **nahi!** (`{}` ek khaali **dictionary** hai) |

````python
s1 = {10, 20, 30, 20, 10}              # duplicates apne aap hat gaye
s2 = set([1, 2, 2, 3, 3, 3])           # list se
s3 = set(range(1, 6))                  # range se
s4 = set("hello")                      # string se -> unique characters

print("s1:", s1)
print("s2:", s2)
print("s3:", s3)
print("s4 (letters):", sorted(s4))     # strings ka order har run mein badal sakta hai, isliye sorted()

print("\n--- ⚠️ Khaali set ka trap ---")
empty_wrong = {}
empty_right = set()
print("{}    ->", type(empty_wrong).__name__)
print("set() ->", type(empty_right).__name__, empty_right)
````

### 3.3 Order Fix Nahi + Indexing Nahi 🎲🚫
Set mein position hoti hi nahi, isliye:
- `s[0]` → `TypeError`
- Strings ka set print karoge toh **order har baar alag** aa sakta hai (numbers mostly same order mein dikhte hain, par guarantee nahi).
- **Ordered dekhna ho toh `sorted(s)`** (jo ek list deta hai).

````python
s = {30, 10, 20}

try:
    print(s[0])
except TypeError as e:
    print("⚠️ TypeError:", e)

print("sorted(s)      ->", sorted(s))
print("sorted(desc)   ->", sorted(s, reverse=True))
print("list(sorted(s))[0] (sabse chhota) ->", sorted(s)[0])

words = {"python", "java", "c", "go"}
print("\nStrings ka set (order alag ho sakta hai):", words)
print("Hamesha same order chahiye? ->", sorted(words))
````

### 3.4 Items Hashable (Immutable) Hone Chahiye 🔒
Set ke andar sirf **immutable** items ja sakte hain. **List** ya **dict** ko set mein daalne par `TypeError`. **Tuple** chalti hai!

````python
ok = {1, "GLA", 3.5, (1, 2), True}     # number, string, float, tuple -> sab theek (True == 1, isliye ek hi gina gaya)
print("Valid set ka size:", len(ok))

try:
    bad = {1, [2, 3]}
except TypeError as e:
    print("⚠️ TypeError:", e)

# Tuple ke jode set mein: seats (row, col)
booked = {(1, 1), (1, 2), (1, 1), (2, 5)}
print("\nUnique booked seats:", len(booked))
print("(1, 2) booked hai?  ->", (1, 2) in booked)
print("(3, 3) booked hai?  ->", (3, 3) in booked)
````

### 3.5 Loop, Membership `in` & Speed ⚡
- `for item in s:` se set par loop chal sakta hai (order fix nahi).
- `x in s` bahut **fast** hai (hashing ki wajah se) — bade data mein list se kaafi tez.

````python
import time

allowed = {101, 102, 103}
for roll in sorted(allowed):                # sorted() taaki output same rahe
    print("Allowed roll:", roll)

print("102 in allowed ->", 102 in allowed)
print("999 in allowed ->", 999 in allowed)

# ⚡ Speed test: 100000 items mein 100 baar search
big_list = list(range(100000))
big_set = set(big_list)

t0 = time.perf_counter()
for _ in range(100):
    99999 in big_list
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
for _ in range(100):
    99999 in big_set
t_set = time.perf_counter() - t0

print(f"\nList mein search: {t_list:.5f} sec")
print(f"Set  mein search: {t_set:.5f} sec")
print("Set tez nikla?   ->", t_set < t_list)
````

### 3.6 Set ki Sabse Badi Takat: Duplicates Hatana + Sorted List 🧹
`sorted(set(data))` — duplicates hatao **aur** order mein lagao, dono ek saath!

````python
marks = [78, 92, 45, 78, 92, 60, 45, 88]
print("Original      :", marks)
print("Unique sorted :", sorted(set(marks)))
print("Kitne unique  :", len(set(marks)))

names = ["Aarav", "Bhavna", "Aarav", "Chirag", "Bhavna"]
print("\nUnique names  :", sorted(set(names)))
print("Duplicate kitne?", len(names) - len(set(names)))
````

## ➕ Group 1: Add Elements
### 1️⃣ `add(x)` — Ek item daalo
**Syntax:** `s.add(item)` • Duplicate daaloge toh kuch nahi hoga (error bhi nahi).

### 2️⃣ `update(iterable)` — Kai items ek saath
**Syntax:** `s.update(list/tuple/set/string...)` • Ek se zyada iterables bhi de sakte ho.

> 🆚 `add()` → **ek** item • `update()` → **kai** items (iterable dena padta hai)

````python
s = {10, 20}
print("Shuru   :", s)

s.add(30)
print("add(30) :", s)

s.add(20)                       # duplicate -> koi asar nahi
print("add(20) :", s, "(duplicate ignore hua)")

s.update([40, 50])
print("update([40, 50])        :", s)

s.update({60}, (70, 80))        # ek saath kai iterables
print("update({60}, (70, 80))  :", s)

# ⚠️ add() mein iterable nahi chalta
try:
    s.add([1, 2])
except TypeError as e:
    print("\n⚠️ add([1,2]) -> TypeError:", e)

# ⚠️ update(string) -> har character alag item banta hai
t = set()
t.update("abc")
print("update('abc') ->", sorted(t))

# add() aur update() None return karte hain
print("Return value:", {1}.add(2))
````

## ➖ Group 2: Remove Elements
### 3️⃣ `remove(x)` — Item hatao (nahi mila toh **`KeyError`**)
### 4️⃣ `discard(x)` — Item hatao (nahi mila toh **chup-chaap**, koi error nahi)
### 5️⃣ `pop()` — **Koi bhi ek** item nikalo aur return karo (set mein index nahi, isliye kaun sa — pata nahi!)
### 6️⃣ `clear()` — Saare items hatao

> 🆚 `remove()` = strict (error deta hai) • `discard()` = relaxed (safe)

````python
s = {10, 20, 30, 40, 50}
print("Shuru:", s)

s.remove(10)
print("remove(10)  :", s)

try:
    s.remove(99)                # hai hi nahi
except KeyError as e:
    print("⚠️ remove(99) -> KeyError:", e)

s.discard(99)                   # koi error nahi!
print("discard(99) : error nahi aaya ->", s)

s.discard(20)
print("discard(20) :", s)

popped = s.pop()                # koi ek item (arbitrary)
print("\npop() ne nikala:", popped, "| bacha hua size:", len(s))
print("popped ab set mein hai?", popped in s)

s.clear()
print("clear() ke baad:", s, "| type:", type(s).__name__)

try:
    set().pop()                 # khaali set par pop
except KeyError as e:
    print("⚠️ Khaali set par pop() -> KeyError:", e)
````

## 🔗 Group 4: Set Algebra — Venn Diagram Wale Tools
Do sets A aur B ke beech ke rishte:

```
     │  101  102     │       │     106  107  │
     │        ┌──────┴───────┴──────┐        │
     │        │   103  104  105     │        │

```

| Tool | Operator | Matlab |
|---|:-:|---|
| **7️⃣ `union()`** | `A \| B` | **Dono** ke saare unique items |

| **🔟 `symmetric_difference()`** | `A ^ B` | **Sirf ek** mein hain (common nahi) |

> ⚠️ `difference` ka order matter karta hai: `A - B` ≠ `B - A`.

````python
hackathon = {101, 102, 103, 104, 105}
quiz      = {103, 104, 105, 106, 107}

print("hackathon:", hackathon)
print("quiz     :", quiz)
print("-" * 55)

# 7. union
print("union()                :", hackathon.union(quiz))
print("hackathon | quiz       :", hackathon | quiz)

# 8. intersection
print("\nintersection()         :", hackathon.intersection(quiz))
print("hackathon & quiz       :", hackathon & quiz)

# 9. difference (order matter karta hai!)
print("\nhackathon.difference(quiz) :", hackathon.difference(quiz), "  <- sirf Hackathon mein")
print("quiz.difference(hackathon) :", quiz.difference(hackathon), "  <- sirf Quiz mein")
print("hackathon - quiz           :", hackathon - quiz)

# 10. symmetric_difference
print("\nsymmetric_difference() :", hackathon.symmetric_difference(quiz))
print("hackathon ^ quiz       :", hackathon ^ quiz)

# Teen sets ek saath
robo = {105, 107, 108}
print("\nTeeno events ka union      :", hackathon.union(quiz, robo))
print("Teeno mein common (intersection):", hackathon.intersection(quiz, robo))
print("Kul unique students        :", len(hackathon | quiz | robo))
````

## 🔍 Group 5: Relationship Check — Subset, Superset, Disjoint
Ye teeno jawab `True` / `False` mein dete hain.

| Tool | Operator | Sawaal |
|---|:-:|---|
| **1️⃣1️⃣ `A.issubset(B)`** | `A <= B` | Kya A ke **saare items B mein** hain? |
| **1️⃣2️⃣ `A.issuperset(B)`** | `A >= B` | Kya A ke andar B ke **saare items** hain? |
| **1️⃣3️⃣ `A.isdisjoint(B)`** | — | Kya A aur B mein **koi bhi common nahi**? |

> 💡 Khaali set har set ka subset hota hai.

````python
all_students = {1, 2, 3, 4, 5, 6}
core_team    = {2, 4}
cse_only     = {1, 3}
it_only      = {5, 6}

print("core_team.issubset(all_students)   ->", core_team.issubset(all_students))
print("core_team <= all_students          ->", core_team <= all_students)

print("\nall_students.issuperset(core_team) ->", all_students.issuperset(core_team))
print("all_students >= core_team          ->", all_students >= core_team)

print("\ncse_only.isdisjoint(it_only)       ->", cse_only.isdisjoint(it_only), " (koi common nahi)")
print("cse_only.isdisjoint({1, 9})       ->", cse_only.isdisjoint({1, 9}), " (1 common hai)")

print("\nset().issubset(all_students)       ->", set().issubset(all_students), " (khaali set sabka subset)")
print("{2, 4}.issubset({2, 4})            ->", {2, 4}.issubset({2, 4}), " (khud ka subset bhi)")
````

### 5.1 Dictionary Kya Hoti Hai?
**Dictionary = `key : value` jodiyon ka collection.** Ye **curly brackets `{ }`** mein likhi jaati hai, har jodi `key: value` ki shakal mein, comma se alag.

```
student = { "name": "Aarav" , "branch": "CSE" , "marks": 94 }
             └key┘ └value┘     └key┘ └value┘
```

| Property | Matlab |
|---|---|
| **Key-Value pairs** | Value tak jaane ka rasta = **key** (index nahi) |
| **Keys unique** | Same key dobara aayi toh **aakhri value** jeet jaati hai |
| **Keys immutable** | String, number, tuple ✅ • list, set, dict ❌ |
| **Values kuch bhi** | Number, string, list, dict, kuch bhi |
| **Insertion order** | Python 3.7+ mein jis order mein daala, usi mein rehti hai |
| **Mutable** ✏️ | Add / change / delete kar sakte ho |

### 5.2 Dictionary Banane ke Tarike
| Tarika | Example |
|---|---|
| Curly brackets | `{"name": "Aarav", "marks": 94}` |
| Khaali dict | `{}` ya `dict()` |
| `dict()` keyword style | `dict(name="Aarav", marks=94)` *(keys bina quotes ke)* |
| Jodiyon ki list se | `dict([("a", 1), ("b", 2)])` |
| Do lists se (`zip`) | `dict(zip(keys_list, values_list))` |

````python
d1 = {"name": "Aarav", "branch": "CSE", "marks": 94}      # 1. curly brackets
d2 = {}                                                   # 2. khaali
d3 = dict(name="Bhavna", branch="IT", marks=91)           # 3. keyword style
d4 = dict([("a", 1), ("b", 2)])                           # 4. jodiyon ki list se
d5 = dict(zip(["x", "y", "z"], [10, 20, 30]))             # 5. do lists se

for d in (d1, d2, d3, d4, d5):
    print(d)

print("\n--- Duplicate key: aakhri value jeet jaati hai ---")
dup = {"a": 1, "b": 2, "a": 99}
print(dup, "| len =", len(dup))
````

### 5.3 Value Access, Add & Update ✏️
| Kaam | Syntax |
|---|---|
| Value padhna | `d[key]` (key nahi mili toh `KeyError`) |
| **Nayi** key add karna | `d[new_key] = value` |
| **Purani** key update karna | `d[old_key] = new_value` |
| Key hai ya nahi? | `key in d` (sirf **keys** check hoti hain) |
| Kitni jodiyan? | `len(d)` |

> 💡 Add aur update ka syntax **ek hi** hai — Python khud dekhta hai ki key pehle se hai ya nahi.

````python
student = {"name": "Aarav", "branch": "CSE", "marks": 94}

# Read
print("Name   :", student["name"])
print("Marks  :", student["marks"])

# Update (key pehle se hai)
student["marks"] = 97
print("\nUpdate ke baad marks:", student["marks"])

# Add (nayi key)
student["email"] = "aarav@gla.ac.in"
student["year"] = 1
print("Add ke baad:", student)

# Membership: sirf KEYS check hoti hain
print("\n'email' in student ->", "email" in student)
print("'phone' in student ->", "phone" in student)
print("'Aarav' in student ->", "Aarav" in student, "  <- 'Aarav' value hai, key nahi!")
print("len(student)       ->", len(student))

# Galat key
try:
    print(student["phone"])
except KeyError as e:
    print("⚠️ KeyError:", e)
````

### 5.4 Dictionary par Loop 🔁
`for key in d:` → sirf **keys** milti hain. Values ya dono ke liye methods (`values()`, `items()`) agle chapter mein — abhi bas `d[key]` se kaam chalate hain.

````python
result = {"Aarav": 94, "Bhavna": 91, "Chirag": 88}

for name in result:                    # default: keys par loop
    print(name, "->", result[name])

total = 0
for name in result:
    total += result[name]
print("\nTotal marks:", total)
````

## 🔍 Group 1: Read — `get()`, `keys()`, `values()`, `items()`
### 1️⃣ `get(key, default=None)` — Safe lookup 🛡️
`d[key]` → key nahi mili toh `KeyError` ❌ • `d.get(key)` → key nahi mili toh `None` ✅ • `d.get(key, default)` → apni pasand ki default value.

### 2️⃣ `keys()` • 3️⃣ `values()` • 4️⃣ `items()`
| Method | Kya deta hai | Loop mein |
|---|---|---|
| `d.keys()` | Saari keys | `for k in d.keys()` |
| `d.values()` | Saari values | `for v in d.values()` |
| `d.items()` | `(key, value)` jodiyan | `for k, v in d.items()` ⭐ |

> Ye **live view** objects hain — dictionary badli toh view bhi badal jaata hai. List chahiye toh `list(d.keys())`.

````python
result = {"Aarav": 94, "Bhavna": 91, "Chirag": 88}

# 1. get()
print("get('Aarav')          ->", result.get("Aarav"))
print("get('Zoya')           ->", result.get("Zoya"))                 # None
print("get('Zoya', 'Absent') ->", result.get("Zoya", "Absent"))       # default

try:
    result["Zoya"]
except KeyError as e:
    print("result['Zoya']       -> ⚠️ KeyError:", e)

# 2, 3, 4. keys(), values(), items()
print("\nkeys()   :", result.keys())
print("values() :", result.values())
print("items()  :", result.items())
print("list(keys()):", list(result.keys()))

print("\n--- items() ke saath loop (sabse zyada use hota hai) ---")
for name, marks in result.items():
    print(f"{name:<8} -> {marks}")

print("\nsum(values) :", sum(result.values()))
print("max(values) :", max(result.values()))

# Live view
view = result.keys()
result["Deepika"] = 79
print("\nNaya student add karne ke baad view bhi badal gaya:", list(view))
````

## ✏️ Group 2: Add / Update — `update()`, `setdefault()`, `fromkeys()`
### 5️⃣ `update(other)` — Doosri dictionary ka data **merge** karo
Nayi keys **add** hoti hain, purani keys ki values **overwrite** ho jaati hain. `update(name=value)` keyword style bhi chalta hai.

### 6️⃣ `setdefault(key, default)` — Key ho toh uski value do, **nahi ho toh default ke saath add karke do**
Pehle se hai → value **badalti nahi**.

### 7️⃣ `dict.fromkeys(keys, value)` — Ek hi value ke saath nayi dictionary
Saari keys ko **ek jaisi starting value** deni ho (jaise attendance "Absent"). *(Ye `dict` class ka method hai.)*

> ⚠️ `fromkeys` mein mutable value (jaise `[]`) mat do — sabhi keys **ek hi list** share karengi!

````python
# 5. update()
marks = {"Aarav": 94, "Bhavna": 91}
new_data = {"Bhavna": 95, "Chirag": 88}          # Bhavna ka marks badla, Chirag naya
marks.update(new_data)
print("update(dict)       :", marks)
marks.update(Deepika=79)                          # keyword style
print("update(Deepika=79) :", marks)

# 6. setdefault()
profile = {"name": "Aarav"}
print("\nsetdefault('year', 1)   ->", profile.setdefault("year", 1), "| profile:", profile)
print("setdefault('name', 'X') ->", profile.setdefault("name", "X"), "| profile:", profile, "(naam nahi badla)")

# 7. fromkeys()
names = ["Aarav", "Bhavna", "Chirag"]
attendance = dict.fromkeys(names, "Absent")
print("\nfromkeys               :", attendance)
print("fromkeys(no value)     :", dict.fromkeys(["a", "b"]))        # default value None

# ⚠️ Mutable value trap
trap = dict.fromkeys(["a", "b"], [])
trap["a"].append(1)
print("\n⚠️ Mutable trap:", trap, "<- dono mein 1 aa gaya (same list share hui)")
````

## ➖ Group 3: Remove — `pop()`, `popitem()`, `del`, `clear()`
| Tool | Kaam |
|---|---|
| **8️⃣ `pop(key)`** | Key hatao aur uski **value return** karo • key na ho toh `KeyError` (`pop(key, default)` se safe) |
| **9️⃣ `popitem()`** | **Aakhri jodi** `(key, value)` nikalo (LIFO) • khaali dict par `KeyError` |
| **🔟 `del d[key]`** | Key hatao (kuch return nahi) • key na ho toh `KeyError` • `del d` poori dict hata deta hai |
| **1️⃣1️⃣ `clear()`** | Saari jodiyan hatao, dictionary khaali `{}` |

````python
d = {"a": 1, "b": 2, "c": 3, "d": 4, "e": 5}
print("Shuru:", d)

# 8. pop()
val = d.pop("a")
print("\npop('a') ne diya:", val, "| dict:", d)
print("pop('zzz', 'nahi mili') ->", d.pop("zzz", "nahi mili"))      # safe pop

try:
    d.pop("zzz")
except KeyError as e:
    print("⚠️ pop('zzz') -> KeyError:", e)

# 9. popitem()
pair = d.popitem()
print("\npopitem() ne diya:", pair, "| dict:", d)

# 10. del
del d["b"]
print("del d['b'] ke baad:", d)

try:
    del d["b"]                         # ab hai hi nahi
except KeyError as e:
    print("⚠️ del dobara -> KeyError:", e)

# 11. clear()
d.clear()
print("\nclear() ke baad:", d)

temp = {"x": 1}
del temp
try:
    print(temp)
except NameError as e:
    print("⚠️ 'del temp' ke baad poori dict gayab -> NameError:", e)
````

### 1️⃣3️⃣ `len(d)` — Kitni jodiyan? • 1️⃣4️⃣ `key in d` — Key hai? (values nahi dekhta)

### 🎯 Real Use: Word Counter (Frequency Dictionary)
Dictionary ka sabse famous use! `get()` ya `setdefault()` se har word ki ginti rakho.

````python
sentence = "python is fun and python is easy"
words = sentence.split()

# Tarika 1: get() se
freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1          # key nahi hai toh 0 se shuru
print("get() se        :", freq)

# Tarika 2: setdefault() se
freq2 = {}
for w in words:
    freq2.setdefault(w, 0)
    freq2[w] += 1
print("setdefault() se :", freq2)
print("Dono same?      :", freq == freq2)

# Sabse zyada aane wala word
top_word = max(freq, key=freq.get)
print("\nTop word:", top_word, "->", freq[top_word], "baar")

# Duplicate hatao, order bacha ke (fromkeys ka mazedaar use)
data = [3, 1, 3, 2, 1, 4]
print("\nOrder ke saath unique:", list(dict.fromkeys(data)))
````

### 7.1 Dictionary Comprehension — Syntax
```
{  key_expr : value_expr   for  item  in  iterable  }
```
List comprehension jaisa hi hai, bas **`[ ]` ki jagah `{ }`** aur **`key: value`** ki jodi.

| Type | Syntax |
|---|---|
| Basic | `{k: v for ... in ...}` |
| Filter | `{k: v for ... in ... if condition}` |
| if-else (value ke liye) | `{k: ("A" if cond else "B") for ...}` |

````python
# 1. Basic: number -> uska square
squares = {n: n ** 2 for n in range(1, 6)}
print("Squares  :", squares)

# 2. Do lists se (zip ke saath)
names = ["Aarav", "Bhavna", "Chirag", "Deepika"]
marks = [94, 91, 38, 79]
result = {name: m for name, m in zip(names, marks)}
print("Result   :", result)

# 3. Filter: sirf pass students
passed = {name: m for name, m in result.items() if m >= 40}
print("Passed   :", passed)

# 4. Value transform: +5 grace (100 se zyada nahi)
graced = {name: min(m + 5, 100) for name, m in result.items()}
print("Graced   :", graced)

# 5. if-else value: Pass / Fail
status = {name: ("Pass" if m >= 40 else "Fail") for name, m in result.items()}
print("Status   :", status)

# 6. Key transform
upper = {name.upper(): m for name, m in result.items()}
print("Upper key:", upper)

# 7. Ulta (invert): key <-> value
inverted = {m: name for name, m in result.items()}
print("Inverted :", inverted)

# 8. String se: har letter kitni baar aaya
word = "banana"
count = {ch: word.count(ch) for ch in word}
print("\nLetter count in 'banana':", count)
````

### 7.2 Dictionary ko Sort Karna 🔀
Dictionary ke paas apna `sort()` nahi hota. Hum **`sorted()`** se kaam chalate hain, aur result ko wapas `dict(...)` mein daalte hain.

| Kaam | Code |
|---|---|
| **Keys** sorted (list) | `sorted(d)` |
| Dict ko **key** ke hisaab se sort | `dict(sorted(d.items()))` |
| Dict ko **value** ke hisaab se sort | `dict(sorted(d.items(), key=lambda kv: kv[1]))` |
| Value ke hisaab se **descending** | `... , reverse=True` |

#### 🤔 Ye `lambda kv: kv[1]` kya hai?
`d.items()` mein har item ek `(key, value)` jodi hai, isliye `kv[0]` = key aur `kv[1]` = value. **`key=`** parameter bolta hai: *"sort karte waqt har item ka **ye hissa** dekho."*
`lambda` ek chhota **one-line function** hai — iska poora detail **afternoon (Chapter 11)** mein aayega. Abhi bas pattern yaad rakho! 🧠

````python
marks = {"Chirag": 88, "Aarav": 94, "Deepika": 79, "Bhavna": 91}
print("Original:", marks)

# --- Key ke hisaab se ---
print("\nsorted(marks)                 :", sorted(marks))                # sirf keys (list)
print("dict(sorted(marks.items()))   :", dict(sorted(marks.items())))   # poori dict, A-Z
print("Z-A order                     :", dict(sorted(marks.items(), reverse=True)))

# --- Value ke hisaab se ---
asc  = dict(sorted(marks.items(), key=lambda kv: kv[1]))
desc = dict(sorted(marks.items(), key=lambda kv: kv[1], reverse=True))
print("\nValue ascending (kam -> zyada) :", asc)
print("Value descending (zyada -> kam):", desc)

# Bina lambda ka shortcut: keys ko values ke hisaab se sort
print("\nsorted(marks, key=marks.get, reverse=True) :", sorted(marks, key=marks.get, reverse=True))

# Top 3 students
top3 = dict(sorted(marks.items(), key=lambda kv: kv[1], reverse=True)[:3])
print("Top 3 :", top3)

print("\nOriginal abhi bhi:", marks)      # sorted() naya banata hai, original safe
````

````python
# 🏆 Leaderboard print karna
marks = {"Chirag": 88, "Aarav": 94, "Deepika": 79, "Bhavna": 91}

ranked = sorted(marks.items(), key=lambda kv: kv[1], reverse=True)

print("🏆 UTKARSH 2026 — LEADERBOARD")
for rank, (name, m) in enumerate(ranked, start=1):
    print(f"  Rank {rank}: {name:<8} {m}")

# Tie-break: marks barabar ho toh naam A-Z
ties = {"Zoya": 90, "Aarav": 90, "Bhavna": 95}
ordered = sorted(ties.items(), key=lambda kv: (-kv[1], kv[0]))   # pehle marks (bade se chhote), phir naam
print("\nTie-break ke saath:", ordered)
````

### 8.2 `return` — Result Wapas Dena 🎁
- `print()` sirf **screen par dikhata** hai — result kahin store nahi hota.
- `return` result ko function ke **bahar bhejta** hai — aap use variable mein rakh sakte ho, aage calculation mein use kar sakte ho.
- `return` chalte hi function **wahin khatam** ho jaata hai (uske neeche ki lines nahi chalti).
- Agar `return` nahi likha toh function automatically **`None`** return karta hai.

````python
def add_print(a, b):
    print("print ne dikhaya:", a + b)         # sirf dikhata hai

def add_return(a, b):
    return a + b                              # wapas bhejta hai

x = add_print(3, 4)
y = add_return(3, 4)

print("\nx (print wale ka result) :", x)       # None!
print("y (return wale ka result):", y)         # 7
print("y ko aage use kar sakte hain: y * 10 =", y * 10)

# return ke baad ki lines nahi chalti
def check(n):
    return "Positive" if n > 0 else "Not positive"
    print("Ye line kabhi nahi chalegi")

print("\ncheck(5)  ->", check(5))
print("check(-2) ->", check(-2))

# Early return — jaldi nikal jao
def safe_divide(a, b):
    if b == 0:
        return "Zero se divide nahi ho sakta!"
    return a / b

print("\nsafe_divide(10, 2) ->", safe_divide(10, 2))
print("safe_divide(10, 0) ->", safe_divide(10, 0))
````

### 8.3 Multiple Values Return Karna 📦
Python mein ek saath **kai values** return kar sakte ho — asal mein wo ek **tuple** hoti hai (Chapter 1 ka packing!). Receive karte waqt **unpacking** kar lo.

````python
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)    # 3 values ek saath

marks = [78, 92, 45, 88, 60]

result = stats(marks)
print("Ek variable mein:", result, "| type:", type(result).__name__)

low, high, avg = stats(marks)                 # unpacking
print(f"Lowest = {low}, Highest = {high}, Average = {avg}")

# Divmod jaisa: quotient aur remainder
def div_rem(a, b):
    return a // b, a % b

q, r = div_rem(17, 5)
print(f"\n17 / 5 -> Quotient = {q}, Remainder = {r}")
````

### 8.4 Docstring — Function ka Parichay 📝
Function ki **pehli line** mein triple quotes `""" ... """` ke andar likha description = **docstring**.
- Dusre programmers (aur future-aap!) ko batata hai function kya karta hai.
- `function.__doc__` se padh sakte ho, ya `help(function)` se.

````python
def rectangle_area(length, breadth):
    """Rectangle ka area return karta hai.

    Parameters:
        length  : rectangle ki lambai
        breadth : rectangle ki chaudai
    Returns:
        length * breadth
    """
    return length * breadth

print("Area:", rectangle_area(5, 3))
print("\n--- __doc__ ---")
print(rectangle_area.__doc__)

print("--- help() ---")
help(rectangle_area)

print("--- Built-in functions ke docstring bhi hote hain ---")
print(len.__doc__)
````

### 8.5 Functions ke Kuch Aur Rang 🎨
- Function **list/string/koi bhi type** return kar sakta hai.
- Ek function **doosre function ko call** kar sakta hai.
- Function ka naam bhi ek variable hi hai — usse kisi aur naam ko de sakte ho.

````python
def grade(marks):
    """Marks se grade batata hai."""
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 40:
        return "B"
    return "Fail"

def report(name, marks):
    """grade() ko call karke report line banata hai."""
    return f"{name} ne {marks} marks laaye -> Grade {grade(marks)}"

for n, m in [("Aarav", 94), ("Bhavna", 78), ("Chirag", 55), ("Deepika", 30)]:
    print(report(n, m))

# Function ek list return karta hai
def evens_upto(n):
    return [i for i in range(1, n + 1) if i % 2 == 0]
print("\nevens_upto(10) ->", evens_upto(10))

# Function ko doosre naam se bulana
my_grade = grade
print("my_grade(80)  ->", my_grade(80))
````

# 🎛️ Chapter 9: Function Arguments — Machine ke Controls

### 🎬 Story
Certificate machine ko ab aur smart banana hai. Kabhi **naam, event, branch** sab dena hai, kabhi **sirf naam** (event default "Participation" hoga), kabhi **10 students** ke marks ek saath total karne hain, kabhi **koi bhi extra detail** bhejni hai...

Bhavna Didi: *"Machine ke alag-alag **controls** (argument types) hote hain. Chalo, saare seekhte hain!"* 🎛️

### 🗂️ 5 Types of Arguments
| # | Type | Example |
|:-:|---|---|
| 1 | **Positional** | `f(10, 20)` — order se milte hain |
| 2 | **Keyword** | `f(b=20, a=10)` — naam se |
| 3 | **Default** | `def f(a, b=5)` — na do toh default |
| 4 | **`*args`** | `def f(*args)` — kitne bhi positional |
| 5 | **`**kwargs`** | `def f(**kwargs)` — kitne bhi keyword |

### 9.1 Positional Arguments — Order Matters! 🔢
Pehla argument pehle parameter ko, doosra doosre ko... **order galat toh answer galat.** Kam ya zyada arguments diye toh `TypeError`.

````python
def introduce(name, branch, year):
    print(f"{name} is in {branch}, year {year}")

introduce("Aarav", "CSE", 1)          # sahi order
introduce("CSE", "Aarav", 1)          # order ulta -> galat matlab, par error nahi!

try:
    introduce("Aarav", "CSE")         # ek argument kam
except TypeError as e:
    print("⚠️ TypeError:", e)

try:
    introduce("Aarav", "CSE", 1, 99)  # ek zyada
except TypeError as e:
    print("⚠️ TypeError:", e)
````

### 9.2 Keyword Arguments — Naam Se Do, Order Free 🏷️
Call karte waqt `parameter_name=value` likho. Order koi matter nahi karta, aur code zyada **readable** hota hai.

> ⚠️ Rule: Positional arguments **pehle**, keyword arguments **baad mein**.

````python
def introduce(name, branch, year):
    print(f"{name} is in {branch}, year {year}")

introduce(name="Aarav", branch="CSE", year=1)
introduce(year=2, name="Bhavna", branch="IT")        # order alag, phir bhi sahi
introduce("Chirag", year=3, branch="ECE")            # positional pehle, keyword baad mein

# ❌ Positional ko keyword ke baad nahi rakh sakte (SyntaxError hota hai), isliye ye comment mein hai:
# introduce(name="Aarav", "CSE", 1)
````

### 9.3 Default Arguments — Na Do Toh Chalega 🎁
`def f(a, b=5)` — agar `b` nahi diya toh `5` use hoga.

> ⚠️ **Rule:** Default wale parameters **hamesha non-default ke baad** aate hain: `def f(a, b=5)` ✅ • `def f(a=5, b)` ❌

#### 🪤 Mutable Default Trap
Default value mein **list / dict (`[]`, `{}`) kabhi mat rakho** — wo function ke har call mein **share** hoti hai!
✅ Sahi tarika: default `None` rakho aur function ke andar nayi list banao.

````python
def register(name, event="Hackathon", fee=100):
    print(f"{name} registered for {event}, fee = ₹{fee}")

register("Aarav")                          # dono default
register("Bhavna", "Quiz")                 # fee default
register("Chirag", fee=250)                # event default, fee keyword se
register("Deepika", "Robo-Race", 300)      # sab diya

# 🪤 Mutable default trap
def add_bad(item, bucket=[]):
    bucket.append(item)
    return bucket

print("\n❌ Bad function:")
print(add_bad("pen"))
print(add_bad("book"))          # 'pen' bhi saath aa gaya! (bucket share hui)

# ✅ Sahi tarika
def add_good(item, bucket=None):
    if bucket is None:
        bucket = []             # har call mein NAYI list
    bucket.append(item)
    return bucket

print("\n✅ Good function:")
print(add_good("pen"))
print(add_good("book"))
````

### 9.4 `*args` — Kitne Bhi Positional Arguments 🎒
Function ko **pata nahi** ki kitne values aayengi? `*args` use karo. Saari extra values ek **tuple** mein aa jaati hain.
(`args` sirf naam hai — asli jaadu `*` hai. Convention: `*args`.)

````python
def total_marks(*args):
    print("   args ka type:", type(args).__name__, "| values:", args)
    return sum(args)

print("3 values :", total_marks(10, 20, 30))
print("5 values :", total_marks(5, 10, 15, 20, 25))
print("0 values :", total_marks())

# *args ke saath normal parameter bhi
def show_team(leader, *members):
    print(f"Leader: {leader} | Members: {members} | Count: {len(members)}")

show_team("Aarav", "Bhavna", "Chirag", "Deepika")
show_team("Esha")

# average function
def average(*nums):
    return sum(nums) / len(nums)
print("\nAverage:", average(80, 90, 100))
````

### 9.5 `**kwargs` — Kitne Bhi Keyword Arguments 🧳
`name=value` wali **kitni bhi extra values** `**kwargs` mein ek **dictionary** ban kar aati hain.

````python
def show_profile(**kwargs):
    print("   kwargs ka type:", type(kwargs).__name__, "| value:", kwargs)
    for key, value in kwargs.items():
        print(f"   {key} -> {value}")

show_profile(name="Aarav", branch="CSE", marks=94)
print()
show_profile(city="Mathura")

# Teeno ek saath: order = normal, *args, **kwargs
def full(a, b=2, *args, **kwargs):
    print(f"a={a}, b={b}, args={args}, kwargs={kwargs}")

full(1)
full(1, 10)
full(1, 10, 20, 30, x=5, y=6)
````

### 9.6 List / Dictionary ko Function mein Bhejna 📨
**Mutable objects (list, dict)** function ko **reference** ke roop mein milte hain — function ke andar unhe badla toh **original bhi badal jaata hai!**
Immutable (number, string, tuple) ke saath aisa nahi hota.

#### 🔓 Unpacking at Call: `*list` aur `**dict`
Ek list ke items ko **alag-alag arguments** bana ke bhejna ho → `f(*my_list)`. Dictionary ko keyword arguments bana ke bhejna ho → `f(**my_dict)`.

````python
# List function ke andar badli -> original bhi badli
def give_grace(marks, grace=5):
    for i in range(len(marks)):
        marks[i] += grace

scores = [78, 92, 45]
give_grace(scores)
print("Original list badal gayi:", scores)

# Original bachana ho toh copy bhejo ya nayi list return karo
def graced_copy(marks, grace=5):
    return [m + grace for m in marks]

scores = [78, 92, 45]
new_scores = graced_copy(scores)
print("Naya:", new_scores, "| Original safe:", scores)

# Dictionary bhi same
def add_email(student):
    student["email"] = student["name"].lower() + "@gla.ac.in"

s = {"name": "Aarav"}
add_email(s)
print("\nDict badal gayi:", s)

# Immutable: number change nahi hota
def try_change(n):
    n = n + 100
    print("   andar n =", n)

x = 5
try_change(x)
print("bahar x =", x, "(waisa hi)")

# 🔓 Unpacking at call
def volume(l, b, h):
    return l * b * h

dims = [2, 3, 4]
print("\nvolume(*dims)  ->", volume(*dims))              # volume(2, 3, 4)

dims_dict = {"l": 2, "b": 3, "h": 5}
print("volume(**dims_dict) ->", volume(**dims_dict))     # volume(l=2, b=3, h=5)
````

### 📌 Arguments Cheat-Sheet
| Type | Define | Call | Andar milta hai |
|---|---|---|---|
| Positional | `def f(a, b)` | `f(1, 2)` | normal variables |
| Keyword | `def f(a, b)` | `f(b=2, a=1)` | normal variables |
| Default | `def f(a, b=5)` | `f(1)` | `b = 5` |
| `*args` | `def f(*args)` | `f(1, 2, 3)` | **tuple** `(1, 2, 3)` |
| `**kwargs` | `def f(**kwargs)` | `f(x=1, y=2)` | **dict** `{'x':1, 'y':2}` |
| Unpack list | — | `f(*lst)` | list ke items alag-alag args |
| Unpack dict | — | `f(**d)` | dict ki jodiyan keyword args |

**Parameter ka order:** `normal → default → *args → **kwargs`

# 🔭 Chapter 10: Scope — Kaun Kya Dekh Sakta Hai?

### 🎬 Story
Fest ke office mein har kamre ke apne rules hain:
- **Registration room** ke andar ka register wahin ka hai — bahar wale use nahi dekh sakte. 🔒
- **Notice board** par jo likha hai wo **sabko** dikhta hai. 📢
- Ek **bade hall ke andar chhota cabin** ho toh cabin wale ko hall ka bhi sab dikhta hai, par hall wale ko cabin ka nahi. 🪆

Variables ke saath bhi bilkul yahi hota hai — isse **Scope** kehte hain.

> **Scope = variable ki "dikhne ki seema"** — program ke kis hisse mein wo variable use ho sakta hai.

### 10.1 Local vs Global 🏠🌍
| | **Local** | **Global** |
|---|---|---|
| Kahan banta hai | **Function ke andar** | Function ke **bahar** (poore program level par) |
| Kahan dikhta hai | Sirf us function ke andar | Poore program mein (functions ke andar bhi **padh** sakte ho) |
| Kab tak jeeta hai | Function ke chalne tak | Program ke end tak |

> ⚠️ Function ke andar kisi variable ko **assign** (`x = ...`) karoge toh Python use **local** bana dega — global ko nahi badlega!

````python
festival = "Utkarsh"              # GLOBAL

def show():
    local_var = "Main sirf andar hoon"      # LOCAL
    print("Andar se global dikha:", festival)
    print("Andar se local dikha :", local_var)

show()

# Local bahar nahi dikhta
try:
    print(local_var)
except NameError as e:
    print("\n⚠️ NameError:", e)

# Same naam, alag variable!
x = "global x"
def demo():
    x = "local x"                  # ye NAYA local x hai, global wala nahi badla
    print("Andar x:", x)

demo()
print("Bahar x:", x)
````

### 10.2 `global` Keyword — Function ke Andar Se Global Badlana 🌍✏️
Function ke andar global variable ko **badalna** ho toh pehle `global variable_name` likhna padta hai.

#### 🪤 `UnboundLocalError`
Agar function mein kahin bhi `x = ...` ya `x += ...` likha hai, toh Python **poore function mein** `x` ko local maanta hai — isliye assign hone se **pehle** padhne par error aata hai.

````python
visitors = 0                       # global counter

# ❌ global ke bina badalne ki koshish
def add_visitor_wrong():
    visitors += 1                  # Python ise local maanta hai, par value pehle se nahi hai

try:
    add_visitor_wrong()
except UnboundLocalError as e:
    print("⚠️ UnboundLocalError:", e)

# ✅ global keyword ke saath
def add_visitor():
    global visitors
    visitors += 1

add_visitor()
add_visitor()
add_visitor()
print("\nTotal visitors:", visitors)

# Sirf padhna ho toh global likhne ki zaroorat nahi
def show_visitors():
    print("Visitors abhi:", visitors)
show_visitors()
````

### 10.3 Nested Functions & Enclosing Scope 🪆
Function ke **andar function** bana sakte hain. Andar wala function (inner) **bahar wale function (outer)** ke variables ko **padh** sakta hai — isse **Enclosing scope** kehte hain.

````python
def outer():
    rate = 18                              # outer ka variable (enclosing)

    def add_gst(amount):                   # inner function
        return amount + amount * rate / 100     # rate outer se mila

    return add_gst(1000)

print("1000 + 18% GST =", outer())

# Inner function bahar se call nahi ho sakta
def outer2():
    def secret():
        return "Main outer2 ke andar hoon"
    return secret()

print(outer2())
try:
    secret()
except NameError as e:
    print("⚠️ NameError:", e)
````

### 10.4 `nonlocal` Keyword — Enclosing Variable Badalna 🪆✏️
Inner function ko **outer function ka variable badalna** ho toh `nonlocal` likhte hain. (`global` = bahar ka sabse upar wala, `nonlocal` = ek level upar wala function ka.)

Isse **closure** banta hai — function apne bahar ke variable ko **yaad rakhta** hai, outer function khatam hone ke baad bhi! 🧠

````python
def make_counter():
    count = 0                       # enclosing variable

    def counter():
        nonlocal count              # outer ka count badlenge
        count += 1
        return count

    return counter                  # function return kiya (call nahi kiya)

c1 = make_counter()
c2 = make_counter()                 # alag counter, alag count

print("c1:", c1(), c1(), c1())
print("c2:", c2(), c2())
print("c1 again:", c1())

# nonlocal ke bina?
def outer():
    x = "outer"
    def inner():
        x = "inner"                 # naya LOCAL x banta hai, outer ka nahi badalta
    inner()
    return x

print("\nnonlocal ke bina outer ka x:", outer())

def outer_nl():
    x = "outer"
    def inner():
        nonlocal x
        x = "inner"
    inner()
    return x

print("nonlocal ke saath outer ka x:", outer_nl())

# Closure: multiplier factory
def make_multiplier(n):
    def multiply(x):
        return x * n
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print("\ndouble(10) =", double(10), "| triple(10) =", triple(10))
````

### 10.5 LEGB Rule — Python Variable Kahan Dhoondhta Hai? 🔎
Jab aap variable ka naam use karte ho, Python **4 jagah, is order mein** dhoondhta hai:

| Letter | Scope | Matlab |
|:-:|---|---|
| **L** | **Local** | Abhi wale function ke andar |
| **E** | **Enclosing** | Bahar wale (outer) function(s) mein |
| **G** | **Global** | File ke top level par |
| **B** | **Built-in** | Python ke ready-made naam (`print`, `len`, `sum`...) |

**Jahan pehle mil gaya, wahin ruk jaata hai.** Kahin nahi mila → `NameError`.

```
┌──────────── Built-in (B) ────────────┐
│  ┌──────── Global (G) ────────────┐  │
│  │  ┌──── Enclosing (E) ───────┐  │  │
│  │  │  ┌──── Local (L) ─────┐  │  │  │
│  │  │  │   yahan se shuru   │  │  │  │
│  │  │  └────────────────────┘  │  │  │
│  │  └──────────────────────────┘  │  │
│  └────────────────────────────────┘  │
└──────────────────────────────────────┘
```

````python
level = "GLOBAL"

def outer():
    level = "ENCLOSING"

    def inner():
        level = "LOCAL"
        print("inner  dekhta hai:", level)        # L mein mil gaya

    def inner_no_local():
        print("inner2 dekhta hai:", level)        # L mein nahi, E mein mila

    inner()
    inner_no_local()
    print("outer  dekhta hai:", level)             # outer ka apna

outer()
print("bahar  dekhta hai:", level)                # G

# Global tak pahunchna
g = "GLOBAL_ONLY"
def outer_b():
    def inner_b():
        print("\nL, E mein nahi mila -> G se mila:", g)
    inner_b()
outer_b()

# Built-in tak: len ko hum ne define nahi kiya
print("len built-in se mila:", len("GLA"))

# ⚠️ Built-in ka naam mat churao (shadowing)
sum = 100                                   # ab 'sum' global variable hai, built-in chhup gaya!
try:
    print(sum([1, 2, 3]))
except TypeError as e:
    print("\n⚠️ TypeError:", e)
del sum                                     # apna 'sum' hata diya -> built-in wapas mil gaya
print("del ke baad sum([1, 2, 3]) =", sum([1, 2, 3]))
````

### 📌 Scope Cheat-Sheet
| Cheez | Rule |
|---|---|
| Function ke andar bana variable | **Local** — bahar nahi dikhta |
| Global ko function mein **padhna** | Seedha chalta hai |
| Global ko function mein **badalna** | `global x` likho |
| Outer function ka variable inner mein **padhna** | Seedha chalta hai |
| Outer function ka variable inner mein **badalna** | `nonlocal x` likho |
| Variable dhoondhne ka order | **L → E → G → B** |
| Built-in naam (`sum`, `list`, `max`...) | Variable naam ke roop mein mat use karo |

> 💡 **Best practice:** `global` ka use kam se kam karo — function ko jo chahiye wo **parameter** se do, aur result **return** se lo.

# ⚡ Chapter 11: Lambda Functions — One-line Functions

### 🎬 Story
Aarav ko sirf **ek chhota kaam** karwana hai: *"marks mein +5 jodo."* Uske liye poora `def`, `return`, 3 lines... 😩

Bhavna Didi muskurayi: *"Chhote kaam ke liye **lambda** hai — **bina naam ka, ek line ka function**!"* ⚡

```python
def add_five(m):            # 3 lines
    return m + 5

add_five = lambda m: m + 5  # 1 line, kaam same!
```

### 11.1 Lambda Syntax
```
lambda  parameters  :  expression
  ↑         ↑              ↑
keyword  input(s)    jo result chahiye (return khud ho jaata hai)
```

| Cheez | Lambda mein |
|---|---|
| `return` keyword | ❌ Nahi likhte — expression ki value apne aap return hoti hai |
| Naam | ❌ Anonymous (bina naam) — chaho toh variable mein rakh lo |
| Kitni lines | ✅ **Sirf ek expression** (if/for/assignment statements nahi) |
| Parameters | 0, 1, ya kitne bhi |
| Kab use karein | Chhota, ek baar ka kaam — `sorted`, `min`, `max`, `map`, `filter`, `reduce` ke saath |

````python
# def vs lambda
def square_def(n):
    return n * n

square = lambda n: n * n

print("def    :", square_def(5))
print("lambda :", square(5))

# Alag-alag parameters
no_arg  = lambda: "Hello GLA!"             # 0 parameter
one_arg = lambda x: x * 2                  # 1 parameter
two_arg = lambda a, b: a + b               # 2 parameters
three   = lambda a, b, c: a * b * c        # 3 parameters
default = lambda name, greeting="Hi": f"{greeting}, {name}!"      # default value bhi chalti hai

print(no_arg())
print(one_arg(21))
print(two_arg(10, 5))
print(three(2, 3, 4))
print(default("Aarav"), "|", default("Aarav", "Namaste"))

# Bina variable ke seedha use (Immediately Invoked)
print("\n(lambda x: x + 100)(5) ->", (lambda x: x + 100)(5))

# String par
first_letter = lambda s: s[0].upper()
print("first_letter('python') ->", first_letter("python"))

# Lambda bhi ek function hi hai
print("type:", type(square).__name__)
````

### 11.2 Lambda with Conditions — `if-else` Expression 🔀
Lambda mein `if` **statement** nahi chalta, par **conditional expression** chalta hai:

```
lambda x:  value_if_true  if  condition  else  value_if_false
```
Kai conditions ke liye chain kar sakte ho (par bahut lambi ho toh `def` better hai).

````python
# Pass / Fail
status = lambda m: "Pass" if m >= 40 else "Fail"
print("78 ->", status(78), "| 33 ->", status(33))

# Even / Odd
parity = lambda n: "Even" if n % 2 == 0 else "Odd"
print("7 ->", parity(7), "| 10 ->", parity(10))

# Maximum of two
bigger = lambda a, b: a if a > b else b
print("bigger(8, 15) ->", bigger(8, 15))

# Chained conditions: grade
grade = lambda m: "A+" if m >= 90 else ("A" if m >= 75 else ("B" if m >= 40 else "Fail"))
for m in [95, 80, 50, 20]:
    print(f"{m} -> {grade(m)}")

# Loop ke saath use
marks = [78, 92, 45, 33, 88]
for m in marks:
    print(m, "->", status(m))

# Absolute value, sign
sign = lambda n: "positive" if n > 0 else ("negative" if n < 0 else "zero")
print([sign(n) for n in [5, -3, 0]])
````

### 11.3 Lambda with `sorted()`, `min()`, `max()` — Sabse Zyada Use! 🏆
In functions mein **`key=`** parameter hota hai: *"Har item ka kaunsa hissa dekh ke compare karna hai?"* — wahan lambda bhaut kaam aata hai.

| Function | Kaam | Example |
|---|---|---|
| `sorted(data, key=lambda ...)` | Custom order mein sort | marks ke hisaab se students |
| `min(data, key=lambda ...)` | Custom rule se **sabse chhota item** | sabse sasta product |
| `max(data, key=lambda ...)` | Custom rule se **sabse bada item** | sabse zyada marks wala student |

> `key=` ka lambda har item ko leta hai aur ek **number / string** return karta hai jiske basis par compare hota hai. Original **item** hi result mein aata hai (key nahi).

````python
students = [("Aarav", 94), ("Bhavna", 91), ("Chirag", 88), ("Deepika", 79)]

# sorted: marks ke hisaab se (index 1)
print("Marks ascending :", sorted(students, key=lambda s: s[1]))
print("Marks descending:", sorted(students, key=lambda s: s[1], reverse=True))

# sorted: naam ki lambai ke hisaab se
names = ["Deepika", "Aarav", "Bhavna", "Chirag", "Esha"]
print("\nLambai ke hisaab se:", sorted(names, key=lambda n: len(n)))
print("Last letter se     :", sorted(names, key=lambda n: n[-1]))
print("Case-insensitive   :", sorted(["banana", "Apple", "cherry"], key=lambda s: s.lower()))

# min / max
print("\nTopper (max)  :", max(students, key=lambda s: s[1]))
print("Lowest (min)  :", min(students, key=lambda s: s[1]))
print("Sabse lamba naam:", max(names, key=lambda n: len(n)))

# List of dictionaries
products = [
    {"name": "Pen",      "price": 10,  "stock": 50},
    {"name": "Notebook", "price": 60,  "stock": 0},
    {"name": "Bag",      "price": 450, "stock": 5},
]
print("\nSabse sasta  :", min(products, key=lambda p: p["price"])["name"])
print("Sabse mehnga :", max(products, key=lambda p: p["price"])["name"])
print("Stock ke hisaab se sorted:", [p["name"] for p in sorted(products, key=lambda p: p["stock"])])

# Multiple conditions: pehle marks (bade se chhote), phir naam (A-Z)
data = [("Zoya", 90), ("Aarav", 90), ("Bhavna", 95)]
print("\nTie-break:", sorted(data, key=lambda s: (-s[1], s[0])))
````

### 11.4 Lambda Ko Variable / Dictionary Mein Rakhna 🗃️
Lambda ek **object** hai — dictionary ki value ban sakta hai. Isse **calculator** jaise menu-driven programs bante hain.

````python
operations = {
    "add":      lambda a, b: a + b,
    "subtract": lambda a, b: a - b,
    "multiply": lambda a, b: a * b,
    "divide":   lambda a, b: a / b if b != 0 else "Error: zero",
}

a, b = 12, 4
for name, func in operations.items():
    print(f"{name:<9}: {a} and {b} -> {func(a, b)}")

print("divide(5, 0) ->", operations["divide"](5, 0))
````

### 📌 Lambda Cheat-Sheet
| Kaam | Code |
|---|---|

| Multiple args | `lambda a, b: a + b` |

| Sort by field | `sorted(data, key=lambda s: s[1])` |
| Max by field | `max(data, key=lambda s: s[1])` |
| Min by field | `min(data, key=lambda s: s[1])` |
| Descending sort | `sorted(data, key=..., reverse=True)` |

> 🤔 **`def` ya `lambda`?** Ek line ka chhota kaam + ek hi jagah use → `lambda`. Lamba ya baar-baar use hone wala logic → `def` (naam aur docstring ke saath).

# 🗺️ Chapter 12: `map()` — Grace Marks for Everyone

### 🎬 Story
Result Day ka dhamaka! Principal sir ne announce kiya: *"Is baar **har student ko +5 grace marks** milenge!"* 🎉

Aarav loop likhne baitha, comprehension bhi likh sakta tha... Chirag bola: *"Python ke paas ek aur tareeka hai — **`map()`**. Bas bata de **kya karna hai (function)** aur **kis par (list)** — baaki Python sambhal lega!"*

### 12.1 `map()` Kya Karta Hai?
**`map(function, iterable)` → iterable ke har item par function lagata hai aur results deta hai.**

```
marks  →  [ 78 ,  92 ,  45 ]
              ↓      ↓      ↓      function: lambda m: m + 5
result →  [ 83 ,  97 ,  50 ]
```

| Cheez | Matlab |
|---|---|
| Syntax | `map(function, iterable)` ya `map(function, iter1, iter2, ...)` |
| Return | **map object** (lazy iterator) — dekhne ke liye `list(...)`, `tuple(...)`, `set(...)` |
| Function kaisa | `def` wala, `lambda`, ya built-in (`len`, `int`, `str.upper`...) |
| Original data | Safe rehta hai |

````python
marks = [78, 92, 45, 88]

# map() seedha print karo toh object dikhta hai
m = map(lambda x: x + 5, marks)
print("map ka type:", type(m).__name__)
print("list(m)    :", list(m))

# 1. lambda ke saath
print("\nGrace +5       :", list(map(lambda x: x + 5, marks)))
print("Doubles        :", list(map(lambda x: x * 2, marks)))

# 2. def function ke saath
def grade(m):
    return "Pass" if m >= 40 else "Fail"
print("Grades (def)   :", list(map(grade, [78, 33, 45])))

# 3. Built-in functions ke saath
names = ["aarav", "bhavna", "chirag"]
print("\nlen            :", list(map(len, names)))
print("str.upper      :", list(map(str.upper, names)))
print("str.title      :", list(map(str.title, names)))
print("str (num->text):", list(map(str, [1, 2, 3])))

# 4. Sabse famous use: text input ko numbers mein badalna
line = "10 20 30 40"
numbers = list(map(int, line.split()))
print("\n'10 20 30 40' -> ", numbers, "| sum =", sum(numbers))

# Original safe
print("\nOriginal marks:", marks)
````

### 12.2 `map()` with Multiple Iterables 🔗
Agar function **2 inputs** leta hai, toh **2 lists** de sakte ho — dono ke **same position wale items** milte hain.
Chhoti list khatam hote hi map ruk jaata hai (jaise `zip`).

````python
prices = [100, 250, 80]
qty    = [2, 1, 5]

totals = list(map(lambda p, q: p * q, prices, qty))
print("Line totals :", totals)
print("Grand total :", sum(totals))

# Do lists ka jod
a = [1, 2, 3]
b = [10, 20, 30]
print("\nElement-wise sum:", list(map(lambda x, y: x + y, a, b)))

# Teen lists
c = [100, 200, 300]
print("Teen lists      :", list(map(lambda x, y, z: x + y + z, a, b, c)))

# Alag lambai: chhoti list tak
print("\nUnequal lengths :", list(map(lambda x, y: x + y, [1, 2, 3, 4], [10, 20])))

# Naam + marks ka sentence
names = ["Aarav", "Bhavna"]
marks = [94, 91]
print(list(map(lambda n, m: f"{n} scored {m}", names, marks)))
````

### 12.3 ⚠️ `map` Ek Baar Hi Chalta Hai (Lazy Iterator)
`map()` result tab banata hai jab aap maango. Ek baar use ho gaya toh **khatam** — dobara `list()` karoge toh **khaali** milega! Dobara chahiye toh result ko **variable mein list bana ke** rakh lo.

````python
m = map(lambda x: x * 10, [1, 2, 3])

print("Pehli baar:", list(m))
print("Doosri baar:", list(m), " <- khaali! (iterator ek baar hi chalta hai)")

# ✅ Sahi tarika: pehle hi list bana lo
result = list(map(lambda x: x * 10, [1, 2, 3]))
print("\nList mein rakha:", result)
print("Dobara dekho   :", result)
````

### 12.4 `map()` vs Loop vs List Comprehension 🤼
Teeno ek hi kaam kar sakte hain — **grace +5**:

````python
marks = [78, 92, 45, 88]

# 1. Loop
result1 = []
for m in marks:
    result1.append(m + 5)

# 2. List comprehension
result2 = [m + 5 for m in marks]

# 3. map
result3 = list(map(lambda m: m + 5, marks))

print("Loop          :", result1)
print("Comprehension :", result2)
print("map           :", result3)
print("Teeno same?   :", result1 == result2 == result3)
````

### 📌 `map()` Cheat-Sheet
| Kaam | Code |
|---|---|
| Har item par function | `list(map(func, data))` |
| Lambda ke saath | `list(map(lambda x: x + 5, data))` |
| Built-in ke saath | `list(map(len, words))` |
| Text → numbers | `list(map(int, text.split()))` |
| 2 lists ek saath | `list(map(lambda a, b: a * b, l1, l2))` |
| ⚠️ Yaad rakho | map object **ek baar** hi use hota hai |

# 🔍 Chapter 13: `filter()` — Sirf Pass Students

### 🎬 Story
Bhavna Didi ne nayi list maangi: *"Sirf wo students jinke marks **40 ya usse zyada** hain — certificate sirf unhi ko milega!"* 🎓

Aarav ke paas ab **`filter()`** hai — ye ek **chhalni (sieve)** ki tarah kaam karta hai: jo condition pass kare wo rukta hai, baaki bahar. 🧹

### 13.1 `filter()` Kya Karta Hai?
**`filter(function, iterable)` → sirf wahi items rakhta hai jinke liye `function(item)` `True` return kare.**

```
marks   →  [ 78 , 33 , 92 , 25 , 60 ]
function:    m >= 40   → T    F    T    F    T
result  →  [ 78 ,      92 ,      60 ]
```

| | `map()` | `filter()` |
|---|---|---|
| Function ka kaam | Item ko **badalna** | Item ko **rakhna ya hatana** (True/False) |
| Result ki length | Hamesha **same** | Same ya **kam** |
| Items badalte hain? | ✅ Haan | ❌ Nahi (wahi original items) |
| Return | map object (lazy) | filter object (lazy) |

````python
marks = [78, 33, 92, 25, 60, 45]

# lambda ke saath
passed = list(filter(lambda m: m >= 40, marks))
print("Passed       :", passed)
print("Distinction  :", list(filter(lambda m: m >= 75, marks)))
print("Failed       :", list(filter(lambda m: m < 40, marks)))
print("Failed count :", len(list(filter(lambda m: m < 40, marks))))

# Even numbers
print("\nEven (1-15)  :", list(filter(lambda n: n % 2 == 0, range(1, 16))))

# def function ke saath
def is_pass(m):
    return m >= 40

print("def ke saath :", list(filter(is_pass, marks)))

# Type par: filter object
f = filter(is_pass, marks)
print("\nfilter ka type:", type(f).__name__)

# Strings par
names = ["Aarav", "Bhavna", "Chirag", "Anil", "Deepika"]
print("\n'A' se shuru:", list(filter(lambda n: n.startswith("A"), names)))
print("Lambai > 5  :", list(filter(lambda n: len(n) > 5, names)))

# Lazy hai, ek baar chalta hai
f2 = filter(is_pass, marks)
print("\nPehli baar :", list(f2))
print("Doosri baar:", list(f2), "<- khaali")
````

### 13.2 `filter(None, data)` — Falsy Values Hatao 🗑️
Function ki jagah **`None`** do toh `filter` sirf **truthy** items rakhta hai. Falsy = `0`, `""`, `None`, `False`, `[]`, `{}`.

*Use: khaali strings, `0` ya `None` saaf karna.*

````python
raw = ["Aarav", "", "Bhavna", None, "Chirag", "  "]
print("Raw           :", raw)
print("filter(None)  :", list(filter(None, raw)))        # '  ' truthy hai (space hai), isliye rehta hai

numbers = [0, 1, 2, 0, 3, False, 4]
print("\nNumbers       :", numbers)
print("Non-zero      :", list(filter(None, numbers)))

# Spaces wale bhi hatane hon toh lambda
print("\nStrip karke  :", list(filter(lambda s: s and s.strip(), raw)))
````

### 13.3 `filter()` with Dictionaries / Records 🗃️
List of dictionaries par bahut kaam aata hai — jaise e-commerce mein filter!

````python
products = [
    {"name": "Pen",      "price": 10,  "stock": 50},
    {"name": "Notebook", "price": 60,  "stock": 0},
    {"name": "Bag",      "price": 450, "stock": 5},
    {"name": "Bottle",   "price": 120, "stock": 0},
    {"name": "Cap",      "price": 200, "stock": 12},
]

in_stock = list(filter(lambda p: p["stock"] > 0, products))
print("In stock      :", [p["name"] for p in in_stock])

cheap = list(filter(lambda p: p["price"] < 100, products))
print("Price < 100   :", [p["name"] for p in cheap])

# Multiple conditions
good = list(filter(lambda p: p["stock"] > 0 and p["price"] < 300, products))
print("In stock & < 300:", [p["name"] for p in good])

# Palindromes (string slicing se)
words = ["madam", "python", "level", "gla", "noon"]
print("\nPalindromes   :", list(filter(lambda w: w == w[::-1], words)))
````

### 13.4 `filter()` vs List Comprehension 🤼

````python
marks = [78, 33, 92, 25, 60, 45]

with_filter = list(filter(lambda m: m >= 40, marks))
with_comp   = [m for m in marks if m >= 40]

print("filter        :", with_filter)
print("comprehension :", with_comp)
print("Same?         :", with_filter == with_comp)
````

| Records filter | `list(filter(lambda p: p["stock"] > 0, products))` |

### 14.1 `reduce()` Kya Karta Hai?
`reduce` **`functools`** module mein hota hai, isliye pehle import karna padta hai:

```python
from functools import reduce
reduce(function, iterable, initial)       # initial optional hai
```

**`function` hamesha 2 parameters leta hai:** `(accumulated, next_item)`.
Ye pehle do items par function lagata hai, result ko agle item ke saath jodta hai, aur aise hi **end tak** chalta hai.

```
reduce(lambda a, b: a + b, [5, 12, 8, 3])

Step 1:  5 + 12  = 17
Step 2: 17 +  8  = 25
Step 3: 25 +  3  = 28   ← final answer
```

| | `map` | `filter` | `reduce` |
|---|---|---|---|
| Kya karta hai | Har item badalta hai | Kuch items rakhta hai | Sab ko **ek value** mein sametta hai |
| Output | Iterator (utni hi items) | Iterator (kam items) | **Ek single value** |
| Import? | Nahi | Nahi | ✅ `from functools import reduce` |

````python
from functools import reduce

fund = [5000, 12000, 8000, 3000]

total = reduce(lambda a, b: a + b, fund)
print("Total fund (reduce):", total)
print("Total fund (sum)   :", sum(fund))

# 👀 Steps dekhna: function mein print lagake
def add_verbose(acc, item):
    result = acc + item
    print(f"   acc = {acc:>6} | item = {item:>6} | result = {result:>6}")
    return result

print("\nStep-by-step:")
final = reduce(add_verbose, fund)
print("Final:", final)
````

### 14.2 Sum, Product, Maximum with `reduce()` 🧰

````python
from functools import reduce

nums = [3, 1, 4, 1, 5, 9, 2, 6]

# 1. Sum
print("Sum     :", reduce(lambda a, b: a + b, nums))

# 2. Product (guna)
print("Product :", reduce(lambda a, b: a * b, nums))

# 3. Maximum
print("Maximum :", reduce(lambda a, b: a if a > b else b, nums))

# 4. Minimum
print("Minimum :", reduce(lambda a, b: a if a < b else b, nums))

# 5. Factorial: 5! = 1*2*3*4*5
print("5!      :", reduce(lambda a, b: a * b, range(1, 6)))

# 6. Strings jodna
words = ["Python", "is", "super", "fun"]
print("Sentence:", reduce(lambda a, b: a + " " + b, words))

# 7. Sabse lamba word
print("Longest :", reduce(lambda a, b: a if len(a) >= len(b) else b, words))
````

### 14.3 Initial Value (Teesra Argument) 🏁
`reduce(func, data, initial)` — **`initial`** pehle `acc` ban jaata hai, aur `data` ka pehla item `item` ban ke aata hai.

**Kab zaroori?**
1. **Khaali list** par bina initial value ke → `TypeError`. Initial ho toh wahi return hota hai.
2. Jab accumulator ka **type alag** ho (jaise tuples ki list se total amount nikalna).

````python
from functools import reduce

# 1. Khaali list
try:
    reduce(lambda a, b: a + b, [])
except TypeError as e:
    print("⚠️ Initial ke bina khaali list -> TypeError:")
    print("  ", e)

print("Initial 0 ke saath khaali list ->", reduce(lambda a, b: a + b, [], 0))

# 2. Initial value shuruaat ka total hai
print("\nInitial 1000 ke saath:", reduce(lambda a, b: a + b, [10, 20, 30], 1000))

# 3. Records se total (acc number hai, item tuple)
donations = [("Aarav", 500), ("Bhavna", 1200), ("Chirag", 300)]
total = reduce(lambda acc, d: acc + d[1], donations, 0)
print("\nDonations total:", total)

# 4. Dictionary banana reduce se (har word ki ginti)
words = ["a", "b", "a", "c", "a", "b"]
def count_word(acc, w):
    acc[w] = acc.get(w, 0) + 1
    return acc
print("Word count:", reduce(count_word, words, {}))
````

### 14.4 `reduce()` vs Loop — Same Kaam, Teen Tarike 🤼
Product of `[2, 3, 4, 5]`:

````python
from functools import reduce

nums = [2, 3, 4, 5]

# 1. Loop
result_loop = 1
for n in nums:
    result_loop *= n

# 2. reduce
result_reduce = reduce(lambda a, b: a * b, nums)

# 3. math.prod (Python ka built-in shortcut)
import math
result_prod = math.prod(nums)

print("Loop        :", result_loop)
print("reduce      :", result_reduce)
print("math.prod   :", result_prod)
print("Teeno same? :", result_loop == result_reduce == result_prod)

print("\nSum ke liye seedha sum() better hai:", sum(nums))
print("Max ke liye seedha max() better hai:", max(nums))
````

### 📌 `reduce()` Cheat-Sheet
| Kaam | Code |
|---|---|
| Import | `from functools import reduce` |
| Sum | `reduce(lambda a, b: a + b, data)` |
| Product | `reduce(lambda a, b: a * b, data)` |
| Maximum | `reduce(lambda a, b: a if a > b else b, data)` |
| Initial value ke saath | `reduce(lambda a, b: a + b, data, 0)` |
| Records ka total | `reduce(lambda acc, r: acc + r[1], records, 0)` |

> 💡 **Kab `reduce`, kab loop / built-in?** Simple sum/max/min/product ke liye **built-ins** (`sum`, `max`, `min`, `math.prod`) behtar aur readable hain. **Custom accumulation** (dictionary banana, kuch complex combine karna) ke liye `reduce` shaandar hai.

# 🏭 Chapter 15: The Pipeline — `map` + `filter` + `reduce` Together

### 🎬 Story (Final Boss 🔥)
Bhavna Didi ka aakhri sawal:

> *"Marks list se pehle **fail walon ko hatao**, phir **bacho ko +5 grace** do, aur aakhir mein **total marks** batao."*

Teen kaam, teen tools — **`filter` ➜ `map` ➜ `reduce`**. Jaise factory ki **assembly line**: ek machine ka output doosri machine ka input! 🏭

```
 marks  ─►  filter(pass?)  ─►  map(+5)  ─►  reduce(total)  ─►  answer
[78,33,92]   [78, 92]          [83, 97]        180
```

### 15.1 Pipeline Step-by-Step

````python
from functools import reduce

marks = [78, 33, 92, 25, 60, 45]

# Step 1: filter — sirf pass
step1 = list(filter(lambda m: m >= 40, marks))
print("Step 1 (filter pass) :", step1)

# Step 2: map — +5 grace
step2 = list(map(lambda m: m + 5, step1))
print("Step 2 (map +5)      :", step2)

# Step 3: reduce — total
step3 = reduce(lambda a, b: a + b, step2)
print("Step 3 (reduce total):", step3)

# Ab sab ek saath (andar se bahar padhte hain)
answer = reduce(
    lambda a, b: a + b,
    map(lambda m: m + 5,
        filter(lambda m: m >= 40, marks))
)
print("\nEk hi expression mein:", answer)
print("Dono same?           :", answer == step3)
````

### 15.2 Aur Pipelines 🛠️
Pipeline ka pattern hamesha wahi: **`filter` (chhaano) → `map` (badlo) → `reduce` / `sum` / `max` (sametto)**.

````python
from functools import reduce

# 1. Even numbers ke squares ka sum (1 se 10)
result = reduce(lambda a, b: a + b,
                map(lambda n: n ** 2,
                    filter(lambda n: n % 2 == 0, range(1, 11))))
print("Even squares ka sum  :", result)          # 4 + 16 + 36 + 64 + 100

# 2. Words: 3 se bade words ko Title case mein
sentence = "the quick brown fox is on the run"
words = list(map(str.title, filter(lambda w: len(w) > 3, sentence.split())))
print("Bade words (Title)   :", words)
print("Joined               :", " ".join(words))

# 3. Shopping cart: in-stock items ka total bill (price * qty)
cart = [
    {"item": "Pen",    "price": 10,  "qty": 5, "in_stock": True},
    {"item": "Bag",    "price": 450, "qty": 1, "in_stock": False},
    {"item": "Notebook", "price": 60, "qty": 3, "in_stock": True},
]
bill = reduce(lambda a, b: a + b,
              map(lambda c: c["price"] * c["qty"],
                  filter(lambda c: c["in_stock"], cart)))
print("\nIn-stock items ka bill:", bill)       # 10*5 + 60*3
````

### 15.3 List Comprehension vs `map` / `filter` 🤼
Python mein ek hi kaam **kai tarike** se ho sakta hai. Dekho **comprehension** aur **map/filter** ek jaisa answer dete hain:

````python
numbers = list(range(1, 11))

# Kaam: even numbers ke squares

# map + filter
way1 = list(map(lambda n: n ** 2, filter(lambda n: n % 2 == 0, numbers)))

# list comprehension
way2 = [n ** 2 for n in numbers if n % 2 == 0]

print("map + filter  :", way1)
print("comprehension :", way2)
print("Same?         :", way1 == way2)

# Pass students ke grace marks
marks = [78, 33, 92, 25, 60]
print("\nmap/filter    :", list(map(lambda m: m + 5, filter(lambda m: m >= 40, marks))))
print("comprehension :", [m + 5 for m in marks if m >= 40])
````

### 📊 Comprehension vs map/filter — Kab Kya?
| | **List Comprehension** | **`map()` / `filter()`** |
|---|---|---|
| Readability | ✅ Zyada natural (English jaisa) | Lambda ki wajah se kabhi bhari |
| Result | Seedha **list** | **Iterator** (list banana padta hai) |
| Memory | Poori list ek baar mein | ✅ Lazy — ek-ek karke (bade data ke liye achha) |
| Pehle se bana function ho | Likhna padega | ✅ `map(str.upper, names)` — seedha |
| `if` + transform ek saath | ✅ Ek hi comprehension mein | Alag `filter` aur `map` chahiye |
| Sametna (reduce) | Nahi (`sum`, `max` use karo) | `reduce` ke saath pipeline |

> 💡 **Thumb rule:** Simple aur readable → **comprehension**. Pehle se bana function ya lazy pipeline → **map/filter**. Custom sametna → **reduce**. Sab sahi hain!

# 🎬 Epilogue: Aarav ka Final Boss — Utkarsh Merit & Registration System 🎓

Aaj ke **saare tools ek saath**: Tuple 🔒, Set 🧺, Dictionary 📒, Function ⚙️, Lambda ⚡, map / filter / reduce 🏭 aur Day 1 ke strings / f-strings.
Ye **demo** hai — har line samjho aur dekho kaunsa chapter kahan use hua. 🔍

````python
from functools import reduce

# 🧺 Ch3-4: Raw registrations mein duplicates (set se saaf)
raw_entries = ["aarav sharma", "BHAVNA VERMA", "aarav sharma", "chirag rao", "Deepika Singh", "bhavna verma"]
unique_students = sorted(set(name.strip().title() for name in raw_entries))
print("Unique students:", unique_students)

# 🔒 Ch1: Fixed event info (tuple)
EVENT = ("Utkarsh 2026", "GLA University", (15, 3, 2026))
fest, college, (day, month, year) = EVENT                    # nested unpacking
print(f"\n{fest} @ {college} on {day}/{month}/{year}")

# 📒 Ch5-6: Marks dictionary
marks = {"Aarav Sharma": 94.5, "Bhavna Verma": 91.25, "Chirag Rao": 38, "Deepika Singh": 79}

# ⚙️ Ch8-9: Function with default arg + docstring + multiple return
def analyse(data, pass_mark=40):
    """Dictionary of marks se (pass_count, fail_count, topper) return karta hai."""
    passed = {n: m for n, m in data.items() if m >= pass_mark}       # Ch7: dict comprehension
    topper = max(data, key=data.get)
    return len(passed), len(data) - len(passed), topper

p, f, topper = analyse(marks)
print(f"\nPass: {p} | Fail: {f} | Topper: {topper}")

# ⚡ Ch11: lambda sort -> Ch7: leaderboard
ranked = sorted(marks.items(), key=lambda kv: kv[1], reverse=True)

# 🏭 Ch12-15: pipeline (filter -> map -> reduce)
graced = list(map(lambda kv: min(kv[1] + 5, 100), filter(lambda kv: kv[1] >= 40, ranked)))
total_graced = reduce(lambda a, b: a + b, graced)

# 🧵 Day 1: f-string formatting
print("\n" + "=" * 46)
print("UTKARSH 2026 - MERIT LIST".center(46))
print("=" * 46)
for rank, (name, m) in enumerate(ranked, start=1):
    status = "PASS" if m >= 40 else "FAIL"
    print(f"{rank}. {name:<16} {m:>6.2f}   {status}")
print("=" * 46)
print(f"Graced marks of passed students: {graced}")
print(f"Total graced marks            : {total_graced}")
print(f"🏆 Topper: {topper} ({marks[topper]}%)")
````

### 🧠 7 Golden Rules
1. **Tuple = locked list** 🔒 • **Set = unique + unordered** 🧺 • **Dict = key → value** 📒
2. Ek item ki tuple mein **comma** (`(5,)`) aur khaali set ke liye **`set()`** (`{}` dictionary hai!).
3. Set mein **index nahi**; order chahiye toh `sorted(s)`.
4. Dictionary mein key nahi mili toh `d[key]` error deta hai — **`d.get(key, default)`** safe hai.
5. `print()` dikhata hai, **`return`** result wapas bhejta hai.
6. Variable dhoondhne ka order: **L → E → G → B**. `global` / `nonlocal` sirf tab, jab badalna ho.
7. **`map` badalta hai, `filter` chhaanta hai, `reduce` sametta hai** — teeno lazy / ek-value ka kaam, aur `lambda` inka sabse achha dost! ⚡
