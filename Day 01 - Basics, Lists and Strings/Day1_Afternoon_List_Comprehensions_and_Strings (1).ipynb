### 1.1 List Kya Hoti Hai?
**List = ek variable mein multiple values store karne ka tarika.** Ye hamesha **square brackets `[ ]`** mein likhi jaati hai aur values **comma `,`** se alag hoti hain.

*Real-life examples*: Shopping list 🛒, Mobile contacts 📱, Cricket team 🏏, Exam marks 📝

| Property | Matlab |
|---|---|
| **Ordered** | Jis order mein daali, usi order mein rahegi |
| **Indexed** | Har item ka ek position number (index) hota hai |
| **Mutable** | Baad mein change / add / delete kar sakte ho |
| **Duplicates allowed** | Same value kai baar rakh sakte ho |
| **Mixed types** | Alag-alag type ki values ek hi list mein ho sakti hain |

````python
# List ke 4 alag roop
numbers = [10, 20, 30, 40, 50]            # 1. Number list
names   = ["Aarav", "Bhavna", "Chirag"]   # 2. String list
mixed   = ["Aarav", 101, 94.5, True]      # 3. Mixed list (alag-alag types)
empty   = []                              # 4. Khaali list

print("Numbers:", numbers)
print("Names  :", names)
print("Mixed  :", mixed)
print("Empty  :", empty)
print("len(mixed) =", len(mixed), "-> list mein kitne items hain")
````

### 1.2 Mutable ka Matlab — List Edit Ho Sakti Hai ✏️

### 1.3 Positive Indexing — Left se Right (0 se shuru!) ➡️
Computer ki ginti **0** se shuru hoti hai, 1 se nahi! Pehla item = index `0`.

| Item | Notebook | Pen | Calculator | ID Card |
|---|---|---|---|---|
| **Index** | 0 | 1 | 2 | 3 |

Agar list mein `N` items hain, toh valid index `0` se `N-1` tak hote hain.

````python
backpack = ["Notebook", "Pen", "Calculator", "ID Card"]

print("backpack[0] ->", backpack[0])
print("backpack[1] ->", backpack[1])
print("backpack[2] ->", backpack[2])
print("backpack[3] ->", backpack[3])

# Galat index dene par kya hota hai? (try/except abhi samajhna zaroori nahi,
# ye bas error ko safely dikha raha hai)
try:
    print(backpack[10])
except IndexError as e:
    print("⚠️ IndexError:", e)
````

### 1.4 Negative Indexing — Right se Left ⬅️
Aakhri item chahiye par list kitni lambi hai pata nahi? **`-1`** use karo!

| Item | Notebook | Pen | Calculator | ID Card |
|---|---|---|---|---|
| **Positive** | 0 | 1 | 2 | 3 |
| **Negative** | -4 | -3 | -2 | -1 |

````python
backpack = ["Notebook", "Pen", "Calculator", "ID Card"]

print("Last item   [-1] ->", backpack[-1])
print("2nd last    [-2] ->", backpack[-2])
print("First (neg) [-4] ->", backpack[-4])

# Dono ek hi cheez dete hain:
print("backpack[3] == backpack[-1] ?", backpack[3] == backpack[-1])
````

### 1.5 Slicing — List ka Ek Hissa Nikalna ✂️
**Syntax:** `list[start : stop : step]`

| Part | Matlab | Default |
|---|---|---|
| `start` | Yahan se shuru (**included**) | 0 |
| `stop`  | Yahan se **pehle** ruk jao (**excluded!**) | end tak |
| `step`  | Kitne ke jump se chalna hai | 1 |

> 🧠 **Yaad rakho:** `scores[2:5]` ka matlab index **2, 3, 4** (5 nahi!).

Bhavna Didi ne **8 students ke marks** diye hain — chalo inhe slice karte hain.

````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]
#  index: 0   1   2   3   4   5   6   7

print("scores[0:3] ->", scores[0:3])   # index 0,1,2
print("scores[2:6] ->", scores[2:6])   # index 2,3,4,5
print("scores[:3]  ->", scores[:3])    # shuru se 3 items
print("scores[4:]  ->", scores[4:])    # index 4 se end tak
print("scores[-3:] ->", scores[-3:])   # last 3 items
print("scores[:]   ->", scores[:])     # poori list (copy)
````

#### Slicing with Step 🪜
Teesra number `step` = kitne-kitne ke jump se items uthane hain. **Negative step = ulta chalo!**

````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

print("scores[::2]   ->", scores[::2])    # har 2nd item (0,2,4,6)
print("scores[1::2]  ->", scores[1::2])   # index 1 se, har 2nd item (1,3,5,7)
print("scores[::3]   ->", scores[::3])    # har 3rd item
print("scores[::-1]  ->", scores[::-1])   # POORI list ulti (reverse)
print("scores[5:1:-1]->", scores[5:1:-1]) # index 5 se peeche ki taraf, index 1 se pehle tak
````

### 1.6 Nested Lists — List ke Andar List 🧩
Fest ke hall ki **seating** ek grid hai: 3 rows, har row mein 3 seats. Ye ek **list of lists** (matrix) se ban jaata hai.

```
          col 0  col 1  col 2
row 0  [   101,   102,   103 ]
row 1  [   201,   202,   203 ]
row 2  [   301,   302,   303 ]
```
Pehla index = **row**, doosra index = **column** → `grid[row][col]`

````python
seating_grid = [
    [101, 102, 103],   # row 0
    [201, 202, 203],   # row 1
    [301, 302, 303],   # row 2
]

print("Poora grid       :", seating_grid)
print("Row 1 (poori)    :", seating_grid[1])
print("Row 1, Col 2     :", seating_grid[1][2])
print("Last row, last col:", seating_grid[-1][-1])

# Nested list ka ek aur example: [naam, marks] ke jodey
records = [["Aarav", 94], ["Bhavna", 91], ["Chirag", 88]]
print("\nBhavna ke marks:", records[1][1])

# Nested list ko row-by-row print karna
for row in seating_grid:
    print(row)
````

### 1.7 List Operators: `+`, `*`, `in` ➕✖️🔍

| Operator | Kaam | Example | Result |
|---|---|---|---|
| `+` | Do lists jodna (concatenation) | `[1,2] + [3]` | `[1, 2, 3]` |
| `*` | List ko repeat karna | `[0] * 3` | `[0, 0, 0]` |
| `in` | Item list mein hai? | `2 in [1,2,3]` | `True` |
| `not in` | Item list mein **nahi** hai? | `5 not in [1,2,3]` | `True` |

> ℹ️ `+` aur `*` **nayi list** banate hain, purani list change nahi hoti.

````python
team_a = ["Aarav", "Bhavna"]
team_b = ["Chirag", "Deepika"]

all_members = team_a + team_b
print("a + b      :", all_members)
print("team_a * 2 :", team_a * 2)
print("Attendance :", ["Absent"] * 4)          # 4 students ki default attendance

backpack = ["Notebook", "Pen", "Calculator", "ID Card"]
print("'Pen' in backpack       ->", "Pen" in backpack)
print("'Laptop' in backpack    ->", "Laptop" in backpack)
print("'Laptop' not in backpack->", "Laptop" not in backpack)

print("team_a abhi bhi:", team_a)   # purani list safe hai
````

# 📋 Chapter 2: 15 List Methods & Functions — Registration Desk

### 🎬 Story
Utkarsh 2026 ka **Registration Desk** khul gaya! Aarav ke haath mein participants ki list hai.
Kabhi koi naya aata hai, kabhi koi cancel karta hai, kabhi list ko sort karna padta hai...

*"Har kaam ke liye naya code?"* — nahi! List ke paas ready-made **tools** hote hain — inhe **methods** kehte hain.

> 🛠️ **Method kya hai?** List ke saath **dot `.`** laga ke likha jaane wala ready-made kaam. Jaise: `participants.append("Esha")`
> Aur **function** (jaise `len(list)`) list ko bracket ke andar leta hai.

**15 tools, 5 groups:**
| Group | Tools |
|---|---|
| ➕ Add | `append()`, `extend()`, `insert()` |
| ➖ Remove | `remove()`, `pop()`, `clear()` |
| 🔍 Search | `index()`, `count()` |
| 🔀 Arrange & Copy | `sort()`, `reverse()`, `copy()` |
| 🧮 Built-in functions | `len()`, `sum()`, `min()`, `max()` |

## ➕ Group 1: Add Elements

### 1️⃣ `append()` — End mein ek item jodo
**Syntax:** `list.append(item)` • Sirf **ek** item, hamesha **last** mein.

````python
participants = ["Aarav", "Bhavna"]
print("Pehle   :", participants)

participants.append("Chirag")
print("append('Chirag') ke baad:", participants)
````

### 2️⃣ `extend()` — Ek saath kai items jodo
**Syntax:** `list.extend(iterable)` • Doosri list ke **saare items** end mein jud jaate hain.

#### ⚠️ #1 Confusion: `append()` vs `extend()`

````python
a = [1, 2, 3]
a.append([4, 5])        # poori list ko EK item bana ke daal diya
print("append([4,5]) ->", a)

b = [1, 2, 3]
b.extend([4, 5])        # 4 aur 5 alag-alag items ban ke jude
print("extend([4,5]) ->", b)

participants = ["Aarav", "Bhavna"]
participants.extend(["Chirag", "Deepika"])
print("\nParticipants:", participants)
````

### 3️⃣ `insert()` — Kisi bhi jagah item daalo
**Syntax:** `list.insert(index, item)` • Us index par item aata hai, baaki items **right ki taraf khisak** jaate hain.

````python
participants = ["Aarav", "Chirag", "Deepika"]
print("Pehle:", participants)

participants.insert(1, "Bhavna")       # index 1 par Bhavna
print("insert(1, 'Bhavna') ->", participants)

participants.insert(0, "Chief Guest")  # index 0 = sabse aage
print("insert(0, 'Chief Guest') ->", participants)
````

## ➖ Group 2: Remove Elements

### 4️⃣ `remove()` — Value batao, wo hat jayegi
**Syntax:** `list.remove(value)` • Sirf **pehli matching value** hatati hai. Value na mile toh `ValueError`.

````python
entries = ["Aarav", "Chirag", "Aarav", "Deepika"]
print("Pehle:", entries)

entries.remove("Aarav")       # sirf PEHLA Aarav hata
print("remove('Aarav') ->", entries)

try:
    entries.remove("Zoya")    # ye list mein hai hi nahi
except ValueError as e:
    print("⚠️ ValueError:", e)
````

### 5️⃣ `pop()` — Index batao, item nikalo (aur wapas bhi pao)
**Syntax:** `list.pop()` → last item • `list.pop(index)` → us index ka item
`pop()` item ko list se hatata hai **aur return bhi karta hai** (aap use kisi variable mein rakh sakte ho).

> 🆚 `remove(value)` → **value** se hatao, kuch return nahi karta • `pop(index)` → **position** se hatao, item return karta hai

````python
numbers = [10, 20, 30, 40]
print("Pehle:", numbers)

last = numbers.pop()          # last item
print("pop() ne diya:", last, "| list:", numbers)

second = numbers.pop(1)       # index 1 wala item
print("pop(1) ne diya:", second, "| list:", numbers)
````

### 6️⃣ `clear()` — Saara saaman saaf!
**Syntax:** `list.clear()` • List ke saare items delete, list khaali `[]` ho jaati hai.

````python
entries = ["Aarav", "Bhavna", "Chirag"]
print("Pehle:", entries)

entries.clear()
print("clear() ke baad:", entries)
print("Length:", len(entries))
````

## 🔍 Group 3: Search Elements

### 7️⃣ `index()` — Item kahan hai?
**Syntax:** `list.index(value)` • Value ka **pehla index** batata hai. Na mile toh `ValueError`.

````python
scores = [78, 92, 45, 88, 92, 60]

print("88 kis index par hai? ->", scores.index(88))
print("92 pehli baar kahan?  ->", scores.index(92))   # duplicate hai, pehla hi milega

try:
    scores.index(100)
except ValueError as e:
    print("⚠️ ValueError:", e)
````

### 8️⃣ `count()` — Kitni baar aaya?
**Syntax:** `list.count(value)` • Value list mein kitni baar hai (na ho toh `0`).

````python
scores = [78, 92, 45, 88, 92, 60, 92]

print("92 kitni baar aaya? ->", scores.count(92))
print("45 kitni baar aaya? ->", scores.count(45))
print("100 kitni baar aaya?->", scores.count(100))
````

## 🔀 Group 4: Arrange & Copy

### 9️⃣ `sort()` — List ko order mein lagao
**Syntax:** `list.sort()` (ascending) • `list.sort(reverse=True)` (descending)

> ⚠️ **Dhyan do:** `sort()` **original list ko hi badal deta hai** (in-place) aur khud `None` return karta hai!

````python
scores = [78, 92, 45, 88, 60]
print("Pehle:", scores)

scores.sort()
print("sort()               ->", scores)

scores.sort(reverse=True)
print("sort(reverse=True)   ->", scores)

# ⚠️ Common galti: sort() ka result kisi variable mein mat rakho
result = [3, 1, 2].sort()
print("\nresult = [3,1,2].sort() ->", result, "(None aata hai!)")

# Strings bhi sort hote hain (alphabetical)
names = ["Deepika", "Aarav", "Chirag", "Bhavna"]
names.sort()
print("Names sorted:", names)
````

### 🔟 `reverse()` — Order ulta karo
**Syntax:** `list.reverse()` • Sort **nahi** karta — bas mojooda order ko **ulta** kar deta hai (in-place).

> 🆚 `sort(reverse=True)` → **bade se chhota** order • `reverse()` → jaisa tha **bas ulta**

````python
numbers = [10, 40, 20, 30]
print("Pehle:", numbers)

numbers.reverse()
print("reverse() ->", numbers)       # [30, 20, 40, 10] — sorted nahi, sirf ulta!
````

### 1️⃣1️⃣ `copy()` — Original ka duplicate banao
**Syntax:** `new_list = old_list.copy()`

#### 🪤 Reference Trap: `b = a` copy **nahi** hai!
`b = a` sirf **ek nayi naam-patti** lagata hai — dono naam **ek hi list** ko point karte hain.

````python
original = [10, 20, 30]

alias = original            # ❌ copy nahi, sirf ek aur naam
alias.append(99)
print("alias se append kiya ->", alias)
print("original ka haal     ->", original, "   <- original bhi badal gayi!")
print("Same list hain?", alias is original)

print("-" * 45)

original = [10, 20, 30]
backup = original.copy()    # ✅ asli copy
backup.append(99)
print("backup se append kiya ->", backup)
print("original ka haal      ->", original, "   <- original safe!")
print("Same list hain?", backup is original)
````

## 🧮 Group 5: Built-in Functions — `len()`, `sum()`, `min()`, `max()`
Ye list ke **methods nahi**, Python ke **built-in functions** hain — list ko bracket ke andar dete hain: `len(list)`.

| Function | Kaam |
|---|---|
| `len(x)` | Kitne items hain |
| `sum(x)` | Saare numbers ka jod |
| `min(x)` | Sabse chhota |

### 3.1 List Comprehension Kya Hai?
Loop + `append()` ko **ek hi line** mein likhne ka compact tarika.

```
[  expression   for   item   in   iterable  ]
   ↑ kya banana hai     ↑ har item   ↑ kahan se
```

Chalo, pehle **purane tarike** se phir **naye tarike** se dekhte hain — marks par **+5 grace marks**:

````python
scores = [78, 92, 45, 88, 60]

# 🐢 Traditional tarika (4 lines)
graced_old = []
for m in scores:
    graced_old.append(m + 5)
print("Loop se        :", graced_old)

# 🚀 List comprehension (1 line)
graced_new = [m + 5 for m in scores]
print("Comprehension  :", graced_new)

print("Dono same?", graced_old == graced_new)
````

#### Expression Ke Saath Khelna 🎨
`expression` kuch bhi ho sakta hai — maths, string methods, function call...

````python
numbers = [1, 2, 3, 4, 5]
print("Squares   :", [n * n for n in numbers])
print("Doubles   :", [n * 2 for n in numbers])

names = ["aarav", "bhavna", "chirag"]
print("Uppercase :", [n.upper() for n in names])
print("Lengths   :", [len(n) for n in names])

# range() ke saath: 1 se 10 tak ke cubes
print("Cubes 1-5 :", [n ** 3 for n in range(1, 6)])
````

### 3.2 Comprehension with Condition (Filter) — `if` end mein 🔍
**Syntax:** `[expression for item in iterable if condition]`
Sirf wahi items aayenge jinke liye condition `True` hai. Bhavna Didi ko **sirf pass students** chahiye (marks ≥ 40).

````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

passed = [m for m in scores if m >= 40]
print("Passed        :", passed)

distinction = [m for m in scores if m >= 75]
print("Distinction   :", distinction)

evens = [n for n in range(1, 11) if n % 2 == 0]
print("Even (1-10)   :", evens)

print("Kitne fail?   :", len([m for m in scores if m < 40]))
````

### 3.3 Comprehension with `if-else` — Ye **Aage** Aata Hai! 🔀
Jab **har item ko rakhna hai par alag-alag value dena hai**, toh `if-else` **expression wale hisse** mein (aage) likhte hain.

| Kaam | Syntax | Position |
|---|---|---|
| **Filter** (kuch items hatao) | `[x for x in data if cond]` | `if` **end** mein |
| **Transform** (sabko value do) | `[A if cond else B for x in data]` | `if-else` **aage** |

````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

status = ["Pass" if m >= 40 else "Fail" for m in scores]
print("Status :", status)

# Cap: 90 se upar ke marks ko 90 rakho
capped = [90 if m > 90 else m for m in scores]
print("Capped :", capped)

# Even/Odd tagging
tags = ["Even" if n % 2 == 0 else "Odd" for n in range(1, 7)]
print("Tags   :", tags)
````

### 3.4 Nested Comprehension — Loop ke Andar Loop 🧩
Seating grid ko **flatten** karna (2D → 1D) ya **table** banana.

```
[ x for row in grid for x in row ]
       └─ bahari loop ─┘ └ andar wala loop ┘      <- jaisa loop likhte ho, waisa hi order!
```

````python
seating_grid = [
    [101, 102, 103],
    [201, 202, 203],
    [301, 302, 303],
]

# 1. Flatten: 2D grid -> 1D list
flat = [seat for row in seating_grid for seat in row]
print("Flattened      :", flat)

# 2. Flatten + filter: sirf even seat numbers
even_seats = [seat for row in seating_grid for seat in row if seat % 2 == 0]
print("Even seats     :", even_seats)

# 3. Nayi 2D list banana: 3x3 multiplication table
table = [[r * c for c in range(1, 4)] for r in range(1, 4)]
print("3x3 table      :", table)
for row in table:
    print(row)
````

### 3.5 `sorted()` — Naya Sorted List, Original Safe ✅
Chapter 2 mein `sort()` seekha tha jo original ko badal deta hai. `sorted()` **nayi list** deta hai aur original ko chhoota bhi nahi. Kisi bhi iterable par chalta hai.

| | `list.sort()` | `sorted(list)` |
|---|---|---|
| Original badalta hai? | ✅ Haan | ❌ Nahi |
| Return karta hai | `None` | Nayi sorted list |
| Kahan chalta hai | Sirf lists par | List, string, tuple... sab par |

````python
scores = [78, 92, 45, 88, 60]

print("sorted(scores)               ->", sorted(scores))
print("sorted(scores, reverse=True) ->", sorted(scores, reverse=True))
print("Original abhi bhi            ->", scores)

# Strings par bhi chalta hai
names = ["Deepika", "Aarav", "Chirag", "Bhavna"]
print("sorted(names)                ->", sorted(names))
print("sorted(names, key=len)       ->", sorted(names, key=len))   # lambai ke hisaab se
print("sorted('python')             ->", sorted("python"))        # string ke letters
````

### 3.6 `enumerate()` — Index + Value Saath Mein 🔢
Loop mein **item aur uska number dono** chahiye? `enumerate()` use karo. `start=1` se ginti 1 se shuru hoti hai.

````python
names = ["Aarav", "Bhavna", "Chirag"]

print("Default (0 se):")
for index, name in enumerate(names):
    print(index, "->", name)

print("\nstart=1 ke saath (Roll No.):")
for roll, name in enumerate(names, start=1):
    print(f"Roll No. {roll}: {name}")

print("\nList mein dekho:", list(enumerate(names, start=1)))
````

### 3.7 `zip()` — Do Lists ko Jodi Mein Silna 🤝
Do (ya zyada) lists ke **same position wale items** ko pair bana deta hai.
- Result ek zip object hota hai — dekhne ke liye `list(...)` mein daalo.
- Agar lists ki lambai alag ho toh **chhoti list ke hisaab se ruk jaata hai**.

````python
names = ["Aarav", "Bhavna", "Chirag"]
marks = [94, 91, 88]

pairs = list(zip(names, marks))
print("zip pairs:", pairs)

for name, mark in zip(names, marks):
    print(f"{name} ne {mark} marks laaye")

# Alag lambai: chhoti list tak hi jodi banegi
print("\nUnequal:", list(zip([1, 2, 3, 4], ["a", "b"])))

# Dictionary banane ka shortcut
print("Dict    :", dict(zip(names, marks)))

# Unzip: jodiyon ko wapas alag karna
n, m = zip(*pairs)
print("Unzip   :", n, m)
````

### 🚀 Boss Move: Teeno Saath Mein! Merit List
`zip()` se jodi banao → `sorted()` se top marks upar → `enumerate()` se rank do.

````python
names = ["Aarav", "Bhavna", "Chirag", "Deepika"]
marks = [88, 94, 79, 91]

# (marks, name) ki jodi banai taaki sorted() marks ke hisaab se sort kare
ranked = sorted(zip(marks, names), reverse=True)

print("🏆 UTKARSH 2026 — MERIT LIST")
for rank, (mark, name) in enumerate(ranked, start=1):
    print(f"  Rank {rank}: {name:<8} {mark}")
````

### 4.5 String Slicing — `string[start : stop : step]` ✂️

````python
word = "PYTHON"

print("word[0:3]  ->", word[0:3])     # PYT
print("word[2:5]  ->", word[2:5])     # THO
print("word[:2]   ->", word[:2])      # PY
print("word[3:]   ->", word[3:])      # HON
print("word[-3:]  ->", word[-3:])     # HON
print("word[::2]  ->", word[::2])     # PTO (har 2nd char)
print("word[::-1] ->", word[::-1])    # NOHTYP (ulta!)

# Real use: email se username nikalna
email = "aarav.sharma@gla.ac.in"
print("\nUsername (pehle 5 chars):", email[:5])
print("Domain (last 9 chars)   :", email[-9:])
````

### 4.6 Immutability — String Badal Nahi Sakte! 🔒
List **mutable** thi (item badal sakte the). String **immutable** hai — ek baar ban gayi toh uske andar ka character **in-place change nahi** ho sakta.

Badalna ho toh **nayi string banao** (slicing + concatenation se).

````python
name = "Aarev"          # typo hai! 'Aarav' hona chahiye

try:
    name[3] = "a"       # ❌ seedha change nahi hoga
except TypeError as e:
    print("⚠️ TypeError:", e)

# ✅ Tarika: nayi string banao
fixed = name[:3] + "a" + name[4:]
print("Fixed name:", fixed)
print("Original   :", name, "(wahi purani rahi)")

# Methods bhi nayi string hi dete hain
word = "python"
print("\nword.upper() ->", word.upper())
print("word abhi bhi ->", word)
````

### 4.7 String Operators: `+`, `*`, `in` ➕✖️🔍
| Operator | Kaam | Example | Result |
|---|---|---|---|
| `+` | Strings jodna | `"Hi" + " GLA"` | `'Hi GLA'` |
| `*` | Repeat karna | `"ab" * 3` | `'ababab'` |
| `in` | Substring hai? | `"GLA" in "Hello GLA"` | `True` |
| `not in` | Substring nahi hai? | `"xyz" not in "abc"` | `True` |

> ⚠️ Number aur string seedha `+` nahi hote: `"Age: " + 18` ❌ → `"Age: " + str(18)` ✅

````python
first = "Utkarsh"
year = 2026

title = first + " " + str(year)     # number ko str() se string banana padta hai
print("Title:", title)
print("Border:", "=" * 20)
print("Wow! " * 3)

banner = "Welcome to GLA University, Mathura"
print("'GLA' in banner     ->", "GLA" in banner)
print("'gla' in banner     ->", "gla" in banner)        # case-sensitive!
print("'Delhi' not in banner->", "Delhi" not in banner)

try:
    print("Year: " + year)
except TypeError as e:
    print("⚠️ TypeError:", e)
````

## 🔠 Group 1: Case Conversion

### 1️⃣ `upper()` • 2️⃣ `lower()` • 3️⃣ `title()` • 4️⃣ `capitalize()` • 5️⃣ `swapcase()`

| Method | Kaam | `"hello gla UNIVERSITY"` ->|

| `upper()` | Sab CAPITAL | `HELLO GLA UNIVERSITY` |

| `swapcase()` | Capital ↔ small ulta-pulta | `HELLO GLA university` |

````python
text = "hello gla UNIVERSITY"

print("upper()      ->", text.upper())
print("lower()      ->", text.lower())
print("title()      ->", text.title())
print("capitalize() ->", text.capitalize())
print("swapcase()   ->", text.swapcase())

print("\nOriginal abhi bhi ->", text)   # immutable!

# Real use: case-insensitive comparison
typed = "YES"
print("typed.lower() == 'yes' ->", typed.lower() == "yes")
````

## 🧽 Group 2: Space Removal
### 6️⃣ `strip()` • 7️⃣ `lstrip()` • 8️⃣ `rstrip()`
- `strip()` → **dono taraf** ke spaces hatao
- `lstrip()` → sirf **left** (shuru) ke
- `rstrip()` → sirf **right** (end) ke

Bonus: bracket mein character dedo toh wo character hatayega (spaces nahi): `"***hi***".strip("*")`

````python
text = "   Python   "

print("Original  :", repr(text))     # repr() se spaces dikhte hain
print("strip()   :", repr(text.strip()))
print("lstrip()  :", repr(text.lstrip()))
print("rstrip()  :", repr(text.rstrip()))

print("\nCustom chars hatana:")
print("'***Hi***'.strip('*')  ->", "***Hi***".strip("*"))
print("'2026!!!'.rstrip('!')  ->", "2026!!!".rstrip("!"))

# strip sirf kinaare ke spaces hatata hai, beech ke nahi
print("'  a   b  '.strip()    ->", repr("  a   b  ".strip()))
````

## ✂️ Group 3: Split & Join
### 9️⃣ `split()` — String ko tukdon mein todo → **list** milti hai
`text.split()` → spaces par todta hai • `text.split(",")` → comma par todta hai
### 🔟 `join()` — List ke items ko ek string mein jodo
`"separator".join(list)` → separator beech mein lagta hai. *(Dhyan do: list ke saare items **strings** hone chahiye.)*

> split ↔ join ek-doosre ke **ulte** hain.

````python
sentence = "I Love Python"
words = sentence.split()
print("split()       ->", words)

row = "Aarav,Bhavna,Chirag"
names = row.split(",")
print("split(',')    ->", names)
print("2 tukde bas   ->", "a-b-c-d".split("-", 1))    # maxsplit=1

print("\n--- join() ---")
print("' '.join   ->", " ".join(words))
print("', '.join  ->", ", ".join(names))
print("'-'.join   ->", "-".join(["2026", "03", "15"]))
print("''.join    ->", "".join(["P", "y", "t", "h", "o", "n"]))

# Trick: extra spaces ek saath saaf karna
messy = "chirag    rao"
print("\nMessy fix ->", " ".join(messy.split()))
````

## 🔁 Group 4: Replace
### 1️⃣1️⃣ `replace(old, new)` — Text badlo
`text.replace("purana", "naya")` → saare matches badal deta hai • `replace(old, new, 1)` → sirf pehla badalta hai.

````python
text = "I Love Java. Java is easy."

print("replace all   ->", text.replace("Java", "Python"))
print("replace first ->", text.replace("Java", "Python", 1))
print("Delete karna  ->", "1-2-3".replace("-", ""))     # new = '' matlab hatao
print("Original      ->", text)
````

## 🔍 Group 5: Searching
### 1️⃣2️⃣ `find()` — Position batao (nahi mila toh **`-1`**)

### 1️⃣4️⃣ `count()` — Kitni baar aaya

````python
text = "Hello Python, Python is fun"

print("find('Python')    ->", text.find("Python"))    # pehla match
print("find('Java')      ->", text.find("Java"))      # -1 (nahi mila)
print("index('fun')      ->", text.index("fun"))
print("count('Python')   ->", text.count("Python"))
print("count('o')        ->", text.count("o"))

try:
    text.index("Java")
except ValueError as e:
    print("⚠️ index('Java') -> ValueError:", e)

# find() ka smart use
if text.find("fun") != -1:
    print("'fun' mil gaya!")
````

### 1️⃣9️⃣ `center(width, fillchar)` — Beech mein lagao
Text ko `width` jitni chaudi jagah ke **center** mein rakhta hai; baaki jagah `fillchar` (default space) se bharta hai.

````python
title = "CERTIFICATE"

print("[" + title.center(21) + "]")
print(title.center(21, "*"))
print(title.center(21, "="))

print("\n" + "=" * 30)
print("UTKARSH 2026".center(30))
print("=" * 30)
````

### 2️⃣0️⃣ `zfill(width)` — Aage zeros lagao
String ke **left** mein `0` bharta hai jab tak `width` poori na ho. *(Number ko pehle `str()` se string banana padega.)*

````python
print("'42'.zfill(6)       ->", "42".zfill(6))
print("'7'.zfill(3)        ->", "7".zfill(3))
print("'-5'.zfill(4)       ->", "-5".zfill(4))        # minus sign aage hi rehta hai

roll = 42
print("Roll no. ->", "GLA2026-" + str(roll).zfill(6))
````

### 2️⃣1️⃣ `isalnum()` — Sirf letters aur digits?
Alphabets **ya** digits (ya dono) ho toh `True`. Space, `@`, `-`, `_` jaisa koi bhi symbol aaya toh `False`.

### 2️⃣2️⃣ `isspace()` — Sirf whitespace?
Sirf spaces / tabs / new-lines ho toh `True`. **Khaali string `""` par `False`!**

````python
print("--- isalnum() ---")
print("'Aarav123'.isalnum()  ->", "Aarav123".isalnum())
print("'Aarav 123'.isalnum() ->", "Aarav 123".isalnum())   # space hai
print("'aarav_01'.isalnum()  ->", "aarav_01".isalnum())    # underscore hai

print("\n--- isspace() ---")
print("'   '.isspace()  ->", "   ".isspace())
print("'  a '.isspace() ->", "  a ".isspace())
print("''.isspace()     ->", "".isspace())          # khaali string: False!
print("'\\t\\n'.isspace()->", "\t\n".isspace())
````

### 2️⃣3️⃣ `partition(separator)` — 3 Hisson Mein Todo
Separator ke **pehle match** par string ko **tuple of 3** mein todta hai: `(pehle ka hissa, separator, baad ka hissa)`.
Separator na mile toh `(poori string, '', '')`.

> 🆚 `split()` → list (kai tukde) • `partition()` → hamesha **exactly 3** tukde (pehle separator par)

````python
setting = "name=Aarav Sharma"

before, sep, after = setting.partition("=")
print("Tuple   :", setting.partition("="))
print("Key     :", before)
print("Sep     :", sep)
print("Value   :", after)

print("\nSeparator nahi mila:", "hello".partition("="))
print("Sirf PEHLE '-' par todta hai:", "Hello-Python-World".partition("-"))
````

## 🎨 String Formatting — Text mein Values Daalna
Message banana hai: *"Aarav ne 94.5% marks laaye"* — name aur marks variables mein hain. Teen tarike hain:

| Tarika | Syntax | Kab |
|---|---|---|
| **f-string** ⭐ | `f"Hello {name}"` | **Modern & best** — hamesha ise use karo |
| `.format()` | `"Hello {}".format(name)` | Purane code mein milta hai |
| `%` operator | `"Hello %s" % name` | Bahut purana, bas pehchaan ke liye |

````python
name = "Aarav"
marks = 94.5
rank = 1

# 1. f-string (best!)
print(f"1. {name} ne {marks} marks laaye aur Rank {rank} paaya!")

# 2. .format()
print("2. {} ne {} marks laaye aur Rank {} paaya!".format(name, marks, rank))
print("2b. {n} ne {m} marks laaye".format(n=name, m=marks))     # naam se

# 3. % operator
print("3. %s ne %.1f marks laaye aur Rank %d paaya!" % (name, marks, rank))
````

### f-string Ki Power: `{value:format}` 💪
`:` ke baad **formatting rules** likhte hain:

| Format | Matlab | Example | Output |
|---|---|---|---|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:,` | Hazaar ka comma | `f"{1500000:,}"` | `1,500,000` |
| `:05d` | 5 width, aage zeros | `f"{42:05d}"` | `00042` |
| `:<10` | Left align (10 width) | `f"{'Hi':<10}"` | `Hi        ` |
| `:>10` | Right align | `f"{'Hi':>10}"` | `        Hi` |
| `:^10` | Center align | `f"{'Hi':^10}"` | `    Hi    ` |
| `:.1%` | Percentage | `f"{0.945:.1%}"` | `94.5%` |
| `{x = }` | Debug (naam + value) | `f"{x = }"` | `x = 5` |

````python
pi = 3.14159
prize = 1500000
x = 5

print(f"{pi:.2f}")                 # 3.14
print(f"{prize:,}")                # 1,500,000
print(f"{42:05d}")                 # 00042
print(f"[{'Hi':<10}]")             # left
print(f"[{'Hi':>10}]")             # right
print(f"[{'Hi':^10}]")             # center
print(f"{0.945:.1%}")              # percentage
print(f"{x = }")                   # debug
print(f"2 + 3 = {2 + 3}")          # {} ke andar expression bhi chalta hai
print(f"{'aarav'.upper()}")        # method call bhi chalta hai

# Mini table
print()
print(f"{'Name':<10}{'Marks':>8}")
print("-" * 18)
for n, m in [("Aarav", 94.5), ("Bhavna", 91.25), ("Chirag", 88)]:
    print(f"{n:<10}{m:>8.2f}")
````

# 🎬 Epilogue: Aarav ka Final Boss — Utkarsh Merit Certificates 🎓

Teeno tools — **lists, comprehensions, strings** — ek saath! Ye program messy registration data leta hai, saaf karta hai,
rank banata hai aur certificate-style output chhaapta hai. Ye **demo** hai — har line samjho aur dekho kaunsa chapter kahan use hua. 🔍

````python
raw_entries = [
    "  aarav   sharma , cse , 94.5 ",
    "BHAVNA verma,it,91.25",
    "  chirag    rao , ece , 88 ",
    "deepika SINGH ,  me,79",
]

# 1) Clean + parse har entry (str methods + list comprehension)
students = []
for entry in raw_entries:
    name, branch, marks = [part.strip() for part in entry.split(",")]    # Ch5: split, strip + Ch3: comprehension
    name = " ".join(name.split()).title()                                # Ch5: split/join/title
    students.append((float(marks), name, branch.upper()))                # float(): text -> number

# 2) Rank banao (Ch3: sorted + enumerate)
ranked = sorted(students, reverse=True)

# 3) Certificate-style output (Ch6: center, zfill, f-string)
print("=" * 44)
print("UTKARSH 2026 - MERIT LIST".center(44))
print("=" * 44)
for rank, (marks, name, branch) in enumerate(ranked, start=1):
    cert_no = "CERT-" + str(rank).zfill(4)
    print(f"{cert_no}  {name:<16} {branch:<4} {marks:>6.2f}")
print("=" * 44)

top = ranked[0]
print(f"🏆 Topper: {top[1]} ({top[2]}) with {top[0]}%")
````
