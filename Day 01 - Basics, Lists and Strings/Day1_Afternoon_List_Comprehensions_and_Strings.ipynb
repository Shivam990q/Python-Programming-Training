### 1.1 `print()` Function — Screen par Text Dikhana 🖨️
Python mein screen par kuch bhi print karne ke liye `print()` function use hota hai.
- Quotes (`"..."` ya `'...'`) ke andar jo bhi likhoge, wo waisa ka waisa print ho jayega.
- Multiple cheezon ko comma `,` se separate karke ek saath print kar sakte hain — Python automatically beech mein space daal deta hai.
- `sep` parameter se hum default space ko kisi bhi character (jaise `-` ya ` | `) se badal sakte hain.
- `end` parameter se hum decide karte hain ki line ke end mein kya aayega (default newline `\n` hoti hai).


````python
# Simple print
print("Hello, GLA University!")
print("Welcome to Utkarsh 2026 Tech Fest 🎉")

# Multiple values with comma
print("Student:", "Aarav Sharma", "Branch:", "CSE", "Year:", 1)

# sep parameter — default space ko badlo
print("GLA", "Mathura", "UP", sep=" -> ")

# end parameter — newline ki jagah space ya arrow
print("Registering student...", end=" ")
print("Done! ✅")
````

### 1.2 Variables & Dynamic Typing — Data ke Labeled Boxes 📦
**Variable kya hai?** Memory ke andar ek labeled box jisme aap koi value store karte ho.
C ya Java ki tarah Python mein pehle se `int x;` ya `String s;` declare nahi karna padta. Python itna smart hai ki value dekhkar khud samajh jaata hai ki data kis type ka hai — isse **Dynamic Typing** kehte hain!

#### 📌 Variable Naming ke 4 Simple Rules:
1. Naam letters (`a-z`, `A-Z`), digits (`0-9`), ya underscore (`_`) se shuru ho sakta hai.
2. Pehla character **kabhi number nahi ho sakta** (e.g. `1score` ❌ galat hai, `score1` ✅ sahi hai).
3. Spaces allow nahi hain (use `student_name`, not `student name`).
4. Python **case-sensitive** hai (`age`, `Age`, aur `AGE` teeno alag-alag variables hain).


````python
# Variables banana — direct value assign karo (= operator se)
student_name = "Aarav Sharma"   # text
roll_number = 260101            # integer
entry_fee = 450.50              # decimal / float
is_fest_volunteer = True        # boolean (True ya False — T aur F capital!)

print("Name:", student_name)
print("Roll No:", roll_number)
print("Fee Paid: ₹", entry_fee)
print("Is Volunteer?:", is_fest_volunteer)

# Dynamic Typing: Variable ki value aur type kabhi bhi badal sakte hain!
badge = 101             # abhi integer hai
print("Badge (int):", badge)
badge = "VIP-Pass-101"  # ab string ban gaya!
print("Badge (str):", badge)
````

### 1.5 Taking User Input & Typecasting — `input()` Function ⌨️
User se data lene ke liye `input()` use hota hai.

### 1.6 Python Operators Playground 🧰
Operators wo symbols hote hain jo data par calculations ya logical faisle karte hain.

#### 1️⃣ Arithmetic Operators (Ganit)
| Operator | Naam | Example (`a=10, b=3`) | Result | Note |
|---|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` | Jodna |
| `-` | Subtraction | `10 - 3` | `7` | Ghatana |
| `*` | Multiplication | `10 * 3` | `30` | Guna |
| `/` | True Division | `10 / 3` | `3.3333...` | **Hamesha float deta hai!** |
| `//` | Floor Division | `10 // 3` | `3` | Decimal hata kar whole integer deta hai |
| `%` | Modulus | `10 % 3` | `1` | Remainder (sheshfal) deta hai |
| `**` | Exponentiation | `2 ** 3` | `8` | Power ($2^3 = 8$) |

#### 2️⃣ Comparison Operators (Tulna — Result hamesha `True` ya `False`)
- `==` (Equal to): `5 == 5` &rarr; `True`
- `!=` (Not equal to): `5 != 3` &rarr; `True`
- `>`, `<`, `>=`, `<=` (Greater than, Less than, etc.)

#### 3️⃣ Logical Operators (Logic jodne ke liye)
- `and`: Dono conditions `True` honi chahiye tabhi `True`
- `or`: Kam se kam ek condition `True` ho toh `True`
- `not`: Ulta kar deta hai (`not True` &rarr; `False`)

#### 4️⃣ Assignment / Shorthand Operators
- `x += 5` matlab `x = x + 5`
- `x -= 2` matlab `x = x - 2`
- `x *= 3` matlab `x = x * 3`


````python
a, b = 10, 3

# Arithmetic Demo
print("10 + 3 =", a + b)
print("10 - 3 =", a - b)
print("10 * 3 =", a * b)
print("10 / 3 =", a / b, " (True division — float)")
print("10 // 3 =", a // b, " (Floor division — integer)")
print("10 % 3 =", a % b, " (Modulus / Remainder)")
print("2 ** 4 =", 2 ** 4, " (Power: 2^4)")

# Comparison Demo
print("10 == 3:", a == b)
print("10 != 3:", a != b)
print("10 > 3 :", a > b)

# Logical Demo
has_id = True
has_pass = False
print("Entry with ID and Pass?:", has_id and has_pass)
print("Entry with ID or Pass?:", has_id or has_pass)
print("Not has pass?:", not has_pass)

# Shorthand Demo
score = 50
score += 25   # score = score + 25
score *= 2    # score = score * 2
print("Final Score after shorthand:", score)
````

### 🎬 Story
Utkarsh Fest ke registration desk par bheed badhti ja rahi hai!
Bhavna Didi ne Aarav ko gate guard ke paas bitha diya:

> *"Aarav, har aane waale student par nazar rakh. Agar uske paas fest pass hai toh entry de; agar nahi hai toh Counter 3 bhej. Agar VIP pass hai toh Audi-1 bhej. Code likh jo ye saare faisle (decisions) **automatically** le sake!"* 🎯

Real life ki tarah programming mein bhi humein har mod par decision lena padta hai — aur Python mein iske liye hain **Conditions**!


### 2.1 `if` Statement — Simple Condition
Agar di gayi condition `True` hai, toh `if` ke andar waala block execute hota hai. Agar `False` hai, toh Python use skip kar deta hai.

> 🚨 **Indentation ka Niyam (The 4-Space Law):**
> C ya Java mein blocks `{ }` se bante hain, par Python mein **Indentation (spaces)** se bante hain!
> `if` ke baad colon `:` lagana zaroori hai, aur uske andar waala code 4 spaces aage khiska hona chahiye!


````python
score = 85

# Agar score 40 ya usse zyada hai toh pass
if score >= 40:
    print("Mubaarak ho! Aap exam mein PASS ho gaye! 🎉")
    print("Apna fest pass collect kar lijiye.")

print("Ye line hamesha chalegi kyunki ye if ke bahar hai.")
````

### 2.2 `if-else` Statement — Do Raste (Ya Yeh, Ya Woh)
Jab aapke paas do mutually exclusive raaste hon:
- Agar condition `True` hui &rarr; `if` block chalega.
- Agar condition `False` hui &rarr; `else` block chalega.
Dono mein se **koi ek hi** chalega, dono kabhi nahi chal sakte!


````python
has_id_card = True

if has_id_card:
    print("✅ Gate Pass: Verified! Welcome to Utkarsh 2026.")
else:
    print("❌ Entry Denied: Pehle security desk se temporary ID lein.")
````

### 2.3 `elif` Ladder — Multiple Conditions (Tiers & Grades)
Jab do se zyada options hon (jaise Marks ke hisaab se Pass ka Tier):
- `if` &rarr; pehli condition check karo.
- `elif` (else if ka short) &rarr; agar upar waali galat thi, toh ye check karo.
- `else` &rarr; agar koi bhi match nahi hui, toh aakhri option.

> 💡 **Order Matters:** Python upar se neeche check karta hai. Pehli condition jo `True` milegi, uska block chalte hi poori ladder se bahar nikal aayega!


````python
marks = 78

if marks >= 90:
    tier = "Diamond VIP Pass 💎 (Front Row Access)"
elif marks >= 75:
    tier = "Gold Pass 🥇 (Auditorium Access)"
elif marks >= 50:
    tier = "Silver Pass 🥈 (Ground Arena Access)"
else:
    tier = "General Entry 🎟️"

print(f"Marks: {marks} -> Allocated: {tier}")
````

### 2.4 Nested `if` — Condition ke Andar Condition 🧩
Kabhi-kabhi pehle gate par verify hone ke baad andar doosra check hota hai.
Jaise: Pehle check karo ki student GLA ka hai ya nahi. Agar GLA ka hai, tab check karo ki Department CSE hai ya Mechanical!


````python
is_gla_student = True
department = "CSE"

if is_gla_student:
    print("College: GLA University Student Verified ✅")
    
    # Nested check — sirf GLA students ke liye
    if department == "CSE":
        print("Special Lab 3 Access Granted: Hackathon Zone 💻")
    else:
        print("General Exhibition Hall Access Granted 🎪")
else:
    print("External Guest: Please collect visitor badge at Gate 1.")
````

### 2.5 Ternary Operator (One-Liner `if-else`) — Pythonic Shortcut ⚡
Agar aapko kisi variable ko choti si condition ke hisaab se value deni hai, toh 4 line ka `if-else` likhne ki jagah **Ternary Operator** use karo:

```python
variable = value_if_true if condition else value_if_false
```


````python
age = 19

# Standard 4-line if-else vs 1-line Ternary:
status = "Eligible to Vote" if age >= 18 else "Minor"
print(f"Age: {age} -> Status: {status}")

# Ek aur example: Even or Odd number
num = 14
kind = "EVEN" if num % 2 == 0 else "ODD"
print(f"{num} is {kind}")
````

### 3.1 `range()` Function — Numbers ki Train 🚂
Loops chalane se pehle Python ke `range()` function ko samajhna zaroori hai. Ye numbers ki ek sequence banata hai:

1. `range(stop)` &rarr; `0` se shuru hokar `stop - 1` tak jaata hai.
   - `range(5)` &rarr; `0, 1, 2, 3, 4` (5 include nahi hota!)
2. `range(start, stop)` &rarr; `start` se shuru hokar `stop - 1` tak.
   - `range(2, 6)` &rarr; `2, 3, 4, 5`
3. `range(start, stop, step)` &rarr; jump / step ke saath!
   - `range(1, 10, 2)` &rarr; `1, 3, 5, 7, 9` (odd numbers!)
   - `range(5, 0, -1)` &rarr; `5, 4, 3, 2, 1` (reverse countdown!)


````python
# range() ko dekhne ke liye list() mein daal kar print karte hain:
print("range(5)        :", list(range(5)))
print("range(2, 7)     :", list(range(2, 7)))
print("range(1, 10, 2) :", list(range(1, 10, 2)))
print("Countdown (-1)  :", list(range(5, 0, -1)))
````

### 3.2 `for` Loop — Collection ya Range par Ghoomna 🔁

````python
# 1 se 5 tak numbers print karna
print("--- 1 se 5 tak numbers ---")
for i in range(1, 6):
    print(f"Participant #{i} registered!")

# Accumulator Pattern: 1 se 10 tak ke numbers ka sum
total_score = 0
for n in range(1, 11):
    total_score += n  # total mein n jodte jao

print(f"\n1 se 10 tak ka Total Sum = {total_score}")
````

### 3.3 `while` Loop — Jab Tak Condition True Hai ⏳
`while` loop tab use hota hai jab aapko **exact count nahi pata**, balki ek **condition** ke adhar par chalna ho (e.g. "jab tak battery khatam na ho jaye, game chalu rakho!").

#### 📌 While Loop ke 3 Khambhe (Pillars):
1. **Initialize:** Counter ko shuru karo (`count = 1`).
2. **Condition:** Check karo (`while count <= 5:`).
3. **Update:** Counter ko aage badhao (`count += 1`).

> ⚠️ **DANGER — Infinite Loop:**
> Agar aap counter ko update karna bhool gaye (`count += 1`), toh condition hamesha `True` rahegi aur aapka computer/browser hang ho jayega! 🛑


````python
battery = 100

print("Fest Speaker Chalu Hai 🔊:")
while battery > 0:
    print(f"Battery at {battery}% — Playing music...")
    battery -= 25   # Har ghante 25% battery discharge

print(f"Battery: {battery}% — Music stopped. Speaker charging pe lagao! 🔌")
````

### 3.4 Loop Control Statements: `break`, `continue`, `pass` 🛑⏭️💤
Kabhi-kabhi loop ke beech mein humein apna plan badalna padta hai:

1. **`break` (Emergency Exit):** Loop ko turant wahi rok kar bahar nikal jao!
2. **`continue` (Skip & Next):** Current round ko wahi chhod kar agle round par jump karo!
3. **`pass` (Kuch Mat Karo):** Empty placeholder. Jab syntax ke liye block chahiye par abhi code nahi likhna, tab `pass` likh do taaki error na aaye.


````python
print("--- 1. BREAK Demo (Emergency Stop on Token 3) ---")
for token in range(1, 6):
    if token == 3:
        print(f"Token {token}: System Error! Halting registration desk.")
        break  # Loop yahin khatam!
    print(f"Serving Token: {token}")

print("\n--- 2. CONTINUE Demo (Skip Token 3) ---")
for token in range(1, 6):
    if token == 3:
        print(f"Token {token} is Absent — Skipping to next!")
        continue  # Bacha hua code skip karo, agle token par jao!
    print(f"Serving Token: {token}")

print("\n--- 3. PASS Demo (Placeholder) ---")
for token in range(1, 4):
    if token == 2:
        pass  # TODO: Baad mein biometric scan implement karenge
    print(f"Processed token: {token}")
````

### 3.5 Loop with `else` — Python ka Secret Superpower 🌟
Bohat kam languages mein loop ke saath `else` hota hai, par Python mein hai!
- Loop ka `else` tabhi chalta hai jab loop **naturally poora finish ho** (bina kisi `break` ke!).
- Agar loop `break` se ruka, toh `else` **nahi chalega**.

Ye **Linear Search** ke liye perfect hai: "List mein item dhoondho. Mil gaya toh `break` kardo. Agar poori list ghoomne par bhi nahi mila, toh `else` mein bolo 'Not Found'!"


````python
# Search in registered roll numbers
registered_rolls = [260101, 260102, 260103, 260104]
search_target = 260103

print(f"Searching for Roll No {search_target}...")
for roll in registered_rolls:
    if roll == search_target:
        print(f"✅ Found! Roll No {roll} is registered.")
        break
else:
    print(f"❌ Not Found: Roll No {search_target} is not in database.")

# Ab ek aisi value dhoondho jo nahi hai:
search_missing = 999999
print(f"\nSearching for Roll No {search_missing}...")
for roll in registered_rolls:
    if roll == search_missing:
        print(f"✅ Found! Roll No {roll} is registered.")
        break
else:
    print(f"❌ Not Found: Roll No {search_missing} is not in database.")
````

### 4.5 Slicing — List ka Ek Hissa Nikalna ✂️
**Syntax:** `list[start : stop : step]`

| Part | Matlab | Default |
|---|---|---|
| `start` | Yahan se shuru (**included**) | 0 |
| `stop`  | Yahan se **pehle** ruk jao (**excluded!**) | end tak |
| `step`  | Kitne ke jump se chalna hai | 1 |

> 🧠 **Yaad rakho:** `scores[2:5]` ka matlab index **2, 3, 4** (5 nahi!).

Bhavna Didi ne **8 students ke marks** diye hain — chalo inhe slice karte hain.


````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]
#  index: 0   1   2   3   4   5   6   7

print("scores[0:3] ->", scores[0:3])   # index 0,1,2
print("scores[2:6] ->", scores[2:6])   # index 2,3,4,5
print("scores[:3]  ->", scores[:3])    # shuru se 3 items
print("scores[4:]  ->", scores[4:])    # index 4 se end tak
print("scores[-3:] ->", scores[-3:])   # last 3 items
print("scores[:]   ->", scores[:])     # poori list (copy)
````

#### Slicing with Step 🪜
Teesra number `step` = kitne-kitne ke jump se items uthane hain. **Negative step = ulta chalo!**


````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

print("scores[::2]   ->", scores[::2])    # har 2nd item (0,2,4,6)
print("scores[1::2]  ->", scores[1::2])   # index 1 se, har 2nd item (1,3,5,7)
print("scores[::3]   ->", scores[::3])    # har 3rd item
print("scores[::-1]  ->", scores[::-1])   # POORI list ulti (reverse)
print("scores[5:1:-1]->", scores[5:1:-1]) # index 5 se peeche ki taraf, index 1 se pehle tak
````

```
          col 0  col 1  col 2

row 1  [   201,   202,   203 ]
row 2  [   301,   302,   303 ]
```
Pehla index = **row**, doosra index = **column** → `grid[row][col]`


````python
seating_grid = [
    [101, 102, 103],   # row 0
    [201, 202, 203],   # row 1
    [301, 302, 303],   # row 2
]

print("Poora grid       :", seating_grid)
print("Row 1 (poori)    :", seating_grid[1])
print("Row 1, Col 2     :", seating_grid[1][2])
print("Last row, last col:", seating_grid[-1][-1])

# Nested list ka ek aur example: [naam, marks] ke jodey
records = [["Aarav", 94], ["Bhavna", 91], ["Chirag", 88]]
print("\nBhavna ke marks:", records[1][1])

# Nested list ko row-by-row print karna
for row in seating_grid:
    print(row)
````

### 4.7 List Operators: `+`, `*`, `in` ➕✖️🔍

## ➕ Group 1: Add Elements


### 1️⃣ `append()` — End mein ek item jodo
**Syntax:** `list.append(item)` • Sirf **ek** item, hamesha **last** mein.


````python
participants = ["Aarav", "Bhavna"]
print("Pehle   :", participants)

participants.append("Chirag")
print("append('Chirag') ke baad:", participants)
````

### 2️⃣ `extend()` — Ek saath kai items jodo
**Syntax:** `list.extend(iterable)` • Doosri list ke **saare items** end mein jud jaate hain.

#### ⚠️ #1 Confusion: `append()` vs `extend()`


````python
a = [1, 2, 3]
a.append([4, 5])        # poori list ko EK item bana ke daal diya
print("append([4,5]) ->", a)

b = [1, 2, 3]
b.extend([4, 5])        # 4 aur 5 alag-alag items ban ke jude
print("extend([4,5]) ->", b)

participants = ["Aarav", "Bhavna"]
participants.extend(["Chirag", "Deepika"])
print("\nParticipants:", participants)
````

### 3️⃣ `insert()` — Kisi bhi jagah item daalo

````python
participants = ["Aarav", "Chirag", "Deepika"]
print("Pehle:", participants)

participants.insert(1, "Bhavna")       # index 1 par Bhavna
print("insert(1, 'Bhavna') ->", participants)

participants.insert(0, "Chief Guest")  # index 0 = sabse aage
print("insert(0, 'Chief Guest') ->", participants)
````

## ➖ Group 2: Remove Elements


### 4️⃣ `remove()` — Value batao, wo hat jayegi
**Syntax:** `list.remove(value)` • Sirf **pehli matching value** hatati hai. Value na mile toh `ValueError`.


````python
entries = ["Aarav", "Chirag", "Aarav", "Deepika"]
print("Pehle:", entries)

entries.remove("Aarav")       # sirf PEHLA Aarav hata
print("remove('Aarav') ->", entries)

try:
    entries.remove("Zoya")    # ye list mein hai hi nahi
except ValueError as e:
    print("⚠️ ValueError:", e)
````

### 5️⃣ `pop()` — Index batao, item nikalo (aur wapas bhi pao)
**Syntax:** `list.pop()` → last item • `list.pop(index)` → us index ka item
`pop()` item ko list se hatata hai **aur return bhi karta hai** (aap use kisi variable mein rakh sakte ho).

> 🆚 `remove(value)` → **value** se hatao, kuch return nahi karta • `pop(index)` → **position** se hatao, item return karta hai


````python
numbers = [10, 20, 30, 40]
print("Pehle:", numbers)

last = numbers.pop()          # last item
print("pop() ne diya:", last, "| list:", numbers)

second = numbers.pop(1)       # index 1 wala item
print("pop(1) ne diya:", second, "| list:", numbers)
````

## 🔀 Group 4: Arrange & Copy


### 9️⃣ `sort()` — List ko order mein lagao
**Syntax:** `list.sort()` (ascending) • `list.sort(reverse=True)` (descending)

> ⚠️ **Dhyan do:** `sort()` **original list ko hi badal deta hai** (in-place) aur khud `None` return karta hai!


````python
scores = [78, 92, 45, 88, 60]
print("Pehle:", scores)

scores.sort()
print("sort()               ->", scores)

scores.sort(reverse=True)
print("sort(reverse=True)   ->", scores)

# ⚠️ Common galti: sort() ka result kisi variable mein mat rakho
result = [3, 1, 2].sort()
print("\nresult = [3,1,2].sort() ->", result, "(None aata hai!)")

# Strings bhi sort hote hain (alphabetical)
names = ["Deepika", "Aarav", "Chirag", "Bhavna"]
names.sort()
print("Names sorted:", names)
````

### 🔟 `reverse()` — Order ulta karo
**Syntax:** `list.reverse()` • Sort **nahi** karta — bas mojooda order ko **ulta** kar deta hai (in-place).

> 🆚 `sort(reverse=True)` → **bade se chhota** order • `reverse()` → jaisa tha **bas ulta**


````python
numbers = [10, 40, 20, 30]
print("Pehle:", numbers)

numbers.reverse()
print("reverse() ->", numbers)       # [30, 20, 40, 10] — sorted nahi, sirf ulta!
````

### 1️⃣1️⃣ `copy()` — Original ka duplicate banao
**Syntax:** `new_list = old_list.copy()`

#### 🪤 Reference Trap: `b = a` copy **nahi** hai!
`b = a` sirf **ek nayi naam-patti** lagata hai — dono naam **ek hi list** ko point karte hain.


````python
original = [10, 20, 30]

alias = original            # ❌ copy nahi, sirf ek aur naam
alias.append(99)
print("alias se append kiya ->", alias)
print("original ka haal     ->", original, "   <- original bhi badal gayi!")
print("Same list hain?", alias is original)

print("-" * 45)

original = [10, 20, 30]
backup = original.copy()    # ✅ asli copy
backup.append(99)
print("backup se append kiya ->", backup)
print("original ka haal      ->", original, "   <- original safe!")
print("Same list hain?", backup is original)
````

# ⚡ Chapter 6: List Comprehensions — Result Day

### 🎬 Story
Utkarsh ka **Result Day**! 200 students ke marks aaye hain. Bhavna Didi bolti hain:
*"Sabko +5 grace marks do, fail walon ko alag nikalo, aur sabka Pass/Fail likh do. Jaldi!"*

Aarav loop likhne baitha... 5 lines, phir 5 lines, phir 5 lines. 😵
Tab Chirag bola: *"Bhai, Python mein ek **one-line magic** hota hai — **List Comprehension**!"*

### 🔁 Forenoon Refresher: `for` loop se Comprehension tak
List comprehension ko samajhne ke liye `for` loop ka idea chahiye: **list ke har item par ek kaam dohrao.**


````python
names = ["Aarav", "Bhavna", "Chirag"]

for name in names:               # har baar 'name' mein agla item aata hai
    print("Hello,", name)

print()
for i in range(1, 6):            # range(1, 6) -> 1,2,3,4,5 (6 nahi!)
    print("Number:", i)
````

### 6.1 List Comprehension Kya Hai?
Loop + `append()` ko **ek hi line** mein likhne ka compact tarika.

```
[  expression   for   item   in   iterable  ]
   ↑ kya banana hai     ↑ har item   ↑ kahan se
```

Chalo, pehle **purane tarike** se phir **naye tarike** se dekhte hain — marks par **+5 grace marks**:


````python
scores = [78, 92, 45, 88, 60]

# 🐢 Traditional tarika (4 lines)
graced_old = []
for m in scores:
    graced_old.append(m + 5)
print("Loop se        :", graced_old)

# 🚀 List comprehension (1 line)
graced_new = [m + 5 for m in scores]
print("Comprehension  :", graced_new)

print("Dono same?", graced_old == graced_new)
````

#### Expression Ke Saath Khelna 🎨
`expression` kuch bhi ho sakta hai — maths, string methods, function call...


````python
numbers = [1, 2, 3, 4, 5]
print("Squares   :", [n * n for n in numbers])
print("Doubles   :", [n * 2 for n in numbers])

names = ["aarav", "bhavna", "chirag"]
print("Uppercase :", [n.upper() for n in names])
print("Lengths   :", [len(n) for n in names])

# range() ke saath: 1 se 10 tak ke cubes
print("Cubes 1-5 :", [n ** 3 for n in range(1, 6)])
````

### 6.2 Comprehension with Condition (Filter) — `if` end mein 🔍
**Syntax:** `[expression for item in iterable if condition]`
Sirf wahi items aayenge jinke liye condition `True` hai. Bhavna Didi ko **sirf pass students** chahiye (marks ≥ 40).


````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

passed = [m for m in scores if m >= 40]
print("Passed        :", passed)

distinction = [m for m in scores if m >= 75]
print("Distinction   :", distinction)

evens = [n for n in range(1, 11) if n % 2 == 0]
print("Even (1-10)   :", evens)

print("Kitne fail?   :", len([m for m in scores if m < 40]))
````

### 6.3 Comprehension with `if-else` — Ye **Aage** Aata Hai! 🔀
Jab **har item ko rakhna hai par alag-alag value dena hai**, toh `if-else` **expression wale hisse** mein (aage) likhte hain.

| Kaam | Syntax | Position |
|---|---|---|
| **Filter** (kuch items hatao) | `[x for x in data if cond]` | `if` **end** mein |
| **Transform** (sabko value do) | `[A if cond else B for x in data]` | `if-else` **aage** |


````python
scores = [78, 92, 45, 88, 60, 95, 33, 82]

status = ["Pass" if m >= 40 else "Fail" for m in scores]
print("Status :", status)

# Cap: 90 se upar ke marks ko 90 rakho
capped = [90 if m > 90 else m for m in scores]
print("Capped :", capped)

# Even/Odd tagging
tags = ["Even" if n % 2 == 0 else "Odd" for n in range(1, 7)]
print("Tags   :", tags)
````

### 6.4 Nested Comprehension — Loop ke Andar Loop 🧩
Seating grid ko **flatten** karna (2D → 1D) ya **table** banana.

```
[ x for row in grid for x in row ]
       └─ bahari loop ─┘ └ andar wala loop ┘      <- jaisa loop likhte ho, waisa hi order!
```


````python
seating_grid = [
    [101, 102, 103],
    [201, 202, 203],
    [301, 302, 303],
]

# 1. Flatten: 2D grid -> 1D list
flat = [seat for row in seating_grid for seat in row]
print("Flattened      :", flat)

# 2. Flatten + filter: sirf even seat numbers
even_seats = [seat for row in seating_grid for seat in row if seat % 2 == 0]
print("Even seats     :", even_seats)

# 3. Nayi 2D list banana: 3x3 multiplication table
table = [[r * c for c in range(1, 4)] for r in range(1, 4)]
print("3x3 table      :", table)
for row in table:
    print(row)
````

### 6.5 `sorted()` — Naya Sorted List, Original Safe ✅
Chapter 5 mein `sort()` seekha tha jo original ko badal deta hai. `sorted()` **nayi list** deta hai aur original ko chhoota bhi nahi. Kisi bhi iterable par chalta hai.

| | `list.sort()` | `sorted(list)` |
|---|---|---|
| Original badalta hai? | ✅ Haan | ❌ Nahi |
| Return karta hai | `None` | Nayi sorted list |
| Kahan chalta hai | Sirf lists par | List, string, tuple... sab par |


````python
scores = [78, 92, 45, 88, 60]

print("sorted(scores)               ->", sorted(scores))
print("sorted(scores, reverse=True) ->", sorted(scores, reverse=True))
print("Original abhi bhi            ->", scores)

# Strings par bhi chalta hai
names = ["Deepika", "Aarav", "Chirag", "Bhavna"]
print("sorted(names)                ->", sorted(names))
print("sorted(names, key=len)       ->", sorted(names, key=len))   # lambai ke hisaab se
print("sorted('python')             ->", sorted("python"))        # string ke letters
````

### 6.6 `enumerate()` — Index + Value Saath Mein 🔢
Loop mein **item aur uska number dono** chahiye? `enumerate()` use karo. `start=1` se ginti 1 se shuru hoti hai.


````python
names = ["Aarav", "Bhavna", "Chirag"]

print("Default (0 se):")
for index, name in enumerate(names):
    print(index, "->", name)

print("\nstart=1 ke saath (Roll No.):")
for roll, name in enumerate(names, start=1):
    print(f"Roll No. {roll}: {name}")

print("\nList mein dekho:", list(enumerate(names, start=1)))
````

### 6.7 `zip()` — Do Lists ko Jodi Mein Silna 🤝
Do (ya zyada) lists ke **same position wale items** ko pair bana deta hai.
- Result ek zip object hota hai — dekhne ke liye `list(...)` mein daalo.
- Agar lists ki lambai alag ho toh **chhoti list ke hisaab se ruk jaata hai**.


````python
names = ["Aarav", "Bhavna", "Chirag"]
marks = [94, 91, 88]

pairs = list(zip(names, marks))
print("zip pairs:", pairs)

for name, mark in zip(names, marks):
    print(f"{name} ne {mark} marks laaye")

# Alag lambai: chhoti list tak hi jodi banegi
print("\nUnequal:", list(zip([1, 2, 3, 4], ["a", "b"])))

# Dictionary banane ka shortcut
print("Dict    :", dict(zip(names, marks)))

# Unzip: jodiyon ko wapas alag karna
n, m = zip(*pairs)
print("Unzip   :", n, m)
````

### 🚀 Boss Move: Teeno Saath Mein! Merit List
`zip()` se jodi banao → `sorted()` se top marks upar → `enumerate()` se rank do.


````python
names = ["Aarav", "Bhavna", "Chirag", "Deepika"]
marks = [88, 94, 79, 91]

# (marks, name) ki jodi banai taaki sorted() marks ke hisaab se sort kare
ranked = sorted(zip(marks, names), reverse=True)

print("🏆 UTKARSH 2026 — MERIT LIST")
for rank, (mark, name) in enumerate(ranked, start=1):
    print(f"  Rank {rank}: {name:<8} {mark}")
````

## 🔠 Group 1: Case Conversion

| Method | Kaam | `"hello gla UNIVERSITY"` ->|

````python
text = "hello gla UNIVERSITY"

print("upper()      ->", text.upper())
print("lower()      ->", text.lower())
print("title()      ->", text.title())
print("capitalize() ->", text.capitalize())
print("swapcase()   ->", text.swapcase())

print("\nOriginal abhi bhi ->", text)   # immutable!

# Real use: case-insensitive comparison
typed = "YES"
print("typed.lower() == 'yes' ->", typed.lower() == "yes")
````

## 🧽 Group 2: Space Removal
### 6️⃣ `strip()` • 7️⃣ `lstrip()` • 8️⃣ `rstrip()`
- `strip()` → **dono taraf** ke spaces hatao
- `lstrip()` → sirf **left** (shuru) ke
- `rstrip()` → sirf **right** (end) ke

Bonus: bracket mein character dedo toh wo character hatayega (spaces nahi): `"***hi***".strip("*")`


````python
text = "   Python   "

print("Original  :", repr(text))     # repr() se spaces dikhte hain
print("strip()   :", repr(text.strip()))
print("lstrip()  :", repr(text.lstrip()))
print("rstrip()  :", repr(text.rstrip()))

print("\nCustom chars hatana:")
print("'***Hi***'.strip('*')  ->", "***Hi***".strip("*"))
print("'2026!!!'.rstrip('!')  ->", "2026!!!".rstrip("!"))

# strip sirf kinaare ke spaces hatata hai, beech ke nahi
print("'  a   b  '.strip()    ->", repr("  a   b  ".strip()))
````

## ✂️ Group 3: Split & Join
### 9️⃣ `split()` — String ko tukdon mein todo → **list** milti hai

### 🔟 `join()` — List ke items ko ek string mein jodo
`"separator".join(list)` → separator beech mein lagta hai. *(Dhyan do: list ke saare items **strings** hone chahiye.)*

> split ↔ join ek-doosre ke **ulte** hain.


````python
sentence = "I Love Python"
words = sentence.split()
print("split()       ->", words)

row = "Aarav,Bhavna,Chirag"
names = row.split(",")
print("split(',')    ->", names)
print("2 tukde bas   ->", "a-b-c-d".split("-", 1))    # maxsplit=1

print("\n--- join() ---")
print("' '.join   ->", " ".join(words))
print("', '.join  ->", ", ".join(names))
print("'-'.join   ->", "-".join(["2026", "03", "15"]))
print("''.join    ->", "".join(["P", "y", "t", "h", "o", "n"]))

# Trick: extra spaces ek saath saaf karna
messy = "chirag    rao"
print("\nMessy fix ->", " ".join(messy.split()))
````

### 1️⃣9️⃣ `center(width, fillchar)` — Beech mein lagao
Text ko `width` jitni chaudi jagah ke **center** mein rakhta hai; baaki jagah `fillchar` (default space) se bharta hai.


````python
title = "CERTIFICATE"

print("[" + title.center(21) + "]")
print(title.center(21, "*"))
print(title.center(21, "="))

print("\n" + "=" * 30)
print("UTKARSH 2026".center(30))
print("=" * 30)
````

### 2️⃣0️⃣ `zfill(width)` — Aage zeros lagao
String ke **left** mein `0` bharta hai jab tak `width` poori na ho. *(Number ko pehle `str()` se string banana padega.)*


````python
print("'42'.zfill(6)       ->", "42".zfill(6))
print("'7'.zfill(3)        ->", "7".zfill(3))
print("'-5'.zfill(4)       ->", "-5".zfill(4))        # minus sign aage hi rehta hai

roll = 42
print("Roll no. ->", "GLA2026-" + str(roll).zfill(6))
````

### 2️⃣1️⃣ `isalnum()` — Sirf letters aur digits?
Alphabets **ya** digits (ya dono) ho toh `True`. Space, `@`, `-`, `_` jaisa koi bhi symbol aaya toh `False`.

### 2️⃣2️⃣ `isspace()` — Sirf whitespace?
Sirf spaces / tabs / new-lines ho toh `True`. **Khaali string `""` par `False`!**


````python
print("--- isalnum() ---")
print("'Aarav123'.isalnum()  ->", "Aarav123".isalnum())
print("'Aarav 123'.isalnum() ->", "Aarav 123".isalnum())   # space hai
print("'aarav_01'.isalnum()  ->", "aarav_01".isalnum())    # underscore hai

print("\n--- isspace() ---")
print("'   '.isspace()  ->", "   ".isspace())
print("'  a '.isspace() ->", "  a ".isspace())
print("''.isspace()     ->", "".isspace())          # khaali string: False!
print("'\\t\\n'.isspace()->", "\t\n".isspace())
````

### 2️⃣3️⃣ `partition(separator)` — 3 Hisson Mein Todo
Separator ke **pehle match** par string ko **tuple of 3** mein todta hai: `(pehle ka hissa, separator, baad ka hissa)`.
Separator na mile toh `(poori string, '', '')`.

> 🆚 `split()` → list (kai tukde) • `partition()` → hamesha **exactly 3** tukde (pehle separator par)


````python
setting = "name=Aarav Sharma"

before, sep, after = setting.partition("=")
print("Tuple   :", setting.partition("="))
print("Key     :", before)
print("Sep     :", sep)
print("Value   :", after)

print("\nSeparator nahi mila:", "hello".partition("="))
print("Sirf PEHLE '-' par todta hai:", "Hello-Python-World".partition("-"))
````

## 🎨 String Formatting — Text mein Values Daalna
Message banana hai: *"Aarav ne 94.5% marks laaye"* — name aur marks variables mein hain. Teen tarike hain:

| Tarika | Syntax | Kab |
|---|---|---|
| **f-string** ⭐ | `f"Hello {name}"` | **Modern & best** — hamesha ise use karo |
| `.format()` | `"Hello {}".format(name)` | Purane code mein milta hai |
| `%` operator | `"Hello %s" % name` | Bahut purana, bas pehchaan ke liye |


````python
name = "Aarav"
marks = 94.5
rank = 1

# 1. f-string (best!)
print(f"1. {name} ne {marks} marks laaye aur Rank {rank} paaya!")

# 2. .format()
print("2. {} ne {} marks laaye aur Rank {} paaya!".format(name, marks, rank))
print("2b. {n} ne {m} marks laaye".format(n=name, m=marks))     # naam se

# 3. % operator
print("3. %s ne %.1f marks laaye aur Rank %d paaya!" % (name, marks, rank))
````

### f-string Ki Power: `{value:format}` 💪
`:` ke baad **formatting rules** likhte hain:

| Format | Matlab | Example | Output |
|---|---|---|---|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:,` | Hazaar ka comma | `f"{1500000:,}"` | `1,500,000` |
| `:05d` | 5 width, aage zeros | `f"{42:05d}"` | `00042` |
| `:<10` | Left align (10 width) | `f"{'Hi':<10}"` | `Hi        ` |
| `:>10` | Right align | `f"{'Hi':>10}"` | `        Hi` |
| `:^10` | Center align | `f"{'Hi':^10}"` | `    Hi    ` |
| `:.1%` | Percentage | `f"{0.945:.1%}"` | `94.5%` |
| `{x = }` | Debug (naam + value) | `f"{x = }"` | `x = 5` |


````python
pi = 3.14159
prize = 1500000
x = 5

print(f"{pi:.2f}")                 # 3.14
print(f"{prize:,}")                # 1,500,000
print(f"{42:05d}")                 # 00042
print(f"[{'Hi':<10}]")             # left
print(f"[{'Hi':>10}]")             # right
print(f"[{'Hi':^10}]")             # center
print(f"{0.945:.1%}")              # percentage
print(f"{x = }")                   # debug
print(f"2 + 3 = {2 + 3}")          # {} ke andar expression bhi chalta hai
print(f"{'aarav'.upper()}")        # method call bhi chalta hai

# Mini table
print()
print(f"{'Name':<10}{'Marks':>8}")
print("-" * 18)
for n, m in [("Aarav", 94.5), ("Bhavna", 91.25), ("Chirag", 88)]:
    print(f"{n:<10}{m:>8.2f}")
````

# 🎬 Epilogue: Aarav ka Final Boss — Utkarsh Merit Certificates 🎓

Teeno tools — **lists, comprehensions, strings** — ek saath! Ye program messy registration data leta hai, saaf karta hai,
rank banata hai aur certificate-style output chhaapta hai. Ye **demo** hai — har line samjho aur dekho kaunsa chapter kahan use hua. 🔍


````python
raw_entries = [
    "  aarav   sharma , cse , 94.5 ",
    "BHAVNA verma,it,91.25",
    "  chirag    rao , ece , 88 ",
    "deepika SINGH ,  me,79",
]

# 1) Clean + parse har entry (str methods + list comprehension)
students = []
for entry in raw_entries:
    name, branch, marks = [part.strip() for part in entry.split(",")]    # Ch5: split, strip + Ch3: comprehension
    name = " ".join(name.split()).title()                                # Ch5: split/join/title
    students.append((float(marks), name, branch.upper()))                # float(): text -> number

# 2) Rank banao (Ch3: sorted + enumerate)
ranked = sorted(students, reverse=True)

# 3) Certificate-style output (Ch6: center, zfill, f-string)
print("=" * 44)
print("UTKARSH 2026 - MERIT LIST".center(44))
print("=" * 44)
for rank, (marks, name, branch) in enumerate(ranked, start=1):
    cert_no = "CERT-" + str(rank).zfill(4)
    print(f"{cert_no}  {name:<16} {branch:<4} {marks:>6.2f}")
print("=" * 44)

top = ranked[0]
print(f"🏆 Topper: {top[1]} ({top[2]}) with {top[0]}%")
````

2. **List mutable hai** (in-place badalti hai), **string immutable hai** (hamesha nayi string banti hai).
3. **Jab bhi atko — `print()` lagao, `type()` check karo, indentation dhyan se dekho, aur error message ko padho!**
