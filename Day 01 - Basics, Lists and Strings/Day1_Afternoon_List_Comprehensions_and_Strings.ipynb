# 📦 Module 1: Lists from Ground Zero (Basics, Indexing, Slicing & Operations)

### What is a List in Python?
Think of a list like a **Student Backpack** or a **Shopping Cart**. 
Instead of creating 10 different variables (`item1`, `item2`, `item3`...), a list lets you store multiple items inside a single variable using **square brackets `[ ]`**.

```
             Index 0       Index 1        Index 2         Index 3
  backpack = [ "Notebook",  "Pen",        "Calculator",   "ID Card" ]
  Backward:   -4            -3            -2              -1
```

### Key Properties of Python Lists:
1. **Ordered**: Items stay in the exact position you put them in.
2. **Heterogeneous**: A list can hold mixed types (e.g. `["Aarav", 101, 94.5, True]`).
3. **Mutable**: You can add, change, or delete items anytime!


### 1.1 List Indexing: Positive (0 to $N-1$) and Negative (-1 to $-N$)
**Trainer Demo**: Access items from our master `backpack` list from both ends.

### 1.2 List Slicing with Step: `list[start : stop : step]`
**Concept**: Slicing lets you take a subset of items.
- `start`: Beginning index (included).
- `stop` : Ending index (**excluded!** Stops right before this index).
- `step` : Jump count (default is 1).
- `list[::-1]`: Reverses the list!

**Trainer Demo**: Slice our master `scores` list.


````python
# 👨‍🏫 TRAINER DEMO: Slicing on master scores
# scores = [78, 92, 45, 88, 60, 95, 33, 82]

print("First 3 scores [0:3]     :", scores[0:3])    # [78, 92, 45]
print("From index 4 onward [4:] :", scores[4:])     # [60, 95, 33, 82]
print("Every 2nd score [::2]    :", scores[::2])    # [78, 45, 60, 33]
print("Reversed scores [::-1]   :", scores[::-1])   # Backwards!
````

### Method 1 & 2: `append()` vs `extend()` (The #1 List Confusion!)
- `append(item)`: Adds **ONE single item** to the very end of the list.
- `extend(iterable)`: Takes all items from another list and adds them **individually** to the end!

```
  ⚠️ THE TRAP:
  backpack.append(["A", "B"]) -> Adds the list AS A SUB-LIST: ["Notebook", ..., ["A", "B"]]
  backpack.extend(["A", "B"]) -> Flattens and adds items: ["Notebook", ..., "A", "B"]
```

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: append() vs extend()
demo_list = backpack.copy()

# 1. append adds a single element:
demo_list.append("Water Bottle")
print("1. After append('Water Bottle'):\n  ", demo_list)

# 2. extend adds multiple elements:
demo_list.extend(["Eraser", "Ruler"])
print("2. After extend(['Eraser', 'Ruler']):\n  ", demo_list)
````

### Method 4 & 5: `remove()` vs `pop()`
- `remove(value)`: Searches for the **first occurrence** of `value` and deletes it. (Does not return anything). Raises `ValueError` if absent!
- `pop(index)`: Deletes the item at `index` (default: last item) **AND returns it to you**!

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: remove() vs pop()
demo_list = backpack.copy()

# 4. remove by value:
demo_list.remove("Pen")
print("4. After remove('Pen'):\n  ", demo_list)

# 5. pop removes from end and hands it back:
removed_item = demo_list.pop()
print(f"5. Popped item: '{removed_item}'")
print("   Remaining list:\n  ", demo_list)

# pop(0) removes the first item:
first_item = demo_list.pop(0)
print(f"   Popped first item: '{first_item}'")
print("   Final list:\n  ", demo_list)
````

### Method 11: `copy()` (The Reference Trap)
- In Python, `list_b = list_a` does **NOT** copy the list! Both variables point to the **same memory box**. Changing `b` will accidentally change `a`!
- Use `list_b = list_a.copy()` to create an independent backup!

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: The Reference Trap vs copy()
original = [10, 20, 30]

# WRONG: Alias (Points to same memory)
alias = original
alias.append(999)
print("After modifying alias, original changed too!:", original)

# RIGHT: Independent copy
original = [10, 20, 30]
safe_backup = original.copy()
safe_backup.append(999)
print("With .copy(), original is safe!:", original)
print("Backup has the new item        :", safe_backup)
````

### 3.3 The Ternary Trap: Filtering (`if`) vs Substitution (`if-else`)
**Concept**:
- **Filtering (Discard items)**: Put `if` at the **END**: `[x for x in nums if x > 40]`
- **Substituting (Keep all items, give different label)**: Put `if-else` at the **FRONT**: `['PASS' if x >= 40 else 'FAIL' for x in nums]`

**Trainer Demo**: Label each score as `'PASS'` or `'FAIL'`.

### 4.2 Indexing, Boundary-Safe Slicing & Immutability
- Indexing: `0` is first, `-1` is last.
- Slicing: `s[start : stop : step]` (Stop is **excluded**!). `s[::-1]` reverses!
- Boundary safety: Slicing beyond length `s[4:999]` never crashes.
- Immutability: Strings **cannot be edited in-place**. Reassigning creates a NEW string object (`id()` changes).

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: Slicing & Immutability
campus = "GLA_UNIVERSITY"

print("First char [0]   :", campus[0])
print("Last char [-1]   :", campus[-1])
print("Slice [0:3]      :", campus[0:3])    # 'GLA'
print("Reversed [::-1]  :", campus[::-1])   # 'YTISREVINU_ALG'
print("Safe slice [4:99]:", repr(campus[4:99]))

# Immutability proof:
word = "GLA"
original_id = id(word)
word = word + " University"
print(f"Old id: {original_id} | New id: {id(word)} (Notice ID changed!)")
````

### Methods 1 to 5: Case Conversion (`upper`, `lower`, `title`, `capitalize`, `swapcase`)
- `upper()`: All uppercase.
- `lower()`: All lowercase.
- `title()`: Capitalizes every word (*Trap: "it's" -> "It'S"*).
- `capitalize()`: Only first character of string capitalized.
- `swapcase()`: Inverts uppercase and lowercase.

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: Case Conversion on Master Banner
clean = banner.strip(" *")
print("1. upper()      :", repr(clean.upper()))
print("2. lower()      :", repr(clean.lower()))
print("3. title()      :", repr(clean.title()))
print("4. capitalize() :", repr(clean.capitalize()))
print("5. swapcase()   :", repr(clean.swapcase()))
````

### Methods 12 to 14: `find()`, `index()`, `count()`

- `count(sub)`: Number of non-overlapping occurrences.

### Methods 15 to 18: `startswith()`, `endswith()`, `isdigit()`, `isalpha()`
- `startswith(prefix)`: Validates beginning of string.
- `endswith(suffix)`: Validates ending of string (accepts tuple of extensions!).
- `isdigit()`: `True` if all characters are digits 0-9. (*False for negatives!*)
- `isalpha()`: `True` if all characters are letters. (*False if spaces exist!*)

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: Affix & Type Checks
clean = banner.strip(" *")

print("15. startswith('welcome') :", clean.startswith("welcome"))
print("16. endswith('batch-2026'):", clean.endswith("batch-2026"))
print("17. '2026'.isdigit()      :", "2026".isdigit())
print("    '-2026'.isdigit()     :", "-2026".isdigit()) # False!
print("18. 'GLA'.isalpha()        :", "GLA".isalpha())
print("    'GLA Univ'.isalpha()  :", "GLA Univ".isalpha()) # False!
````

# 🎨 Module 6: 5 Other String Methods & Modern Formatting
Methods: `center()`, `zfill()`, `isalnum()`, `isspace()`, `partition()`, and **f-strings**!


### Methods 19 to 23: `center()`, `zfill()`, `isalnum()`, `isspace()`, `partition()`
- `center(width, fillchar)`: Centers text padded with characters.
- `zfill(width)`: Pads number strings with leading zeros (sign-aware!).
- `isalnum()`: Letters and/or numbers (no spaces/symbols).
- `isspace()`: All whitespace.
- `partition(sep)`: Guaranteed 3-tuple: `(before, sep, after)`.

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: 5 Other Methods
print("19. center(30, '=') :", repr("GLA".center(30, "=")))
print("20. '42'.zfill(6)   :", "42".zfill(6))
print("    '-42'.zfill(6)  :", "-42".zfill(6)) # Sign preserved!
print("21. 'GLA2026'.isalnum():", "GLA2026".isalnum())
print("22. '   \\t\\n '.isspace() :", "   \t\n ".isspace())

# 23. partition:
user, at, domain = "student@gla.ac.in".partition("@")
print(f"23. partition('@') -> User='{user}', Domain='{domain}'")
````

### Method 24: Modern String Formatting (f-strings)
**Concept**: In Python 3.6+, place an `f` before quotes and insert variables in `{curly_brackets}`!
- `{cgpa:.2f}`: 2 decimal places.
- `{fee:,}`: Thousands comma.
- `{name:<15}`: Left alignment.

**Trainer Demo**:


````python
# 👨‍🏫 TRAINER DEMO: Modern F-Strings
st_name = "Priya Sharma"
st_cgpa = 9.4582
fee = 145000

print(f"Student: {st_name:<16} | CGPA: {st_cgpa:.2f} | Fee: ₹{fee:,}")
````
