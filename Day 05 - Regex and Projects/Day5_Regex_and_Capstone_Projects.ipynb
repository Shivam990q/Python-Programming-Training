# 🔎 Chapter 1: Regular Expressions — Pattern ka Jaadu

### 🎬 Story
Aarav ke saamne 2,000 lines thi, jaise:

> `Aarav Sharma, aarav@gla.ac.in, 9876543210, joined 12-03-2026`

Usne phone number nikalne ke liye `"9876543210" in line` likha... par **har line ka number alag hai!** Use number nahi, sirf **shakal** pata hai: *"10 digits ek saath"*.

Bhavna Didi: *"**Regular Expression (regex)** = text ki **shakal** ka description. Tu bata: `\d{10}` — matlab '10 digits' — aur Python **saari matching jagahein** dhundh dega!"* 🔍

### 🆚 `str` Methods vs Regex
| Sawaal | `str` method | Regex |
|---|:-:|:-:|
| Kya `"Utkarsh"` text mein hai? | ✅ `"Utkarsh" in text` | ✅ |
| **Koi bhi** 10-digit number? | ❌ | ✅ `\d{10}` |
| Email jaisa kuch? | ❌ | ✅ |
| Tarikh `DD-MM-YYYY` ki shakal? | ❌ | ✅ |

### 📦 `re` Module — Pehla Kadam
```python
import re                                   # built-in hai, install nahi karna
m = re.search(r"\d{10}", text)              # pattern, text
```
- `re.search(pattern, text)` text mein **pehli matching jagah** dhundhta hai.
- Mila toh **Match object** milta hai; **na mile toh `None`**.

### 🔬 Match Object ke Andar
| Kya chahiye | Code |
|---|---|
| Jo text mila | `m.group()` |
| Kahan se shuru | `m.start()` |
| Kahan par khatam (is index se **pehle**) | `m.end()` |
| Dono ek saath | `m.span()` → `(start, end)` |

````python
import re

text = "Aarav Sharma, aarav@gla.ac.in, 9876543210, joined 12-03-2026"

# ❌ str method: number PEHLE se pata hona chahiye
print("str method  :", "9876543210" in text, " <- par agar number pata hi na ho toh?")

# ✅ regex: sirf SHAKAL batao — '10 digits ek saath' (\d = digit)
m = re.search(r"\d{10}", text)
print("regex       :", m.group(), " <- shakal se mil gaya!")

# Match object ke andar
print("\nm.group()   :", m.group())
print("m.start()   :", m.start())
print("m.end()     :", m.end())
print("m.span()    :", m.span())
print("text[m.start():m.end()] :", text[m.start():m.end()])
````

### 1.1 Raw Strings — `r"..."` Hamesha! 🛡️
Regex mein **backslash `\`** bahut aata hai (`\d`, `\s`, `\.`). Par Python ke **normal string** mein bhi `\` ka apna matlab hai (`\n` = newline, `\t` = tab). Dono takra jaate hain!

| Likha | Python kya samajhta hai |
|---|---|
| `"\n"` | **1 character** — newline |
| `r"\n"` | **2 characters** — `\` aur `n` (jaisa likha waisa) |
| `"\d"` | ⚠️ Galat escape — warning (`SyntaxWarning`) |
| `r"\d"` | ✅ Seedha regex tak pahunchta hai |

> 🧠 **Golden Rule:** regex pattern **hamesha raw string** mein likho: `r"..."`. (`r` = *raw*)

````python
import re, warnings

print('len("\\n")  normal :', len("\n"), " <- sirf 1 character (newline)")
print('len(r"\\n") raw    :', len(r"\n"), " <- 2 characters: \\ aur n")
print('"\\\\d" == r"\\d"    :', "\\d" == r"\d", " <- normal mein \\\\ likhna padta, raw mein ek hi")

# Normal string mein \d likhoge toh Python complain karta hai
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    compile('pattern = "\\d+"', "<demo>", "exec")
print("\nNormal string mein \\d par warning aayi?", bool(caught), "->", caught[0].category.__name__ if caught else "")

# Raw string se pattern bilkul saaf
print("r'\\d+' se search :", re.search(r"\d+", "Room 204").group())
````

### 1.2 ⚠️ `None` ka Jaal — Hamesha Check Karo!
`re.search()` jab kuch **na mile** toh **`None`** deta hai. `None.group()` karoge toh **`AttributeError`** 💥. Isliye pehle `if m:` check karo.

````python
import re

for text in ["Room 204", "No digits here"]:
    m = re.search(r"\d+", text)
    if m:                                           # ✅ pehle check
        print(f"{text!r:18} -> mila: {m.group()}")
    else:
        print(f"{text!r:18} -> kuch nahi mila")

# ❌ Bina check ke
try:
    print(re.search(r"\d+", "No digits here").group())
except AttributeError as e:
    print("\n💥 Bina check:", type(e).__name__ + ":", e)
````

### 1.3 Regex Case-Sensitive Hota Hai 🔠
`re.search("fest", "Fest")` → **None** (`F` aur `f` alag). Case ignore karna ho toh **flag** do: `re.IGNORECASE` (chhota naam: `re.I`).

````python
import re

text = "Utkarsh FEST 2026"
print("search('fest')          :", re.search("fest", text))
print("search('fest', re.I)    :", re.search("fest", text, re.I))
print("m.group() asli text deta hai:", re.search("fest", text, re.IGNORECASE).group())
````

# 🧩 Chapter 2: Metacharacters — Jaadui Symbols

### 🎬 Story
Bhavna Didi ne Aarav ko ek **jaadui pitaara** diya jisme **12 jaadui symbols** the. *"Inme se har ek apna **normal matlab bhool jaata hai** aur kuch khaas kaam karta hai. Inhe **metacharacters** kehte hain. Aaj ye 12 yaad ho gaye, toh regex tera ghulaam!"* ✨

Normal letters (`a`, `Z`, `5`) **khud ko hi match** karte hain: `r"cat"` sirf `"cat"` dhundhta hai. Par metacharacters **shakal** batate hain.

### 🔣 Metacharacters ki List
| Symbol | Matlab | Example → kya match hota hai |
|:-:|---|---|
| **`.`** | **Koi bhi ek** character (newline chhodke) | `c.t` → `cat`, `cut`, `c9t`, `c t` |
| **`^`** | String ka **shuru** | `^Hello` → `Hello world` ✅, `Say Hello` ❌ |
| **`$`** | String ka **aakhir** | `world$` → `Hello world` ✅ |
| **`*`** | Pichhla **0 ya zyada** baar | `ab*c` → `ac`, `abc`, `abbbc` |
| **`+`** | Pichhla **1 ya zyada** baar | `ab+c` → `abc`, `abbc` (`ac` ❌) |
| **`?`** | Pichhla **0 ya 1** baar (optional) | `colou?r` → `color`, `colour` |
| **`{n}`** | **Ek dum `n`** baar | `[0-9]{3}` → `204` |
| **`{n,m}`** | **`n` se `m`** baar | `a{2,4}` → `aa`, `aaa`, `aaaa` |
| **`[ ]`** | **Set** — inme se koi ek | `[aeiou]`, `[0-9]`, `[A-Z]` |
| **`\`** | **Escape** — special ko normal banao (`\.`) | `3\.14` |
| **`( )`** | **Group** (Chapter 6 mein!) | `(ab)+` |

> ➕ Ek aur: **`|`** (pipe) = **ya (OR)** — `cat|dog` matlab `cat` **ya** `dog`.

> 🧠 **Abhi `\d` nahi!** Digit ke liye hum `[0-9]` likhenge. Chapter 3 mein `\d` shortcut seekhenge.

Neeche ke cells mein ek chhota `check()` helper hai jo har text par pattern chalake dikhata hai.

````python
import re

def check(pattern, texts, how="search"):
    """Har text par pattern chalao aur dikhao kya mila."""
    func = {"search": re.search, "fullmatch": re.fullmatch}[how]
    print(f"Pattern: r'{pattern}'  ({how})")
    for t in texts:
        m = func(pattern, t)
        result = repr(m.group()) if m else "no match"
        print(f"   {t!r:12} -> {result}")
    print()

print("═══ 1️⃣ DOT  . = koi bhi ek character ═══")
check(r"c.t", ["cat", "cut", "c9t", "c t", "ct", "coat"])

print("═══ 2️⃣ ANCHORS  ^ shuru  $ aakhir ═══")
check(r"^Hello", ["Hello world", "Say Hello"])
check(r"world$", ["Hello world", "world peace"])
check(r"^Hello world$", ["Hello world", "Hello world!"])
````

### 2.1 Quantifiers — `*` `+` `?` `{}` 🔢
Ye **pichhle character** (ya group) ko **kitni baar** repeat karna hai ye batate hain.

| Quantifier | Kitni baar | `ab?c` jaisa trick |
|:-:|---|---|
| `*` | **0 ya zyada** | `ab*c` — `b` ho bhi sakta hai, nahi bhi, kitne bhi |
| `+` | **1 ya zyada** | `ab+c` — `b` **kam se kam ek** |
| `?` | **0 ya 1** | `ab?c` — `b` **optional** |
| `{3}` | **ek dum 3** | `[0-9]{3}` |
| `{2,4}` | **2 se 4** | `a{2,4}` |
| `{2,}` | **2 ya zyada** | `a{2,}` |

> 🧠 **Yaad rakhne ki trick:** `*` = *star = kitne bhi (zero bhi)*, `+` = *plus = kam se kam ek*, `?` = *question = ho ya na ho*.

````python
import re

def check(pattern, texts, how="search"):
    """Har text par pattern chalao aur dikhao kya mila."""
    func = {"search": re.search, "fullmatch": re.fullmatch}[how]
    print(f"Pattern: r'{pattern}'  ({how})")
    for t in texts:
        m = func(pattern, t)
        result = repr(m.group()) if m else "no match"
        print(f"   {t!r:12} -> {result}")
    print()

STRINGS = ["ac", "abc", "abbc", "abbbc"]
print("═══ * (0 ya zyada) ═══")
check(r"^ab*c$", STRINGS)
print("═══ + (1 ya zyada) ═══")
check(r"^ab+c$", STRINGS)
print("═══ ? (0 ya 1) ═══")
check(r"^ab?c$", STRINGS)

print("═══ ? se spelling ki do shakalein ═══")
check(r"colou?r", ["color", "colour", "colouur"])

print("═══ {n} aur {n,m} ═══")
check(r"^[0-9]{3}$", ["204", "20", "2045"])
check(r"^a{2,4}$", ["a", "aa", "aaaa", "aaaaa"])
````

### 2.2 Set `[ ]` aur OR `|` 🎯
| Pattern | Matlab |
|---|---|
| `[aeiou]` | Koi ek **vowel** |
| `[a-z]` | Koi ek **chhota letter** (range) |
| `[A-Za-z]` | Koi bhi ek **letter** |
| `[0-9]` | Koi ek **digit** |
| `[^0-9]` | **Digit ke alawa** koi bhi (`^` set ke **andar** = "ulta / NOT") |
| `gr[ae]y` | `gray` **ya** `grey` |
| `cat\|dog` *(pipe)* | `cat` **ya** `dog` |

> ⚠️ `^` ke **do matlab**: set ke **bahar** = string ka shuru • set ke **andar pehla** = NOT.

````python
import re

def check(pattern, texts, how="search"):
    """Har text par pattern chalao aur dikhao kya mila."""
    func = {"search": re.search, "fullmatch": re.fullmatch}[how]
    print(f"Pattern: r'{pattern}'  ({how})")
    for t in texts:
        m = func(pattern, t)
        result = repr(m.group()) if m else "no match"
        print(f"   {t!r:12} -> {result}")
    print()

print("═══ [ ] set ═══")
check(r"gr[ae]y", ["gray", "grey", "groy"])
check(r"^[A-Z][a-z]+$", ["Utkarsh", "utkarsh", "UTKARSH", "U"])
check(r"[^0-9]", ["123", "12a", "abc"])

print("═══ | (OR) ═══")
check(r"cat|dog", ["my cat", "hot dog", "bird"])
check(r"^(Mr|Ms|Dr)\.", ["Dr. Sharma", "Ms. Verma", "Prof. Rao"])
````

### 2.3 Greedy vs Lazy — Zyada ya Kam Khane Wale 🍽️
`*`, `+`, `{}` **greedy (lalchi)** hote hain — jitna **zyada** ho sake utna match karte hain. Peeche **`?`** laga do toh **lazy (santushth)** — **kam se kam** match.

| | Pattern | `<b>bold</b> and <i>italic</i>` par |
|---|---|---|
| Greedy | `<.*>` | **Poora** `<b>...</i>` ek hi match |
| Lazy | `<.*?>` | Har tag **alag-alag**: `<b>`, `</b>`, `<i>`, `</i>` |

````python
import re

html = "<b>bold</b> and <i>italic</i>"

print("Greedy  <.*>  :", re.findall(r"<.*>", html))
print("Lazy    <.*?> :", re.findall(r"<.*?>", html))

# Greedy ka asli nuksaan: pehle quote se aakhri quote tak
text = 'name="Aarav" city="Meerut"'
print("\nGreedy  \".*\"  :", re.findall(r'".*"', text))
print("Lazy    \".*?\" :", re.findall(r'".*?"', text))
````

### 2.4 Escape `\` — Jaadui Symbol Ko Normal Banao 🔓
Agar tumhe **sachmuch `.` ya `$` ya `+`** dhundhna ho, toh peeche `\` lagao:
| Chahiye | Pattern |
|---|---|
| Literal dot `.` | `\.` |
| Literal `$` | `\$` |
| Literal `+` | `\+` |
| Literal `(` `)` | `\(` `\)` |
| Literal `\` | `\\` |

> 🛡️ **Shortcut:** `re.escape(text)` kisi bhi text ke saare special characters ko **apne aap escape** kar deta hai — jab user ka diya text pattern ban raha ho tab bahut kaam aata hai.

````python
import re

def check(pattern, texts, how="search"):
    """Har text par pattern chalao aur dikhao kya mila."""
    func = {"search": re.search, "fullmatch": re.fullmatch}[how]
    print(f"Pattern: r'{pattern}'  ({how})")
    for t in texts:
        m = func(pattern, t)
        result = repr(m.group()) if m else "no match"
        print(f"   {t!r:12} -> {result}")
    print()

print("═══ Bina escape: . koi bhi character ═══")
check(r"3.14", ["3.14", "3x14", "3 14"])

print("═══ Escape ke saath: \\. sirf asli dot ═══")
check(r"3\.14", ["3.14", "3x14", "3 14"])

print("═══ Price: Rs 250+  (plus asli hai) ═══")
check(r"250\+", ["250+", "2500", "250"])

import re
user_text = "a.b*c"
print("re.escape('a.b*c') ->", re.escape(user_text))
print("Escaped pattern se search:", re.search(re.escape(user_text), "xx a.b*c yy").group())
````

# 🔤 Chapter 3: Character Classes — `\d` `\w` `\s`

### 🎬 Story

*"Beta, regex ke paas **shortcuts** hain! `[0-9]` ki jagah sirf **`\d`**. Ye **character classes** hain — ek chhota sa code jo **characters ke poore group** ko represent karta hai."* ⚡

### 🔤 Teen Shortcuts aur Unke Ulte
| Shortcut | Matlab | Equivalent | **Ulta (NOT)** | Matlab |
|:-:|---|---|:-:|---|
| **`\d`** | **D**igit | `[0-9]` | **`\D`** | Digit **nahi** |
| **`\w`** | **W**ord character (letter, digit, `_`) | `[A-Za-z0-9_]` | **`\W`** | Word character **nahi** |
| **`\s`** | **S**pace-type (space, tab `\t`, newline `\n`) | `[ \t\n\r\f\v]` | **`\S`** | Whitespace **nahi** |

> 🧠 **Trick:** **Chhota akshar** = *hai* (`\d \w \s`) • **Bada akshar** = *nahi hai* (`\D \W \S`).

### 📏 Bonus: `\b` — Word Boundary
`\b` koi character **nahi** hai — ek **jagah (position)** hai: jahan **word ka shuru ya aakhir** hota hai (`\w` aur `\W` ke beech). Isse **poora word** match karte hain, kisi word ka **hissa** nahi.

| Pattern | `"This island is"` mein |
|---|---|
| `is` | **3 jagah**: `This`, `island`, `is` |
| `\bis\b` | Sirf **1**: akela `is` |

````python
import re

text = "Order 66: 3 pizzas at Rs 250 each, total Rs 750."

print("\\d+  (digits ke tukde)    :", re.findall(r"\d+", text))
print("\\D+  (digits ke alawa)    :", re.findall(r"\D+", text)[:3], "...")
print("\\d{2} (ek dum 2 digits)   :", re.findall(r"\d{2}", text))

nums = [int(n) for n in re.findall(r"\d+", text)]
print("\nInt list     :", nums)
print("Total        :", sum(nums))
print("Sabse bada   :", max(nums))
````

### 3.1 Sets `[ ]` ke Andar Shortcuts + Special Rules 🧰
| Pattern | Matlab |
|---|---|
| `[\d.]` | Digit **ya** dot (set ke andar `.` **normal dot** hota hai — escape nahi chahiye) |
| `[\w.-]` | Word char, dot, **ya** hyphen (`-` ko **aakhir mein** likho) |
| `[^\s]` | Whitespace ke alawa koi (`\S` jaisa) |
| `[A-Za-z]` | Sirf letters (digits aur `_` nahi) |

> ⚠️ **Set ke andar** sirf `^` (shuru mein), `-` (beech mein = range) aur `\` special hain. Baaki (`.`, `+`, `*`, `$`) **normal** hote hain.

````python
import re

print("[\\d.]+   :", re.findall(r"[\d.]+", "Prices 12.50 and 7.25 only"), " <- decimal numbers")
print("[A-Za-z]+ :", re.findall(r"[A-Za-z]+", "Room_204 Block-C"), " <- digits aur _ nahi aaye")
print("[^\\s]+   :", re.findall(r"[^\s]+", "Aarav  Bhavna"), " <- \\S+ jaisa")

# Email ki shakal ka jhalak: [\w.-]+ @ [\w.-]+
print("\nEmail   :", re.findall(r"[\w.-]+@[\w.-]+", "Mail aarav.sharma@gla.ac.in or ask@utkarsh.in now"))

# Punctuation nikalna
print("Punctuation:", re.findall(r"[.,:!?]", "Wow! Fee: Rs 250, ok."))

# Time, date shakalein \d se kitni saaf!
sched = "Quiz at 14:30 on 12-03-2026"
print("\nTime  \\d{2}:\\d{2}        :", re.search(r"\d{2}:\d{2}", sched).group())
print("Date  \\d{2}-\\d{2}-\\d{4}  :", re.search(r"\d{2}-\d{2}-\d{4}", sched).group())
````

### 3.2 ⚠️ Hindi / Unicode ka Gotcha: `\w` Devanagari Todta Hai!
Python 3 mein `\w` **Unicode letters** (jaise `é`, `ö`, `क`) bhi pehchanta hai. Par Hindi mein **matra (ा ि ी े) aur halant (्)** *letter nahi, "mark"* hote hain — isliye `\w` unpar **ruk jaata hai** aur shabd **tut jaata hai**!

| Text | `\w+` | `\S+` |
|---|---|---|
| `नमस्ते` | `['नमस', 'त']` ❌ | `['नमस्ते']` ✅ |

> 💡 **Fix:** Hindi / mixed text mein **`\S+`** (jo whitespace nahi hai) ya `[^\s]+` use karo. Sirf English chahiye toh `re.ASCII` flag (`re.A`) se `\w` ko `[A-Za-z0-9_]` tak seemit karo.

````python
import re

hindi = "नमस्ते Utkarsh"
print("\\w+ :", re.findall(r"\w+", hindi), " <- नमस्ते toot gaya!")
print("\\S+ :", re.findall(r"\S+", hindi), " <- sahi")

# é jaise letters: \w pehchanta hai; re.A se sirf ASCII
cafe = "Zoë drinks café 2026"
print("\n\\w+          :", re.findall(r"\w+", cafe))
print("\\w+ (re.A)   :", re.findall(r"\w+", cafe, re.A), " <- ë, é kat gaye")
````

# 🎯 Chapter 4: `match()` `search()` `findall()` `finditer()`

### 🎬 Story
Aarav ke paas ab pattern likhna to aa gaya... par **`re` ke functions** ka confusion tha:
- *"`match` aur `search` mein farq kya hai?"*
- *"Saare phone numbers kaise nikalun — sirf pehla nahi!"*

Bhavna Didi ne ek chhoti tarkeeb batayi: *"Socho ek **detective** hai. **`match`** sirf **darwaze (shuru)** par dekhta hai. **`search`** poora ghar chaan marta hai par **pehla suraag** milte hi ruk jaata hai. **`findall`** **sab suraag** ek list mein deta hai. **`finditer`** har suraag ke saath **jagah ka nakshaa** bhi deta hai."* 🕵️

### 🧭 Paanch Functions Ek Nazar Mein
| Function | **Kahan** dhundhta hai | **Kitne** | Return |
|---|---|---|---|
| **`re.match(p, s)`** | Sirf string ke **shuru** se | Pehla | Match / `None` |
| **`re.search(p, s)`** | **Kahin bhi** | Pehla | Match / `None` |
| **`re.fullmatch(p, s)`** | **Poori string** ek pattern ho | — | Match / `None` |
| **`re.findall(p, s)`** | Kahin bhi | **Saare** | **`list`** of strings |
| **`re.finditer(p, s)`** | Kahin bhi | **Saare** | **Iterator** of Match objects (position ke saath) |

> 🧠 **Yaad rakhne ki trick:**
> - **match** = *shuru se mel* (jaise `^` laga ho)
> - **search** = *pehla kahin bhi*
> - **fullmatch** = *poori string, shuru se aakhir* (jaise `^...$`) — **validation** ke liye best!
> - **findall** = *sab ki list* • **finditer** = *sab, positions ke saath*

````python
import re

text = "Hello Utkarsh"
pattern = r"Utkarsh"

print("text    :", repr(text))
print("pattern :", pattern)
print()
print("re.match    :", re.match(pattern, text), "  <- 'Hello' se shuru hota hai, Utkarsh se nahi!")
print("re.search   :", re.search(pattern, text))
print("re.fullmatch:", re.fullmatch(pattern, text), "  <- poori string sirf 'Utkarsh' nahi hai")
print()
print("fullmatch('Utkarsh', 'Utkarsh') :", re.fullmatch(pattern, "Utkarsh"))

# match mein ^ ki zaroorat nahi — apne aap shuru se
print("\nmatch(r'\\d+', '204 Block') :", re.match(r"\d+", "204 Block"))
print("match(r'\\d+', 'Room 204')  :", re.match(r"\d+", "Room 204"), " <- shuru mein digit nahi")
print("search(r'\\d+', 'Room 204') :", re.search(r"\d+", "Room 204"))
````

### 4.1 `findall()` — Saare Matches ki List 📋
```python
re.findall(pattern, text)      →   ['9876543210', '9123456780']
```
- Kuch na mile toh **khaali list `[]`** (`None` nahi!) — isliye `if` check ki zaroorat nahi.
- Pattern mein **groups `()`** ho toh alag cheez return hoti hai (Chapter 6).

> 💡 **Lambi digits wala jaal:** `\d{10}` lambe number ka bhi hissa pakad leta hai. Poora number chahiye toh **`\b\d{10}\b`**.

````python
import re

text = "Aarav: 9876543210, Bhavna: 9123456780, wrong: 12345, long: 98765432101"

print("search  (sirf pehla) :", re.search(r"\b\d{10}\b", text).group())
print("findall (sab)        :", re.findall(r"\b\d{10}\b", text))
print("kuch na mile toh     :", re.findall(r"\b\d{10}\b", "no phone here"))

phones = re.findall(r"\b\d{10}\b", text)
print("\nKul phone numbers:", len(phones))
for i, p in enumerate(phones, start=1):
    print(f"  {i}. {p}")
````

### 4.2 `finditer()` — Har Match ka Pata-Thikana 🗺️
`findall` sirf **text** deta hai. Par kabhi **kahan mila** (`start`, `end`) bhi chahiye hota hai — tab **`finditer`**. Ye **Match objects** ka iterator deta hai, isliye `m.group()`, `m.start()`, `m.end()`, `m.span()` sab chalte hain.

````python
import re

text = "Fest fever! The fest, the FEST..."

for m in re.finditer(r"fest", text, re.I):
    print(f"{m.group()!r:8} -> start={m.start():2}  end={m.end():2}  span={m.span()}")

# Positions ki madad se text highlight karna
marked = text
for m in reversed(list(re.finditer(r"fest", text, re.I))):         # peeche se, taaki index na bigdein
    marked = marked[:m.start()] + "[" + m.group() + "]" + marked[m.end():]
print("\nHighlighted:", marked)

# finditer lazy hai
it = re.finditer(r"fest", text, re.I)
print("\nIterator hai (list nahi)?", hasattr(it, "__next__"), "| pehla match:", next(it).group())
````

### 4.3 Flags — Pattern ka Behaviour Badalna 🚩
`re.search(pattern, text, flags)` ka teesra parameter. Ek se zyada flags ko `|` se jodo (`re.I | re.M`).

| Flag | Chhota naam | Kaam |
|---|:-:|---|
| `re.IGNORECASE` | `re.I` | Capital / small ko **barabar** maano |
| `re.MULTILINE` | `re.M` | `^` aur `$` **har line** ke shuru / aakhir par lagein (sirf poori string ke nahi) |
| `re.DOTALL` | `re.S` | `.` **newline ko bhi** match kare |
| `re.ASCII` | `re.A` | `\w \d \s` sirf ASCII ke liye |

````python
import re

log = """INFO app started
ERROR db down
INFO user login
ERROR disk full
WARNING slow response"""

print("re.M ke bina  ^ERROR.*$ :", re.findall(r"^ERROR.*$", log))
print("re.M ke saath ^ERROR.*$ :", re.findall(r"^ERROR.*$", log, re.M))

# re.S: dot newline ko bhi khaata hai
print("\n. bina re.S  :", re.findall(r"INFO.*started", log)[:1], "(ek hi line mein)")
print("re.S ke saath:", len(re.findall(r"INFO.*full", log, re.S)[0].splitlines()), "lines ka ek match")

# Flags jodna
print("\nI | M :", re.findall(r"^error.*$", log, re.I | re.M))
````

### 4.4 Validation Ka Mantra: `fullmatch()` ✅
Kisi value ko **check karna** ho (poora PIN? poora email?) toh **`re.fullmatch`** best — kyunki ye **poori string** match karta hai, `^` / `$` likhne ki zaroorat nahi.

````python
import re

def is_pin(value):
    return re.fullmatch(r"\d{6}", value) is not None

for v in ["201001", "20100", "2010011", "ABC123", " 201001"]:
    print(f"{v!r:12} -> valid PIN? {is_pin(v)}")

# search se galti: poori string check nahi hoti
print("\nsearch  se 'ab201001cd' :", re.search(r"\d{6}", "ab201001cd") is not None, " <- galat tarike se 'valid' dikha!")
print("fullmatch se 'ab201001cd':", re.fullmatch(r"\d{6}", "ab201001cd") is not None)
````

# ✂️ Chapter 5: `split()` aur `sub()` — Kaato aur Badlo

### 🎬 Story
Aarav ko volunteers ki list mili — par har kisi ne apne tareeke se likhi thi:

> `"Aarav,Bhavna;Chirag | Deepika ,Esha"`

Koi **comma**, koi **semicolon**, koi **pipe** (`|`) aur beech mein **extra spaces!** `"...".split(",")` sirf **ek** separator jaanta hai. Bhavna Didi bolin:

*"Jab separator **ek fixed cheez** nahi balki **ek pattern** ho — tab **`re.split()`**. Aur `replace()` ki tarah badalna ho par **pattern se** — tab **`re.sub()`**!"* 🔧

### ✂️ `re.split()` — Pattern par Kaato
```python
re.split(pattern, text, maxsplit=0)      →   list of strings
```
| | `str.split()` | **`re.split()`** |
|---|---|---|
| Separator | Ek **fixed text** | Koi bhi **pattern** |
| Example | `"a,b".split(",")` | `re.split(r"[,;]", "a,b;c")` |

### 🔁 `re.sub()` — Pattern Dhundho, Badal Do
```python
re.sub(pattern, replacement, text, count=0)      →   naya string
```
| | `str.replace()` | **`re.sub()`** |
|---|---|---|
| Kya dhundhta hai | Fixed text | **Pattern** |
| Example | `"color".replace("color", "colour")` | `re.sub(r"colou?r", "shade", text)` |

> 🧠 **Python mein "regex replace" ka naam `sub` hai** (*substitute*). `re.sub` **naya string** deta hai — original **nahi badalta** (strings immutable hain).

### 🧰 `sub` ke Teen Roop
| Replacement | Kaam |
|---|---|
| **String** `"***"` | Har match ko wahi string se badlo |
| **Group reference** `r"\2, \1"` | Match ke **hisse** reorder karo (Chapter 6) |
| **Function** `lambda m: ...` | Har match ke liye **code chalao**, jo return kare wo rakho ✨ |
| **`re.subn()`** | Naya string **aur kitne badle** — tuple `(text, count)` |

````python
import re

names = "Aarav,Bhavna;Chirag | Deepika ,Esha"

print("str.split(',')     :", names.split(","), " <- adhura!")
print("re.split pattern   :", re.split(r"\s*[,;|]\s*", names))

# maxsplit: sirf pehle N baar kaato
print("\nmaxsplit=2         :", re.split(r"\s*[,;|]\s*", names, maxsplit=2))

# Whitespace par: " " vs \s+
messy = "a  b   c"
print("\nsplit(' ')         :", messy.split(" "), " <- khaali strings aa gaye")
print("re.split(r'\\s+')   :", re.split(r"\s+", messy))

# Delimiter bhi chahiye? Group () mein daalo
print("\nGroup ke saath     :", re.split(r"(\+|-)", "10+5-3"), " <- operators bhi bach gaye")
````

### 5.1 ⚠️ `split` ke Khaali Tukde
Agar text **separator se shuru ya khatam** hota hai, toh list mein **khaali string `""`** aa jaati hai. `filter(None, ...)` ya list comprehension se hata do.

````python
import re

line = ",Aarav,,Bhavna,"
parts = re.split(r",", line)
print("raw split        :", parts)
print("khaali hataye    :", [p for p in parts if p])
print("filter(None, ..) :", list(filter(None, parts)))

# Sentences todna
text = "Hello! How are you? I am fine. Thanks"
print("\nSentences        :", re.split(r"[.!?]\s*", text))
````

### 5.2 `re.sub()` Basics 🔁

````python
import re

text = "Room 204, Phone 98765"

print("sab digits -> #        :", re.sub(r"\d", "#", text))
print("numbers -> <num>       :", re.sub(r"\d+", "<num>", text))
print("sirf pehla 1 badlo     :", re.sub(r"\d+", "<num>", text, count=1))

# subn: kitne badle bhi batata hai
new_text, n = re.subn(r"\d", "#", text)
print("\nsubn ->", repr(new_text), "|", n, "badle")

# Original nahi badalta
print("Original ab bhi        :", text)

# replace() vs sub()
british = "colour, color, colouur"
print("\nreplace (2 baar likhna padta):", british.replace("colour", "shade").replace("color", "shade"))
print("sub (ek pattern, dono)       :", re.sub(r"colou?r", "shade", british))
````

### 5.3 `sub` ke Saath Function ✨ — Sabse Powerful Trick
Replacement ki jagah **function** do. Python **har match** par us function ko **Match object** dega, aur jo wo **return** kare wahi nayi value hogi.
```python
re.sub(r"\d+", lambda m: str(int(m.group()) * 2), "3 pizzas")   →   "6 pizzas"
```

````python
import re

# 1️⃣ Har number double
print(re.sub(r"\d+", lambda m: str(int(m.group()) * 2), "3 pizzas and 12 samosas"))

# 2️⃣ Har shabd ka pehla akshar bada
print(re.sub(r"\b[a-z]", lambda m: m.group().upper(), "utkarsh fest is here"))

# 3️⃣ Phone mask: aakhri 4 digits chhodke baaki *
def mask(m):
    number = m.group()
    return "*" * (len(number) - 4) + number[-4:]

print(re.sub(r"\b\d{10}\b", mask, "Call 9876543210 or 9123456780"))

# 4️⃣ Normal function bhi chalta hai
def bracket(m):
    return f"[{m.group().upper()}]"

print(re.sub(r"fest", bracket, "fest fever, fest fun", flags=re.I))
````

# 👥 Chapter 6: Groups aur `compile()` — Match ke Hisse Nikalo

### 🎬 Story
Aarav ko ab sirf **match** nahi chahiye tha, **uske hisse** chahiye the. Tarikh `12-03-2026` mein se alag-alag **din**, **mahina** aur **saal**; email `aarav@gla.ac.in` mein se **naam** aur **domain**.

Bhavna Didi ne pattern ke kuch hisson ko **gol bracket `( )`** mein daal diya: *"Ye **group** hai. Python har group ka match **alag yaad rakhta** hai — jaise ek lunchbox ke alag-alag khaane!"* 🍱

### 👥 Capturing Groups `( )`
```python
m = re.search(r"(\d{2})-(\d{2})-(\d{4})", "Fest on 12-03-2026")
m.group(0)   →  '12-03-2026'     # poora match
m.group(1)   →  '12'             # pehla ( ) — din
m.group(2)   →  '03'             # doosra ( ) — mahina
m.group(3)   →  '2026'           # teesra ( ) — saal
m.groups()   →  ('12', '03', '2026')
```
> 🧠 Groups ki ginti **khulte bracket `(` se, left-to-right** hoti hai. **`group(0)`** hamesha **poora match**.

### 🏷️ Named Groups `(?P<naam>...)`
Number yaad rakhna mushkil — **naam** do!
```python
m = re.search(r"(?P<day>\d{2})-(?P<month>\d{2})-(?P<year>\d{4})", text)
m.group("day")  →  '12'          m.groupdict()  →  {'day': '12', 'month': '03', 'year': '2026'}
```

### 🧰 Groups ke Prakar
| Syntax | Naam | Kaam |
|---|---|---|
| `( ... )` | **Capturing** | Group banao aur **yaad rakho** |
| `(?P<naam> ... )` | **Named** | Naam se yaad rakho |
| `(?: ... )` | **Non-capturing** | Sirf **grouping** ke liye (OR / quantifier), **yaad nahi** rakhta |
| `( ... )?` | **Optional group** | Group na mile toh `m.group(n)` = **`None`** |

````python
import re

text = "Fest on 12-03-2026"
m = re.search(r"(\d{2})-(\d{2})-(\d{4})", text)

print("group(0) poora    :", m.group(0))
print("group(1) din      :", m.group(1))
print("group(2) mahina   :", m.group(2))
print("group(3) saal     :", m.group(3))
print("groups()          :", m.groups())
print("group(1, 3)       :", m.group(1, 3))
print("span(2) mahine ki jagah:", m.span(2))

day, month, year = m.groups()                 # tuple unpacking
print(f"\nUnpack: {day=} {month=} {year=}")
print("Ulta format      :", f"{year}/{month}/{day}")
````

````python
import re

# Nested groups: ginti BAHAR wale ( se shuru
m = re.search(r"((\d{3})-(\d{4}))", "Call 022-2345")
for i in range(4):
    print(f"group({i}) = {m.group(i)!r}")
print("groups()   =", m.groups())

# Group par quantifier
print("\n(ab)+ :", re.search(r"(ab)+", "xxababab").group(), "| aakhri group:", re.search(r"(ab)+", "xxababab").group(1))

# Non-capturing (?:...) vs capturing
print("\ncapturing   (Mr|Ms|Dr)\\. (\\w+) :", re.search(r"(Mr|Ms|Dr)\. (\w+)", "Dr. Sharma").groups())
print("non-capturing (?:Mr|Ms|Dr)\\. (\\w+):", re.search(r"(?:Mr|Ms|Dr)\. (\w+)", "Dr. Sharma").groups(), " <- sirf naam bacha")

# Optional group: na mile toh None
for t in ["12-03-2026", "12-03"]:
    m = re.fullmatch(r"(\d{2})-(\d{2})(?:-(\d{4}))?", t)
    print(f"{t!r:14} -> {m.groups()}")
````

### 6.1 `findall()` aur Groups — Dhyaan Do! ⚠️
Pattern mein **groups** ho toh **`findall` ka return badal jaata hai:**

| Groups kitne | `findall` kya deta hai |
|---|---|
| **0** groups | List of **strings** (poora match) |
| **1** group | List of strings (sirf **us group** ka text!) |
| **2+** groups | List of **tuples** (har group ek element) |

Poora match **aur** groups ek saath chahiye? → **`finditer`** (Match objects).

````python
import re

text = "1-2 and 33-44 and 555-666"

print("0 groups  \\d+-\\d+     :", re.findall(r"\d+-\d+", text))
print("1 group   (\\d+)-\\d+   :", re.findall(r"(\d+)-\d+", text), " <- sirf group ka text!")
print("2 groups  (\\d+)-(\\d+) :", re.findall(r"(\d+)-(\d+)", text), " <- tuples")

# finditer: poora match + groups dono
print("\nfinditer se:")
for m in re.finditer(r"(\d+)-(\d+)", text):
    print(f"   poora={m.group()!r:9} pehla={m.group(1)!r:6} doosra={m.group(2)!r:6} span={m.span()}")

# Tuples ko dictionary mein
pairs = re.findall(r"(\w+)=(\d+)", "a=1, b=22, c=333")
print("\nkey=value ->", {k: int(v) for k, v in pairs})
````

### 6.2 Groups + `sub()` — Reorder Karne ka Jaadu 🔀
Replacement mein **`\1`, `\2`** (ya **`\g<naam>`**) likho — matlab "pehle group ka text", "doosre group ka text". Isse text ke **hisse idhar-udhar** ho sakte hain.

````python
import re

# 1️⃣ Tarikh DD-MM-YYYY -> YYYY/MM/DD
print(re.sub(r"(\d{2})-(\d{2})-(\d{4})", r"\3/\2/\1", "Fest on 12-03-2026"))

# 2️⃣ Naam ulta: "Aarav Sharma" -> "Sharma, Aarav"
print(re.sub(r"(\w+) (\w+)", r"\2, \1", "Aarav Sharma"))

# 3️⃣ Named groups ke saath \g<naam>
print(re.sub(r"(?P<first>\w+) (?P<last>\w+)", r"\g<last> \g<first>", "Bhavna Verma"))

# 4️⃣ Pattern ke andar group ko dohrana: \1 = pehle group jaisa hi
print("\nDoubled word dhundo (\\b(\\w+) \\1\\b):", re.findall(r"\b(\w+) \1\b", "this is is a test test ok"))
print("Doubled word hatao:", re.sub(r"\b(\w+) \1\b", r"\1", "this is is a test test ok"))
````

### 📜 `re.VERBOSE` (`re.X`) — Lamba Pattern Samjhane Layak Banao

````python
import re

PHONE = re.compile(r"^[6-9]\d{9}$")                 # ek baar compile

print("type            :", type(PHONE).__name__)
print("PHONE.pattern   :", PHONE.pattern)
print("PHONE.groups    :", PHONE.groups, "(kitne capturing groups)")

for value in ["9876543210", "1234567890", "6123456789", "98765"]:
    print(f"{value:12} -> {PHONE.search(value) is not None}")

# Compiled object ke methods
DATE = re.compile(r"(?P<d>\d{2})-(?P<m>\d{2})-(?P<y>\d{4})")
text = "Quiz 12-03-2026, Dance 13-03-2026"
print("\nfindall :", DATE.findall(text))
print("sub     :", DATE.sub(r"\g<y>/\g<m>/\g<d>", text))
print("groupindex:", DATE.groupindex)

# Flags compile mein ek baar
YES = re.compile(r"^(y|yes|haan)$", re.I)
print("\n'HAAN' yes hai? ", bool(YES.match("HAAN")), "| 'nahi' yes hai?", bool(YES.match("nahi")))
````

### 🆚 Flat File (`.txt` / `.json`) vs Database (`sqlite3`)
| Feature | Text / JSON File | SQLite Database (`sqlite3`) |
|---|---|---|
| **Storage** | Poori file ek saath padho/likho | Tables (Rows & Columns) mein structured |
| **Search Speed** | Slow ($O(n)$ full scan) | Super fast (`WHERE`, Indexing) |
| **Safety** | Crash hone par file corrupt ho sakti hai | **ACID Transactions** (`commit` / `rollback`) |
| **Setup** | `open("data.json")` | `sqlite3.connect("fest.db")` ya `":memory:"` |

### 🪜 Python Database Connectivity ke 5 Kadam
1. **Connect:** `conn = sqlite3.connect("utkarsh.db")` (ya testing ke liye `":memory:"` jo RAM mein chalta hai)
2. **Cursor:** `cur = conn.cursor()` (SQL commands chalane wala pointer)
3. **Execute:** `cur.execute("SQL QUERY", params)`
4. **Fetch / Commit:** Data padhna ho toh `cur.fetchall()` / `cur.fetchone()`; data badalna ho toh `conn.commit()`
5. **Close:** `conn.close()` (ya `with sqlite3.connect(...) as conn:` use karo!)

````python
import sqlite3

# ":memory:" RAM mein ek temporary database banata hai — seekhne aur test karne ke liye best!
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

# 1. CREATE TABLE — Participants ki table banao
cur.execute("""
CREATE TABLE participants (
    roll INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    branch TEXT,
    score REAL
)
""")

# 2. INSERT — Data daalo (Hamesha '?' placeholders use karo!)
cur.execute("INSERT INTO participants VALUES (?, ?, ?, ?)", (101, "Aarav Sharma", "CSE", 94.5))

many_rows = [
    (102, "Bhavna Verma", "IT",  91.0),
    (103, "Chirag Rao",   "CSE", 76.5),
    (104, "Divya Singh",  "ECE", 88.0),
]
cur.executemany("INSERT INTO participants VALUES (?, ?, ?, ?)", many_rows)
conn.commit()

# 3. SELECT — Data padho
cur.execute("SELECT roll, name, score FROM participants WHERE score >= ? ORDER BY score DESC", (85.0,))
toppers = cur.fetchall()
print("🏆 85+ Toppers:", toppers)

conn.close()
````

### 7.1 🛡️ Parameterized Queries (`?`) — SQL Injection se Bachav!
Kabhi bhi f-string se SQL query mein user input mat jodo:
- ❌ **Khatarnaak (SQL Injection Trap):** `cur.execute(f"SELECT * FROM users WHERE name = '{user_input}'")`
- ✅ **Surakshit (Parameterized Tuple):** `cur.execute("SELECT * FROM users WHERE name = ?", (user_input,))`

> ⚠️ **Single-item tuple yaad hai (Day 2)?** Ek parameter bhejte waqt comma lagana zaroori hai: `(85.0,)`.

### 7.2 `UPDATE`, `DELETE` aur Aggregate Functions (`COUNT`, `AVG`, `MAX`) 🔧

````python
import sqlite3

with sqlite3.connect(":memory:") as conn:
    cur = conn.cursor()
    cur.execute("CREATE TABLE stalls (id INTEGER PRIMARY KEY, title TEXT, sales INTEGER)")
    cur.executemany("INSERT INTO stalls VALUES (?, ?, ?)", [
        (1, "Robotics", 4500),
        (2, "Food Court", 8200),
        (3, "Gaming", 3100),
    ])

    # UPDATE — Gaming stall ki sales badhao
    cur.execute("UPDATE stalls SET sales = sales + ? WHERE title = ?", (1900, "Gaming"))
    print("Updated rows:", cur.rowcount)

    # DELETE — Kam sales wala stall hatao (yahan sab 4500+ hain toh 0 hatenge)
    cur.execute("DELETE FROM stalls WHERE sales < ?", (2000,))

    # Aggregate query: COUNT, SUM, AVG, MAX
    cur.execute("SELECT COUNT(*), SUM(sales), ROUND(AVG(sales), 2), MAX(sales) FROM stalls")
    count, total, avg, highest = cur.fetchone()
    print(f"Stalls: {count} | Total: ₹{total} | Avg: ₹{avg} | Max: ₹{highest}")
````

### 📦 Step 1: Poora Program — `contact_book.py` banao
Neeche ka cell `%%writefile` se `contact_book.py` file banata hai. **Code padho** — har section ke upar comment hai. (Cell chalane par kuch print nahi hota — bas file ban jaati hai.)

````python
%%writefile contact_book.py
"""📒 Contact Book — Day 5 Project 1
Add • Search • Update • Delete • File mein save (JSON)"""
import json
import os
import re

PHONE_RE = re.compile(r"[6-9]\d{9}")                       # 10 digits, 6-9 se shuru
EMAIL_RE = re.compile(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+")


# ───────────── Custom exceptions (Day 4) ─────────────
class ContactError(Exception):
    """Contact Book ki saari errors ka baap."""

class InvalidContactError(ContactError):
    pass

class DuplicateContactError(ContactError):
    pass

class ContactNotFoundError(ContactError):
    pass


# ───────────── File: save / load ─────────────
def load_contacts(path):
    """File se contacts lo. File na ho ya kharab ho toh khaali list."""
    try:
        with open(path, encoding="utf-8") as f:
            return json.load(f)
    except (FileNotFoundError, json.JSONDecodeError):
        return []


def save_contacts(contacts, path):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(contacts, f, indent=2, ensure_ascii=False)


# ───────────── Validation helpers (regex) ─────────────
def _clean_phone(phone):
    phone = phone.strip()
    if not PHONE_RE.fullmatch(phone):
        raise InvalidContactError("Phone must be 10 digits and start with 6-9")
    return phone


def _clean_email(email):
    email = email.strip()
    if email and not EMAIL_RE.fullmatch(email):
        raise InvalidContactError("Email is not valid")
    return email


# ───────────── Core functions (koi input / print nahi) ─────────────
def find_index(contacts, name):
    """Naam (capital-small ignore) se contact ki position. Na mile toh -1."""
    for i, contact in enumerate(contacts):
        if contact["name"].lower() == name.strip().lower():
            return i
    return -1


def add_contact(contacts, name, phone, email=""):
    name = name.strip()
    if not name:
        raise InvalidContactError("Name is required")
    phone = _clean_phone(phone)
    email = _clean_email(email)
    if find_index(contacts, name) != -1:
        raise DuplicateContactError(f"{name} already exists")
    contact = {"name": name, "phone": phone, "email": email}
    contacts.append(contact)
    return contact


def search_contacts(contacts, term):
    term = term.strip().lower()
    return [c for c in contacts
            if term in c["name"].lower() or term in c["phone"] or term in c["email"].lower()]


def update_contact(contacts, name, phone=None, email=None):
    i = find_index(contacts, name)
    if i == -1:
        raise ContactNotFoundError(f"{name.strip()} not found")
    contact = contacts[i]
    if phone is not None:
        contact["phone"] = _clean_phone(phone)
    if email is not None:
        contact["email"] = _clean_email(email)
    return contact


def delete_contact(contacts, name):
    i = find_index(contacts, name)
    if i == -1:
        raise ContactNotFoundError(f"{name.strip()} not found")
    return contacts.pop(i)


def format_contact(contact):
    return f"{contact['name']:<14}{contact['phone']}  {contact['email'] or '-'}"


# ───────────── Menu (input / print yahin) ─────────────
MENU = "1 Add | 2 Search | 3 Update | 4 Delete | 5 List | 0 Exit"


def main(path="contacts.json", ask=input):
    contacts = load_contacts(path)
    print(f"📒 Contact Book — {len(contacts)} contact(s) loaded from {os.path.basename(path)}")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        try:
            if choice == "1":
                c = add_contact(contacts, ask("Name: "), ask("Phone: "), ask("Email (optional): "))
                save_contacts(contacts, path)
                print("✅ Added  :", format_contact(c))
            elif choice == "2":
                results = search_contacts(contacts, ask("Search: "))
                print(f"🔎 {len(results)} found")
                for c in results:
                    print("   ", format_contact(c))
            elif choice == "3":
                name = ask("Name to update: ")
                phone = ask("New phone (Enter = keep): ")
                email = ask("New email (Enter = keep): ")
                c = update_contact(contacts, name, phone or None, email or None)
                save_contacts(contacts, path)
                print("✅ Updated:", format_contact(c))
            elif choice == "4":
                c = delete_contact(contacts, ask("Name to delete: "))
                save_contacts(contacts, path)
                print("🗑️  Deleted:", c["name"])
            elif choice == "5":
                print(f"📋 {len(contacts)} contact(s):")
                for c in sorted(contacts, key=lambda c: c["name"].lower()):
                    print("   ", format_contact(c))
            elif choice == "0":
                print("👋 Bye! Contacts saved in", os.path.basename(path))
                break
            else:
                print("⚠️ Invalid choice — menu se 0 se 5 chuno")
        except ContactError as e:
            print("⚠️", e)


if __name__ == "__main__":
    main()
````

# 🎓 Chapter 9: Project 2 — Student Grade Analyzer

### 🎬 Story
Utkarsh ke baad semester result ka season aa gaya. Aarav ke dost **6 students ke marks** (3 subjects) leke aaye: *"Bhai, average, topper, kaun fail hua, aur sabka grade — jaldi nikaal de!"* 🎯

Aarav ne calculator uthaya... Bhavna Didi ne uska haath rok liya: *"Marks **list aur dictionary** mein daal. Phir **`lambda`, `map`, `filter`, `reduce`** — chaaron ek line mein poora hisaab kar denge!"* ⚡

### ✅ Requirements
| # | Feature | Kaise |
|:-:|---|---|
| 1️⃣ | Har student ka **total, average, grade** | `map` + `lambda` |
| 2️⃣ | **Topper** | `max(..., key=lambda ...)` |
| 3️⃣ | **Fail** hue students (kisi bhi subject mein `< 40`) | `filter` + `lambda` |
| 4️⃣ | **Class ka total** / **subject-wise average** | `reduce` / `zip` + `map` |
| 5️⃣ | **Ranking** aur **grade distribution** | `sorted(key=lambda)` / `reduce` |
| 6️⃣ | Naya student jodna (marks ki jaanch ke saath) | custom exception `MarksError` |

### 🧠 Design
**Data:** `dict` — **naam → marks ki list** (Maths, Physics, Chemistry):
```python
{"Aarav": [85, 90, 78], "Bhavna": [92, 88, 95], ...}
```
**Grade rule:** average `≥ 90` → **A+**, `≥ 80` → **A**, `≥ 70` → **B**, `≥ 60` → **C**, `≥ 40` → **D**, warna **F**. Pass hone ke liye **har subject mein `≥ 40`**.

### 🧰 Naye Auzaar: `lambda`, `map`, `filter`, `reduce` — Ek Nazar Mein
| Tool | Kaam | Example |
|---|---|---|
| **`lambda`** | **Chhota, bina naam ka function** | `lambda x: x * 2` |
| **`map(f, data)`** | Har item par `f` lagao → **naya item** | `map(lambda m: sum(m), marks_lists)` |
| **`filter(f, data)`** | Jin par `f` **True** wo **rakho** | `filter(lambda m: m >= 40, marks)` |
| **`reduce(f, data, start)`** | Sab ko **ek value** mein sameto | `reduce(lambda a, b: a + b, [1, 2, 3])` → `6` |
| `key=lambda` | `max` / `min` / `sorted` ko **kis cheez se compare** karna hai | `max(items, key=lambda it: it[1])` |

> 🧠 `map` aur `filter` **lazy** hain (Day 3) — `list()` se kholo. **`reduce`** `functools` mein hai: `from functools import reduce`.

````python
from functools import reduce

marks = [85, 90, 78, 35, 99]

squared = list(map(lambda m: m * m, marks))
passed = list(filter(lambda m: m >= 40, marks))
total = reduce(lambda acc, m: acc + m, marks)                  # ((((85+90)+78)+35)+99)
biggest = reduce(lambda a, b: a if a > b else b, marks)       # reduce se max
students = {"Aarav": 253, "Bhavna": 275, "Chirag": 143}
topper = max(students.items(), key=lambda item: item[1])      # (naam, total) ka jodi

print("map    (squares)     :", squared)
print("filter (>= 40)       :", passed)
print("reduce (jod)         :", total, "| sum() se barabar?", total == sum(marks))
print("reduce (max)         :", biggest)
print("max(key=lambda)      :", topper)
print("sorted(key=lambda)   :", sorted(students, key=lambda n: students[n], reverse=True))

# reduce ka teesra argument = shuruaati value
print("\nreduce(start=100)    :", reduce(lambda a, b: a + b, [1, 2, 3], 100))
````

### 📦 Step 1: Poora Program — `grade_analyzer.py` banao

````python
%%writefile grade_analyzer.py
"""🎓 Student Grade Analyzer — Day 5 Project 2
Marks lists / dicts mein • average • topper • grade (lambda, map, filter, reduce)"""
from functools import reduce

SUBJECTS = ["Maths", "Physics", "Chemistry"]
PASS_MARK = 40
GRADE_RULES = [(90, "A+"), (80, "A"), (70, "B"), (60, "C"), (40, "D")]     # (kam se kam average, grade)

SAMPLE_CLASS = {
    "Aarav":   [85, 90, 78],
    "Bhavna":  [92, 88, 95],
    "Chirag":  [45, 38, 60],
    "Deepika": [70, 65, 80],
    "Esha":    [55, 60, 35],
    "Farhan":  [99, 97, 100],
}


class MarksError(Exception):
    """Marks ya naam galat hone par."""


# ───────────── Core functions (koi input / print nahi) ─────────────
def average(marks):
    return round(sum(marks) / len(marks), 2)


def grade_of(avg):
    for minimum, grade in GRADE_RULES:
        if avg >= minimum:
            return grade
    return "F"


def parse_marks(text):
    """'80, 75, 90' -> [80, 75, 90]  (galat ho toh MarksError)"""
    try:
        return [int(part) for part in text.split(",")]
    except ValueError:
        raise MarksError("Marks must be whole numbers like 80,75,90") from None


def add_student(students, name, marks):
    name = name.strip()
    if not name:
        raise MarksError("Name is required")
    if len(marks) != len(SUBJECTS):
        raise MarksError(f"Need exactly {len(SUBJECTS)} marks")
    for m in marks:
        if not 0 <= m <= 100:
            raise MarksError(f"Mark {m} is outside 0-100")
    students[name] = list(marks)
    return students[name]


def report(students):
    """Har student ka total, average, grade, passed — dict of dicts."""
    return {
        name: {
            "total": sum(marks),
            "average": average(marks),
            "grade": grade_of(average(marks)),
            "passed": min(marks) >= PASS_MARK,
        }
        for name, marks in students.items()
    }


def topper(students):
    """(naam, total) — sabse zyada total wala."""
    name, marks = max(students.items(), key=lambda item: sum(item[1]))
    return name, sum(marks)


def failed_students(students):
    """Jis kisi subject mein PASS_MARK se kam — filter + lambda."""
    failed = filter(lambda item: min(item[1]) < PASS_MARK, students.items())
    return [name for name, _ in failed]


def class_total(students):
    """Poori class ke sabhi marks ka jod — reduce."""
    return reduce(lambda acc, marks: acc + sum(marks), students.values(), 0)


def subject_averages(students):
    """Har subject ka class average — zip(*) + map."""
    columns = zip(*students.values())                      # har subject ke marks ek saath
    return dict(zip(SUBJECTS, map(lambda col: round(sum(col) / len(col), 2), columns)))


def ranking(students):
    return sorted(students, key=lambda name: sum(students[name]), reverse=True)


def grade_distribution(students):
    grades = map(lambda marks: grade_of(average(marks)), students.values())
    return reduce(lambda counts, g: {**counts, g: counts.get(g, 0) + 1}, grades, {})


def print_report(students):
    head = f"{'Name':<9}" + "".join(f"{s[:4]:>6}" for s in SUBJECTS) + f"{'Total':>7}{'Avg':>8}  Grade  Result"
    print(head)
    print("-" * len(head))
    for name, r in report(students).items():
        marks = "".join(f"{m:>6}" for m in students[name])
        print(f"{name:<9}{marks}{r['total']:>7}{r['average']:>8}  {r['grade']:<5}  {'Pass' if r['passed'] else 'FAIL'}")


# ───────────── Menu ─────────────
MENU = "1 Add student | 2 Report | 3 Topper | 4 Failed | 5 Subject avg | 6 Grades | 0 Exit"


def main(ask=input):
    students = dict(SAMPLE_CLASS)
    print(f"🎓 Grade Analyzer — sample class mein {len(students)} students")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        try:
            if choice == "1":
                name = ask("Name: ")
                marks = parse_marks(ask(f"Marks ({', '.join(SUBJECTS)}) e.g. 80,75,90: "))
                add_student(students, name, marks)
                print("✅ Added", name.strip())
            elif choice == "2":
                print_report(students)
            elif choice == "3":
                print("🏆 Topper: %s (total %d)" % topper(students))
            elif choice == "4":
                print("❌ Failed:", failed_students(students) or "koi nahi")
            elif choice == "5":
                print("📊 Subject averages:", subject_averages(students))
            elif choice == "6":
                print("🎖️  Grades:", dict(sorted(grade_distribution(students).items())))
            elif choice == "0":
                print("👋 Bye!")
                break
            else:
                print("⚠️ Invalid choice")
        except MarksError as e:
            print("⚠️", e)


if __name__ == "__main__":
    main()
````

# 🎮 Chapter 10: Project 3 — Quiz Game

### 🎬 Story
Utkarsh ke **Tech Quiz** mein 500 log aaye. Quiz-master ne kagaz ke sawaal baant diye... aur phir **answers check karne mein 3 ghante** lag gaye! 😵

Bhavna Didi: *"Chal, ek **Quiz Game** bana — **random sawaal**, **turant score**, aur har player ka **score history file mein** — taaki leaderboard bhi ban sake!"* 🏆

### ✅ Requirements
| # | Feature | Kaise |
|:-:|---|---|
| 1️⃣ | **Random** sawaal har baar alag | `random.sample` (Day 3) |
| 2️⃣ | A / B / C / D se jawab, **galat letter par dobara pucho** | custom exception + `while` retry |
| 3️⃣ | **Score**: sahi = 10 points (optional negative marking) | `calculate_score()` |
| 4️⃣ | Percentage aur **rating message** | `if / elif` |
| 5️⃣ | **Score history file mein append** | file mode `"a"` (Day 4) |
| 6️⃣ | History aur **Leaderboard** dikhana | file padho → `sorted(key=lambda)` |

### 🧠 Design
**Question ka format** — dictionary:
```python
{"q": "len([1, 2, 3]) ka output?", "options": ["2", "3", "4", "Error"], "answer": 1}      # answer = sahi option ka index
```
**History file** (`quiz_history.txt`) — **har line ek game**, comma se alag:
```
2026-03-12 10:00,Aarav,30,40
2026-03-12 10:05,Bhavna,40,40
```
| Function | Kaam |
|---|---|
| `pick_questions(questions, n, rng)` | `n` random sawaal (`rng.sample`) |
| `is_correct(question, letter)` | Letter sahi hai? (galat letter → `QuizError`) |
| `calculate_score(results, ...)` | `[True, False, ...]` → score |
| `rating(percent)` | Message |
| `save_score(...)` / `load_history(path)` | History file |
| `leaderboard(history, top)` | Top players |
| `play_quiz(...)` / `main(...)` | Khelna / menu |

> 🎲 **`rng` parameter:** hum `random.Random(seed)` bhi de sakte hain — taaki **test mein sawaal wahi aayein** (reproducible), aur asli game mein `random` module.

````python
%%writefile quiz_game.py
"""🎮 Quiz Game — Day 5 Project 3
Random questions • score • history file mein save"""
import random
from datetime import datetime

LETTERS = "abcd"
POINTS = 10

QUESTIONS = [
    {"q": "Python mein comment kis symbol se shuru hota hai?", "options": ["//", "#", "/*", "--"], "answer": 1},
    {"q": "len([1, 2, 3]) ka output kya hoga?", "options": ["2", "3", "4", "Error"], "answer": 1},
    {"q": "Kaunsa data type immutable hai?", "options": ["list", "dict", "tuple", "set"], "answer": 2},
    {"q": "10 // 3 ka result kya hai?", "options": ["3.33", "3", "4", "1"], "answer": 1},
    {"q": "File padhne ke liye kaunsa mode use hota hai?", "options": ["w", "a", "r", "x"], "answer": 2},
    {"q": "ZeroDivisionError kab aata hai?", "options": ["Zero se bhaag dene par", "Zero se guna karne par", "Zero jodne par", "Zero print karne par"], "answer": 0},
    {"q": "Regex mein \\d kya match karta hai?", "options": ["Letter", "Digit", "Space", "Symbol"], "answer": 1},
    {"q": "Exception pakadne ke liye kaunsa block use hota hai?", "options": ["try-except", "if-else", "for-in", "def-return"], "answer": 0},
    {"q": "random.randint(1, 6) kaunsi values de sakta hai?", "options": ["1 se 5", "1 se 6 (dono shamil)", "2 se 6", "0 se 6"], "answer": 1},
    {"q": "'Utkarsh'[0] ka output kya hoga?", "options": ["U", "t", "k", "Error"], "answer": 0},
]


class QuizError(Exception):
    """Quiz mein galat input ya setting."""


# ───────────── Core functions (koi input / print nahi) ─────────────
def pick_questions(questions, n, rng=random):
    if not 1 <= n <= len(questions):
        raise QuizError(f"Choose between 1 and {len(questions)} questions")
    return rng.sample(questions, n)


def is_correct(question, letter):
    letter = letter.strip().lower()
    valid = LETTERS[:len(question["options"])]
    if len(letter) != 1 or letter not in valid:
        raise QuizError(f"Please answer with one of: {', '.join(valid)}")
    return LETTERS.index(letter) == question["answer"]


def calculate_score(results, correct_points=POINTS, wrong_penalty=0):
    """results = [True, False, ...] -> score"""
    correct = sum(results)
    wrong = len(results) - correct
    return correct * correct_points - wrong * wrong_penalty


def rating(percent):
    if percent >= 90:
        return "Genius"
    if percent >= 70:
        return "Great"
    if percent >= 50:
        return "Good"
    return "Keep practising"


# ───────────── File: score history ─────────────
def save_score(path, name, score, total, when=None):
    when = when or datetime.now().strftime("%Y-%m-%d %H:%M")
    name = name.replace(",", " ").strip() or "Guest"
    with open(path, "a", encoding="utf-8") as f:                 # "a" = peeche jodo
        f.write(f"{when},{name},{score},{total}\n")


def load_history(path):
    history = []
    try:
        with open(path, encoding="utf-8") as f:
            for line in f:
                try:
                    when, name, score, total = line.strip().split(",")
                    history.append({"when": when, "name": name, "score": int(score), "total": int(total)})
                except ValueError:
                    continue                                       # kharab line skip
    except FileNotFoundError:
        pass
    return history


def leaderboard(history, top=3):
    return sorted(history, key=lambda h: (-h["score"], h["when"]))[:top]


# ───────────── Khelna ─────────────
def ask_question(question, number, ask=input):
    print(f"\nQ{number}. {question['q']}")
    for letter, option in zip(LETTERS, question["options"]):
        print(f"   {letter}) {option}")
    while True:
        try:
            correct = is_correct(question, ask("Your answer: "))
            break
        except QuizError as e:
            print("⚠️", e)
    if correct:
        print("✅ Correct!")
    else:
        right = question["answer"]
        print(f"❌ Wrong! Sahi jawab: {LETTERS[right]}) {question['options'][right]}")
    return correct


def play_quiz(name, questions, n, path, ask=input, rng=random):
    chosen = pick_questions(questions, n, rng)
    results = [ask_question(q, i, ask) for i, q in enumerate(chosen, start=1)]
    score = calculate_score(results)
    total = n * POINTS
    percent = round(score / total * 100)
    print(f"\n🏁 {name}: {sum(results)}/{n} sahi | Score {score}/{total} ({percent}%) | {rating(percent)}")
    save_score(path, name, score, total)
    return score


# ───────────── Menu ─────────────
MENU = "1 Play | 2 History | 3 Leaderboard | 0 Exit"


def main(path="quiz_history.txt", ask=input, rng=random):
    print("🎮 Utkarsh Quiz!")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        try:
            if choice == "1":
                name = ask("Your name: ").strip() or "Guest"
                try:
                    n = int(ask(f"Kitne questions (1-{len(QUESTIONS)})? "))
                except ValueError:
                    raise QuizError("Enter a number") from None
                play_quiz(name, QUESTIONS, n, path, ask, rng)
            elif choice == "2":
                history = load_history(path)
                print(f"📜 {len(history)} game(s) khele gaye")
                for h in history[-5:]:
                    print(f"   {h['when']}  {h['name']:<10} {h['score']}/{h['total']}")
            elif choice == "3":
                for i, h in enumerate(leaderboard(load_history(path)), start=1):
                    print(f"   {i}. {h['name']:<10} {h['score']}/{h['total']}")
            elif choice == "0":
                print("👋 Thanks for playing!")
                break
            else:
                print("⚠️ Invalid choice")
        except QuizError as e:
            print("⚠️", e)


if __name__ == "__main__":
    main()
````

# 💰 Chapter 11: Project 4 — Expense Tracker

### 🎬 Story
Utkarsh ke baad **fest ka budget** dekhne ka time aaya. Treasurer ne ek chit pakdayi: *"Bhai, kitna kharcha hua? Kis cheez par sabse zyada? Kal tak batana!"* Aarav ne kharche **WhatsApp ke 40 messages** mein bikhre paaye. 🤯

Bhavna Didi: *"Chal, ek **Expense Tracker** bana — har kharcha **category** ke saath likh, **category-wise total** dekh, aur sab **file mein save** kar. Fest ka hisaab bhi, tera apna bhi!"* 📊

### ✅ Requirements
| # | Feature | Kaise |
|:-:|---|---|
| 1️⃣ | Kharcha **add**: date, category, amount, note | validation + custom exception |
| 2️⃣ | **Date** `YYYY-MM-DD` ki jaanch | `datetime.strptime` (Day 3) |
| 3️⃣ | Category **fixed list** mein se | `Food`, `Travel`, `Study`, `Fun`, `Other` |
| 4️⃣ | **Category-wise total**, grand total, **top category** | dictionary + `max(key=...)` |
| 5️⃣ | **Bar chart** (text) | `"█" * n` |
| 6️⃣ | Kharcha **delete** | number se |
| 7️⃣ | **File mein save** — har badlaav ke baad | `json` |

### 🧠 Design
```python
{"date": "2026-03-12", "category": "Food", "amount": 250.0, "note": "Fest food stall"}
```
| Function | Kaam |
|---|---|
| `add_expense(expenses, date_text, category, amount_text, note)` | Jaanch + jodo |
| `total(expenses)` | Grand total |
| `totals_by_category(expenses)` | `{"Food": 450.5, ...}` (jo categories mein kharcha hua) |
| `top_category(expenses)` | Sabse zyada kharcha kahan |
| `expenses_in_month(expenses, "2026-03")` | Mahine ka filter |
| `delete_expense(expenses, number)` | 1-based number se hatao |
| `print_summary(expenses)` | Totals + bar chart |

````python
%%writefile expense_tracker.py
"""💰 Expense Tracker — Day 5 Project 4
Category-wise totals • JSON file mein save"""
import json
from datetime import datetime

CATEGORIES = ["Food", "Travel", "Study", "Fun", "Other"]


class ExpenseError(Exception):
    """Expense Tracker ki saari errors."""

class InvalidExpenseError(ExpenseError):
    pass


# ───────────── File: save / load ─────────────
def load_expenses(path):
    try:
        with open(path, encoding="utf-8") as f:
            return json.load(f)
    except (FileNotFoundError, json.JSONDecodeError):
        return []


def save_expenses(expenses, path):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(expenses, f, indent=2, ensure_ascii=False)


# ───────────── Core functions (koi input / print nahi) ─────────────
def add_expense(expenses, date_text, category, amount_text, note=""):
    try:
        date = datetime.strptime(date_text.strip(), "%Y-%m-%d").strftime("%Y-%m-%d")
    except ValueError:
        raise InvalidExpenseError("Date must look like 2026-03-12") from None

    category = category.strip().title()
    if category not in CATEGORIES:
        raise InvalidExpenseError(f"Category must be one of: {', '.join(CATEGORIES)}")

    try:
        amount = round(float(amount_text), 2)
    except ValueError:
        raise InvalidExpenseError("Amount must be a number") from None
    if amount <= 0:
        raise InvalidExpenseError("Amount must be greater than 0")

    expense = {"date": date, "category": category, "amount": amount, "note": note.strip()}
    expenses.append(expense)
    return expense


def total(expenses):
    return round(sum(e["amount"] for e in expenses), 2)


def totals_by_category(expenses):
    """{'Food': 450.5, ...} — sirf wahi categories jahan kharcha hua, CATEGORIES ke order mein."""
    sums = {}
    for e in expenses:
        sums[e["category"]] = sums.get(e["category"], 0) + e["amount"]
    return {cat: round(sums[cat], 2) for cat in CATEGORIES if cat in sums}


def top_category(expenses):
    sums = totals_by_category(expenses)
    if not sums:
        raise ExpenseError("No expenses yet")
    name = max(sums, key=lambda cat: sums[cat])
    return name, sums[name]


def expenses_in_month(expenses, month):
    """month = '2026-03'"""
    return [e for e in expenses if e["date"].startswith(month)]


def delete_expense(expenses, number):
    if not 1 <= number <= len(expenses):
        raise InvalidExpenseError(f"Choose a number between 1 and {len(expenses)}")
    return expenses.pop(number - 1)


def format_expense(e):
    return f"{e['date']}  {e['category']:<7}{e['amount']:>9.2f}  {e['note']}"


def print_summary(expenses):
    sums = totals_by_category(expenses)
    if not sums:
        print("Koi kharcha nahi abhi tak.")
        return
    biggest = max(sums.values())
    for cat, amount in sums.items():
        bar = "█" * max(1, round(amount / biggest * 20))
        print(f"{cat:<7}{amount:>9.2f}  {bar}")
    print("-" * 36)
    print(f"{'TOTAL':<7}{total(expenses):>9.2f}")
    name, amount = top_category(expenses)
    print(f"🔝 Sabse zyada: {name} ({amount:.2f})")


# ───────────── Menu ─────────────
MENU = "1 Add | 2 List | 3 Summary | 4 Delete | 0 Exit"


def main(path="expenses.json", ask=input):
    expenses = load_expenses(path)
    print(f"💰 Expense Tracker — {len(expenses)} kharche load hue")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        try:
            if choice == "1":
                e = add_expense(expenses, ask("Date (YYYY-MM-DD): "),
                                ask(f"Category ({'/'.join(CATEGORIES)}): "),
                                ask("Amount: "), ask("Note: "))
                save_expenses(expenses, path)
                print("✅ Added:", format_expense(e))
            elif choice == "2":
                for i, e in enumerate(expenses, start=1):
                    print(f"{i:>3}. {format_expense(e)}")
                print(f"     Total: {total(expenses):.2f}")
            elif choice == "3":
                print_summary(expenses)
            elif choice == "4":
                try:
                    number = int(ask("Number to delete: "))
                except ValueError:
                    raise InvalidExpenseError("Enter a number") from None
                e = delete_expense(expenses, number)
                save_expenses(expenses, path)
                print("🗑️  Deleted:", format_expense(e))
            elif choice == "0":
                print("👋 Bye! Data saved.")
                break
            else:
                print("⚠️ Invalid choice")
        except ExpenseError as e:
            print("⚠️", e)


if __name__ == "__main__":
    main()
````

# ✉️ Chapter 12: Project 5 — Email & Log Validator

### 🎬 Story
Utkarsh 2026 ke server par **registration CSV lines** aur **application logs** aa rahe hain. Kisi ne email mein `@` bhool diya, kisi ne phone number 8 digits ka likha, aur server log mein `ERROR` aur `WARNING` chhupi hui hain! 🕵️‍♂️

Bhavna Didi: *"Aarav, Forenoon mein jo **Regex (`re.compile`, `fullmatch`, named groups)** seekha hai, ab usse ek **Email, Phone & Log Validator** module bana jo har galat line ka **exact error report** banaye!"* ✅

### ✅ Requirements
| # | Feature | Regex Tool |
|:-:|---|---|
| 1️⃣ | **Email Validator** (`user@domain.ext`) | `EMAIL_RE.fullmatch()` |
| 2️⃣ | **Indian Phone Validator** (`10 digits`, starts with `6–9`, optional `+91`) | `PHONE_RE.fullmatch()` |
| 3️⃣ | **Registration Row Validator** (`Name, email, phone`) | Error list ke saath line-by-line audit |
| 4️⃣ | **Log Line Parser & Error Reporter** (`[YYYY-MM-DD HH:MM:SS] LEVEL: msg`) | Named groups `(?P<date>...)` + level filter |

### 📦 Step 1: Poora Program — `email_log_validator.py` banao

````python
%%writefile email_log_validator.py
"""✉️ Email & Log Validator — Day 5 Project 5
Check emails and phone numbers with regex and report errors"""
import re

EMAIL_RE = re.compile(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9-]+(?:\.[A-Za-z]{2,})+")
PHONE_RE = re.compile(r"(?:\+91[-\s]?)?([6-9]\d{9})")
LOG_RE = re.compile(
    r"\[(?P<date>\d{4}-\d{2}-\d{2})\s+(?P<time>\d{2}:\d{2}:\d{2})\]\s+"
    r"(?P<level>INFO|WARNING|ERROR):\s+(?P<msg>.+)"
)


def is_valid_email(text):
    return bool(EMAIL_RE.fullmatch(text.strip()))


def normalize_phone(text):
    """Valid 10-digit mobile number लौटाओ, warna None."""
    m = PHONE_RE.fullmatch(text.strip())
    return m.group(1) if m else None


def validate_record(line, line_no=1):
    """'Name, email, phone' line check karo -> (is_ok, result_or_errors)."""
    parts = [p.strip() for p in line.split(",")]
    if len(parts) != 3:
        return False, [f"Line {line_no}: expected 3 comma-separated fields, got {len(parts)}"]
    name, email, phone = parts
    errors = []
    if not name:
        errors.append(f"Line {line_no}: missing name")
    if not is_valid_email(email):
        errors.append(f"Line {line_no}: invalid email '{email}'")
    clean_phone = normalize_phone(phone)
    if not clean_phone:
        errors.append(f"Line {line_no}: invalid phone '{phone}'")
    if errors:
        return False, errors
    return True, {"name": name, "email": email.lower(), "phone": clean_phone}


def validate_batch(lines):
    valid, errors = [], []
    for idx, raw in enumerate(lines, start=1):
        if not raw.strip():
            continue
        ok, info = validate_record(raw, idx)
        if ok:
            valid.append(info)
        else:
            errors.extend(info)
    return valid, errors


def parse_log_line(line):
    m = LOG_RE.fullmatch(line.strip())
    return m.groupdict() if m else None


def analyze_logs(lines):
    counts = {"INFO": 0, "WARNING": 0, "ERROR": 0, "MALFORMED": 0}
    error_messages = []
    for idx, line in enumerate(lines, start=1):
        if not line.strip():
            continue
        parsed = parse_log_line(line)
        if not parsed:
            counts["MALFORMED"] += 1
            error_messages.append(f"Line {idx}: malformed log -> {line.strip()}")
        else:
            lvl = parsed["level"]
            counts[lvl] += 1
            if lvl == "ERROR":
                error_messages.append(f"Line {idx} [{parsed['date']} {parsed['time']}]: {parsed['msg']}")
    return counts, error_messages


MENU = "1 Check Email | 2 Check Phone | 3 Validate CSV Row | 4 Parse Log Line | 0 Exit"


def main(ask=input):
    print("✉️ Email & Log Validator Ready")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        if choice == "1":
            em = ask("Email: ").strip()
            print("✅ Valid email" if is_valid_email(em) else "❌ Invalid email")
        elif choice == "2":
            ph = ask("Phone: ").strip()
            norm = normalize_phone(ph)
            print(f"✅ Valid phone: {norm}" if norm else "❌ Invalid phone")
        elif choice == "3":
            row = ask("Row (Name, email, phone): ")
            ok, info = validate_record(row, 1)
            print("✅ Clean record:", info if ok else "❌ Errors: " + "; ".join(info))
        elif choice == "4":
            lg = ask("Log line: ")
            parsed = parse_log_line(lg)
            print("✅ Parsed:", parsed if parsed else "❌ Malformed log format")
        elif choice == "0":
            print("👋 Validator closed.")
            break
        else:
            print("⚠️ Invalid choice")


if __name__ == "__main__":
    main()
````

# 🏧 Chapter 13: Project 6 — ATM & Library Management Simulation

### 🎬 Story
Utkarsh 2026 ka aakhri aur sabse bada challenge! 🏁
Fest campus mein **Cashless Campus ATM Kiosk** aur **Department Library Desk** ka simulation banana hai jahan:
- Galat PIN, zero/negative amount, ya balance se zyada withdrawal par **Custom Exceptions (Day 4)** uthein!
- Har transaction aur book issue/return **clean functions (Day 2)** aur **JSON file persistence (Day 4)** ke saath chale!

### ✅ Requirements
| # | Feature | Concepts Used |
|:-:|---|---|
| 1️⃣ | **Custom Exception Family** (`ATMError`, `InvalidPINError`, `InsufficientFundsError`, `InvalidAmountError`, `LibraryError`) | Day 4 Custom Exceptions |
| 2️⃣ | **ATM Operations** (PIN Check, Deposit, Withdraw, Transfer, Mini-Statement) | Day 2 Functions + Dictionaries |
| 3️⃣ | **Library Simulation** (Issue Book, Return Book with Fine Calculation) | Sets/Dicts + Guard Clauses |

### 📦 Step 1: Poora Program — `atm_simulator.py` banao

````python
%%writefile atm_simulator.py
"""🏧 ATM & Library Management Simulation — Day 5 Project 6
Functions + Custom Exception Handling + File Persistence"""
import json


# ───────────── Custom Exceptions (Day 4) ─────────────
class ATMError(Exception):
    """ATM simulation ki base exception."""

class InvalidPINError(ATMError):
    pass

class InvalidAmountError(ATMError):
    pass

class InsufficientFundsError(ATMError):
    pass

class AccountNotFoundError(ATMError):
    pass

class LibraryError(Exception):
    """Library simulation ki base exception."""


# ───────────── ATM Functions ─────────────
def create_account(acc_no, name, pin, balance=0.0):
    return {
        "acc_no": str(acc_no),
        "name": name.strip(),
        "pin": str(pin),
        "balance": round(float(balance), 2),
        "history": [f"OPENED with ₹{float(balance):.2f}"]
    }


def verify_pin(account, entered_pin):
    if str(entered_pin).strip() != account["pin"]:
        raise InvalidPINError("Incorrect PIN entered")
    return True


def parse_positive_amount(amount_val):
    try:
        amt = round(float(amount_val), 2)
    except (ValueError, TypeError):
        raise InvalidAmountError("Amount must be a valid number") from None
    if amt <= 0:
        raise InvalidAmountError("Amount must be greater than 0")
    return amt


def deposit(account, amount_val):
    amt = parse_positive_amount(amount_val)
    account["balance"] = round(account["balance"] + amt, 2)
    account["history"].append(f"DEPOSIT +₹{amt:.2f} -> Bal ₹{account['balance']:.2f}")
    return account["balance"]


def withdraw(account, pin, amount_val, min_balance=0.0):
    verify_pin(account, pin)
    amt = parse_positive_amount(amount_val)
    if account["balance"] - amt < min_balance:
        raise InsufficientFundsError(
            f"Insufficient funds: balance ₹{account['balance']:.2f}, requested ₹{amt:.2f}"
        )
    account["balance"] = round(account["balance"] - amt, 2)
    account["history"].append(f"WITHDRAW -₹{amt:.2f} -> Bal ₹{account['balance']:.2f}")
    return account["balance"]


def transfer(sender, receiver, pin, amount_val):
    verify_pin(sender, pin)
    amt = parse_positive_amount(amount_val)
    if sender["balance"] < amt:
        raise InsufficientFundsError("Insufficient balance for transfer")
    sender["balance"] = round(sender["balance"] - amt, 2)
    receiver["balance"] = round(receiver["balance"] + amt, 2)
    sender["history"].append(f"TRANSFER_OUT -₹{amt:.2f} to {receiver['name']}")
    receiver["history"].append(f"TRANSFER_IN +₹{amt:.2f} from {sender['name']}")
    return sender["balance"], receiver["balance"]


# ───────────── Library Management Simulation Functions ─────────────
def issue_book(catalog, book_id, student_name):
    if book_id not in catalog:
        raise LibraryError(f"Book '{book_id}' not found in catalog")
    book = catalog[book_id]
    if book["issued_to"] is not None:
        raise LibraryError(f"Book '{book['title']}' is already issued to {book['issued_to']}")
    book["issued_to"] = student_name.strip()
    return book


def return_book(catalog, book_id, days_kept, allowed_days=14, fine_per_day=5):
    if book_id not in catalog:
        raise LibraryError(f"Book '{book_id}' not found in catalog")
    book = catalog[book_id]
    if book["issued_to"] is None:
        raise LibraryError(f"Book '{book['title']}' was not issued")
    borrower = book["issued_to"]
    book["issued_to"] = None
    late_days = max(0, int(days_kept) - allowed_days)
    fine = late_days * fine_per_day
    return {"book": book["title"], "borrower": borrower, "late_days": late_days, "fine": fine}


MENU = "1 Balance | 2 Deposit | 3 Withdraw | 4 Mini-Statement | 0 Exit"


def main(ask=input):
    acc = create_account("1001", "Aarav Sharma", "1234", 5000.0)
    print(f"🏧 Welcome {acc['name']} (Acc #{acc['acc_no']})")
    while True:
        print("\n" + MENU)
        choice = ask("Choice: ").strip()
        try:
            if choice == "1":
                verify_pin(acc, ask("PIN: "))
                print(f"💰 Current Balance: ₹{acc['balance']:.2f}")
            elif choice == "2":
                new_bal = deposit(acc, ask("Deposit Amount: "))
                print(f"✅ Deposited! New Balance: ₹{new_bal:.2f}")
            elif choice == "3":
                pin = ask("PIN: ")
                amt = ask("Withdraw Amount: ")
                new_bal = withdraw(acc, pin, amt)
                print(f"✅ Withdrawn! New Balance: ₹{new_bal:.2f}")
            elif choice == "4":
                verify_pin(acc, ask("PIN: "))
                print("📜 Mini-Statement (last 5):")
                for entry in acc["history"][-5:]:
                    print("  •", entry)
            elif choice == "0":
                print("👋 Thank you for using GLA Campus ATM!")
                break
            else:
                print("⚠️ Invalid option")
        except ATMError as e:
            print("⚠️ ATM Alert:", e)


if __name__ == "__main__":
    main()
````

````python
amount = 3800
notes = [2000, 500, 200, 100]

rem = amount
dispensed = {}
for n in notes:
    cnt = rem // n
    if cnt > 0:
        dispensed[n] = cnt
        rem %= n
print(f"Notes for {amount}: {dispensed}")
````
