## 🧰 Ground Zero 101: What is a Module? (The Hardware Store Analogy)

Imagine you want to fix a bicycle. You don't go to an iron mine, dig up ore, and melt it down to build a screwdriver from scratch! You simply go to a **hardware store** and buy a screwdriver.

In Python:
- A **Module** is like a toolbox that someone else already built and tested for you.
- `import math` opens your math toolbox (square roots, pi, trigonometry).
- `import random` opens your game toolbox (rolling dice, picking lottery winners).
- `import os` lets your code explore your computer's folders and files!
- `import numpy` gives you a mathematical supercomputer inside Python!

Let's learn how to use these tools step-by-step.

## 📦 Module 1: Inbuilt Functions Deep Dive

### ⚠️ Top Placement Trap: Banker's Rounding in `round()`!
Python 3 does NOT always round `.5` upwards! It uses **Half to Even** (Banker's Rounding) to minimize statistical drift:
- `round(2.5)` evaluates to `2` (even!)
- `round(3.5)` evaluates to `4` (even!)


````python
# 1. Inbuilt Functions & Banker's Rounding
print("round(2.5):", round(2.5))  # 2 (Rounds to nearest EVEN number)
print("round(3.5):", round(3.5))  # 4 (Rounds to nearest EVEN number)
print("round(2.567, 2):", round(2.567, 2))

# 2. abs(), type(), eval() security warning
print("abs(-42.8):", abs(-42.8))
print("type(42):", type(42))

# eval() parses Python expressions dynamically:
expression = "3 * 10 + 5"
result = eval(expression)
print(f"eval('{expression}'):", result)
# ⚠️ Warning: Never use eval() with untrusted user input due to code injection risks!
````

## 📦 Module 2: Modules, Packages & `__name__ == '__main__'`

When Python imports a file, it executes the code at top-level. The special variable `__name__` is set to:
- `"__main__"` when executed directly as a standalone script.
- The module's filename (e.g. `"math_utils"`) when imported by another script.


````python
# Demonstrating __name__
print("Current module name:", __name__)
if __name__ == "__main__":
    print("Executing directly as the main application script.")
````

### The 12 Essential Math Functions:
`sqrt()`, `pow()`, `ceil()`, `floor()`, `factorial()`, `gcd()`, `fabs()`, `log()`, `sin()`, `cos()`, `pi`, `e`.


````python
import math, random

# 1. The 12 Math Functions
print("math.sqrt(144):", math.sqrt(144))
print("math.pow(2, 5):", math.pow(2, 5))
print("math.ceil(4.1):", math.ceil(4.1))
print("math.floor(4.9):", math.floor(4.9))
print("math.factorial(6):", math.factorial(6))
print("math.gcd(48, 180):", math.gcd(48, 180))
print("math.fabs(-12.5):", math.fabs(-12.5))
print("math.log(math.e):", math.log(math.e))
print("math.sin(math.pi / 2):", math.sin(math.pi / 2))
print("math.cos(0):", math.cos(0))
print("math.pi:", math.pi)
print("math.e:", math.e)

# 2. Random Module: Reproducibility with seed()
random.seed(42)
print("\nrandom.random() [0.0 to 1.0):", random.random())
print("random.randint(1, 10) [both inclusive!]:", random.randint(1, 10))

items = ["Aarav", "Diya", "Rohan", "Sneha", "Karan"]
print("random.choice(items):", random.choice(items))
print("random.sample(items, 3) [no duplicates]:", random.sample(items, 3))
random.shuffle(items)
print("Shuffled items in-place:", items)
````

## 📦 Module 4: `time` (8 Functions), `os` (12 Functions) & `datetime`

### Time Functions:
`time()`, `sleep()`, `ctime()`, `localtime()`, `gmtime()`, `strftime()`, `mktime()`, `perf_counter()` (high-resolution benchmarking).

### OS Functions:
`getcwd()`, `listdir()`, `mkdir()`, `makedirs()`, `rmdir()`, `remove()`, `rename()`, `path.exists()`, `path.join()`, `path.isfile()`, `path.isdir()`, `path.getsize()`.


````python
import time, os
from datetime import datetime, date, timedelta

# 1. High-resolution benchmarking with time.perf_counter()
t0 = time.perf_counter()
total = sum(i ** 2 for i in range(100000))
elapsed = time.perf_counter() - t0
print(f"Computed sum in {elapsed * 1000:.3f} ms")

# Time formatting
print("Current Epoch time:", time.time())
print("time.ctime():", time.ctime())

# 2. OS Module File Operations
current_dir = os.getcwd()
print("\nCurrent working directory:", current_dir)
dummy_path = os.path.join(current_dir, "test_file.tmp")
print("Boundary-safe path.join:", dummy_path)
print("path.exists(current_dir):", os.path.exists(current_dir))
print("path.isdir(current_dir):", os.path.isdir(current_dir))

# 3. Datetime and Timedelta Arithmetic
now = datetime.now()
print("\nFormatted Datetime:", now.strftime("%Y-%m-%d %H:%M:%S"))
future_deadline = now + timedelta(days=7, hours=4)
print("Deadline (+7 days, 4 hrs):", future_deadline.strftime("%A, %d %B %Y"))
difference = future_deadline - now
print("Time delta days:", difference.days)
````

## 📦 Module 5: Introduction to Scientific Computing with NumPy

NumPy introduces continuous, memory-aligned **C-arrays (`ndarray`)** that allow blazing fast SIMD vectorization without Python loop overhead.


````python
import numpy as np

# 1. Array creation & Reshaping
arr_1d = np.arange(1, 13)
print("1D Array:", arr_1d)

# Reshape into 3x4 Matrix
matrix_2d = arr_1d.reshape(3, 4)
print("\n2D Matrix (3x4):\n", matrix_2d)

# 2. Vectorized Arithmetic (No for-loops!)
doubled = matrix_2d * 2
print("\nVectorized Multiply by 2:\n", doubled)

# 3. Statistical Aggregation
print(f"\nMatrix Mean: {matrix_2d.mean():.2f}")
print(f"Matrix Sum: {matrix_2d.sum()}")
print(f"Column-wise Sum (axis=0): {matrix_2d.sum(axis=0)}")
print(f"Row-wise Sum (axis=1): {matrix_2d.sum(axis=1)}")
````

````python
# Challenge 1: Safe Mathematical Distance
import math
def euclidean_distance(p1, p2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))

assert round(euclidean_distance((0, 0), (3, 4)), 2) == 5.0
assert round(euclidean_distance((1, 2, 3), (4, 6, 8)), 2) == 7.07
print("✅ Challenge 1 Passed!")
````

````python
# Challenge 2: Deterministic Lucky Draw
import random
def pick_winners(participants, count, seed_val=42):
    random.seed(seed_val)
    if count > len(participants):
        raise ValueError("Count exceeds candidate pool")
    return random.sample(sorted(participants), count)

cands = ["Aarav", "Diya", "Rohan", "Sneha", "Karan", "Pooja"]
assert pick_winners(cands, 2, seed_val=10) == pick_winners(cands, 2, seed_val=10)
print("✅ Challenge 2 Passed!")
````

````python
# Challenge 3: Path Normalizer
import os
def normalize_and_validate(base, *subpaths):
    full = os.path.normpath(os.path.join(base, *subpaths))
    return full

p = normalize_and_validate("C:/projects", "gla", "..", "gla", "main.py")
assert p.endswith(os.path.join("gla", "main.py"))
print("✅ Challenge 3 Passed!")
````

````python
# Challenge 4: Age in Days Calculator
from datetime import date
def calculate_days_alive(birth_date_str):
    # format: YYYY-MM-DD
    parts = [int(x) for x in birth_date_str.split("-")]
    b_date = date(parts[0], parts[1], parts[2])
    target = date(2026, 11, 16)
    return (target - b_date).days

assert calculate_days_alive("2026-11-06") == 10
print("✅ Challenge 4 Passed!")
````

````python
# Challenge 5: Performance Profiler Function
import time
def benchmark_callable(fn, *args, **kwargs):
    t0 = time.perf_counter()
    res = fn(*args, **kwargs)
    t1 = time.perf_counter()
    return res, (t1 - t0)

res, duration = benchmark_callable(sum, [1, 2, 3, 4, 5])
assert res == 15
assert duration >= 0.0
print("✅ Challenge 5 Passed!")
````

````python
# Challenge 6: Banker's Rounding Formatter
def format_currency_round(values):
    return [round(v) for v in values]

# Tests Banker's even rounding:
assert format_currency_round([1.5, 2.5, 3.5, 4.5]) == [2, 2, 4, 4]
print("✅ Challenge 6 Passed!")
````

````python
# Challenge 7: NumPy Matrix Normalizer
import numpy as np
def min_max_normalize(matrix):
    min_val = matrix.min()
    max_val = matrix.max()
    if min_val == max_val:
        return np.zeros_like(matrix)
    return (matrix - min_val) / (max_val - min_val)

m = np.array([[10, 20], [30, 40]])
norm = min_max_normalize(m)
assert norm[0, 0] == 0.0 and norm[1, 1] == 1.0
print("✅ Challenge 7 Passed!")
````

````python
# Challenge 8: Safe Dynamic Calculator
def safe_eval_arithmetic(expr):
    # Only allow digits, spaces, and + - * /
    allowed = set("0123456789+-*/. ()")
    if not all(ch in allowed for ch in expr):
        raise ValueError("Untrusted characters detected")
    return eval(expr, {"__builtins__": None}, {})

assert safe_eval_arithmetic("10 + 20 * 3") == 70
try:
    safe_eval_arithmetic("__import__('os').system('ls')")
    assert False
except ValueError:
    pass
print("✅ Challenge 8 Passed!")
````

````python
# Challenge 9: NumPy 2D Flatten and Top-K
import numpy as np
def top_k_elements(matrix, k):
    flat = matrix.flatten()
    flat.sort()
    return list(flat[-k:][::-1])

mat = np.array([[12, 45, 2], [67, 89, 23]])
assert top_k_elements(mat, 3) == [89, 67, 45]
print("✅ Challenge 9 Passed!")
````

````python
# Challenge 10: Date Range Generator
from datetime import date, timedelta
def generate_workdays(start_date_str, num_days):
    parts = [int(x) for x in start_date_str.split("-")]
    curr = date(parts[0], parts[1], parts[2])
    workdays = []
    while len(workdays) < num_days:
        if curr.weekday() < 5:  # Mon=0, Fri=4
            workdays.append(curr.strftime("%Y-%m-%d"))
        curr += timedelta(days=1)
    return workdays

# Starting on Friday 2026-11-13, 3 workdays -> Fri, Mon (16), Tue (17)
w_days = generate_workdays("2026-11-13", 3)
assert w_days == ["2026-11-13", "2026-11-16", "2026-11-17"]
print("✅ Challenge 10 Passed! All 10 Challenges Verified Successfully!")
````
