### 🎬 Story
Utkarsh ke gift-stall par **Russian Matryoshka dolls** 🪆 bikne aaye. Badi doll kholo → andar ek chhoti doll. Use kholo → andar usse bhi chhoti... Aise **tab tak khulte jao jab tak sabse chhoti, thos doll na mil jaye.**

### 1.1 Recursion Kya Hai?
**Recursion = ek function jo apne aap ko hi call karta hai** — har baar problem ko thoda **chhota** karke.

| Doll ki kahani | Programming |
|---|---|
| Doll kholo, andar nayi doll | Function chalao, andar wahi function dobara |
| Har doll pehli se chhoti | Har call mein input chhota hota hai (`n - 1`) |
| Sabse chhoti thos doll | **Base case** — yahan ruk jao |

**Har recursive function ke 2 hisse hote hain:**

| Hissa | Kaam | Zaroori? |
|---|---|:-:|
| 🛑 **Base case** | Sabse chhota / aasaan case jiska jawab **seedha** pata hai — yahan recursion **rukta** hai | ✅ **Bilkul zaroori** |
| 🔁 **Recursive case** | Function **khud ko** chhote input ke saath call karta hai | ✅ |

```python
def solve(n):
    if n == 0:            # 🛑 BASE CASE   -> rukne ki shart
        return ...
    return ... solve(n - 1) ...    # 🔁 RECURSIVE CASE -> chhota problem, khud ko bulao
```

> ⚠️ **Base case ke bina function kabhi rukega nahi** → Python `RecursionError` de dega.

````python
# 🚀 Sabse seedha example: Countdown
def countdown(n):
    if n == 0:                    # 🛑 base case
        print("🚀 Liftoff!")
        return
    print(n)
    countdown(n - 1)              # 🔁 recursive case: chhota number

countdown(5)
````

````python
# 🪆 Doll kholne wala example
def open_doll(size):
    if size == 1:                                   # 🛑 sabse chhoti doll
        print("  " * (5 - size) + f"🪆 Doll {size}: thos doll mili! Kaam khatam.")
        return
    print("  " * (5 - size) + f"🪆 Doll {size}: kholo... andar doll {size - 1} hai")
    open_doll(size - 1)                             # 🔁 andar wali doll kholo

open_doll(5)
````

### 1.2 Call Stack — Python Andar Kya Karta Hai? 📚
Jab function doosre function ko bulata hai, Python purane function ko **rok kar** (pause) naye function ke liye ek naya **frame (dibba)** banata hai aur sabko ek **stack** (plates ki dher) mein rakhta hai.

- 🔽 **Winding (andar jana):** har recursive call par ek naya frame **upar** push hota hai.
- 🛑 **Base case** par sabse upar wala frame seedha jawab deta hai.
- 🔼 **Unwinding (bahar aana):** frame **ek-ek karke pop** hote hain, har ek apna kaam poora karta hai.

```
 factorial(3) ka call stack (Last In, First Out)

 Step 1: call            Step 2: call            Step 3: call           Step 4: base case
 ┌────────────┐          ┌────────────┐          ┌────────────┐         ┌────────────┐
 │            │          │            │          │            │         │ fact(0)=1  │ ◄── yahan se
 │            │          │            │          │ fact(1)    │         │ fact(1)    │     wapas
 │            │          │ fact(2)    │          │ fact(2)    │         │ fact(2)    │     aana
 │ fact(3)    │          │ fact(3)    │          │ fact(3)    │         │ fact(3)    │     shuru
 └────────────┘          └────────────┘          └────────────┘         └────────────┘

 Unwinding:  fact(0)=1 → fact(1)=1*1=1 → fact(2)=2*1=2 → fact(3)=3*2=6 ✅
```

````python
# 👀 Going IN aur coming OUT dekhte hain
def show(n):
    if n == 0:
        print("   🛑 base case mila (n = 0)")
        return
    print(f"  ⬇️ andar jaa raha hai : n = {n}")
    show(n - 1)
    print(f"  ⬆️ bahar aa raha hai  : n = {n}")

show(3)
````

````python
# 📚 Factorial ka poora call-stack trace (indentation = stack ki gehraai)
def fact(n, depth=0):
    pad = "    " * depth
    print(f"{pad}fact({n}) called")
    if n == 0:
        print(f"{pad}fact(0) -> base case, return 1")
        return 1
    result = n * fact(n - 1, depth + 1)
    print(f"{pad}fact({n}) -> {n} * fact({n - 1}) = {result}")
    return result

print("Final answer:", fact(3))
````

### 1.3 Stack Overflow & Recursion Limit 💥
Stack ki jagah **limited** hoti hai. Python ne ek **safety limit** lagayi hai (usually **1000** calls). Isse zyada gehrai par `RecursionError` aata hai — ye Python ka **stack overflow** alarm hai.

| Cheez | Kaam |
|---|---|
| `sys.getrecursionlimit()` | Abhi ki limit batata hai |
| `sys.setrecursionlimit(n)` | Limit badal deta hai |

#### ⚠️ Limit badhana — dhyaan se!
- Bahut zyada limit kar di aur real stack khatam ho gaya toh **Python crash** ho sakta hai.
- **Behtar solution:** base case sahi rakho, ya loop / memoization (Chapter 6) use karo.
- Limit badhao toh kaam ke baad **purani limit wapas** kar do.

````python
import sys

print("Abhi ki limit:", sys.getrecursionlimit())

# ❌ Base case ke bina — kabhi rukega nahi
depth = 0
def no_base_case():
    global depth
    depth += 1
    no_base_case()               # koi rukne ki shart nahi!

try:
    no_base_case()
except RecursionError as e:
    print("\n💥 RecursionError aaya!")
    print("   Message          :", e)
    print("   Kitni gehraai tak :", "limit ke aas-paas ->", depth > 500)
````

````python
import sys

def sum_to(n):
    if n == 0:
        return 0
    return n + sum_to(n - 1)

old_limit = sys.getrecursionlimit()

# Chhota kaam theek chalta hai
print("sum_to(100)  =", sum_to(100))

# Gehra kaam — default limit par fail
try:
    print(sum_to(2500))
except RecursionError:
    print("sum_to(2500) -> ❌ RecursionError (default limit se zyada gehra)")

# Limit badhao, kaam karo, limit wapas karo
sys.setrecursionlimit(5000)
print("Nayi limit    :", sys.getrecursionlimit())
print("sum_to(2500) =", sum_to(2500), "✅")
sys.setrecursionlimit(old_limit)          # 🔙 purani limit wapas
print("Limit wapas   :", sys.getrecursionlimit() == old_limit)
````

### 2.1 Factorial — `n! = n × (n-1)!`
| | |
|---|---|
| 🧠 Chhota version | `n! = n × (n-1)!` |
| 🛑 Base case | `0! = 1` (aur `1! = 1`) |

````python
import math

def factorial(n):
    if n == 0:                       # 🛑 base case
        return 1
    return n * factorial(n - 1)      # 🔁 n x (n-1)!

for n in [0, 1, 5, 10]:
    print(f"{n}! = {factorial(n)}")

# Python ke ready-made math.factorial se verify
print("\nmath.factorial(10) se match?", factorial(10) == math.factorial(10))

print("\n--- Expansion: factorial(4) ---")
print("factorial(4) = 4 * factorial(3)")
print("             = 4 * (3 * factorial(2))")
print("             = 4 * (3 * (2 * factorial(1)))")
print("             = 4 * (3 * (2 * (1 * factorial(0))))")
print("             = 4 * 3 * 2 * 1 * 1 =", factorial(4))
````

### 2.2 Sum of Natural Numbers — `S(n) = n + S(n-1)`
| | |
|---|---|
| 🧠 Chhota version | `1 + 2 + ... + n = n + (1 + ... + (n-1))` |
| 🛑 Base case | `S(0) = 0` |

````python
def sum_natural(n):
    if n == 0:
        return 0
    return n + sum_natural(n - 1)

print("sum_natural(5)   =", sum_natural(5))
print("sum_natural(100) =", sum_natural(100))

# Formula n(n+1)/2 se verify
n = 100
print("Formula n(n+1)/2 =", n * (n + 1) // 2)
print("Dono same?       =", sum_natural(n) == n * (n + 1) // 2)

# 👀 Expansion
print("\nsum_natural(4) = 4 + sum_natural(3)")
print("               = 4 + (3 + sum_natural(2))")
print("               = 4 + (3 + (2 + (1 + sum_natural(0))))")
print("               = 4 + 3 + 2 + 1 + 0 =", sum_natural(4))
````

### 2.3 Sum of Digits — Last digit alag, baaki number alag 🔢
Trick: **`n % 10`** = last digit • **`n // 10`** = last digit hata ke bacha number

| | |
|---|---|
| 🧠 Chhota version | `sum_digits(1234) = 4 + sum_digits(123)` |
| 🛑 Base case | `n == 0` → `0` |

````python
def sum_digits(n):
    if n == 0:
        return 0
    return (n % 10) + sum_digits(n // 10)

def count_digits(n):
    if n < 10:                        # ek digit ka number -> 1
        return 1
    return 1 + count_digits(n // 10)

print("sum_digits(1234)   =", sum_digits(1234))
print("sum_digits(98765)  =", sum_digits(98765))
print("count_digits(98765)=", count_digits(98765))

print("\n--- n % 10 aur n // 10 ---")
n = 1234
print(f"{n} % 10  = {n % 10}   (last digit)")
print(f"{n} // 10 = {n // 10}  (last digit hata diya)")

print("\n--- Step-by-step: sum_digits(1234) ---")
for x in [1234, 123, 12, 1]:
    print(f"{x} -> last digit {x % 10}, bacha {x // 10}")
````

### 2.4 Power — `b^e = b × b^(e-1)`
| | |
|---|---|
| 🧠 Chhota version | `b^e = b × b^(e-1)` |
| 🛑 Base case | `b^0 = 1` |

#### ⚡ Fast Power (Divide & Conquer)
`b^e = (b^(e/2))²` — har baar exponent **aadha**! `2^1000` ke liye 1000 ki jagah sirf **~10 calls**.

````python
def power(b, e):
    if e == 0:
        return 1
    return b * power(b, e - 1)

print("power(2, 5)  =", power(2, 5))
print("power(5, 3)  =", power(5, 3))
print("power(7, 0)  =", power(7, 0))

# ⚡ Fast power: exponent har baar aadha
calls = 0
def fast_power(b, e):
    global calls
    calls += 1
    if e == 0:
        return 1
    half = fast_power(b, e // 2)
    if e % 2 == 0:
        return half * half
    return b * half * half

print("\nfast_power(2, 30) =", fast_power(2, 30), "| calls:", calls)
calls = 0
print("fast_power(3, 13) =", fast_power(3, 13), "| calls:", calls)

print("\nVerify: 2 ** 30 =", 2 ** 30, "| 3 ** 13 =", 3 ** 13)
````

### 2.5 GCD — Euclid ka Jaadu ✨
**GCD (Greatest Common Divisor)** = dono numbers ko divide karne wala sabse bada number.

Euclid ka rule: **`gcd(a, b) = gcd(b, a % b)`**

| | |
|---|---|
| 🧠 Chhota version | `gcd(48, 18) = gcd(18, 12) = gcd(12, 6) = gcd(6, 0)` |
| 🛑 Base case | `b == 0` → `a` hi GCD hai |

````python
import math

def gcd(a, b):
    if b == 0:                # 🛑 base case
        return a
    return gcd(b, a % b)      # 🔁 chhota problem

print("gcd(48, 18) =", gcd(48, 18))
print("gcd(17, 5)  =", gcd(17, 5))
print("gcd(100, 75)=", gcd(100, 75))
print("math.gcd se match?", gcd(48, 18) == math.gcd(48, 18))

# LCM = (a x b) / GCD
a, b = 12, 18
print(f"\nLCM({a}, {b}) = {a * b // gcd(a, b)}")

# 👀 Trace
def gcd_trace(a, b):
    print(f"gcd({a}, {b})")
    if b == 0:
        print(f"  -> base case, answer = {a}")
        return a
    return gcd_trace(b, a % b)

print("\n--- Trace ---")
gcd_trace(48, 18)
````

### 2.6 Fibonacci — Do Pichhle Numbers ka Jod 🐇
Series: `0, 1, 1, 2, 3, 5, 8, 13, 21, ...`

| | |
|---|---|
| 🧠 Chhota version | `fib(n) = fib(n-1) + fib(n-2)` |
| 🛑 Base cases | `fib(0) = 0`, `fib(1) = 1` *(do base cases!)* |

```
                     fib(4)
                  /          \
             fib(3)          fib(2)
            /     \          /    \
        fib(2)   fib(1)   fib(1)  fib(0)
        /   \
    fib(1) fib(0)
```
> 👀 Dekho: `fib(2)` **kai baar** compute ho raha hai! Bade `n` par ye bahut dheema ho jaata hai — **Chapter 6 (Memoization)** mein iska ilaaj hai. 💊

````python
def fib(n):
    if n == 0:             # 🛑 base case 1
        return 0
    if n == 1:             # 🛑 base case 2
        return 1
    return fib(n - 1) + fib(n - 2)

print("fib(10) =", fib(10))
print("Pehle 12 Fibonacci numbers:", [fib(i) for i in range(12)])

# 👀 Kitni baar call hua?
calls = 0
def fib_counted(n):
    global calls
    calls += 1
    if n < 2:
        return n
    return fib_counted(n - 1) + fib_counted(n - 2)

for n in [5, 10, 15, 20]:
    calls = 0
    value = fib_counted(n)
    print(f"fib({n:>2}) = {value:>5}  | function calls = {calls}")
````

### 3.1 Recursion Ka Pattern: **Head + Tail** ✂️
Strings aur lists ko do hisson mein tod do:

| Hissa | String / List mein |
|---|---|
| **Head** (pehla item) | `s[0]` |
| **Tail** (baaki sab) | `s[1:]` ← ye hamesha **chhota** hota hai! |
| **Base case** | Khaali (`""` / `[]`) ya ek hi item |

```
 "GLA"  ──►  head = "G"  +  tail = "LA"
 "LA"   ──►  head = "L"  +  tail = "A"
 "A"    ──►  head = "A"  +  tail = ""      ← base case: khaali!
```

### 3.2 String Reversal — `reverse(s) = reverse(tail) + head`
| | |
|---|---|
| 🧠 Chhota version | `reverse("GLA") = reverse("LA") + "G"` |
| 🛑 Base case | `s == ""` → `""` |

````python
def reverse(s):
    if s == "":                    # 🛑 base case
        return ""
    return reverse(s[1:]) + s[0]   # 🔁 tail ulta karo + head peeche lagao

print(reverse("GLA"))
print(reverse("Utkarsh"))
print(reverse("Python is fun"))
print("Slicing se match?", reverse("Utkarsh") == "Utkarsh"[::-1])

# 👀 Expansion
print("\nreverse('GLA') = reverse('LA') + 'G'")
print("               = (reverse('A') + 'L') + 'G'")
print("               = ((reverse('') + 'A') + 'L') + 'G'")
print("               = (('' + 'A') + 'L') + 'G' =", reverse("GLA"))
````

````python
# 📚 Trace: unwinding mein string kaise ban rahi hai
def reverse_trace(s, depth=0):
    pad = "  " * depth
    print(f"{pad}reverse({s!r})")
    if s == "":
        print(f"{pad}  -> base case: ''")
        return ""
    result = reverse_trace(s[1:], depth + 1) + s[0]
    print(f"{pad}  -> {result!r}")
    return result

reverse_trace("GLA")
````

### 3.3 Palindrome Check — Pehla aur Aakhri Character Milao 🪞
Palindrome = seedha aur ulta same (`madam`, `level`, `noon`).

| | |
|---|---|
| 🧠 Chhota version | Pehla == aakhri ho toh **beech wali string** ko check karo → `s[1:-1]` |
| 🛑 Base case 1 | Length `0` ya `1` → `True` |
| 🛑 Base case 2 | `s[0] != s[-1]` → `False` (jaldi bahar!) |

````python
def is_palindrome(s):
    if len(s) <= 1:                 # 🛑 khaali ya ek akshar -> palindrome
        return True
    if s[0] != s[-1]:               # 🛑 pehla aur aakhri alag -> nahi
        return False
    return is_palindrome(s[1:-1])   # 🔁 beech wala hissa check karo

for w in ["madam", "level", "python", "noon", "a", ""]:
    print(f"{w!r:10} -> {is_palindrome(w)}")

# Sentence: spaces, capital aur punctuation hatana padega
def clean(text):
    return "".join(ch.lower() for ch in text if ch.isalnum())

sentence = "A man, a plan, a canal: Panama"
print("\nSentence       :", sentence)
print("Saaf karke     :", clean(sentence))
print("Palindrome?    :", is_palindrome(clean(sentence)))
````

### 3.4 Counting Characters — Head match hua toh +1
| | |
|---|---|
| 🧠 Chhota version | `count("banana", "a") = (1 agar head == 'a' warna 0) + count(tail, "a")` |
| 🛑 Base case | `s == ""` → `0` |

````python
def count_char(s, ch):
    if s == "":
        return 0
    if s[0] == ch:
        return 1 + count_char(s[1:], ch)
    return count_char(s[1:], ch)

print("'a' in banana          :", count_char("banana", "a"))
print("'n' in banana          :", count_char("banana", "n"))
print("'z' in banana          :", count_char("banana", "z"))
print("Slicing .count() se    :", "banana".count("a"))

# Vowels ginna (helper ke saath)
def count_vowels(s):
    if s == "":
        return 0
    first = 1 if s[0].lower() in "aeiou" else 0
    return first + count_vowels(s[1:])

print("\nVowels in 'Recursion is Awesome':", count_vowels("Recursion is Awesome"))
````

### 3.5 Recursion on Lists — Sum, Length, Maximum 🔢
Head = `lst[0]`, Tail = `lst[1:]`, Base = `[]` (ya ek hi item).

| Kaam | Chhota version | Base case |
|---|---|---|
| Sum | `lst[0] + sum(tail)` | `[]` → `0` |
| Length | `1 + length(tail)` | `[]` → `0` |
| Maximum | `head` aur `max(tail)` mein se bada | ek item → wahi item |

````python
def sum_list(lst):
    if lst == []:
        return 0
    return lst[0] + sum_list(lst[1:])

def length(lst):
    if lst == []:
        return 0
    return 1 + length(lst[1:])

def max_list(lst):
    if len(lst) == 1:                 # 🛑 ek hi item -> wahi maximum
        return lst[0]
    rest_max = max_list(lst[1:])      # baaki ka maximum
    return lst[0] if lst[0] > rest_max else rest_max

scores = [34, 7, 89, 12, 56]
print("List    :", scores)
print("Sum     :", sum_list(scores), "| built-in sum:", sum(scores))
print("Length  :", length(scores), "| built-in len:", len(scores))
print("Maximum :", max_list(scores), "| built-in max:", max(scores))

# 👀 Sum ka expansion
print("\nsum_list([1, 2, 3, 4])")
print("= 1 + sum_list([2, 3, 4])")
print("= 1 + (2 + sum_list([3, 4]))")
print("= 1 + (2 + (3 + sum_list([4])))")
print("= 1 + (2 + (3 + (4 + sum_list([]))))")
print("= 1 + 2 + 3 + 4 + 0 =", sum_list([1, 2, 3, 4]))
````

### 3.6 Bonus Boss 🔥: Nested List Flatten
Loop se `[1, [2, [3, 4]], 5]` jaisi **kitni bhi gehri** list ko seedha karna mushkil hai — recursion ke liye aasaan! Agar item khud **list** hai, toh us par bhi wahi function lagao.

````python
def flatten(lst):
    if lst == []:
        return []
    head, tail = lst[0], lst[1:]
    if isinstance(head, list):                   # head khud list hai -> use bhi flatten karo
        return flatten(head) + flatten(tail)
    return [head] + flatten(tail)

nested = [1, [2, [3, 4]], 5, [[6], 7]]
print("Nested :", nested)
print("Flat   :", flatten(nested))

# Gehrai (depth) bhi recursion se
def depth(x):
    if not isinstance(x, list):
        return 0
    if x == []:
        return 1
    return 1 + max(depth(item) for item in x)

print("Depth  :", depth(nested))
````

# ⚖️ Chapter 4: Recursion vs Iteration — Loop Ya Khud Ko Bulana?

### 🎬 Story
Aarav ke dost Chirag ne bola: *"Bhai, recursion itna sundar hai toh loop kyun seekhe?"*
Bhavna Didi: *"Dono ek hi kaam karte hain. Par **kab kaun behtar** — ye samajhna asli skill hai!"* 🥊

**Iteration** = loop (`for` / `while`) se repeat karna.
**Recursion** = function ka khud ko call karke repeat karna.

````python
# Same kaam, do tareeke: Factorial
def fact_recursive(n):
    if n == 0:
        return 1
    return n * fact_recursive(n - 1)

def fact_iterative(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print("Recursive :", fact_recursive(10))
print("Iterative :", fact_iterative(10))
print("Same?     :", fact_recursive(10) == fact_iterative(10))

# Sum of list
def sum_recursive(lst):
    return 0 if lst == [] else lst[0] + sum_recursive(lst[1:])

def sum_iterative(lst):
    total = 0
    for x in lst:
        total += x
    return total

data = [3, 6, 9, 12]
print("\nList sum recursive:", sum_recursive(data), "| iterative:", sum_iterative(data))
````

### 4.1 Teen Maapdand: Readability, Speed, Memory 📏
| | **Recursion 🪆** | **Iteration 🔁** |
|---|---|---|
| **Readability** | Tree jaisi / "chhota version" wali problems mein **bahut saaf** (Hanoi, flatten, tree) | Simple repeat wale kaam (sum, count) mein **seedha aur chhota** |
| **Speed** ⏱️ | Har call ka overhead → **dheema** | **Tez** (sirf ek loop) |
| **Memory** 🧠 | Har call ke liye **naya stack frame** → `O(n)` memory | **Constant** memory (ek-do variables) |
| **Limit** | `RecursionError` (~1000 gehrai) | Koi limit nahi |
| **Kab use karein** | Naturally recursive structure: trees, nested data, divide & conquer, backtracking | Simple counting / accumulating, bade `n` ke liye |

> 💡 **Thumb rule:** Pehle **saaf** likho. Agar recursion se code **zyada saaf** ho, toh recursion. Agar bahut gehra / bahut tez chahiye, toh loop (ya memoization).

### 4.2 Speed Test ⏱️
Hum `time.perf_counter()` (Chapter 11 mein poora seekhenge) se samay naapenge.
Naive recursive Fibonacci **bahut dheema** hai kyunki wahi kaam baar-baar hota hai.

````python
import time

def fib_recursive(n):
    if n < 2:
        return n
    return fib_recursive(n - 1) + fib_recursive(n - 2)

def fib_iterative(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

n = 25

t0 = time.perf_counter()
r1 = fib_recursive(n)
t_rec = time.perf_counter() - t0

t0 = time.perf_counter()
r2 = fib_iterative(n)
t_iter = time.perf_counter() - t0

print(f"fib({n}) recursive : {r1}  in {t_rec:.5f} sec")
print(f"fib({n}) iterative : {r2}  in {t_iter:.7f} sec")
print("Same answer?     :", r1 == r2)
print(f"Iterative kitna tez? ~{t_rec / t_iter:,.0f} guna")
````

### 4.3 Memory Test 🧠 — Stack ki Gehraai
`inspect.stack()` se dekhte hain function ke andar **kitne frames** chal rahe hain.

````python
import inspect

def depth_here():
    return len(inspect.stack(0))

baseline = depth_here()           # ek function ke andar ki gehraai

def sum_rec(n):
    if n == 0:
        return depth_here()       # sabse gehre point par stack ki ginti
    return sum_rec(n - 1)

def sum_loop(n):
    total = 0
    for i in range(n):
        total += i
    return depth_here()           # loop mein stack nahi badhta

n = 200
print("Baseline frames           :", baseline)
print(f"Recursion (n={n}) gehraai :", sum_rec(n) - baseline, "extra frames")
print(f"Loop (n={n}) gehraai      :", sum_loop(n) - baseline, "extra frames")
print("\n👉 Recursion ne har call ke liye ek frame liya, loop ne koi nahi!")
````

### 5.1 Binary Search — Sorted List ko Aadha-Aadha Karo ✂️
**Zaroori shart: list pehle se SORTED honi chahiye!**

| Step | Kya karo |
|---|---|
| 1 | `lo` (shuru) aur `hi` (end) rakho |
| 2 | `mid = (lo + hi) // 2` |
| 3 | `arr[mid] == target` → **mil gaya!** ✅ |
| 4 | `target < arr[mid]` → **left half** mein dhoondho (`hi = mid - 1`) |
| 5 | `target > arr[mid]` → **right half** mein dhoondho (`lo = mid + 1`) |
| 🛑 Base case | `lo > hi` → **nahi mila** (`-1`) |

```
 [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]      target = 23
  lo=0                          hi=9
  mid=4 -> 16   23 > 16  ► right half
                 lo=5            hi=9
                 mid=7 -> 56   23 < 56  ► left half
                 lo=5  hi=6
                 mid=5 -> 23  ✅ MIL GAYA index 5
```

````python
def binary_search(arr, target, lo, hi):
    if lo > hi:                                   # 🛑 base case: nahi mila
        return -1
    mid = (lo + hi) // 2
    if arr[mid] == target:                        # 🛑 base case: mil gaya
        return mid
    if target < arr[mid]:
        return binary_search(arr, target, lo, mid - 1)      # 🔁 left half
    return binary_search(arr, target, mid + 1, hi)          # 🔁 right half

arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]

for t in [23, 91, 2, 50]:
    idx = binary_search(arr, t, 0, len(arr) - 1)
    print(f"{t:>3} ->", f"index {idx}" if idx != -1 else "nahi mila ❌")
````

````python
# 👀 Trace: har step par kya ho raha hai
def binary_search_trace(arr, target, lo, hi, step=1):
    if lo > hi:
        print(f"Step {step}: lo={lo} > hi={hi}  -> nahi mila")
        return -1
    mid = (lo + hi) // 2
    print(f"Step {step}: lo={lo}, hi={hi}, mid={mid}, arr[mid]={arr[mid]}")
    if arr[mid] == target:
        print(f"        ✅ mil gaya index {mid} par!")
        return mid
    if target < arr[mid]:
        print(f"        {target} < {arr[mid]}  -> left half")
        return binary_search_trace(arr, target, lo, mid - 1, step + 1)
    print(f"        {target} > {arr[mid]}  -> right half")
    return binary_search_trace(arr, target, mid + 1, hi, step + 1)

arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
binary_search_trace(arr, 23, 0, len(arr) - 1)
print()
binary_search_trace(arr, 50, 0, len(arr) - 1)
````

### 5.2 Binary Search kitna Tez? ⚡
Har step mein list **aadhi** hoti hai → maximum steps ≈ **log₂(n)**.

| Items (n) | Linear search (worst) | Binary search (worst) |
|---:|---:|---:|
| 1,000 | 1,000 | ~10 |
| 1,000,000 | 1,000,000 | ~20 |
| 1,000,000,000 | 1,000,000,000 | ~30 |

````python
import math

calls = 0
def bs(arr, target, lo, hi):
    global calls
    calls += 1
    if lo > hi:
        return -1
    mid = (lo + hi) // 2
    if arr[mid] == target:
        return mid
    if target < arr[mid]:
        return bs(arr, target, lo, mid - 1)
    return bs(arr, target, mid + 1, hi)

for n in [1_000, 1_000_000]:
    big = list(range(1, n + 1))
    calls = 0
    idx = bs(big, n, 0, n - 1)          # sabse peeche wala item
    print(f"n = {n:>9,} | binary calls = {calls:>2} | log2(n) = {math.log2(n):.1f} | linear steps = {n:,}")

# Ek asli example
roll_numbers = list(range(1, 1001))
calls = 0
idx = bs(roll_numbers, 742, 0, 999)
print(f"\nRoll 742 mila index {idx} par, sirf {calls} calls mein! 🚀")
````

### 5.3 Tower of Hanoi 🗼
**Paheli:** 3 pegs hain — **A (source)**, **B (helper)**, **C (destination)**. A par `n` disks hain (bada neeche, chhota upar). Sabhi disks A se C par le jaane hain.

**Rules:**
1. Ek baar mein **ek hi disk** move kar sakte ho.
2. **Bada disk kabhi chhote ke upar nahi** rakh sakte.

```
      |        |        |
     (1)       |        |          n = 3
    (2 2)      |        |
   (3 3 3)     |        |
  ─────────  ─────────  ─────────
      A          B          C
```

#### 🧠 Recursive Soch
`n` disks ko A se C le jaana =
1. **Upar ke `n-1` disks** A se **B** par le jao (C ko helper banake)
2. **Sabse bade disk** ko A se **C** par rakho
3. B wale **`n-1` disks** ko B se **C** par le jao (A ko helper banake)

🛑 **Base case:** `n == 0` → kuch nahi karna.

**Total moves = 2ⁿ − 1**

````python
def hanoi(n, source, target, helper):
    if n == 0:                                        # 🛑 base case
        return
    hanoi(n - 1, source, helper, target)              # 1. n-1 disks: source -> helper
    print(f"Move disk {n}: {source} -> {target}")     # 2. sabse bada: source -> target
    hanoi(n - 1, helper, target, source)              # 3. n-1 disks: helper -> target

print("=== n = 1 ===")
hanoi(1, "A", "C", "B")
print("\n=== n = 2 ===")
hanoi(2, "A", "C", "B")
print("\n=== n = 3 (7 moves) ===")
hanoi(3, "A", "C", "B")
````

````python
# 🎬 Pegs ki state har move ke baad dekho (visual simulation)
pegs = {"A": [3, 2, 1], "B": [], "C": []}      # list ka last item = sabse upar ka disk
move_no = 0

def show():
    print(f"   A:{pegs['A']}   B:{pegs['B']}   C:{pegs['C']}")

def hanoi_visual(n, source, target, helper):
    global move_no
    if n == 0:
        return
    hanoi_visual(n - 1, source, helper, target)
    disk = pegs[source].pop()
    pegs[target].append(disk)
    move_no += 1
    print(f"Move {move_no}: disk {disk} {source} -> {target}")
    show()
    hanoi_visual(n - 1, helper, target, source)

print("Shuruaat:")
show()
hanoi_visual(3, "A", "C", "B")
print("\n✅ Saare disks C par pahunch gaye!")
````

````python
# 📈 Moves ka formula: 2^n - 1
def hanoi_count(n):
    if n == 0:
        return 0
    return 2 * hanoi_count(n - 1) + 1      # do baar (n-1) + ek bada move

print(f"{'Disks':>5} | {'Moves (recursion)':>17} | {'2^n - 1':>8}")
for n in range(1, 11):
    print(f"{n:>5} | {hanoi_count(n):>17} | {2 ** n - 1:>8}")

# Legend: 64 disks
moves = 2 ** 64 - 1
years = moves / (60 * 60 * 24 * 365)
print(f"\n🏛️ Legend: 64 disks = {moves:,} moves")
print(f"   Ek move/second par = {years:,.0f} saal (universe ki umar se bhi zyada!)")
````

### 6.1 Problem: Overlapping Subproblems
```
                     fib(5)
                   /        \
              fib(4)         fib(3)  ◄── dobara!
             /     \         /    \
         fib(3)   fib(2)  fib(2)  fib(1)    ◄── fib(3), fib(2) kai baar!
```

````python
import time

calls = 0
def fib_naive(n):
    global calls
    calls += 1
    if n < 2:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)

print(f"{'n':>3} | {'fib(n)':>10} | {'function calls':>14}")
for n in [10, 15, 20, 25]:
    calls = 0
    v = fib_naive(n)
    print(f"{n:>3} | {v:>10} | {calls:>14,}")

t0 = time.perf_counter()
calls = 0
fib_naive(27)
print(f"\nfib(27) mein {calls:,} calls aur {time.perf_counter() - t0:.2f} second laga")
print("fib(40) mein toh 33 crore+ calls lagenge... bahut der! 🐌")
````

### 6.2 Memoization with Dictionary Cache 📒
Idea: ek **dictionary (`cache`)** rakho — `{n: fib(n)}`.
1. Call aaye → pehle dekho `n` **cache mein hai?** Haan toh seedha wahi jawab. 🎯
2. Nahi hai → compute karo, **cache mein save karo**, phir return.

````python
calls = 0

def fib_memo(n, cache=None):
    global calls
    calls += 1
    if cache is None:                 # har naye bahari call par nayi dictionary
        cache = {}
    if n in cache:                    # ✅ pehle se pata hai!
        return cache[n]
    if n < 2:
        result = n
    else:
        result = fib_memo(n - 1, cache) + fib_memo(n - 2, cache)
    cache[n] = result                 # 💾 yaad kar lo
    return result

for n in [10, 20, 30, 50, 80]:
    calls = 0
    print(f"fib({n:>2}) = {fib_memo(n):>20}  | calls = {calls}")

print("\n🤯 fib(80) sirf 159 calls mein! Naive mein ye kabhi khatam nahi hota.")
print("fib(100) =", fib_memo(100))
````

````python
# Dictionary cache ko andar se dekhna
def fib_show(n, cache):
    if n in cache:
        print(f"  cache hit : fib({n}) = {cache[n]}")
        return cache[n]
    if n < 2:
        cache[n] = n
        return n
    value = fib_show(n - 1, cache) + fib_show(n - 2, cache)
    cache[n] = value
    print(f"  saved     : fib({n}) = {value}")
    return value

cache = {}
print("fib(6) ki journey:")
fib_show(6, cache)
print("\nFinal cache:", cache)
````

### 6.3 `functools.lru_cache` — Python ka Ready-made Yaaddasht 🪄
Khud dictionary banane ki zaroorat nahi! Function ke upar **`@lru_cache`** lagao — Python apne aap cache sambhal leta hai.

| Cheez | Kaam |
|---|---|
| `@lru_cache(maxsize=None)` | Unlimited cache (`maxsize=128` default) |
| `@functools.cache` | `lru_cache(maxsize=None)` ka chhota rup (Python 3.9+) |
| `f.cache_info()` | `hits`, `misses`, `currsize`, `maxsize` dikhata hai |
| `f.cache_clear()` | Cache khaali kar deta hai |

> **LRU = Least Recently Used** → `maxsize` bhar jaye toh sabse purana unused jawab hata deta hai.
> `@` wali line ko **decorator** kehte hain — function ko ek superpower deta hai. (Detail aage ke course mein.)

````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print("fib(10) =", fib(10))
print("fib(50) =", fib(50))
print("fib(90) =", fib(90))

print("\nCache report:", fib.cache_info())
print("  hits    -> cache se mile jawab")
print("  misses  -> naye compute hue")

fib.cache_clear()
print("\ncache_clear() ke baad:", fib.cache_info())

# Chhota maxsize
@lru_cache(maxsize=3)
def square(n):
    return n * n

for n in [1, 2, 3, 4]:
    square(n)
print("\nmaxsize=3 par 4 alag inputs:", square.cache_info())
````

````python
import time
from functools import lru_cache

def fib_slow(n):
    return n if n < 2 else fib_slow(n - 1) + fib_slow(n - 2)

@lru_cache(maxsize=None)
def fib_fast(n):
    return n if n < 2 else fib_fast(n - 1) + fib_fast(n - 2)

n = 28
t0 = time.perf_counter(); a = fib_slow(n); t_slow = time.perf_counter() - t0
t0 = time.perf_counter(); b = fib_fast(n); t_fast = time.perf_counter() - t0

print(f"Naive    fib({n}) = {a}  -> {t_slow:.4f} sec")
print(f"lru_cache fib({n}) = {b}  -> {t_fast:.6f} sec")
print("Answers same?:", a == b)
print(f"lru_cache ~{t_slow / t_fast:,.0f} guna tez!")
````

### 6.4 Memoization Aur Kahan Kaam Aata Hai? 🧩
Jahan bhi **same sub-problem baar-baar** aata ho:

| Problem | Recursive soch |
|---|---|
| 🪜 **Seedhiyan chadhna** (1 ya 2 steps) | `ways(n) = ways(n-1) + ways(n-2)` |
| 🧭 **Grid paths** (sirf right / down) | `paths(r, c) = paths(r-1, c) + paths(r, c-1)` |

````python
import math
from functools import lru_cache

# 🪜 Seedhiyan: n steps, ek baar mein 1 ya 2 step
@lru_cache(maxsize=None)
def stairs(n):
    if n <= 1:
        return 1
    return stairs(n - 1) + stairs(n - 2)

print("Seedhiyan (n=1..8):", [stairs(n) for n in range(1, 9)])
print("40 seedhiyan chadhne ke tareeke:", stairs(40))

# 🧭 Grid paths
@lru_cache(maxsize=None)
def paths(r, c):
    if r == 1 or c == 1:           # sirf ek hi raasta
        return 1
    return paths(r - 1, c) + paths(r, c - 1)

print("\n3x3 grid  :", paths(3, 3))
print("5x5 grid  :", paths(5, 5))
print("18x18 grid:", paths(18, 18))
print("math.comb se verify:", paths(18, 18) == math.comb(34, 17))

# ⚠️ lru_cache ke arguments hashable hone chahiye
try:
    stairs([1, 2])
except TypeError as e:
    print("\n⚠️ TypeError:", e)
````

# 🧰 Chapter 8: `zip()`, `enumerate()`, `map()`, `filter()` — Zip & Dost

### 🎬 Story
Aarav ke paas **do alag lists** thi: `names` aur `marks`. Bhavna Didi ne jacket ki **zip** dikhayi: *"Jaise zip ke dono taraf ke daant ek-ek karke jud jaate hain, waise hi `zip()` do lists ke items ko **jod** deta hai!"* 🧥

**`map()` aur `filter()`** Day 2 mein seekhe the — aaj hum unhe **inbuilt functions ke saath** aur **`zip` / `enumerate` ke saath** milakar dekhte hain.

### 8.1 `zip()` — Lists ko Jodna 🤐
`zip(a, b, c)` → har list ka **pehla item saath**, phir **doosra item saath**...

```
names  = ["Aarav", "Bhavna", "Chirag"]
marks  = [94, 91, 88]
            ↓
zip → ("Aarav", 94), ("Bhavna", 91), ("Chirag", 88)
```
| Baat | Detail |
|---|---|
| Return | **zip object** (lazy) — `list(...)`, `dict(...)` se dekho |
| Lambai alag ho | **Sabse chhoti** list tak ruk jaata hai |
| Alag-alag karna (unzip) | `zip(*pairs)` |
| Lambi list tak jaana ho | `itertools.zip_longest` |

````python
names = ["Aarav", "Bhavna", "Chirag"]
marks = [94, 91, 88]
city  = ["Mathura", "Agra", "Delhi"]

print("list(zip(names, marks))      :", list(zip(names, marks)))
print("dict(zip(names, marks))      :", dict(zip(names, marks)))

print("\n--- Loop mein ---")
for n, m in zip(names, marks):
    print(f"{n} scored {m}")

print("\n--- Teen lists ---")
for n, m, c in zip(names, marks, city):
    print(f"{n} ({c}) -> {m}")

print("\n--- Lambai alag: chhoti list par ruk jaata hai ---")
print(list(zip([1, 2, 3, 4, 5], ["a", "b"])))

# zip_longest
from itertools import zip_longest
print("zip_longest:", list(zip_longest([1, 2, 3], ["a"], fillvalue="-")))
````

### 8.2 Unzip & Transpose — `zip(*data)` 🔄
`*` (star) list ko **alag-alag arguments** mein todta hai. `zip(*pairs)` = jodiyon ko wapas alag-alag lists mein.
**Matrix transpose** (rows ↔ columns) bhi isi se!

````python
pairs = [("Aarav", 94), ("Bhavna", 91), ("Chirag", 88)]

names, marks = zip(*pairs)              # unzip
print("names :", names)
print("marks :", marks)

# Matrix transpose
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]
transposed = [list(row) for row in zip(*matrix)]
print("\nMatrix    :", matrix)
print("Transpose :", transposed)

# Do lists ke elements ka jod
a = [1, 2, 3]
b = [10, 20, 30]
print("\nElement-wise sum:", [x + y for x, y in zip(a, b)])
print("Dot product     :", sum(x * y for x, y in zip(a, b)))
````

### 8.3 `enumerate()` — Index + Value Saath 🔢
`enumerate(iterable, start=0)` → `(index, value)` jodiyan. `range(len(...))` ka **behtar** vikalp!

````python
names = ["Aarav", "Bhavna", "Chirag"]

# ❌ Purana tarika
for i in range(len(names)):
    print(i, names[i])

print("-" * 25)

# ✅ enumerate se
for i, name in enumerate(names):
    print(i, name)

print("\nRoll number 1 se shuru:")
for roll, name in enumerate(names, start=1):
    print(f"  Roll {roll}: {name}")

# String par
print("\nString par:", list(enumerate("GLA")))

# Index ke saath dictionary banana
print("Index map  :", dict(enumerate(names, start=1)))

# Maximum marks ka index
marks = [78, 92, 45, 88]
best_i = max(range(len(marks)), key=lambda i: marks[i])
print("\nTopper ka index (range+max):", best_i)

# enumerate + zip: dono ek saath
scores = [94, 91, 88]
for rank, (n, m) in enumerate(zip(names, scores), start=1):
    print(f"Rank {rank}: {n} - {m}")
````

### 8.4 `map()` aur `filter()` — Inbuilt Functions ke Saath 🗺️🔍
Recap: **`map(func, data)`** har item **badalta** hai • **`filter(func, data)`** sirf kuch items **rakhta** hai.
Aaj dekhte hain inhe **inbuilt functions** (`int`, `abs`, `round`, `len`, `str.upper`, `str.isdigit`) ke saath — **lambda ki zaroorat nahi!**

````python
# --- map() with inbuilt functions ---
print("int   :", list(map(int, "10 20 30".split())))
print("abs   :", list(map(abs, [-5, 3, -8, 0])))
print("len   :", list(map(len, ["GLA", "Utkarsh", "Python"])))
print("upper :", list(map(str.upper, ["aarav", "bhavna"])))
print("round :", list(map(round, [1.2, 2.5, 3.5, 4.7])))
print("str   :", list(map(str, [1, 2, 3])))

# map + zip: do lists par
prices = [100, 250, 80]
qty = [2, 1, 5]
print("\nLine totals (map+zip):", list(map(lambda p, q: p * q, prices, qty)))
print("Same with map(mul)   :", list(map(int.__mul__, prices, qty)))

# --- filter() with inbuilt functions ---
tokens = ["10", "abc", "25", "x9", "7"]
print("\nfilter(str.isdigit):", list(filter(str.isdigit, tokens)))
print("filter(None)       :", list(filter(None, [0, 1, "", "a", None, [], [1]])))

names = ["Aarav", "Bhavna", "Chirag", "Esha"]
print("filter(len > 5)    :", list(filter(lambda n: len(n) > 5, names)))

# zip + filter: sirf pass students
marks = [94, 33, 88, 25]
passed = [n for n, m in filter(lambda p: p[1] >= 40, zip(names, marks))]
print("\nPass students (zip + filter):", passed)
````

### 📌 Chapter 8 Cheat-Sheet
| Function | Kaam | Return |
|---|---|---|
| `zip(a, b)` | Items jodna | zip object (lazy) |
| `zip(*pairs)` | Unzip / transpose | zip object |
| `enumerate(x, start)` | `(index, value)` | enumerate object |
| `map(f, x)` | Har item par `f` | map object |
| `filter(f, x)` | `f(item)` True wale | filter object |

> Ye sab **lazy** hain — dekhne ke liye `list()`, `dict()`, `tuple()` lagao. Aur **ek baar hi** use ho sakte hain!

### 9.2 Import Karne ke 5 Tarike 📥
| Style | Syntax | Use kaise karein | Kab |
|---|---|---|---|
| 1️⃣ Poora module | `import math` | `math.sqrt(16)` | **Sabse safe** — kaun sa module saaf dikhta hai |
| 2️⃣ Alias (chhota naam) | `import math as m` | `m.sqrt(16)` | Lamba naam: `import numpy as np` |
| 3️⃣ Specific cheez | `from math import sqrt, pi` | `sqrt(16)` | Kam cheezein chahiye |
| 4️⃣ Specific + alias | `from math import sqrt as root` | `root(16)` | Naam clash se bachne ke liye |
| 5️⃣ Sab kuch (`*`) | `from math import *` | `sqrt(16)` | ⚠️ **Avoid** — naam takra sakte hain |

````python
# 1. import module
import math
print("1. math.sqrt(16)       =", math.sqrt(16))

# 2. import module as alias
import math as m
print("2. m.sqrt(25)          =", m.sqrt(25))

# 3. from module import name
from math import sqrt, pi
print("3. sqrt(36), pi        =", sqrt(36), round(pi, 4))

# 4. from module import name as alias
from math import factorial as fact
print("4. fact(5)             =", fact(5))

# 5. from module import *  (⚠️ avoid) — hum ise alag namespace mein dikhate hain
ns = {}
exec("from math import *", ns)
public = [n for n in ns if not n.startswith("_")]
print("\n5. 'from math import *' ne itne naam ghused diye:", len(public))
print("   Ek namuna :", public[:8])
print("   ⚠️ Isme pow bhi aata hai jo Python ke apne pow() ko chhupa deta hai:")
print("   built-in pow(2, 3) =", pow(2, 3), "| math wala pow =", ns["pow"](2, 3))
````

### 9.4 Python Module Kahan Dhoondhta Hai? 🔎
`import xyz` karne par Python **is order mein** dhoondhta hai:
1. `sys.modules` (pehle se import ho chuke modules — **cache**)
2. Built-in modules

#### ⚠️ Radians vs Degrees
Python ke trig functions **degrees nahi, radians** lete hain.
Conversion: **`math.radians(deg)`** aur **`math.degrees(rad)`**. (`180° = π radians`)

> `sin(30°)` bilkul `0.5` hona chahiye, par computer ke float ki wajah se `0.49999999999999994` aa sakta hai — isliye `round()` use karo.

````python
import math

print("math.pi =", math.pi)
print("math.e  =", math.e)

# Degrees -> Radians
print("\nradians(180) =", math.radians(180), "(= pi)")
print("degrees(pi)  =", math.degrees(math.pi))

print("\n--- Trig table (degrees -> radians -> sin, cos) ---")
for deg in [0, 30, 45, 60, 90]:
    rad = math.radians(deg)
    print(f"{deg:>3}° | sin = {round(math.sin(rad), 4):>6} | cos = {round(math.cos(rad), 4):>6}")

# Float ki limitation
print("\nsin(30°) bina round :", math.sin(math.radians(30)))
print("sin(30°) round ke saath:", round(math.sin(math.radians(30)), 10))

# sin^2 + cos^2 = 1 (identity)
x = math.radians(37)
print("\nsin²(37°) + cos²(37°) =", round(math.sin(x) ** 2 + math.cos(x) ** 2, 10))

# 🎪 Gol stage ka area
r = 7
print(f"\nStage radius {r} m -> area = pi x r² = {round(math.pi * r ** 2, 2)} m²")
print(f"                  -> circumference = 2 x pi x r = {round(2 * math.pi * r, 2)} m")

# e ka use: compound growth (continuous)
print("\nRs 1000, 5% continuous growth, 3 saal:", round(1000 * math.e ** (0.05 * 3), 2))
````

## ⏰ Group 1: `time()`, `ctime()`, `localtime()`, `gmtime()`
| Tool | Kaam | Return |
|---|---|---|
| **1️⃣ `time.time()`** | Epoch se ab tak ke **seconds** | `float` |
| **2️⃣ `time.ctime(secs)`** | Seconds ko **padhne layak string** mein (secs na do toh abhi ka) | `str` (24 chars) |
| **3️⃣ `time.localtime(secs)`** | Seconds ko **local timezone** ke `struct_time` mein | `struct_time` |

`struct_time` ke fields: `tm_year, tm_mon, tm_mday, tm_hour, tm_min, tm_sec, tm_wday` (Monday = 0), `tm_yday` (saal ka din), `tm_isdst`.

> 🌍 Colab ka timezone aksar **UTC** hota hai; India (IST) UTC se **+5:30** aage hai. Isliye `localtime()` aapke laptop par alag dikh sakta hai.

````python
import time

now = time.time()
print("time.time()  =", now)
print("Type         :", type(now).__name__)
print("Saal (approx) 1970 se:", round(now / (365.25 * 24 * 3600), 1))

print("\ntime.ctime() (abhi)        :", time.ctime())
print("ctime ki lambai            :", len(time.ctime()), "characters")

print("\n--- struct_time (abhi, local) ---")
lt = time.localtime()
print(lt)
print(f"Saal={lt.tm_year}, Mahina={lt.tm_mon}, Din={lt.tm_mday}, Ghanta={lt.tm_hour}, Minute={lt.tm_min}")
print("Hafte ka din (Mon=0) :", lt.tm_wday, "| Saal ka din:", lt.tm_yday)

# Fixed seconds par gmtime — hamesha ek jaisa
print("\n--- gmtime(0): epoch ka pehla pal ---")
g = time.gmtime(0)
print(g.tm_year, g.tm_mon, g.tm_mday, g.tm_hour, g.tm_min, g.tm_sec)
print("gmtime(1_700_000_000):", time.gmtime(1_700_000_000))

# Tuple ki tarah index
print("\nstruct_time ko tuple ki tarah bhi padh sakte hain:", g[0], g[1], g[2])
````

### 6️⃣ `time.mktime(struct_time)` — `localtime()` ka ulta: **struct_time → seconds** (local timezone ke hisaab se)

````python
import time

t = time.gmtime(1_767_225_600)       # ek fixed pal: 1 Jan 2026, 00:00:00 UTC

print("Date           :", time.strftime("%d-%m-%Y", t))
print("Time           :", time.strftime("%H:%M:%S", t))
print("Date + Time    :", time.strftime("%Y-%m-%d %H:%M:%S", t))
print("Poora likha    :", time.strftime("%A, %d %B %Y", t))
print("12-hour clock  :", time.strftime("%I:%M %p", t))
print("Chhota rup     :", time.strftime("%a, %d %b %y", t))
print("Saal ka din    :", time.strftime("%j", t))

# Abhi ka samay apne format mein
print("\nAbhi (local)   :", time.strftime("%d %B %Y, %I:%M %p"))

# --- mktime: struct -> seconds (local) ---
# Format: (saal, mahina, din, ghanta, minute, second, wday, yday, isdst)
struct = (2026, 1, 15, 10, 30, 0, 0, 0, -1)
secs = time.mktime(struct)
print("\nmktime ka result (seconds) type:", type(secs).__name__)

back = time.localtime(secs)               # wapas struct banao
print("Wapas padha     :", back.tm_year, back.tm_mon, back.tm_mday, back.tm_hour, back.tm_min)

# Do samay ke beech ka fark
d1 = time.mktime((2026, 1, 15, 10, 30, 0, 0, 0, -1))
d2 = time.mktime((2026, 1, 16, 10, 30, 0, 0, 0, -1))
print("\n15 Jan se 16 Jan ke beech seconds:", d2 - d1)
print("Ghante mein                      :", (d2 - d1) / 3600)
````

## ⏳ Group 3: `sleep()` aur `perf_counter()`
| Tool | Kaam |
|---|---|
| **7️⃣ `time.sleep(seconds)`** | Program ko itne seconds ke liye **rok do** (float bhi chalega: `0.5`) |
| **8️⃣ `time.perf_counter()`** | **High-precision timer** — code ka time naapne ke liye sabse behtar |

**Time naapne ka formula:**
```python
start = time.perf_counter()
# ... koi kaam ...
elapsed = time.perf_counter() - start
```
> 🆚 `time.time()` system clock hai (badal sakti hai) • `perf_counter()` sirf **duration** ke liye bana hai, zyada **sahi** hai.

````python
import time

# ⏳ sleep + countdown
print("Robo-Race shuru hone wali hai...")
for n in [3, 2, 1]:
    print(n)
    time.sleep(0.4)
print("🏁 GO!")

# ⏱️ Stopwatch with perf_counter
start = time.perf_counter()
time.sleep(0.5)
elapsed = time.perf_counter() - start
print(f"\nsleep(0.5) ne actual mein liya: {elapsed:.3f} sec")
print("Kam se kam 0.5 sec?            :", elapsed >= 0.5)

# Code ka time naapna
start = time.perf_counter()
total = sum(i * i for i in range(1_000_000))
elapsed = time.perf_counter() - start
print(f"\n10 lakh squares ka sum = {total}")
print(f"Lagne wala samay       = {elapsed:.4f} sec")

# Do tareeke compare: loop vs sum()
start = time.perf_counter()
t = 0
for i in range(1_000_000):
    t += i
loop_time = time.perf_counter() - start

start = time.perf_counter()
t2 = sum(range(1_000_000))
sum_time = time.perf_counter() - start
print(f"\nLoop se : {loop_time:.4f} sec")
print(f"sum() se: {sum_time:.4f} sec")
print("Same answer?:", t == t2)
````

## 📁 Group 2: Folder Banana aur Hatana
| Tool | Kaam | Dhyaan do |

### 6.1 `randint`, `random` aur Range Ka Khel 🎯
- `randint(a, b)` mein **dono end shamil** (`randint(1, 6)` mein `6` bhi aa sakta hai) — `range(a, b)` se alag!
- Kisi bhi range mein float: `a + (b - a) * random.random()` (ya `random.uniform(a, b)`).

````python
import random

random.seed(1)                          # taaki neeche ke checks hamesha sahi aayein

# 1000 baar dice pheko
rolls = [random.randint(1, 6) for _ in range(1000)]
counts = {face: rolls.count(face) for face in range(1, 7)}
print("Dice rolls ki ginti:", counts)
print("Saare rolls 1 se 6 ke beech?:", all(1 <= r <= 6 for r in rolls))
print("Har face ~167 baar aana chahiye (120-220 ke beech?):", all(120 <= c <= 220 for c in counts.values()))

# random() float
values = [random.random() for _ in range(5)]
print("\nrandom() ke 5 values 0 se 1 ke beech?:", all(0 <= v < 1 for v in values))

# Kisi bhi range mein float
temp = 20 + (35 - 20) * random.random()
print("20 se 35 ke beech temperature:", round(temp, 1))
print("random.uniform(20, 35)       :", round(random.uniform(20, 35), 1))
````

### 6.2 `shuffle` vs `sample` 🃏
| | `shuffle(list)` | `sample(seq, k)` |
|---|---|---|
| Original badalta hai? | ✅ **Haan** (in-place) | ❌ Nahi — nayi list deta hai |
| Return | `None` | Nayi list |
| Tuple / string par | ❌ Nahi (immutable) | ✅ Chalta hai |
| Kaam | Poori list ka order badalna | Kuch items **bina repeat** chunna |

> `random.choices(seq, k=3)` — **repeat ke saath** chunta hai (alag function).

````python
import random

random.seed(5)

names = ["Aarav", "Bhavna", "Chirag", "Deepika", "Esha", "Farhan", "Gauri", "Hari"]

# 🎁 Lucky draw: 3 winners, koi repeat nahi
winners = random.sample(names, 3)
print("Winners             :", winners)
print("Teen alag-alag?     :", len(set(winners)) == 3)
print("Sab list mein se?   :", all(w in names for w in winners))
print("Original badla?     :", names == ["Aarav", "Bhavna", "Chirag", "Deepika", "Esha", "Farhan", "Gauri", "Hari"], "(sample original nahi badalta)")

# 🎤 Chief guest
print("\nChief guest in names?:", random.choice(names) in names)

# 🔀 Shuffle: schedule ka order
events = ["Hackathon", "Quiz", "Robo-Race", "Dance", "Music"]
random.shuffle(events)
print("\nShuffled schedule:", events)
print("Wahi 5 events?   :", sorted(events) == sorted(["Hackathon", "Quiz", "Robo-Race", "Dance", "Music"]))

# Tuple par shuffle nahi chalta
try:
    random.shuffle(("a", "b", "c"))
except TypeError as e:
    print("\n⚠️ Tuple par shuffle ->", e)
print("Tuple par sample chalta hai:", len(random.sample(("a", "b", "c"), 2)) == 2)
````

### 6.3 `seed()` — Random ko Fix Karna 🔒
`random.seed(42)` ke baad **jo bhi random numbers** aayenge wo **har baar same order** mein aayenge. Fayda:
- **Debugging** (bug ek hi sequence mein dobara dikhta hai)
- **Reproducible experiments** (research / exams mein result dobara nikalna)
- Teacher ko sabke answers **same** dikhane ke liye 😄

````python
import random

random.seed(42)
first_run = [random.randint(1, 100) for _ in range(5)]

random.seed(42)                         # wahi seed dobara
second_run = [random.randint(1, 100) for _ in range(5)]

random.seed(99)                         # alag seed
third_run = [random.randint(1, 100) for _ in range(5)]

print("seed(42) pehli baar :", first_run)
print("seed(42) doosri baar:", second_run)
print("seed(99)            :", third_run)
print("\nSame seed -> same list? :", first_run == second_run)
print("Alag seed -> alag list? :", first_run != third_run)

# random() ke saath bhi
random.seed(7)
a = random.random()
random.seed(7)
b = random.random()
print("\nseed(7) par random() dobara same?:", a == b)
````

### 6.4 Real-Life Use: OTP, Password, Monte Carlo 🔐
Random se bahut kuch bante hain: **OTP**, **password**, **games**, aur **simulations** (jaise π ka andaza!).

````python
import random, string, math

# 🔢 4-digit OTP
otp = random.randint(1000, 9999)
print("OTP:", otp, "| 4 digits?", 1000 <= otp <= 9999)

# 🔑 8 character password
chars = string.ascii_letters + string.digits
password = "".join(random.choice(chars) for _ in range(8))
print("Password:", password, "| lambai 8?", len(password) == 8)

# 🎯 Monte Carlo: pi ka andaza (square mein random points, quarter circle ke andar kitne?)
random.seed(7)
n = 100_000
inside = 0
for _ in range(n):
    x, y = random.random(), random.random()
    if x * x + y * y <= 1:
        inside += 1
estimate = 4 * inside / n
print(f"\n{n:,} random points se pi ka andaza = {estimate}")
print("Asli pi                          =", math.pi)
print("Fark 0.05 se kam?                :", abs(estimate - math.pi) < 0.05)
````

### 14.6 Date Arithmetic — 6 Kaam ke Formule 🧮
| # | Kaam | Formula |
|:-:|---|---|
| 1️⃣ | Kitne din bache? | `(target - today).days` |
| 2️⃣ | N din baad / pehle ki date | `d + timedelta(days=N)` |
| 3️⃣ | Umar (age) | `on.year - b.year - ((on.month, on.day) < (b.month, b.day))` |
| 4️⃣ | Weekend hai? | `d.weekday() >= 5` |
| 5️⃣ | Kaam ke din (Mon–Fri) ginna | loop + `d += timedelta(days=1)` |
| 6️⃣ | Mahine ka aakhri din | agle mahine ki 1 tarikh **minus 1 din** |

````python
from datetime import date, timedelta
import calendar

today = date(2026, 2, 1)             # 👈 fixed "aaj" taaki output hamesha same aaye
fest  = date(2026, 3, 12)

# 1️⃣ Countdown
left = (fest - today).days
print(f"1) Fest mein {left} din bache = {left // 7} hafte aur {left % 7} din")

# 2️⃣ Deadline
deadline = fest - timedelta(days=7)
print("2) Registration deadline:", deadline, deadline.strftime("(%A)"))

# 3️⃣ Age
def age_on(birth, on):
    return on.year - birth.year - ((on.month, on.day) < (birth.month, birth.day))

print("3) Janam 15-Aug-2006 -> fest par umar:", age_on(date(2006, 8, 15), fest))
print("   Janam 05-Mar-2006 -> fest par umar:", age_on(date(2006, 3, 5), fest))

# 4️⃣ Weekend?
print("\n4) Fest ke 5 din:")
for i in range(5):
    d = fest + timedelta(days=i)
    print(f"   {d.strftime('%a %d %b')} -> {'🎉 weekend' if d.weekday() >= 5 else 'working day'}")

# 5️⃣ Kaam ke din
def working_days(start, end):                     # dono din shamil
    count, d = 0, start
    while d <= end:
        if d.weekday() < 5:
            count += 1
        d += timedelta(days=1)
    return count

print("\n5) 2 Mar se 12 Mar 2026 ke working days:", working_days(date(2026, 3, 2), date(2026, 3, 12)))

# 6️⃣ Mahine ka aakhri din
def last_day(year, month):
    first_of_next = date(year + (month == 12), month % 12 + 1, 1)
    return (first_of_next - timedelta(days=1)).day

for y, m in [(2026, 2), (2028, 2), (2026, 4), (2026, 12)]:
    print(f"6) {calendar.month_abbr[m]} {y} ka aakhri din: {last_day(y, m)}",
          "| calendar se match?", last_day(y, m) == calendar.monthrange(y, m)[1])
````

### 14.7 Bonus: Timezone ki Jhalak 🌐 (Naive vs Aware)
Ab tak ke saare `datetime` **naive** the — unhe pata nahi ki wo kis desh ki ghadi hai. **Aware** datetime mein timezone judta hai (`tzinfo`).

| Cheez | Kaam |
|---|---|
| `timezone.utc` | UTC (world standard) |
| `timezone(timedelta(hours=5, minutes=30))` | Apna offset — **India (IST) = UTC + 5:30** |
| `dt.astimezone(tz)` | Wahi **pal**, doosre timezone ki ghadi mein |

> ⚠️ **Naive aur aware ko mix** karke minus/compare nahi kar sakte → `TypeError`.

````python
from datetime import datetime, timedelta, timezone

IST = timezone(timedelta(hours=5, minutes=30), "IST")

utc_time = datetime(2026, 3, 12, 9, 0, tzinfo=timezone.utc)
ist_time = utc_time.astimezone(IST)

print("UTC mein :", utc_time)
print("IST mein :", ist_time)
print("Alag dikhte hain, par ek hi pal?", utc_time == ist_time)
print("Fark     :", ist_time.utcoffset())

# Abhi ka India time (Colab kahin bhi chale, ye IST dega)
now_ist = datetime.now(timezone.utc).astimezone(IST)
print("\nAbhi IST ka offset +5:30 hai?", now_ist.utcoffset() == timedelta(hours=5, minutes=30))

# ⚠️ naive - aware
try:
    datetime(2026, 3, 12, 9, 0) - utc_time
except TypeError as e:
    print("\n⚠️ naive - aware -> TypeError:", e)
````

<details><summary>💡 Hint (pehle khud try karo!)</summary>

````python
#@title 🔑 Trainer Solution — WAP 14.5 (code hidden — dekhne ke liye 'Show code')
from datetime import date, timedelta

birth = date(2006, 8, 15)
on = date(2026, 3, 12)

age = on.year - birth.year - ((on.month, on.day) < (birth.month, birth.day))
print("Age:", age)
print("Days lived:", (on - birth).days)

def working_days(start, end):
    count, d = 0, start
    while d <= end:
        if d.weekday() < 5:
            count += 1
        d += timedelta(days=1)
    return count

print("Working days:", working_days(date(2026, 3, 2), date(2026, 3, 12)))

def last_day(year, month):
    first_of_next = date(year + (month == 12), month % 12 + 1, 1)
    return (first_of_next - timedelta(days=1)).day

print("Last day of Feb 2026:", last_day(2026, 2))
print("Last day of Feb 2028:", last_day(2028, 2))
````

### 15.1 NumPy Kyun? — Speed + Chhota Code ⚡
| | Python `list` | NumPy `array` |
|---|---|---|
| **Operation** | Loop chahiye | **Seedha poore array par** (*vectorization*) |
| **Speed** | Dheema | **10–100 guna tez** (C mein bana hai) |
| **Memory** | Har item ek poora object | Ek hi type ke numbers **saath-saath** |
| **Type** | Kuch bhi mix kar sakte ho | **Ek hi `dtype`** (sab int ya sab float) |

> 🧠 **Vectorization:** `a * 2`, `a + b`, `a ** 2` — operation **har element par apne aap** lagta hai. Loop likhne ki zaroorat nahi!

````python
import numpy as np, time

n = 1_000_000
numbers_list = list(range(n))
numbers_arr  = np.arange(n)

# 📋 List + loop se: saare squares ka jod
t0 = time.perf_counter()
total_list = sum(x * x for x in numbers_list)
t_list = time.perf_counter() - t0

# ⚡ NumPy se
t0 = time.perf_counter()
total_np = int((numbers_arr * numbers_arr).sum())
t_np = time.perf_counter() - t0

# Speedup calculate karein
speedup = t_list / t_np

print(f"Dono ka jawab same?  : {total_list == total_np}")
print(f"Python List time     : {t_list * 1000:.2f} ms")
print(f"NumPy Array time     : {t_np * 1000:.2f} ms")
print(f"Actual Speedup       : NumPy {speedup:.1f}x zyada tez hai!")
````

### 15.5 `sum()` aur `mean()` — Total aur Average 🧮

> 🧠 **axis ki trick:** jis `axis` ka naam lo, **wahi dimension "gayab" ho jaata hai** (usi par jod / average hota hai).

```
                 │      │      │
              axis=0 ▼
                245    270    265
```

> 🆕 **NumPy 2 ka tip:** `print(a.mean())` → `65.0` hi dikhata hai, par `repr` / list ke andar `np.float64(65.0)` dikh sakta hai. Saaf output ke liye `float(x)` ya `arr.tolist()` use karo.

````python
import math
import numpy as np
from scipy import integrate, optimize

# ∫ integrate.quad(function, start, end)  ->  (answer, error-estimate)
area, err = integrate.quad(lambda x: x ** 2, 0, 3)
print("∫ x² dx (0 se 3)    =", round(area, 6), "| exact 9 se match?", abs(area - 9) < 1e-9)

area2, _ = integrate.quad(np.sin, 0, np.pi)
print("∫ sin(x) dx (0 se π) =", round(area2, 6))

# 🚗 Speed v(t) = 2t + 1 ho toh 0-10 second mein kitna distance?
distance, _ = integrate.quad(lambda t: 2 * t + 1, 0, 10)
print("Distance (0-10 s)    =", round(distance, 6), "m")

# 🔽 optimize.minimize_scalar: function ka minimum
res = optimize.minimize_scalar(lambda x: (x - 3) ** 2 + 2)
print("\n(x-3)² + 2 ka minimum: x =", round(res.x, 4), "| value =", round(res.fun, 4))

# 🎯 optimize.brentq: root (jahan function = 0)
root = optimize.brentq(lambda x: x ** 2 - 2, 0, 2)
print("x² - 2 = 0 ka root   =", round(root, 6), "| math.sqrt(2) =", round(math.sqrt(2), 6))

# 💰 Stall ka profit: profit(p) = 400 - (p - 50)²  -> kis price par sabse zyada?
best = optimize.minimize_scalar(lambda p: (p - 50) ** 2 - 400)     # minimize(-profit)
print("\nSabse zyada profit price Rs", round(best.x, 2), "par, profit = Rs", round(-best.fun, 2))
````

````python
import numpy as np
from scipy import stats, linalg

# 📈 stats.pearsonr: do cheezon ka rishta (correlation)
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])                     # padhai ke ghante
marks = np.array([35, 42, 50, 58, 63, 72, 78, 90])             # marks
r = stats.pearsonr(hours, marks)
print("Correlation r        :", round(r.statistic, 4), "(+1 = perfect saath-saath badhna)")
print("Bahut strong (r > 0.9)?", r.statistic > 0.9)
print("p-value < 0.01 ?     :", r.pvalue < 0.01)

# 📊 describe + zscore
d = stats.describe(marks)
print("\nstats.describe -> nobs:", d.nobs, "| mean:", float(d.mean), "| variance:", round(float(d.variance), 2))
print("z-scores [50, 60, 70] :", stats.zscore(np.array([50, 60, 70])).round(3), "(mean se kitne std door)")

# 🔔 Normal distribution
print("\nP(Z <= 0)    :", stats.norm.cdf(0), "(aadha)")
print("P(Z <= 1.96) :", round(stats.norm.cdf(1.96), 4), "(~97.5%)")

# 🧩 linalg.solve: equations hal karna
# Combo 1: 2 samosa + 1 chai = Rs 35
# Combo 2: 1 samosa + 3 chai = Rs 55
A = np.array([[2, 1],
              [1, 3]])
b = np.array([35, 55])
samosa, chai = linalg.solve(A, b)
print("\nSamosa = Rs", round(float(samosa), 6), "| Chai = Rs", round(float(chai), 6))
print("Verify A @ x == b ?    :", np.allclose(A @ np.array([samosa, chai]), b))
print("det(A)                 :", round(float(linalg.det(A)), 6))
````

````python
# ✍️ WAP 15.4 — Marks Analysis (sum & mean)
# Apna code is cell mein likho 👇

import numpy as np
names = ["Aarav", "Bhavna", "Chirag", "Deepika"]
marks = np.array([[70, 80, 90],
                  [60, 50, 40],
                  [85, 95, 75],
                  [30, 45, 60]])
````

````python
#@title 🔑 Trainer Solution — WAP 15.4 (code hidden — dekhne ke liye 'Show code')
import numpy as np
names = ["Aarav", "Bhavna", "Chirag", "Deepika"]
marks = np.array([[70, 80, 90],
                  [60, 50, 40],
                  [85, 95, 75],
                  [30, 45, 60]])

overall_mean = marks.mean()
student_means = marks.mean(axis=1)

print("Total of all marks:", marks.sum())
print("Overall mean:", overall_mean)
print("Subject totals:", marks.sum(axis=0).tolist())
print("Student means:", student_means.tolist())
print("Topper:", names[int(marks.sum(axis=1).argmax())])
print("Students above overall mean:", (student_means > overall_mean).sum())
print("Marks >= 80:", marks[marks >= 80].tolist())
````

### 🏋️ WAP 15.5: SciPy Lab

SciPy ke 4 tools chalao aur print karo:
1. `integrate.quad` se `∫ 3x² dx` (0 se 2) → `Integral: 8.0` (`round(value, 6)`)
2. `optimize.minimize_scalar` se `(x - 4)² + 1` ka minimum → `Min at x: 4.0` aur `Min value: 1.0` (dono `round(..., 4)`)
3. `special.comb(6, 2, exact=True)` → `Ways to pick 2 of 6: 15`
4. `linalg.solve` se: `2s + c = 35` aur `s + 3c = 55` → `Samosa: 10.0 | Chai: 15.0` (`round(float(..), 6)`)
5. `stats.pearsonr` se `hours = [1,2,3,4,5]` aur `marks = [40,50,65,70,85]` ka correlation → `Strong positive: True` (`r.statistic > 0.9`)

<details><summary>💡 Hint (pehle khud try karo!)</summary>

`quad` ek **tuple** `(value, error)` deta hai — `value, _ = ...` se unpack karo. `solve(A, b)` ka answer ek array hai: `s, c = ...`.

</details>

````python
# ✍️ WAP 15.5 — SciPy Lab
# Apna code is cell mein likho 👇

import numpy as np
from scipy import integrate, optimize, special, linalg, stats
````

````python
#@title 🔑 Trainer Solution — WAP 15.5 (code hidden — dekhne ke liye 'Show code')
import numpy as np
from scipy import integrate, optimize, special, linalg, stats

value, _ = integrate.quad(lambda x: 3 * x ** 2, 0, 2)
print("Integral:", round(value, 6))

res = optimize.minimize_scalar(lambda x: (x - 4) ** 2 + 1)
print("Min at x:", round(res.x, 4))
print("Min value:", round(res.fun, 4))

print("Ways to pick 2 of 6:", special.comb(6, 2, exact=True))

A = np.array([[2, 1], [1, 3]])
b = np.array([35, 55])
s, c = linalg.solve(A, b)
print("Samosa:", round(float(s), 6), "| Chai:", round(float(c), 6))

hours = np.array([1, 2, 3, 4, 5])
marks = np.array([40, 50, 65, 70, 85])
r = stats.pearsonr(hours, marks)
print("Strong positive:", bool(r.statistic > 0.9))
````

````python
# 🏆 UTKARSH 2026 — FEST MANAGER  (Day 3 ke saare tools ek saath)
import os, sys, shutil, importlib, math, random, time
from datetime import date, datetime, timedelta
import numpy as np
from scipy import stats

t_start = time.perf_counter()                                        # ⏱️ stopwatch shuru

SANDBOX = "fest_manager"
shutil.rmtree(SANDBOX, ignore_errors=True)
os.makedirs(os.path.join(SANDBOX, "festkit"))                         # 📁 package folder
os.makedirs(os.path.join(SANDBOX, "reports"))                         # 📁 reports folder

# ══════════ 📦 STEP 1: apna PACKAGE `festkit` banao ══════════
TOOLS_SOURCE = """
import math
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

def volunteers_needed(participants, per_volunteer=25):
    return math.ceil(participants / per_volunteer)

def binary_search(items, target, low=0, high=None):
    if high is None:
        high = len(items) - 1
    if low > high:
        return -1
    mid = (low + high) // 2
    if items[mid] == target:
        return mid
    if items[mid] < target:
        return binary_search(items, target, mid + 1, high)
    return binary_search(items, target, low, mid - 1)

if __name__ == "__main__":
    print("festkit.tools seedha chalaya gaya")
"""
with open(os.path.join(SANDBOX, "festkit", "__init__.py"), "w") as f:
    f.write("")                                                       # khaali __init__.py = package ban gaya
with open(os.path.join(SANDBOX, "festkit", "tools.py"), "w") as f:
    f.write(TOOLS_SOURCE)

sys.path.insert(0, os.path.abspath(SANDBOX))
importlib.invalidate_caches()
from festkit.tools import fib, volunteers_needed, binary_search
print("📦 Package import ho gaya:", [fib.__name__, volunteers_needed.__name__, binary_search.__name__])

# ══════════ 🪆 STEP 2: recursion + memoization ══════════
prizes = [fib(n) * 1000 for n in range(5, 10)]                        # Fibonacci prize tiers
print("\n🏆 Prize tiers (Rs):", prizes, "| Total pool: Rs", sum(prizes))
print("   fib(60) bina atke (lru_cache):", fib(60))

# ══════════ 📅 STEP 3: datetime — 3 din ka schedule ══════════
fest_start = date(2026, 3, 12)
events = [                              # (naam, din offset, shuru "HH:MM", minute)
    ("Hackathon",   0, "09:00", 180),
    ("Quiz",        0, "14:00",  60),
    ("Robo-Race",   1, "10:30", 120),
    ("Dance Off",   2, "17:00",  90),
    ("Music Night", 2, "19:00", 120),
]
deadline = fest_start - timedelta(days=7)
print("\n📅 Registration deadline:", deadline.strftime("%d %B %Y (%A)"))
print("📅 SCHEDULE")
schedule = []
for name, offset, hhmm, minutes in events:
    day = fest_start + timedelta(days=offset)
    start = datetime.strptime(f"{day} {hhmm}", "%Y-%m-%d %H:%M")
    end = start + timedelta(minutes=minutes)
    schedule.append((name, start, end))
    print(f"   {start:%a %d %b}  {start:%I:%M %p} -> {end:%I:%M %p}   {name}")

# ══════════ 📊 STEP 4: numpy + scipy — registrations ══════════
names = [e[0] for e in events]
regs = np.array([180, 95, 60, 140, 210])
z = stats.zscore(regs)
popular = [n for n, v in zip(names, z) if v > 0]
print("\n📊 REGISTRATIONS")
print("   Total:", regs.sum(), "| Average:", regs.mean(), "| Sabse popular:", names[int(regs.argmax())])
print("   Average se upar ke events:", popular)

# ══════════ 👷 STEP 5: math — volunteers ══════════
need = [volunteers_needed(int(r)) for r in regs]
print("\n👷 Volunteers (25 participants par 1):")
for i, (n, v) in enumerate(zip(names, need), start=1):
    print(f"   {i}. {n:12} -> {v}")
print("   Total volunteers:", sum(need))

# ══════════ 🎁 STEP 6: random — lucky draw ══════════
random.seed(2026)
ids = list(range(1001, 1201))                                         # 200 participants ke sorted IDs
winners = random.sample(ids, 3)
chief_guest = random.choice(["Dr. Sharma", "Ms. Verma", "Mr. Khan"])
positions = [binary_search(ids, w) for w in winners]                  # recursion se wapas dhoondho
print("\n🎁 Lucky draw winners (ID):", winners, "| 3 alag-alag?", len(set(winners)) == 3)
print("   Binary search se mile?  :", all(p >= 0 for p in positions), "| ID 9999 ->", binary_search(ids, 9999))
print("   Chief guest             :", chief_guest)

# ══════════ 📁 STEP 7: os — report likho, check karo ══════════
report_path = os.path.join(SANDBOX, "reports", "summary.txt")
lines = [f"UTKARSH 2026 SUMMARY (deadline {deadline})"]
lines += [f"{n}: {s:%d %b %H:%M} - {e:%H:%M}" for n, s, e in schedule]
lines += [f"Total registrations: {regs.sum()}", f"Volunteers: {sum(need)}", f"Winners: {winners}"]
with open(report_path, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

print("\n📁 Report ban gayi?     :", os.path.exists(report_path))
print("   File hai (folder nahi)?:", os.path.isfile(report_path), "| Size > 0:", os.path.getsize(report_path) > 0)
print("   reports/ ke andar     :", os.listdir(os.path.join(SANDBOX, "reports")))
with open(report_path, encoding="utf-8") as f:
    print("   Pehli 2 lines padho   :", f.read().splitlines()[:2])

# ══════════ ⏱️ STEP 8: stopwatch band + safai ══════════
elapsed = time.perf_counter() - t_start
print("\n⏱️ Poora program 5 second se kam mein chala?", elapsed < 5)

sys.path.remove(os.path.abspath(SANDBOX))
for mod in [m for m in sys.modules if m == "festkit" or m.startswith("festkit.")]:
    sys.modules.pop(mod)
shutil.rmtree(SANDBOX, ignore_errors=True)
print("🧹 Sandbox saaf ho gaya?", not os.path.exists(SANDBOX))
print("\n✅ Utkarsh 2026 ka Fest Manager tayyar!")
````
