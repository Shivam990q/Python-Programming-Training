# 🚨 Chapter 1: Errors vs Exceptions — Kiosk Crash

### 🎬 Story
Utkarsh ka registration kiosk chal raha tha. Ek student ne **"Umar:"** ke saamne type kiya: `twenty`. Program ne `int("twenty")` karne ki koshish ki... aur **BOOM** 💥 — screen par laal rang ka traceback aaya aur kiosk band!

Bhavna Didi: *"Aarav, har error ek jaisa nahi hota. Pehle **error ki jaat** pehchan — phir ilaaj samajh aayega."* 🩺

### 🧬 Error ki 3 Jaatein
| Jaat | Kab pakda jaata hai? | Python kya karta hai? | Example |
|---|---|---|---|
| 1️⃣ **Syntax Error** | Program **shuru hone se pehle** (padhte hi) | Program **chalta hi nahi** | `print("hi"` (bracket band nahi) |
| 2️⃣ **Runtime Error (Exception)** | Program **chalte-chalte** | **Wahin ruk jaata hai**, traceback dikhata hai | `10 / 0` |
| 3️⃣ **Logical Error** | **Kabhi nahi!** 😶 | Chalta hai, par **jawab galat** | `a + b / 2` (average ke liye) |

> 🧠 **Exception** = wo **runtime error** jise Python ek **object** bana ke "phenk" (`raise`) deta hai. Agar koi use **pakde** (`except`) nahi, toh program **crash** ho jaata hai.

> ⚠️ **Sabse khatarnaak Logical Error hai** — Python ek lafz nahi bolta, aur jawab chupchaap galat aata hai!

````python
# 1️⃣ SYNTAX ERROR — Python padhte hi pakad leta hai (chalne se pehle)
bad_code = "print('Namaste'"                      # bracket band karna bhool gaye!
try:
    compile(bad_code, "<demo>", "exec")           # sirf CHECK karta hai, chalata nahi
except SyntaxError as e:
    print("1️⃣ Syntax Error   ->", type(e).__name__)

# 2️⃣ RUNTIME ERROR (EXCEPTION) — chalte-chalte
try:
    fees = 500 / 0
except ZeroDivisionError as e:
    print("2️⃣ Runtime Error  ->", type(e).__name__, "|", e)

# 3️⃣ LOGICAL ERROR — koi error nahi, par jawab galat!
def average(a, b):
    return a + b / 2                              # ❌ (a + b) / 2 hona chahiye tha

print("3️⃣ Logical Error  -> average(10, 20) =", average(10, 20), "(sahi jawab 15.0 hai!)")
````

### 1.1 Jab Exception Pakdi Nahi Jaati — Asli Crash 💥
Neeche ka cell ek **alag Python process** mein ek program chalata hai jisme exception **koi pakadta nahi**. Aap dekhoge:
- Exception ke **baad wali lines kabhi nahi chalti**,
- Python ek **traceback** chhapta hai,
- Program ka **exit code 1** hota hai (0 = sab theek).

> 💡 Hum alag process isliye use kar rahe hain taaki **aapki notebook crash na ho** aur aap asli traceback dekh sako.

````python
import subprocess, sys

program = """
print("Step 1: registration shuru")
print("Step 2: fees =", 500 / 0)
print("Step 3: certificate bhejna")
"""

result = subprocess.run([sys.executable, "-c", program], capture_output=True, text=True)

print("──── Program ne kya print kiya (stdout) ────")
print(result.stdout)
print("──── Program ka crash message (stderr) ────")
print(result.stderr)
print("Exit code:", result.returncode, "(1 = exception se crash)")
print("\n👀 Kya 'Step 3' kabhi print hua?", "Step 3" in result.stdout, "(exception ke baad ki lines nahi chalti!)")
````

### 1.2 Traceback Padhna — Neeche se Upar! 🔍
```
Traceback (most recent call last):          ← Header (call chain shuru)
  File "kiosk.py", line 9, in <module>       ← 1️⃣ Pehle ye call hua (bahar wala)
    show_fee(registration)
  File "kiosk.py", line 5, in show_fee       ← 2️⃣ Phir ye (andar wala) — yahin error aaya
    return amount / people
ZeroDivisionError: division by zero          ← ⭐ ASLI JAWAB: kya hua? (Type: Message)
```
| Kahan dekhna | Kya milta hai |
|---|---|
| **Aakhri line** ⭐ | **Exception ka type** + **message** — pehle ise padho! |
| Uske **upar** ki `File ... line N, in naam` | **Kahan** hua (kaunsi file, kaunsi line, kaunsa function) |
| Sabse **neeche wali** `File` line | **Wo jagah jahan asli error aaya** |
| Upar wali `File` lines | Wo **raasta** (call chain) jisse yahan pahunche — Day 3 ka **call stack**! |

> 🧠 **Rule:** Traceback ko **neeche se upar** padho. Aakhri line = *kya*, uske upar = *kahan*.

````python
import traceback

def per_head(amount, people):
    return amount / people                        # yahan error aayega

def show_fee(amount, people):
    return per_head(amount, people)

try:
    show_fee(500, 0)
except Exception as e:
    frames = traceback.extract_tb(e.__traceback__)          # call chain ke saare frames
    print("Exception type   :", type(e).__name__)
    print("Message          :", e)
    print("Error kis function mein aaya:", frames[-1].name, "  (sabse andar wala frame)")
    print("Call chain mein functions   :", [f.name for f in frames if f.name in ("show_fee", "per_head")])
````

### 1.3 Exception Family Tree 🌳
Saari exceptions **classes** hain aur ek **family tree** mein judi hain. Ye tree **Chapter 3** (`except` ka order) aur **Chapter 6** (apni exception) mein bahut kaam aayega.

```
BaseException
 ├── KeyboardInterrupt          (Ctrl+C)    ← Exception ki family mein NAHI
 ├── SystemExit
 └── Exception                  ← 👑 Aam errors — hum yahi pakadte hain
      ├── ArithmeticError
      │     └── ZeroDivisionError
      ├── LookupError
      │     ├── IndexError
      │     └── KeyError
      ├── ValueError
      ├── TypeError
      ├── NameError
      ├── AttributeError
      └── OSError
            └── FileNotFoundError
```
> 🧠 **Baap pakde toh bachche bhi pakde jaate hain:** `except ArithmeticError` → `ZeroDivisionError` bhi pakad lega. `except LookupError` → `IndexError` aur `KeyError` dono.

````python
# Har exception ka "shajra" (family chain) — __mro__ se
six = [ZeroDivisionError, ValueError, TypeError, IndexError, KeyError, FileNotFoundError]
for exc in six:
    chain = " -> ".join(c.__name__ for c in exc.__mro__)
    print(f"{exc.__name__:18}: {chain}")

print("\n--- issubclass(bachcha, baap) ---")
print("ZeroDivisionError ek ArithmeticError hai? :", issubclass(ZeroDivisionError, ArithmeticError))
print("KeyError ek LookupError hai?              :", issubclass(KeyError, LookupError))
print("ValueError ek LookupError hai?            :", issubclass(ValueError, LookupError))
print("FileNotFoundError ek OSError hai?         :", issubclass(FileNotFoundError, OSError))
print("KeyboardInterrupt ek Exception hai?       :", issubclass(KeyboardInterrupt, Exception))
````

### 1.4 Ek Jhalak: `try / except` 👀
Abhi hum sirf **pakadna** dekh rahe hain — **poori detail Chapter 3 mein**:
```python
try:
    ...risky kaam...           # jahan error aa sakta hai
except ZeroDivisionError:
    ...plan B...               # error aaye toh ye chalega
```
`except Exception as e` = *"koi bhi aam error pakdo aur use `e` naam do"* — `type(e).__name__` se uska naam, `e` (ya `str(e)`) se message milta hai.

````python
# Bina try/except: crash.   try/except ke saath: program zinda!
ages = ["19", "twenty", "21", "", "18"]

for text in ages:
    try:
        age = int(text)
        print(f"✅ {text!r:10} -> umar {age}")
    except Exception as e:
        print(f"⚠️  {text!r:10} -> {type(e).__name__}: {e}")

print("\n🎉 Poori list process ho gayi — kiosk crash nahi hua!")
````

### 🎬 Story

### 🖼️ Wanted Poster — 6 Mujrim
| Exception | Kab aata hai? | Example | Pehchan (message) |
|---|---|---|---|
| ➗ **`ZeroDivisionError`** | Zero se bhaag diya | `10 / 0` | `division by zero` |
| 🔢 **`ValueError`** | **Type sahi, value galat** | `int("abc")` | `invalid literal for int()...` |
| 🔀 **`TypeError`** | **Type hi galat** | `"5" + 5` | `can only concatenate str...` |
| 📍 **`IndexError`** | List/string ka **index bahar** | `[1,2,3][10]` | `list index out of range` |
| 🔑 **`KeyError`** | Dictionary mein **key nahi** | `d["xyz"]` | `'xyz'` |
| 📁 **`FileNotFoundError`** | **File / folder mila hi nahi** | `open("ghost.txt")` | `[Errno 2] No such file...` |

> 🧠 Neeche ke cells mein ek chhota `show()` helper hai — ye kisi bhi kaam ko chalata hai aur **error aaye toh uska naam + message** dikhata hai (notebook crash nahi hoti). Helper har cell mein diya gaya hai, taaki **koi bhi cell akela** chala sako.

````python
def show(label, func):
    """func ko chalao. Error aaye toh uska naam aur message dikhao."""
    try:
        value = func()
        print(f"✅ {label:26} -> {value!r}")
    except Exception as e:
        print(f"💥 {label:26} -> {type(e).__name__}: {e}")

# ➗ ZeroDivisionError — "zero se bhaag"
print("──── ZeroDivisionError ────")
show("10 / 0", lambda: 10 / 0)
show("10.5 / 0", lambda: 10.5 / 0)
show("10 // 0", lambda: 10 // 0)
show("10 % 0", lambda: 10 % 0)
show("0 / 10  (theek hai!)", lambda: 0 / 10)
````

### 2.1 ➗ `ZeroDivisionError` — Zero se Bhaagna
`/`, `//` aur `%` — **teeno** mein divisor `0` nahi ho sakta. (`0 / 10` bilkul theek hai — **numerator** zero ho sakta hai, **divisor** nahi.)

> 🛠️ **Bachao:** divide karne se pehle `if divisor != 0` check karo, ya `try/except ZeroDivisionError` lagao.

### 2.2 🔢 `ValueError` vs 🔀 `TypeError` — Sabse Zyada Confuse Hone Wali Jodi
| | **`ValueError`** | **`TypeError`** |
|---|---|---|
| Matlab | Type **sahi hai**, par **value** kaam ki nahi | **Type hi galat** hai |
| Example | `int("abc")` — `str` chalta hai, par `"abc"` number nahi | `int([1, 2])` — `list` se int ban hi nahi sakta |
| Yaad rakhne ki trick | *"Cheez sahi, **maal kharab**"* | *"**Cheez hi** galat"* |

````python
def show(label, func):
    """func ko chalao. Error aaye toh uska naam aur message dikhao."""
    try:
        value = func()
        print(f"✅ {label:26} -> {value!r}")
    except Exception as e:
        print(f"💥 {label:26} -> {type(e).__name__}: {e}")

import math

print("──── ValueError (type sahi, value galat) ────")
show('int("abc")', lambda: int("abc"))
show('int("12.5")', lambda: int("12.5"))
show('float("1,5")', lambda: float("1,5"))
show("math.sqrt(-4)", lambda: math.sqrt(-4))
show("[1, 2, 3].index(9)", lambda: [1, 2, 3].index(9))
show('int("  42  ")  (theek hai!)', lambda: int("  42  "))

print("\n──── TypeError (type hi galat) ────")
show('"5" + 5', lambda: "5" + 5)
show("5 + '5'", lambda: 5 + "5")
show("len(100)", lambda: len(100))
show("None + 1", lambda: None + 1)
show("int([1, 2])", lambda: int([1, 2]))
show('"3" * "3"', lambda: "3" * "3")
show('"3" * 3  (theek hai!)', lambda: "3" * 3)
````

### 2.3 📍 `IndexError` — Index Range ke Bahar
List mein `n` items hain toh **valid index** `0` se `n-1` (ya `-1` se `-n`).
```
index :   0        1        2        3
        ["Hack", "Quiz", "Dance", "Music"]
index :  -4       -3       -2       -1
```
`fest[4]` ya `fest[-5]` → **IndexError**. String aur tuple par bhi yahi hota hai.

### 2.4 🔑 `KeyError` — Dictionary mein Key Nahi
`d["key"]` mein key na mile toh **KeyError**. Teen bachav ke tareeke:
| Tareeka | Code | Key na ho toh |
|---|---|---|
| Seedha | `d["x"]` | 💥 `KeyError` |
| `.get()` | `d.get("x")` | `None` |
| `.get()` + default | `d.get("x", 0)` | `0` |
| `in` check | `if "x" in d:` | Bina error ke check |

### 2.7 Do Soch: Pehle Dekho (LBYL) vs Pehle Karo (EAFP) 🧠
| Soch | Naam | Kaise |
|---|---|---|
| **"Kudne se pehle dekh lo"** | **LBYL** — *Look Before You Leap* | `if b != 0: result = a / b` |
| **"Kud jao, girne par sambhal lo"** | **EAFP** — *Easier to Ask Forgiveness than Permission* | `try: a / b` `except ZeroDivisionError: ...` |

Python mein **dono** chalte hain. Simple checks ke liye `if`, aur jahan **kai cheezein galat ho sakti hain** (jaise file kholna) wahan `try/except` behtar hai.

````python
data = {"marks": [80, 90, 70]}

# 🅰️ LBYL — pehle check
def average_lbyl(d):
    if "marks" in d and len(d["marks"]) > 0:
        return sum(d["marks"]) / len(d["marks"])
    return None

# 🅱️ EAFP — karo, error aaye toh sambhalo
def average_eafp(d):
    try:
        return sum(d["marks"]) / len(d["marks"])
    except (KeyError, ZeroDivisionError):
        return None

for d in [data, {"marks": []}, {}]:
    print(f"{str(d):22} -> LBYL: {average_lbyl(d)} | EAFP: {average_eafp(d)}")
````

# 🕸️ Chapter 3: `try-except` — Safety Net

### 🎬 Story
Circus mein trapeze artist upar jhoolta hai — par neeche ek **jaal (safety net)** bichha hota hai. Gir bhi gaya toh **show chalta rehta hai!** 🎪

Bhavna Didi: *"Aarav, tera kiosk bhi aise hi chalega. **Risky kaam `try` mein** — aur neeche **`except` ka jaal**. Galti hui toh program **girega nahi, sambhal jayega.**"*

### 🧱 Syntax
```python
try:
    # 🎯 risky code — yahan error aa sakta hai
except ErrorType:
    # 🛟 plan B — error aaye toh ye chalega
```

### 🔄 Flow — Kya, Kab, Kaise?
| Situation | Kya hota hai |
|---|---|
| `try` mein **koi error nahi** | Poora `try` chalta hai → `except` **skip** → aage badho |
| `try` mein **error aaya** aur `except` match hua | `try` ki **baaki lines skip** → matching `except` chalta hai → aage badho |
| Error aaya par **koi `except` match nahi hua** | Exception **bahar bhaag** jaati hai → agar kisi ne nahi pakdi, **crash** 💥 |

> 🧠 **`except ... as e`** → exception object ko naam `e` mil jaata hai: `type(e).__name__` (naam), `str(e)` (message), `e.args` (arguments ka tuple).

````python
print("1. Program shuru")

try:
    print("2. try shuru hua")
    x = 10 / 0                                    # 💥 yahan error!
    print("3. ye line SKIP ho jaayegi")
except ZeroDivisionError:
    print("4. except chala — error sambhal liya")

print("5. try-except ke baad program jaari rehta hai ✅")
````

````python
# ⌨️ INTERACTIVE: Umar poochho — galat likhoge toh bhi crash nahi hoga
# (pehle "twenty" type karke dekho, phir 19)
try:
    age = int(input("Aapki umar: "))
    print(f"✅ Agle saal aap {age + 1} saal ke honge.")
except ValueError:
    print("⚠️ Please number likho, jaise 19")

print("Kiosk chalta raha 🎉")
````

### 3.1 `as e` — Exception Object Ke Andar Kya Hai?
| Kya chahiye | Code | Example output |
|---|---|---|
| Exception ka **naam** | `type(e).__name__` | `ValueError` |
| **Message** | `str(e)` ya `print(e)` | `invalid literal for int()...` |
| **Arguments** (tuple) | `e.args` | `("invalid literal...",)` |
| **Developer-style** | `repr(e)` | `ValueError("invalid literal...")` |

````python
try:
    int("twenty")
except ValueError as e:
    print("type(e).__name__ :", type(e).__name__)
    print("str(e)           :", str(e))
    print("e.args           :", e.args)
    print("repr(e)          :", repr(e))
    print("isinstance(e, Exception):", isinstance(e, Exception))
````

### 3.2 ⚠️ `except:` aur `except Exception:` — Jaal Itna Bada Mat Karo!
| Style | Kya pakadta hai | Raay |
|---|---|---|
| `except ZeroDivisionError:` | **Sirf** ye ek | ✅ **Sabse behtar** — jo expect karte ho wahi pakdo |
| `except Exception:` | Lagbhag **saari** aam errors | ⚠️ Sirf **aakhri jaal** ya logging ke liye |
| `except:` *(bina naam)* | **Sab kuch**, `Ctrl+C` bhi! | ❌ **Mat use karo** |

> 😱 **Bahut bada khatra:** bahut chauda `except` **asli bugs ko chhupa deta hai!** Neeche dekho — typo ki wajah se jawab galat aa raha hai, par kisi ko pata hi nahi chala.

````python
def total_price(prices):
    try:
        return sum(prces)                         # ❌ typo: prces (prices likhna tha)
    except Exception:
        return 0                                  # 😱 error nigal gaya!

print("total_price([100, 200]) =", total_price([100, 200]), " <- GALAT (300 hona chahiye), par koi error dikha nahi!")

# ✅ Sirf wahi pakdo jo expect karte ho — typo wali NameError bahar aa jaati hai
def total_price_safe(prices):
    try:
        return sum(prces)
    except TypeError:                             # sirf TypeError pakdenge
        return 0

try:
    total_price_safe([100, 200])
except NameError as e:
    print("✅ Ab bug pakda gaya:", e)
````

### 3.3 Multiple `except` Blocks — Alag Error, Alag Ilaaj
```python
try:
    ...
except ZeroDivisionError:
    ...              # zero se bhaag ka ilaaj
except ValueError:
    ...              # galat value ka ilaaj
except Exception as e:
    ...              # baaki sab (aakhri jaal)
```
- Python **upar se neeche** check karta hai aur **pehla match** chunta hai — baaki **skip**.
- Aur **ek hi `except`** chalta hai (ek exception, ek ilaaj).

### 3.4 Kai Exceptions Ek Saath: Tuple ✨
Jab **ilaaj ek hi** ho, toh tuple mein likho: `except (KeyError, IndexError) as e:`

````python
def run(kind):
    try:
        if kind == "zero":  return 10 / 0
        if kind == "value": return int("abc")
        if kind == "key":   return {"a": 1}["b"]
        if kind == "index": return [1, 2][5]
        if kind == "type":  return "5" + 5
        return "sab theek"
    except ZeroDivisionError:
        return "➗ Zero se bhaag nahi sakte"
    except ValueError:
        return "🔢 Value galat hai"
    except (KeyError, IndexError) as e:          # ✨ do exceptions, ek ilaaj
        return f"🔎 Nahi mila ({type(e).__name__})"
    except Exception as e:                       # aakhri jaal
        return f"❓ Anjaan error: {type(e).__name__}"

for kind in ["zero", "value", "key", "index", "type", "ok"]:
    print(f"{kind:6} -> {run(kind)}")
````

### 3.5 🥇 `except` ka **Order** Matter Karta Hai!
Python **pehla match** chunta hai. Agar **baap (`Exception`) pehle** likh diya, toh **bachche (`ZeroDivisionError`) tak baat pahunchti hi nahi!**

| Order | Result |
|---|---|
| ✅ **Specific pehle**, general baad mein | Har error ko apna ilaaj mila |
| ❌ **General pehle** | Neeche wale `except` **kabhi nahi chalte** (par Python koi warning nahi deta!) |

> 🌳 Chapter 1 ka family tree yaad karo: `ArithmeticError` ⊃ `ZeroDivisionError`, `LookupError` ⊃ `IndexError`, `KeyError`.

````python
def wrong_order():
    try:
        10 / 0
    except Exception:                    # ❌ baap pehle
        return "Exception wala handler"
    except ZeroDivisionError:            # 😢 ye kabhi nahi chalega
        return "ZeroDivisionError wala handler"

def right_order():
    try:
        10 / 0
    except ZeroDivisionError:            # ✅ specific pehle
        return "ZeroDivisionError wala handler"
    except Exception:
        return "Exception wala handler"

print("Galat order :", wrong_order())
print("Sahi order  :", right_order())

# 🌳 Baap pakde toh bachche bhi pakde jaate hain
for func in [lambda: {}["k"], lambda: [][0], lambda: int("x")]:
    try:
        func()
    except LookupError as e:
        print(f"LookupError ne pakda -> {type(e).__name__}")
    except Exception as e:
        print(f"LookupError ne NAHI pakda -> {type(e).__name__}")
````

### 3.6 🔁 Retry Loop — "Jab Tak Sahi Na Ho, Poochte Raho"
Kiosk ke liye sabse kaam ki trick: **`while True` + `try/except` + `break`**.

````python
# ⌨️ INTERACTIVE: Sahi umar aane tak poochho
# (pehle "abc", phir "200" aur aakhir mein "19" try karo)
while True:
    try:
        age = int(input("Umar (1-120): "))
        if not 1 <= age <= 120:
            print("⚠️ 1 se 120 ke beech likho")
            continue
        break                                     # ✅ sahi mila — loop khatam
    except ValueError:
        print("⚠️ Number likho!")

print(f"✅ Umar mil gayi: {age}")
````

### 3.7 Exception Upar Bhaagti Hai (Propagation) 🏃
Agar function ke andar exception **pakdi nahi gayi**, toh wo apne **caller** ke paas bhaagti hai... phir uske caller ke paas... jab tak **koi pakde** ya **program crash**. (Day 3 ka **call stack** yaad hai? Wahi raasta!)

````python
def level3():
    print("   level3: kaam shuru")
    return 1 / 0                                   # 💥 yahan error
    print("   level3: ye kabhi nahi chalega")

def level2():
    print("  level2: level3 ko bulaya")
    level3()
    print("  level2: ye bhi SKIP hogi")

def level1():
    print(" level1: level2 ko bulaya")
    level2()
    print(" level1: ye bhi SKIP hogi")

try:
    level1()
except ZeroDivisionError:
    print("🛟 Sabse bahar wale try ne exception pakad li!")
````

### 3.8 Loop ke Andar `try/except` — Ek Gandi Entry Poori List Nahi Bigaadti
Jab **bahut saara data** process karna ho, toh **har item ke liye alag `try/except`** lagao — ek kharab item baaki sab ko nahi rokega.

````python
raw_marks = ["88", "abc", "95", "", "72", "7.5", "60"]
valid, invalid = [], []

for item in raw_marks:
    try:
        valid.append(int(item))
    except ValueError:
        invalid.append(item)

print("Valid marks   :", valid)
print("Invalid items :", invalid)
print("Average       :", sum(valid) / len(valid))
````

# 🧹 Chapter 4: `else` aur `finally` — Safai Abhiyan

### 🎬 Story
Utkarsh ke canteen mein **chai** ☕ ban rahi hai. Bhavna Didi ne `try` ke **4 hisse** samjhaye:

| Scene | Python |
|---|---|
| ☕ Chai banane ki **koshish** | `try` |
| 🥛 Doodh ukal gaya — **Plan B** | `except` |
| 🎉 Sab theek raha — **serve karo!** | `else` |
| 🧽 Chai bani ya bigdi — **bartan dhona zaroori** | `finally` |

### 🗺️ 4 Hisse — Kaun Kab Chalta Hai?
| Block | Kab chalta hai? |
|---|---|
| `try` | **Hamesha pehle** |
| `except` | **Sirf jab** `try` mein exception aaye aur match ho |
| `else` | **Sirf jab** `try` mein **koi exception nahi** aayi |
| `finally` | **HAMESHA** — error ho ya na ho, pakdi gayi ho ya nahi ✅ |

```
              ┌── error NAHI aayi ──► else ──────────┐
   try ───────┤                                       ├──► finally ──► aage
              └── error aayi ───────► except ─────────┘
                       └─ (match NAHI hua) ──────────────► finally ──► 💥 crash/upar bhaagi
```

````python
def divide(a, b):
    print(f"\n--- divide({a}, {b}) ---")
    try:
        result = a / b
    except ZeroDivisionError:
        print("except : ➗ zero se bhaag nahi sakte")
    else:
        print(f"else   : 🎉 sab theek! result = {result}")
    finally:
        print("finally: 🧽 safai ho gayi (HAMESHA chalta hai)")

divide(10, 2)        # koi error nahi  -> try, else, finally
divide(10, 0)        # error aayi      -> try (adhura), except, finally
````

### 4.1 `else` Kyun? Seedha `try` mein kyun nahi? 🤔
`else` ka matlab: **"sirf success par chalne wala code"**. Isse `try` **chhota** rehta hai — aur `try` mein **sirf wahi** likhna chahiye jahan error **aa sakta hai**.

> ⚠️ **`else` ke andar ka error uske apne `except` se pakda NAHI jaata!** Wo bahar chala jaata hai. Ye galti nahi — **design** hai: `except` sirf `try` ke code ke liye hai.

````python
def parse_then_use(text):
    try:
        n = int(text)                       # 🎯 sirf yahan ValueError aa sakta hai
    except ValueError:
        return "⚠️ number nahi hai"
    else:
        return 100 / n                      # n = 0 ho toh ZeroDivisionError — ValueError-handler nahi pakdega

print('parse_then_use("4")   ->', parse_then_use("4"))
print('parse_then_use("abc") ->', parse_then_use("abc"))

try:
    parse_then_use("0")
except ZeroDivisionError:
    print('parse_then_use("0")   -> ZeroDivisionError bahar nikal aayi (else ke andar ka error `except ValueError` ne nahi pakda)')
````

### 4.2 `finally` — "Chahe Kuch Bhi Ho, Ye Chalega" 🛡️
`finally` in **sabhi** situations mein chalta hai:
| Situation | `finally` chalta hai? |
|---|---|
| `try` poora chala, koi error nahi | ✅ |
| Error aayi aur `except` ne pakdi | ✅ |
| Error aayi par **kisi ne nahi pakdi** (crash hone wala) | ✅ (crash se **pehle**) |
| `try` mein **`return`** hai | ✅ (return poora hone se **pehle**) |
| Loop mein `break` / `continue` | ✅ |

Isliye **safai ke kaam** (file band karna, connection chhodna) `finally` mein rakhte hain.

````python
def trace(mode):
    print(f"\n▶ mode = {mode!r}")
    try:
        print("  try: shuru")
        if mode == "ok":
            return "try ne return kiya"
        if mode == "zero":
            1 / 0
        if mode == "value":
            int("abc")
        print("  try: poora hua")
    except ZeroDivisionError:
        print("  except: ZeroDivisionError pakdi")
    finally:
        print("  finally: 🧽 chal gaya")

print("1. return value:", trace("ok"))     # return ke saath bhi finally chalega
print("2. return value:", trace("zero"))   # error pakdi gayi (return nahi hua, isliye None)

# 3. Error pakdi NAHI gayi — finally phir bhi chalta hai, phir exception bahar jaati hai
try:
    trace("value")
except ValueError:
    print("  (bahar wale try ne ValueError pakdi — finally pehle hi chal chuka tha)")

# 4. Loop ke andar break
for i in range(3):
    try:
        if i == 1:
            break
    finally:
        print(f"  loop finally: i = {i}")
````

### 4.3 🪤 Gotcha: `finally` Mein `return` Mat Likho!
`finally` mein `return` likhoge toh wo **`try` ke return ko (aur exceptions ko bhi) overwrite** kar dega — bugs chhup jaate hain.

````python
def good():
    try:
        return "try ka return"
    finally:
        print("   (good: finally ne sirf safai ki)")

def bad():
    try:
        return "try ka return"
    finally:
        return "finally ka return"          # ❌ try ke return ko dabaa diya!

print("good() ->", good())
print("bad()  ->", bad(), "  <- try wala jawab gum!")
````

### 4.4 Real Safai: Resource Band Karna 📁
Jab bhi hum koi **resource** kholte hain (file, connection), use **band karna zaroori** hai — **chahe beech mein error aaye**. Yahan hum file ka **sneak peek** dekh rahe hain (poori detail Chapter 7 mein!). Abhi bas ye dekho ki `finally` har haal mein file band karta hai:

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)
os.makedirs("lab")

f = None
try:
    f = open("lab/data.txt", "w")
    f.write("Utkarsh 2026")
    print("Likh diya. File band hai?", f.closed)
    1 / 0                                      # 💥 beech mein error!
except ZeroDivisionError:
    print("⚠️ Error aayi!")
finally:
    if f is not None:
        f.close()                              # 🧽 chahe kuch bhi ho, file band hogi
        print("finally ne file band ki. File band hai?", f.closed)

shutil.rmtree("lab")
````

### 4.5 Nested `try` — Andar Wala Pehle! 🪆
Andar wala `finally` **pehle** chalta hai, phir bahar wala `except`, phir bahar wala `finally` — bilkul **call stack** ki tarah (andar se bahar).

````python
try:
    try:
        print("1. inner try")
        1 / 0
    finally:
        print("2. inner finally  (exception ko roka nahi, bas safai ki)")
except ZeroDivisionError:
    print("3. outer except   (ab exception pakdi gayi)")
finally:
    print("4. outer finally")
````

### 4.6 Poora Template: `try / except / else / finally` 🧾

````python
# ⌨️ INTERACTIVE: Poora template — number likho (try "abc" aur phir "0" aur phir "4" alag-alag run mein)
try:
    n = int(input("Ek number likho: "))
    answer = 100 / n
except ValueError:
    print("⚠️ Number likho, text nahi!")
except ZeroDivisionError:
    print("⚠️ Zero se 100 divide nahi hota!")
else:
    print(f"✅ 100 / {n} = {answer}")
finally:
    print("🙏 Dhanyavaad! (finally hamesha chalta hai)")
````

# 📣 Chapter 5: `raise` aur `assert` — Alarm Bajao

### 🎬 Story
Fee-counter ka program chal raha tha. Ek volunteer ne galti se fees likh di: **`-500`** — aur program ne **chupchaap accept kar li!** 😱 Kisi ko pata hi nahi chala ki "negative fees" ka matlab kya hota hai.

Bhavna Didi: *"Aarav, ab tak tu **doosron ki** exceptions pakad raha tha. Kabhi-kabhi **tujhe khud alarm bajana** padta hai — jab data galat ho, toh **kaam rok do aur shor machao!** Iske liye hai **`raise`**."* 🚨

### 🧱 `raise` ke 4 Roop
| Syntax | Matlab |
|---|---|
| `raise ValueError("message")` | Exception **banao aur phenko** (sabse aam) |
| `raise ValueError` | Class do — Python khud object banata hai (bina message) |
| `raise` *(akela, `except` ke andar)* | Abhi pakdi hui exception ko **dobara upar phenko** (re-raise) |
| `raise NewError("...") from e` | **Nayi** exception phenko, par **purani wajah** (`e`) saath rakho (chaining) |

### 🤔 Kab `raise` Karein?
| Situation | Kya karein |
|---|---|
| Function ko **galat input** mila | `raise ValueError` / `TypeError` |
| Kaam **aage badh hi nahi sakta** | `raise` — chupchaap `None` return mat karo |
| Sirf user ko batana hai | `print` (par function ke andar `raise` zyada saaf hai) |

### 🏷️ Sahi Exception Type Chunno
| Problem | Exception |
|---|---|
| Value galat (negative fees, bahar ki range) | `ValueError` |
| Type galat (number ki jagah text) | `TypeError` |
| Kuch aur gadbad (generic) | `RuntimeError` |

````python
def set_fees(amount):
    if amount < 0:
        raise ValueError(f"Fees negative nahi ho sakti: {amount}")
    return amount

print("set_fees(500)  ->", set_fees(500))

try:
    set_fees(-500)
except ValueError as e:
    print("🚨 Alarm pakda gaya:", e)

print("\nProgram zinda hai ✅")
````

### 5.1 Arguments ko Validate Karna ✅
Function ki **pehli kuch lines** mein check karo — galat mila toh **turant `raise`**. Isse bug **jahan hua wahin** pakda jaata hai, kahin door jaake nahi.

````python
def check_age(age):
    if not isinstance(age, int):
        raise TypeError("Age must be an integer")
    if not 0 <= age <= 120:
        raise ValueError("Age must be between 0 and 120")
    return age

for value in [19, -5, 200, "twenty", 3.5]:
    try:
        print(f"{value!r:10} -> OK ({check_age(value)})")
    except (TypeError, ValueError) as e:
        print(f"{value!r:10} -> {type(e).__name__}: {e}")
````

### 5.2 Bare `raise` — Pakdo, Note Karo, Phir Aage Phenko 🔁
Kabhi **pakadna** nahi chahte — bas **log** likhna chahte ho aur exception **upar jaane do**. `except` ke andar akela `raise` yehi karta hai.

````python
log = []

def process(x):
    try:
        return 100 / x
    except ZeroDivisionError as e:
        log.append(f"ERROR: {e}")              # 📝 sirf note kiya
        raise                                   # 🔁 exception ko wapas upar phenko

print("process(4) =", process(4))

try:
    process(0)
except ZeroDivisionError:
    print("Bahar wale ne pakda: ZeroDivisionError")

print("Log:", log)
````

### 5.3 `raise ... from e` — Chaining: Asli Wajah Saath Rakho 🔗
Kabhi andar ki exception (jaise `KeyError`) user ke liye **samajhne layak nahi** hoti. Tab hum **apni, saaf exception** phenkte hain — par **asli wajah `from e`** se saath rakhte hain.

````python
PRICES = {"Hackathon": 200, "Quiz": 50}

def get_price(event):
    try:
        return PRICES[event]
    except KeyError as e:
        raise ValueError(f"Unknown event: {event}") from e       # saaf message + asli wajah

print("Quiz ka price:", get_price("Quiz"))

try:
    get_price("Chess")
except ValueError as e:
    print("Error :", e)
    print("Cause :", type(e.__cause__).__name__, "->", e.__cause__)
````

### 5.4 `assert` — Apne Aap Se Sawaal 🧐
```python
assert condition, "error message"
```
- `condition` **True** → kuch nahi hota, aage badho.
- `condition` **False** → **`AssertionError`** (message ke saath).

### ⚖️ `raise` vs `assert` — Sabse Zaroori Farq
| | **`raise`** | **`assert`** |
|---|---|---|
| Kaam | **Asli galtiyon** ko roko (user input, files...) | **Developer ki apni galti** pakdo ("ye toh kabhi hona hi nahi chahiye") |
| Hamesha chalta hai? | ✅ **Haan** | ⚠️ **Nahi!** `python -O` se **hata diye jaate hain** |
| Exception | Jo chaho (`ValueError`, ...) | Sirf `AssertionError` |
| User ke input ke liye? | ✅ Haan | ❌ **Kabhi nahi** |

> 🧠 **Rule:** *User ya bahar ki duniya* se aane wala data → **`raise`**. *Apne code ke andar ka bharosa* ("is waqt list khaali nahi honi chahiye") → **`assert`**.

````python
def average(marks):
    assert len(marks) > 0, "Marks list empty"
    assert all(0 <= m <= 100 for m in marks), "Marks must be between 0 and 100"
    return sum(marks) / len(marks)

for marks in [[80, 90, 70], [], [50, 105]]:
    try:
        print(f"{marks!s:14} -> Average: {average(marks)}")
    except AssertionError as e:
        print(f"{marks!s:14} -> AssertionError: {e}")
````

### 5.5 😱 `assert` Ko Band Kiya Ja Sakta Hai!
Python ko `-O` (optimize) flag ke saath chalao toh **saare `assert` ignore ho jaate hain**. Isliye **user input kabhi `assert` se check mat karo** — warna production mein check gayab!

````python
import subprocess, sys

program = 'assert 1 == 2, "yahan fail hona chahiye"\nprint("assert chala hi nahi — aage badh gaye!")'

normal = subprocess.run([sys.executable, "-c", program], capture_output=True, text=True)
optimized = subprocess.run([sys.executable, "-O", "-c", program], capture_output=True, text=True)

print("🔹 Normal chalaya   -> exit code:", normal.returncode, "| AssertionError aaya?", "AssertionError" in normal.stderr)
print("🔹 python -O se     -> exit code:", optimized.returncode, "| output:", optimized.stdout.strip())
````

### 🧠 Kyun Apni Exception?
| Fayda | Matlab |
|---|---|
| 🏷️ **Saaf naam** | `DuplicateEntryError` padhte hi samajh aata hai |
| 👨‍👩‍👧 **Group mein pakadna** | Ek `except FestError` se saare fest-errors pakdo |
| 📦 **Extra data** | Exception ke saath `balance`, `field` jaisi cheezein bhi bhejo |
| 🔒 **Alag pehchan** | Python ki apni `ValueError` se mix-up nahi hota |

> 📘 **Class ka chhota primer:** *Class* ka poora chapter aage OOP mein aayega. Aaj bas **do pattern** yaad rakhne hain — (1) `class NaamError(Exception): pass` aur (2) `__init__` mein extra data rakhna (neeche dikhaya hai). Bas itna!

### 6.1 Pattern 1: Sabse Simple — Ek Line ki Exception
```python
class NegativeAmountError(Exception):    # 👈 Exception ka bachcha
    pass                                  # 👈 kuch aur nahi chahiye!
```
- Naam ka **ant `Error` se** karo (convention).
- Parent hamesha **`Exception`** (ya uska koi bachcha) — `BaseException` **nahi**.
- Phir `raise NegativeAmountError("message")` aur `except NegativeAmountError:` — bilkul built-in jaisa!

````python
class NegativeAmountError(Exception):
    pass

def add_fees(total, amount):
    if amount < 0:
        raise NegativeAmountError(f"Negative fees not allowed: {amount}")
    return total + amount

print("add_fees(100, 50) =", add_fees(100, 50))

try:
    add_fees(100, -50)
except NegativeAmountError as e:
    print("Caught :", e)
    print("Type   :", type(e).__name__)
    print("Is an Exception? :", isinstance(e, Exception))
    print("Is a ValueError? :", isinstance(e, ValueError), "(nahi! alag hi pehchan hai)")
    print("Family :", " -> ".join(c.__name__ for c in NegativeAmountError.__mro__))
````

### 6.2 Pattern 2: Extra Data Ke Saath ✨
Jab exception ke saath **jaankari** bhejni ho (kitna balance tha, kitna chahiye tha), toh `__init__` banao:
```python
class InsufficientFundsError(Exception):
```

| Line | Kaam |
|---|---|
| `self.balance = balance` | Data ko exception object mein **save** karo |

````python
class InsufficientFundsError(Exception):
    def __init__(self, balance, needed):
        self.balance = balance
        self.needed = needed
        self.shortfall = needed - balance
        super().__init__(f"Insufficient funds: balance={balance}, needed={needed}")

def withdraw(balance, needed):
    if needed > balance:
        raise InsufficientFundsError(balance, needed)
    return balance - needed

print("withdraw(500, 300) ->", withdraw(500, 300))

try:
    withdraw(500, 800)
except InsufficientFundsError as e:
    print("Message  :", e)
    print("Balance  :", e.balance)
    print("Needed   :", e.needed)
    print("Shortfall:", e.shortfall, " <- exception ke saath data bhi aaya!")
````

### 6.3 Exception Family 👨‍👩‍👧‍👦 — Ek Baap, Kai Bachche
Pehle ek **base class** banao (`FestError`), phir har problem uska **bachcha**:
```
Exception
 └── FestError                          ← 👑 Fest ki saari errors ka baap
      ├── RegistrationClosedError
      ├── DuplicateEntryError
      └── PaymentFailedError
```
| Tareeka | Kya pakdega |
|---|---|
| `except DuplicateEntryError:` | **Sirf** duplicate |
| `except FestError:` | **Teeno** (baap pakde toh bachche bhi!) |
| `except Exception:` | Fest ki bhi, Python ki bhi — sab |

````python
class FestError(Exception):                         # 👑 base
    pass

class RegistrationClosedError(FestError):
    pass

class DuplicateEntryError(FestError):
    pass

class PaymentFailedError(FestError):
    pass

registered = set()

def register(name, day, paid):
    if day not in (1, 2, 3):
        raise RegistrationClosedError(f"Day {day} is closed")
    if name in registered:
        raise DuplicateEntryError(f"{name} already registered")
    if not paid:
        raise PaymentFailedError(f"{name} has not paid")
    registered.add(name)
    return f"Registered: {name}"

requests = [("Aarav", 1, True), ("Aarav", 2, True), ("Bhavna", 5, True),
            ("Chirag", 2, False), ("Deepika", 3, True)]

for name, day, paid in requests:
    try:
        print(register(name, day, paid))
    except FestError as e:                          # 👑 ek hi except se teeno
        print(f"{type(e).__name__}: {e}")

print("\nSabhi FestError ke bachche hain?",
      all(issubclass(c, FestError) for c in (RegistrationClosedError, DuplicateEntryError, PaymentFailedError)))
````

### 6.5 Custom Exception + Chaining — Asli Wajah Ke Saath 🔗

### 6.6 Real Pattern: Jaankari Wali Exception + Loop ✨
Ek **seat-booking** system — har galti ke saath **kaunsi seat** aur **kya wajah** thi wo bhi aata hai, aur loop **rukta nahi**.

````python
class SeatError(Exception):
    def __init__(self, seat, reason):
        self.seat = seat
        self.reason = reason
        super().__init__(f"Seat {seat}: {reason}")

taken = {"A1", "B2"}

def book(seat):
    if len(seat) != 2 or seat[0] not in "ABC" or not seat[1].isdigit():
        raise SeatError(seat, "invalid seat code")
    if seat in taken:
        raise SeatError(seat, "already taken")
    taken.add(seat)
    return f"Seat {seat} booked"

booked, failed = 0, []
for seat in ["C3", "A1", "Z9", "B2", "A2"]:
    try:
        print("✅", book(seat))
        booked += 1
    except SeatError as e:
        print("⚠️ ", e)
        failed.append((e.seat, e.reason))

print(f"\nBooked: {booked} | Failed: {failed}")
````

### 7.1 `with` Statement — Apne Aap Band Hone Wala Darwaza 🚪✨
`close()` bhoolna **bahut aam galti** hai — aur beech mein error aaye toh `close()` tak code pahunchta hi nahi! Isliye Python ne diya **`with`**:
```python
with open("guests.txt", "w", encoding="utf-8") as f:
    f.write("Aarav")
# 👆 yahan aate hi file AUTOMATICALLY band — chahe error aaye ya na aaye!
```
| | `open()` + `close()` | **`with open()`** ✅ |
|---|---|---|
| Band karna | Khud likhna padta hai | **Apne aap** |
| Beech mein error aaye | File **khuli reh sakti hai** | File **phir bhi band** hoti hai |
| Code | Lamba (`try/finally` chahiye) | Chhota aur saaf |

> 🧠 **`with` = Chapter 4 ke `try/finally` ka shortcut!** Andar se `finally: f.close()` apne aap lag jaata hai.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/guests.txt", "w", encoding="utf-8") as f:
    f.write("Aarav\nBhavna\n")
    print("with ke ANDAR band hai?", f.closed)

print("with ke BAAD band hai?  ", f.closed, " <- apne aap band ho gayi ✅")

# with ke andar error aaye tab bhi file band hoti hai
try:
    with open("lab/guests.txt", "a", encoding="utf-8") as f:
        f.write("Chirag\n")
        1 / 0                                      # 💥 beech mein error!
except ZeroDivisionError:
    print("\nError aayi... par file band hai?", f.closed, "✅")

shutil.rmtree("lab")
````

### 7.4 Aam Galtiyan — Asli Error Messages Ke Saath 💥

|---|---|

| File ki jagah **folder** kholna | `IsADirectoryError` |

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

def attempt(label, func):
    try:
        func()
        print(f"✅ {label:38} -> theek chala")
    except Exception as e:
        print(f"💥 {label:38} -> {type(e).__name__}: {e}")

def read_closed():
    with open("lab/x.txt", "w") as f:
        pass
    f.read()                                       # band file par padhna

attempt('open("lab/missing.txt")', lambda: open("lab/missing.txt"))
attempt('open("lab/nofolder/x.txt", "w")', lambda: open("lab/nofolder/x.txt", "w"))
attempt('open("lab")  (folder!)', lambda: open("lab"))
attempt("band file par read()", read_closed)
attempt('open("lab/x.txt", "z")', lambda: open("lab/x.txt", "z"))

shutil.rmtree("lab")
````

# 🎛️ Chapter 8: File Modes — Mode Ka Mode

### 🎬 Story
Aarav ke paas ek **guest diary** thi. Teen kaam the: **naya page** shuru karna, **purane pages mein naya naam jodna**, aur **sirf padhna**. Usne `"w"` mode lagaya sab jagah...

Bhavna Didi: *"Mode galat chuna toh **data udd jaata hai**. Chaar mode ratt le: **r, w, a, r+**."* 🎛️

### 🎛️ Mode Table
| Mode | Matlab | File **nahi** hai toh | File **hai** toh | Pointer |
|:-:|---|---|---|---|
| **`r`** | **Read** (default) | ❌ `FileNotFoundError` | Padho | Shuru |
| **`w`** | **Write** | ✅ Naya banao | ⚠️ **Purana SAAF!** | Shuru |
| **`a`** | **Append** (peeche jodo) | ✅ Naya banao | ✅ Purana **safe**, naya peeche | **Aakhir** |
| **`r+`** | Read **aur** Write | ❌ `FileNotFoundError` | Purana safe, likhoge toh **overwrite** (pointer se) | Shuru |
| **`rb`** | Read **binary** | ❌ Error | Bytes padho | Shuru |
| **`wb`** | Write **binary** | ✅ Naya | ⚠️ Saaf | Shuru |

### 🧬 Mode String ki Anatomy
```
   r  /  w  /  a  /  x     +  (optional)     t  /  b  (optional)
   └── kaam ──────────┘    └ dono (read+write) └ text (default) / binary
```
| Mode | Kaam |
|:-:|---|
| `x` *(bonus)* | **Naya** file banao — file pehle se hai toh `FileExistsError` (safe!) |
| `w+` / `a+` *(bonus)* | Likhne ke saath padhna bhi |
| `rt` = `r` | `t` (text) default hai |
| `rb`, `wb`, `ab` | Binary versions |

> ⚠️ **Sabse bada rule:** `"w"` **hamesha** file ko **khaali karke** shuru karta hai — bina warning ke! Jodna ho toh `"a"`.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

# ⚠️ 'w' — likhta hai, par PURANA SAB SAAF kar deta hai
with open("lab/guests.txt", "w", encoding="utf-8") as f:
    f.write("Aarav\nBhavna\nChirag\n")
print("1️⃣ Pehli baar 'w' (file bani):")
print(cat("lab/guests.txt"))

with open("lab/guests.txt", "w", encoding="utf-8") as f:     # 😱 dobara 'w'
    f.write("Deepika\n")
print("2️⃣ Dobara 'w' ke baad (purane naam gaye!):")
print(cat("lab/guests.txt"))

shutil.rmtree("lab")
````

### 8.1 Mode ke Hisaab se Kya Allowed Hai? 🚦
Har mode kuch cheezein **rokta** hai:

| Mode | `read()` | `write()` |
|:-:|:-:|:-:|
| `r` | ✅ | ❌ |
| `w` | ❌ | ✅ |
| `a` | ❌ | ✅ |
| `r+` | ✅ | ✅ |

Galat kaam karoge toh **`io.UnsupportedOperation`** milta hai.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/data.txt", "w", encoding="utf-8") as f:
    f.write("sample")

def attempt(label, func):
    try:
        func()
        print(f"✅ {label}")
    except Exception as e:
        print(f"💥 {label:32} -> {type(e).__name__}: {e}")

def write_in_r():
    with open("lab/data.txt", "r") as f:
        f.write("x")

def read_in_w():
    with open("lab/data.txt", "w") as f:
        f.read()

def read_in_a():
    with open("lab/data.txt", "a") as f:
        f.read()

attempt("write() in 'r' mode", write_in_r)
attempt("read()  in 'w' mode", read_in_w)
attempt("read()  in 'a' mode", read_in_a)
attempt("open 'r' on missing file", lambda: open("lab/ghost.txt", "r"))

shutil.rmtree("lab")
````

### 9.1 ⭐ `for line in f` — Sabse Behtar Tareeka
File object **khud iterable** hai — seedha `for` mein daalo. Ek baar mein **sirf ek line** memory mein hoti hai, isliye **lakhon lines** ki file bhi chal jaati hai!

> 💡 `print(line)` **do baar** newline deta hai (ek `line` ka apna `\n`, ek `print` ka). Isliye `print(line, end="")` ya `line.rstrip("\n")` use karo.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

# 🛠️ Setup: ek sample schedule file banate hain
with open("lab/schedule.txt", "w", encoding="utf-8") as f:
    f.write("Utkarsh 2026 Schedule\n"
            "Day 1: Hackathon\n"
            "Day 1: Quiz\n"
            "Day 2: Robo-Race\n"
            "Day 3: Dance Off\n"
            "Day 3: Music Night\n")

print("--- numbering ke saath (enumerate, Day 3) ---")
with open("lab/schedule.txt", encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        print(f"{number}: {line.rstrip()}")

print("\n--- sirf 'Day 3' wali lines ---")
with open("lab/schedule.txt", encoding="utf-8") as f:
    for line in f:
        if line.startswith("Day 3"):
            print("  ", line.strip())

print("\n--- end='' ka trick ---")
with open("lab/schedule.txt", encoding="utf-8") as f:
    for i, line in enumerate(f):
        if i < 2:
            print(line, end="")

shutil.rmtree("lab")
````

### 9.2 End-of-File (EOF) Loop 🔚
`readline()` file khatam hone par **`""`** deta hai (khaali string). Isse hum **loop rok** sakte hain:
```python
while True:
    line = f.readline()
    if not line:         # "" = file khatam
        break
    ...
```
Aur `read(n)` se **chunks** mein padhna bhi isi tarah:

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

# 🛠️ Setup: ek sample schedule file banate hain
with open("lab/schedule.txt", "w", encoding="utf-8") as f:
    f.write("Utkarsh 2026 Schedule\n"
            "Day 1: Hackathon\n"
            "Day 1: Quiz\n"
            "Day 2: Robo-Race\n"
            "Day 3: Dance Off\n"
            "Day 3: Music Night\n")
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

# 1️⃣ readline() ka EOF loop
count = 0
with open("lab/schedule.txt", encoding="utf-8") as f:
    while True:
        line = f.readline()
        if not line:                         # "" matlab file khatam
            break
        count += 1
print("readline loop ne", count, "lines padhi")

# 2️⃣ read(n) se chunks (10-10 characters)
chunks = []
with open("lab/schedule.txt", encoding="utf-8") as f:
    while True:
        chunk = f.read(10)
        if not chunk:
            break
        chunks.append(chunk)

print("chunks :", len(chunks))
print("pehla  :", repr(chunks[0]))
print("aakhri :", repr(chunks[-1]))
print("jodne par poori file wapas milti hai?", "".join(chunks) == cat("lab/schedule.txt"))

shutil.rmtree("lab")
````

|---|---|---|
| `f.read()` | `str` | Poori file • chhoti file ke liye |
| `f.read(n)` | `str` | Agle `n` chars • aakhir mein `""` |
| `f.readline()` | `str` | Ek line (`\n` ke saath) • EOF par `""` |
| `f.readlines()` | `list` | Saari lines (`\n` ke saath) |
| `for line in f` | har `line` | ✅ Badi files ke liye |
| `read().splitlines()` | `list` | Lines **bina `\n`** ke |

# ✍️ Chapter 10: Writing Files — Likhne Wale

### 🎬 Story
Aarav ne guest list likhne ke liye `f.write("Aarav")`, `f.write("Bhavna")`, `f.write("Chirag")` chalaya aur file kholi to dekha:

> `AaravBhavnaChirag`

*"Didi! Teeno naam ek hi line mein chipak gaye!"* 😵

Bhavna Didi: *"`write()` **kuch bhi apne aap nahi jodta** — na space, na newline. **Likhne ka matlab: jo diya, wahi likha.** Aur ek aur raaz — tu jo likhta hai wo **turant disk par nahi** jaata..."* 🤫

### ✍️ Likhne ke Tareeke
| Method | Kaam | Return |
|---|---|---|
| **`f.write(s)`** | **Ek string** likhta hai | Kitne **characters** likhe (`int`) |
| **`f.writelines(list)`** | **Strings ki list** likhta hai | `None` |
| **`f.flush()`** | Buffer ka data **abhi disk par** bhejo (file khuli rehti hai) | `None` |
| **`f.close()`** | Flush + file **band** | `None` |
| `print(..., file=f)` | `print` jaisa, par file mein — **`\n` apne aap** jodta hai | `None` |

> ⚠️ **Do bade rules:**
> 1. `write()` / `writelines()` **newline (`\n`) nahi jodte** — khud lagao!
> 2. `write()` ko **sirf `str`** chahiye — number do toh **`TypeError`** (pehle `str(n)` ya f-string).

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

with open("lab/notes.txt", "w", encoding="utf-8") as f:
    n1 = f.write("Hello")
    n2 = f.write(" Utkarsh")
    n3 = f.write("\n")
    f.write("Line 2\n")
    print("write() ne return kiya:", n1, n2, n3, " <- kitne characters likhe")

print("\nFile (repr se):", repr(cat("lab/notes.txt")))

# ⚠️ Number seedha nahi likh sakte
try:
    with open("lab/notes.txt", "w", encoding="utf-8") as f:
        f.write(2026)
except TypeError as e:
    print("\n💥 f.write(2026) ->", type(e).__name__ + ":", e)

# ✅ Pehle string banao
with open("lab/notes.txt", "w", encoding="utf-8") as f:
    f.write(str(2026) + "\n")
    f.write(f"Fee: Rs {250}\n")
print("✅ str() / f-string ke baad:", repr(cat("lab/notes.txt")))

shutil.rmtree("lab")
````

### 10.1 Newline Khud Lagao! ↩️
`write()` sirf wahi likhta hai jo diya. Alag lines chahiye toh **har naam ke saath `"\n"`**.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

names = ["Aarav", "Bhavna", "Chirag"]

with open("lab/oops.txt", "w", encoding="utf-8") as f:
    for name in names:
        f.write(name)                      # ❌ newline nahi
print("❌ Bina \\n :", repr(cat("lab/oops.txt")), "->", len(cat("lab/oops.txt").splitlines()), "line")

with open("lab/good.txt", "w", encoding="utf-8") as f:
    for name in names:
        f.write(name + "\n")               # ✅ har naam ke baad \n
print("✅ \\n ke saath:", repr(cat("lab/good.txt")), "->", len(cat("lab/good.txt").splitlines()), "lines")

shutil.rmtree("lab")
````

### 10.2 `writelines()` — List Ek Saath, Par Newline Nahi! 🪤
`writelines(list)` = list ke **har item par `write()`** — par **beech mein kuch nahi jodta**. Naam mein bhram hai ("lines"), asal mein ye sirf **strings** likhta hai.

| Tareeka | Result |
|---|---|
| `f.writelines(names)` | ❌ `AaravBhavnaChirag` (ek line) |
| `f.writelines([n + "\n" for n in names])` | ✅ Teen lines |
| `f.write("\n".join(names))` | ✅ Teen lines (par **aakhir mein `\n` nahi**) |

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

names = ["Aarav", "Bhavna", "Chirag"]

with open("lab/a.txt", "w", encoding="utf-8") as f:
    result = f.writelines(names)
print("1️⃣ writelines(names)           :", repr(cat("lab/a.txt")), "| return:", result)

with open("lab/b.txt", "w", encoding="utf-8") as f:
    f.writelines([n + "\n" for n in names])
print("2️⃣ writelines([n + '\\n' ...])  :", repr(cat("lab/b.txt")))

with open("lab/c.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(names))
print("3️⃣ write('\\n'.join(names))     :", repr(cat("lab/c.txt")), "<- aakhri \\n nahi!")

shutil.rmtree("lab")
````

### 10.3 `print(..., file=f)` — Jaana-Pehchana Dost 🖨️
Hum jis `print()` ko jaante hain, wo file mein bhi likh sakta hai: `print(..., file=f)`. Faayda: **`\n` apne aap**, aur `sep=` / `end=` bhi chalte hain.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

with open("lab/p.txt", "w", encoding="utf-8") as f:
    print("Aarav", file=f)
    print("Bhavna", 92, file=f)                     # number bhi chalega (print str bana deta hai)
    print("Chirag", 78, 85, sep=",", file=f)         # sep ke saath
    print("bina newline", end="", file=f)           # end="" -> \n nahi

print(repr(cat("lab/p.txt")))
shutil.rmtree("lab")
````

### 10.4 Buffer aur `flush()` — Likha Hua Turant Disk Par Nahi Jaata! 🧠
Disk **dheema** hota hai, isliye Python likha hua data pehle **buffer (RAM ka chhota dabba)** mein jodta hai aur **baad mein ek saath** disk par bhejta hai.

```
f.write("draft")  ──►  [ BUFFER (RAM) ]  ── flush() / close() / buffer bhar jaye ──►  [ DISK 💾 ]
```
| Kab buffer disk par jaata hai | |
|---|---|
| `f.flush()` | **Abhi** (file khuli rehti hai) |
| `f.close()` / `with` ka end | Band karte waqt (apne aap flush) |
| Buffer **bhar jaye** (aam taur par ~8 KB) | Apne aap |

> 😱 **Khatra:** `flush()`/`close()` se pehle program **crash** ya bijli chali gayi, toh buffer ka data **gaya!** Isliye `with` use karo.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

f = open("lab/live.txt", "w", encoding="utf-8")
f.write("draft")

# Dusre "padhne wale" se dekhte hain ki DISK par kya pahuncha
with open("lab/live.txt", encoding="utf-8") as reader:
    print("flush() se pehle :", repr(reader.read()), " <- abhi buffer mein hai, disk par nahi!")

f.flush()                                              # 🚿 buffer -> disk

with open("lab/live.txt", encoding="utf-8") as reader:
    print("flush() ke baad  :", repr(reader.read()), " <- ab disk par hai (file abhi bhi khuli hai)")

print("File abhi khuli hai?", not f.closed)
f.close()
shutil.rmtree("lab")
````

# 🧭 Chapter 11: `seek()`, `tell()`, `truncate()` — Cursor ka Safar

### 🎬 Story
Aarav ne file ek baar `for` loop se padhi. Phir **dobara** wahi loop chalaya... aur **kuch bhi print nahi hua!** 😳

*"Didi, file khaali ho gayi kya?"*

Bhavna Didi hansi: *"Nahi beta! Tape recorder ki tarah file ka bhi ek **play-head (cursor)** hota hai. Ek baar padh liya toh cursor **aakhir** par pahunch gaya — ab wahan se aage kuch hai hi nahi. **Rewind** karna padega!"* ⏪

### 🧭 Teen Naye Method
| Method | Kaam | Example |
|---|---|---|
| **`f.tell()`** | Cursor **kahan** hai? (shuru se kitne bytes aage) | `f.tell()` → `19` |
| **`f.seek(offset, whence)`** | Cursor ko **kahin bhi** le jao | `f.seek(0)` = rewind |
| **`f.truncate(size)`** | File ko **`size` bytes par kaat do** | `f.truncate(5)` |

### 🎚️ `seek(offset, whence)` — whence kya hai?
| `whence` | Kahan se naapna | Example | Text mode mein? |
|:-:|---|---|:-:|
| **`0`** *(default)* | File ke **shuru** se | `f.seek(5)` — 5 byte aage | ✅ |
| **`1`** | **Abhi ki position** se | `f.seek(3, 1)` — 3 aage | ❌ (sirf binary) |
| **`2`** | File ke **aakhir** se | `f.seek(-3, 2)` — aakhir se 3 peeche | ❌ (sirf binary) |

> 🔒 **Text mode ki limit:** `seek(0)`, `seek(0, 2)` aur `seek(tell_ki_value)` hi pakke chalte hain. Kahin bhi bytes-gin kar jaana ho toh **binary mode (`rb`)** use karo.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/sample.txt", "w", encoding="utf-8") as f:
    f.write("Hello Utkarsh 2026\nDay 4 Files\n")          # 19 + 12 = 31 characters

with open("lab/sample.txt", encoding="utf-8") as f:
    print("open karte hi      tell() =", f.tell())
    f.read(5)
    print("read(5) ke baad    tell() =", f.tell())
    f.readline()
    print("readline() ke baad tell() =", f.tell(), " <- pehli line (19 chars) poori padh li")
    f.read()
    print("read() ke baad     tell() =", f.tell(), " <- file ka aakhir (31)")

shutil.rmtree("lab")
````

### 11.1 `seek(0)` — Rewind ⏪
Cursor aakhir par hai toh dobara padhne par `""` milta hai. `seek(0)` se **shuru par wapas** aao — phir se padho!

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/sample.txt", "w", encoding="utf-8") as f:
    f.write("Hello Utkarsh 2026\nDay 4 Files\n")          # 19 + 12 = 31 characters

with open("lab/sample.txt", encoding="utf-8") as f:
    first_pass = f.read()
    second_pass = f.read()
    print("1st read():", len(first_pass), "characters")
    print("2nd read():", repr(second_pass), " <- cursor aakhir par tha!")

    f.seek(0)                                    # ⏪ rewind
    print("\nseek(0) ke baad:", repr(f.readline()))

    f.seek(6)                                    # seedha 6th character par
    print("seek(6) ke baad :", repr(f.read(7)))

# for-loop wali problem bhi yahi hai
with open("lab/sample.txt", encoding="utf-8") as f:
    pass1 = [line.strip() for line in f]
    pass2 = [line.strip() for line in f]         # cursor aakhir par -> kuch nahi
    f.seek(0)
    pass3 = [line.strip() for line in f]
print("\nfor-loop #1:", pass1)
print("for-loop #2:", pass2, "<- khaali!")
print("seek(0) ke baad for-loop #3:", pass3)

shutil.rmtree("lab")
````

### 11.2 Binary Mode mein Poora `seek(offset, whence)` 🧬
Binary file mein **teeno `whence`** chalte hain. Neeche `A` se `Z` tak 26 bytes ki file hai (A = byte 0, B = byte 1 ...).

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/alpha.bin", "wb") as f:
    f.write(bytes(range(65, 91)))                # b"ABCDEFGHIJKLMNOPQRSTUVWXYZ"

with open("lab/alpha.bin", "rb") as f:
    f.seek(5)                                    # whence=0: shuru se 5
    print("seek(5)      ->", f.read(1), " (shuru se 5 aage = 'F')")

    f.seek(2, 1)                                 # whence=1: abhi se 2 aage
    print("seek(2, 1)   ->", f.read(1), " (abhi 6 par tha, +2 = 8 = 'I')")

    f.seek(-3, 2)                                # whence=2: aakhir se 3 peeche
    print("seek(-3, 2)  ->", f.read(), " (aakhri 3 bytes)")

    print("tell() ab    =", f.tell())

shutil.rmtree("lab")
````

### 11.3 File ka Size `seek` Se 📏
`seek(0, 2)` cursor ko **aakhir** par le jaata hai — aur tab `tell()` = **file ka size**!

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/sample.txt", "w", encoding="utf-8") as f:
    f.write("Hello Utkarsh 2026\nDay 4 Files\n")          # 19 + 12 = 31 characters

with open("lab/sample.txt", "rb") as f:
    f.seek(0, 2)                                  # aakhir par
    size = f.tell()

print("seek(0, 2) + tell() se size:", size)
print("os.path.getsize() se size  :", os.path.getsize("lab/sample.txt"))
print("Dono barabar?", size == os.path.getsize("lab/sample.txt"))

# ⚠️ Text mode ki limit
with open("lab/sample.txt", encoding="utf-8") as f:
    for args in [(3, 1), (-3, 2)]:
        try:
            f.seek(*args)
        except Exception as e:
            print(f"\n💥 text mode mein seek{args} ->", type(e).__name__ + ":", e)

shutil.rmtree("lab")
````

### 11.4 `r+` + `seek()` — File Mein Beech Mein Badalna ✏️
`seek(position)` se cursor wahan le jao jahan badalna hai, phir `write()`. **Overwrite** hota hai (insert nahi) — isliye **same length** ka text rakho warna aage ka text bigad jaayega.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

with open("lab/greet.txt", "w", encoding="utf-8") as f:
    f.write("Hello World")

with open("lab/greet.txt", "r+", encoding="utf-8") as f:
    f.seek(6)                                    # "World" yahan se shuru hota hai
    f.write("Earth")                             # same length (5) — safe

print("Pehle :", "Hello World")
print("Baad  :", cat("lab/greet.txt"))

# ⚠️ Alag length ka text overwrite karta hai — insert nahi!
with open("lab/greet.txt", "w", encoding="utf-8") as f:
    f.write("Hello World")
with open("lab/greet.txt", "r+", encoding="utf-8") as f:
    f.seek(6)
    f.write("Python!!")                          # 8 chars, "World" sirf 5 tha
print("Lamba text:", cat("lab/greet.txt"), " <- peeche ka kuch nahi bacha, poora overwrite hua")

shutil.rmtree("lab")
````

### 11.5 `truncate()` — File Ko Kaat Do ✂️
| Call | Kya hota hai |
|---|---|
| `f.truncate(5)` | File **5 bytes par kat** jaati hai |
| `f.truncate()` | **Abhi ke cursor** par kat jaati hai |
| `f.truncate(0)` | File **poori khaali** |

> 🪤 **Gotcha:** `truncate()` **cursor ko nahi hilata!** `truncate(0)` ke baad cursor purani position par hi rehta hai — agle `write()` se file mein **`\x00` ke khaali jagah** ban jaati hai. Isliye `truncate(0)` ke baad **`seek(0)` zaroor karo**.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

def reset():
    with open("lab/t.txt", "w", encoding="utf-8") as f:
        f.write("abcdefghij")                    # 10 characters

reset()
with open("lab/t.txt", "r+", encoding="utf-8") as f:
    f.truncate(5)
print("truncate(5)       ->", repr(cat("lab/t.txt")), "| size:", os.path.getsize("lab/t.txt"))

reset()
with open("lab/t.txt", "r+", encoding="utf-8") as f:
    f.seek(3)
    f.truncate()                                 # cursor (3) par kato
print("seek(3)+truncate()->", repr(cat("lab/t.txt")))

reset()
with open("lab/t.txt", "r+", encoding="utf-8") as f:
    f.truncate(0)
print("truncate(0)       ->", repr(cat("lab/t.txt")), "| size:", os.path.getsize("lab/t.txt"))

# 🪤 Gotcha: truncate(0) ke baad cursor purani jagah par hota hai
with open("lab/t.txt", "w+", encoding="utf-8") as f:
    f.write("hello world")                       # cursor = 11
    f.truncate(0)                                # file khaali, par cursor abhi bhi 11
    f.write("X")
    f.seek(0)
    data = f.read()
print("\nGotcha: truncate(0) ke baad seek(0) bhoole...")
print("   file mein", len(data), "characters; null bytes (\\x00):", data.count("\x00"), "| aakhri:", repr(data[-1]))

shutil.rmtree("lab")
````

### 📌 Chapter 11 Cheat-Sheet
| Kaam | Code |
|---|---|
| Cursor kahan? | `f.tell()` |
| Rewind (shuru par) | `f.seek(0)` |
| Seedha 6th byte par | `f.seek(6)` |
| Aakhir par jao | `f.seek(0, 2)` |
| File ka size | `f.seek(0, 2)` → `f.tell()` |

| Kaat do | `f.truncate(size)` / `f.truncate()` / `f.truncate(0)` |
| 🪤 | `truncate(0)` ke baad `seek(0)` karo |

# 📝 Chapter 12: Working with Text Files — Text Ka Kamaal

### 🔤 Encoding — Text Bytes Kaise Banta Hai?
Disk par sab **bytes** hota hai. Text ↔ bytes ka **dictionary** hi **encoding** hai. **`UTF-8`** duniya ki lagbhag **har bhasha** (Hindi, emoji 😀 sab) sambhalta hai.

| Encoding | Hindi chal sakti hai? |
|---|:-:|
| `utf-8` | ✅ |
| `ascii` | ❌ (sirf English) |
| `latin-1` | ❌ (galat padhega — "mojibake") |

> 💡 **Rule:** hamesha **`encoding="utf-8"`** likho — likhne **aur** padhne dono mein.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

hindi = "नमस्ते"
with open("lab/hindi.txt", "w", encoding="utf-8") as f:
    f.write(hindi)

print("Characters (len)      :", len(hindi))
print("Bytes on disk         :", os.path.getsize("lab/hindi.txt"), " <- har Hindi akshar = 3 bytes!")
print("utf-8 se padha        :", cat("lab/hindi.txt"), "| sahi?", cat("lab/hindi.txt") == hindi)

# ❌ Galat encoding
with open("lab/hindi.txt", encoding="latin-1") as f:
    wrong = f.read()
print("\nlatin-1 se padha      :", len(wrong), "characters (kachra!) | sahi?", wrong == hindi)

try:
    with open("lab/hindi.txt", encoding="ascii") as f:
        f.read()
except UnicodeDecodeError as e:
    print("ascii se padha        : 💥 UnicodeDecodeError:", e)

shutil.rmtree("lab")
````

### 12.1 Newline ka Jhol — `\n` vs `\r\n` ↩️
Windows files mein line ka ant **`\r\n`** hota hai, Linux / Colab mein **`\n`**. Python text mode mein **dono ko `\n` bana deta hai** (*universal newlines*) — isliye aapko fikar nahi karni. Asli bytes dekhne ho toh `newline=""`.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/win.txt", "wb") as f:                      # Windows jaisi file (bytes)
    f.write(b"line1\r\nline2\r\n")

with open("lab/win.txt", encoding="utf-8") as f:
    print("Text mode (default)  :", repr(f.read()), " <- \\r\\n -> \\n ban gaya")

with open("lab/win.txt", encoding="utf-8", newline="") as f:
    print("newline=''  (raw)    :", repr(f.read()))

shutil.rmtree("lab")
````

### 12.2 Recipe 1️⃣: Word Frequency 📊
File padho → words todo → dictionary mein ginti → sort karo.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/speech.txt", "w", encoding="utf-8") as f:
    f.write("to be or not to be\nthat is the question\nto be\n")

freq = {}
with open("lab/speech.txt", encoding="utf-8") as f:
    for line in f:
        for word in line.split():
            freq[word] = freq.get(word, 0) + 1

# ginti ke hisaab se ulta, barabar ho toh alphabetical
ranked = sorted(freq.items(), key=lambda item: (-item[1], item[0]))

print("Unique words:", len(freq))
for word, count in ranked[:3]:
    print(f"  {word:5} {'█' * count} ({count})")

shutil.rmtree("lab")
````

### 12.3 Recipe 2️⃣: Find & Replace 🔁
**Read → Modify → Write** pattern: poori file padho, `replace()` karo, wapas `"w"` se likho.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

with open("lab/notes.txt", "w", encoding="utf-8") as f:
    f.write("Day 3 was fun.\nDay 3 had recursion.\nSee you on Day 4!\n")

text = cat("lab/notes.txt")
new_text = text.replace("Day 3", "Day Three")
print("Kitne replace hue:", text.count("Day 3"))

with open("lab/notes.txt", "w", encoding="utf-8") as f:       # wapas likho
    f.write(new_text)

print(cat("lab/notes.txt"))
shutil.rmtree("lab")
````

### 12.4 Recipe 3️⃣: Log Analyzer 🔍
Server / program ki **log file** mein har line ek event hoti hai — `INFO`, `WARNING`, `ERROR`. Dictionary se ginti karo.

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/app.log", "w", encoding="utf-8") as f:
    f.write("INFO start\nWARNING disk 90%\nERROR db timeout\nINFO user login\nERROR file missing\nINFO logout\n")

counts, errors = {}, []
with open("lab/app.log", encoding="utf-8") as f:
    for line in f:
        level, message = line.strip().split(" ", 1)       # pehla shabd = level
        counts[level] = counts.get(level, 0) + 1
        if level == "ERROR":
            errors.append(message)

print("Levels :", counts)
print("Errors :", errors)
shutil.rmtree("lab")
````

### 12.5 Recipe 4️⃣: Config File Padhna ⚙️
`key=value` wali lines — **comments (`#`)** aur **khaali lines** chhodni hain. `split("=", 1)` mein `1` zaroori hai (value ke andar bhi `=` ho sakta hai).

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox

with open("lab/fest.cfg", "w", encoding="utf-8") as f:
    f.write("# Fest config\nname=Utkarsh 2026\ndays=3\n\nfee=250\n# end\n")

config = {}
with open("lab/fest.cfg", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line or line.startswith("#"):              # khaali / comment skip
            continue
        key, value = line.split("=", 1)
        config[key] = int(value) if value.isdigit() else value

print("Config:", config)
print("Total fee:", config["days"] * config["fee"])
shutil.rmtree("lab")
````

### 12.6 Recipe 5️⃣: Safe Save — Pehle Temp, Phir Rename 🛡️
`"w"` kholte hi purani file **saaf**. Agar likhte waqt crash hua toh **na purana bacha, na naya!** Safe tareeka: **nayi file temp naam se likho**, poori ban jaye tab **`os.replace()`** se asli file ki jagah rakh do (ek jhatke mein).

````python
import os, shutil

shutil.rmtree("lab", ignore_errors=True)       # 🧹 purana sandbox saaf
os.makedirs("lab")                             # 📁 naya chhota sandbox
def cat(path):
    """File ka poora text wapas deta hai (with ke saath padhta hai)."""
    with open(path, encoding="utf-8") as f:
        return f.read()

path = "lab/important.txt"
with open(path, "w", encoding="utf-8") as f:
    f.write("OLD DATA")

def safe_save(path, text):
    tmp = path + ".tmp"
    with open(tmp, "w", encoding="utf-8") as f:           # 1️⃣ nayi file temp naam se
        f.write(text)
    os.replace(tmp, path)                                 # 2️⃣ ek jhatke mein badlo

safe_save(path, "NEW DATA")
print("Content        :", cat(path))
print("Temp file bachi?", os.path.exists(path + ".tmp"))
shutil.rmtree("lab")
````

### 📌 Chapter 12 Cheat-Sheet
| Kaam | Pattern |
|---|---|
| Encoding | `encoding="utf-8"` **hamesha** |
| Hindi / emoji | 1 akshar = kai bytes → `len(text)` ≠ file size |
| Word count | `len(text.split())` |
| Line count | `len(text.splitlines())` |
| Frequency | `d[w] = d.get(w, 0) + 1` |
| Find & Replace | `read` → `.replace()` → `write("w")` |
| `key=value` | `line.split("=", 1)` |
| Skip | `if not line or line.startswith("#"): continue` |
| Safe save | temp file → `os.replace(tmp, path)` |
