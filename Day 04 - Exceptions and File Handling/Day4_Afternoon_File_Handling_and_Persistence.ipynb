## 📦 Module 1: File Architecture & The Context Manager (`with`)

Opening files consumes operating system file descriptors. Always use the **`with` statement** to guarantee that the file descriptor is closed automatically, even if an unhandled exception crashes the block!

```python
# ❌ RISKY (Leaked descriptor on exception):
f = open('data.txt', 'r')
content = f.read()
f.close()

# ✅ PRODUCTION IDIOM:
with open('data.txt', 'r', encoding='utf-8') as f:
    content = f.read()
```


````python
# 1. Demonstrating Context Manager Safety
import os

test_file = "sample_test.txt"

# Writing using context manager
with open(test_file, "w", encoding="utf-8") as f:
    f.write("Line 1: GLA University Python Masterclass\n")
    f.write("Line 2: Edulateral Foundation Training\n")

# Verify file is closed automatically
print("File closed outside with block?", f.closed)
assert f.closed == True

# Reading content
with open(test_file, "r", encoding="utf-8") as f:
    lines = f.readlines()
    print("Read lines:", lines)
````

## 📦 Module 2: The Complete File Modes Matrix

| Mode | Meaning | Cursor Start | Truncates Existing? | Creates if Missing? |
| :--- | :--- | :--- | :--- | :--- |
| **`'r'`** | Read-only | Beginning | ❌ No | ❌ Error (`FileNotFoundError`) |
| **`'w'`** | Write-only | Beginning | ✅ **YES (Erases file!)** | ✅ Yes |
| **`'a'`** | Append-only | End | ❌ No | ✅ Yes |
| **`'r+'`** | Read + Write | Beginning | ❌ No | ❌ Error (`FileNotFoundError`) |
| **`'w+'`** | Write + Read | Beginning | ✅ **YES (Erases file!)** | ✅ Yes |
| **`'a+'`** | Append + Read | End | ❌ No | ✅ Yes |
| **`'rb'`** | Binary Read | Beginning | ❌ No | ❌ Error |
| **`'wb'`** | Binary Write | Beginning | ✅ **YES** | ✅ Yes |


````python
# Demonstrating r+ vs w+
with open("mode_demo.txt", "w", encoding="utf-8") as f:
    f.write("ABCDEFGH")

# In 'r+' mode: overwrites at cursor without erasing the whole file
with open("mode_demo.txt", "r+", encoding="utf-8") as f:
    f.seek(3)  # Move cursor to index 3
    f.write("999")  # Replaces 'DEF'

with open("mode_demo.txt", "r", encoding="utf-8") as f:
    print("Content after r+ seek(3) overwrite:", f.read())  # 'ABC999GH'
assert open("mode_demo.txt").read() == "ABC999GH"
````

## 📦 Module 3: The 10 Essential File Methods Deep Dive

1. `read([size])`: Reads entire file or up to `size` characters.
2. `readline([size])`: Reads single line including trailing `\n`.
3. `readlines()`: Reads all lines into a `list` of strings.
4. `write(str)`: Writes string to file.
5. `writelines(iterable)`: Writes sequence of strings (does NOT automatically append `\n`!).
6. `seek(offset, whence)`: Repositions byte cursor (`whence`: 0=start, 1=current, 2=end).
7. `tell()`: Returns current byte offset of cursor.
8. `flush()`: Flushes internal C-buffer to physical disk without closing.
9. `truncate([size])`: Resizes file to specified size.
10. `close()`: Flushes and terminates file descriptor.


````python
# Demonstrating all 10 file methods
with open("methods_demo.txt", "w+", encoding="utf-8") as f:
    # 1. write()
    f.write("First Line\n")
    
    # 2. writelines()
    f.writelines(["Second Line\n", "Third Line\n"])
    
    # 3. flush()
    f.flush()
    
    # 4. tell()
    print("Cursor offset after writing:", f.tell())
    
    # 5. seek()
    f.seek(0)
    print("Cursor offset after seek(0):", f.tell())
    
    # 6. readline()
    print("First line:", f.readline().strip())
    
    # 7. readlines()
    print("Remaining lines:", f.readlines())
    
    # 8. truncate()
    f.seek(5)
    f.truncate()  # Truncates file at byte 5

with open("methods_demo.txt", "r", encoding="utf-8") as f:
    # 9. read()
    content = f.read()
    print("Content after truncate(5):", content)
    assert content == "First"
    
# 10. close() is called automatically by with
````

## 📦 Module 4: Structured Data Persistence (CSV & Binary I/O)


````python
import csv

# 1. Writing and Reading CSV Data
csv_filename = "students.csv"
students_data = [
    ["Roll", "Name", "Branch", "CGPA"],
    [101, "Aarav", "CSE", 9.2],
    [102, "Diya", "AI", 9.5],
    [103, "Rohan", "ECE", 8.8]
]

with open(csv_filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(students_data)

# Reading with csv.DictReader
with open(csv_filename, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    print("Parsed CSV Records:")
    for row in reader:
        print(f"Student: {row['Name']} ({row['Branch']}) - CGPA: {row['CGPA']}")

# 2. Binary File I/O
with open("binary_sample.bin", "wb") as f:
    f.write(bytes([0x47, 0x4C, 0x41, 0x0A]))  # "GLA\n" in ASCII hex

with open("binary_sample.bin", "rb") as f:
    data = f.read()
    print("Read raw bytes:", data)
    assert data == b"GLA\n"
````

## 🎯 Hands-On Placement Coding Challenges (10 Automated Tasks)


````python
# Challenge 1: Line Counter Function
def count_file_stats(filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        lines = f.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

with open("c1.txt", "w", encoding="utf-8") as f:
    f.write("Hello GLA University\nPython Masterclass Training\n")

l, w, c = count_file_stats("c1.txt")
assert l == 2
assert w == 6
print("✅ Challenge 1 Passed!")
````

````python
# Challenge 2: Safe Append Logger
import os
def append_log_entry(filepath, log_message):
    with open(filepath, "a", encoding="utf-8") as f:
        f.write(log_message + "\n")

if os.path.exists("app.log"):
    os.remove("app.log")
append_log_entry("app.log", "INFO: Server started")
append_log_entry("app.log", "ERROR: Connection timeout")
with open("app.log", "r", encoding="utf-8") as f:
    logs = f.readlines()
assert len(logs) == 2
assert logs[1].strip() == "ERROR: Connection timeout"
print("✅ Challenge 2 Passed!")
````

````python
# Challenge 3: In-Place Word Replacer
def replace_word_in_file(filepath, target_word, replacement):
    with open(filepath, "r", encoding="utf-8") as f:
        content = f.read()
    new_content = content.replace(target_word, replacement)
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(new_content)

with open("c3.txt", "w", encoding="utf-8") as f:
    f.write("Python is slow, but Python is fun")
replace_word_in_file("c3.txt", "slow", "fast")
assert open("c3.txt").read() == "Python is fast, but Python is fun"
print("✅ Challenge 3 Passed!")
````

````python
# Challenge 4: File Head and Tail Utility
def head_and_tail(filepath, n=2):
    with open(filepath, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f.readlines()]
    return lines[:n], lines[-n:]

with open("c4.txt", "w", encoding="utf-8") as f:
    for i in range(1, 11):
        f.write(f"Line {i}\n")

head, tail = head_and_tail("c4.txt", 2)
assert head == ["Line 1", "Line 2"]
assert tail == ["Line 9", "Line 10"]
print("✅ Challenge 4 Passed!")
````

````python
# Challenge 5: Reverse File Lines
def reverse_file_contents(input_file, output_file):
    with open(input_file, "r", encoding="utf-8") as f:
        lines = f.readlines()
    with open(output_file, "w", encoding="utf-8") as f:
        f.writelines(reversed(lines))

with open("c5_in.txt", "w", encoding="utf-8") as f:
    f.write("Alpha\nBeta\nGamma\n")
reverse_file_contents("c5_in.txt", "c5_out.txt")
assert open("c5_out.txt").read() == "Gamma\nBeta\nAlpha\n"
print("✅ Challenge 5 Passed!")
````

````python
# Challenge 6: Binary File Duplicator
def copy_binary_file(src, dest):
    with open(src, "rb") as f_in, open(dest, "wb") as f_out:
        while True:
            chunk = f_in.read(1024)
            if not chunk:
                break
            f_out.write(chunk)

with open("c6_src.bin", "wb") as f:
    f.write(b"BINARY_PAYLOAD_TEST_123")
copy_binary_file("c6_src.bin", "c6_dest.bin")
assert open("c6_dest.bin", "rb").read() == b"BINARY_PAYLOAD_TEST_123"
print("✅ Challenge 6 Passed!")
````

````python
# Challenge 7: CSV Column Filter
import csv
def filter_csv_column(csv_file, column_name):
    values = []
    with open(csv_file, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            values.append(row[column_name])
    return values

with open("c7.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["id", "city", "score"])
    writer.writerow(["1", "Mathura", "90"])
    writer.writerow(["2", "Agra", "85"])

assert filter_csv_column("c7.csv", "city") == ["Mathura", "Agra"]
print("✅ Challenge 7 Passed!")
````

````python
# Challenge 8: Safe Seek and Truncate
def truncate_at_word(filepath, word):
    with open(filepath, "r+", encoding="utf-8") as f:
        content = f.read()
        idx = content.find(word)
        if idx != -1:
            f.seek(idx)
            f.truncate()

with open("c8.txt", "w", encoding="utf-8") as f:
    f.write("Keep this part. DELETE_AFTER_THIS.")
truncate_at_word("c8.txt", "DELETE")
assert open("c8.txt").read() == "Keep this part. "
print("✅ Challenge 8 Passed!")
````

````python
# Challenge 9: Merge Multiple Files
def merge_files(output_path, *input_paths):
    with open(output_path, "w", encoding="utf-8") as f_out:
        for p in input_paths:
            with open(p, "r", encoding="utf-8") as f_in:
                f_out.write(f_in.read().strip() + "\n")

with open("m1.txt", "w") as f: f.write("Part 1")
with open("m2.txt", "w") as f: f.write("Part 2")
merge_files("m_merged.txt", "m1.txt", "m2.txt")
assert open("m_merged.txt").read() == "Part 1\nPart 2\n"
print("✅ Challenge 9 Passed!")
````

````python
# Challenge 10: Master CSV Record Aggregator
import csv
def calculate_department_averages(csv_file):
    dept_totals = {}
    with open(csv_file, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            dept = row["dept"]
            sal = float(row["salary"])
            if dept not in dept_totals:
                dept_totals[dept] = []
            dept_totals[dept].append(sal)
    return {k: round(sum(v)/len(v), 2) for k, v in dept_totals.items()}

with open("c10.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "dept", "salary"])
    writer.writerow(["Aarav", "CSE", "80000"])
    writer.writerow(["Diya", "ECE", "90000"])
    writer.writerow(["Rohan", "CSE", "100000"])

avgs = calculate_department_averages("c10.csv")
assert avgs["CSE"] == 90000.0
assert avgs["ECE"] == 90000.0
print("✅ Challenge 10 Passed! All 10 Challenges Verified Successfully!")
````
